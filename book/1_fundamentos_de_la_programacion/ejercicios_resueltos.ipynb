## Ejercicio 1.1 — Pintura para un tanque de agua

Una municipalidad debe pintar por fuera un **tanque cilíndrico** de agua potable: la pared lateral y la tapa superior (el fondo está apoyado en el suelo). Las áreas son:

$$A_{lateral} = 2 \pi r h \qquad A_{tapa} = \pi r^2 \qquad \pi = 3.1416$$

Con un litro de pintura se cubren **8 m²** por mano, y se aplican **2 manos**. Cada litro cuesta **₡6500**.

Se necesita un programa que, dados el radio y la altura del tanque en metros, calcule el área a pintar, los litros de pintura y el costo.

Aplique la metodología de pensamiento computacional:

A. **Comprensión:** identifique las entradas, las salidas y las fórmulas necesarias.

B. **Descomposición:** divida el problema en subproblemas.

C. **Especificación del algoritmo:** escriba el algoritmo en pseudocódigo.

D. **Codificación:** escriba el programa en Python.

E. **Validación:** proponga un caso de prueba cuyo resultado pueda calcular a mano.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Radio del tanque (m): <b>2.5</b>
Altura del tanque (m): <b>4</b>
Área a pintar: 82.467 m²
Pintura necesaria: 20.61675 L
Costo: 134008.875 colones
</pre>

<pre>
Radio del tanque (m): <b>1</b>
Altura del tanque (m): <b>1</b>
Área a pintar: 9.4248 m²
Pintura necesaria: 2.3562 L
Costo: 15315.3 colones
</pre>

**Solución**

**A. Comprensión**

* **Entradas:** radio $r$ (m) y altura $h$ (m).
* **Salidas:** área a pintar (m²), litros de pintura y costo (₡).
* **Fórmulas:** $A = 2\pi r h + \pi r^2$; litros $= \frac{A \cdot 2}{8}$; costo $=$ litros $\cdot 6500$.

**B. Descomposición**

1. Obtener las dimensiones del tanque.
2. Calcular el área lateral y el área de la tapa.
3. Calcular la pintura necesaria para dos manos.
4. Calcular el costo.
5. Mostrar los resultados.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER radio
    LEER altura
    area_lateral <- 2 * PI * radio * altura
    area_tapa <- PI * radio * radio
    area_total <- area_lateral + area_tapa
    litros <- area_total * 2 / 8
    costo <- litros * 6500
    MOSTRAR area_total, litros, costo
FIN
```

**E. Validación:** con un tanque de radio 1 m y altura 1 m: $A = 2\pi + \pi = 3\pi = 9.4248$ m² (con $\pi = 3.1416$), litros $= 9.4248 \cdot 2 / 8 = 2.3562$ L y costo $= 2.3562 \cdot 6500 =$ ₡15 315.3. El programa debe dar esos valores (segundo ejemplo de ejecución).

**D. Codificación**

````python
#@title Solución
PI = 3.1416

radio = float(input("Radio del tanque (m): "))
altura = float(input("Altura del tanque (m): "))

area_lateral = 2 * PI * radio * altura
area_tapa = PI * radio * radio
area_total = area_lateral + area_tapa
litros = area_total * 2 / 8
costo = litros * 6500

print("Área a pintar:", area_total, "m²")
print("Pintura necesaria:", litros, "L")
print("Costo:", costo, "colones")
````

1. Leer los datos.
2. Convertir la velocidad a cm/min.
3. Calcular botellas por minuto.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER velocidad_m_min

    velocidad_cm_min <- velocidad_m_min * 100
    botellas_minuto <- velocidad_cm_min / separacion_cm

```

## Ejercicio 1.3 — Presión en el fondo de un tanque

Un ingeniero químico necesita conocer la **presión hidrostática** en el fondo de un tanque de almacenamiento para elegir el sensor de presión adecuado:

$$P = \rho \cdot g \cdot h$$

donde $\rho$ es la densidad del líquido (kg/m³), $g = 9.81$ m/s² y $h$ la altura del líquido (m). El resultado queda en pascales (Pa); 1 kPa = 1000 Pa y 1 atm = 101 325 Pa.

Aplique la metodología de pensamiento computacional:

A. **Comprensión:** identifique las entradas, las salidas y las fórmulas necesarias.

B. **Descomposición:** divida el problema en subproblemas.

C. **Especificación del algoritmo:** escriba el algoritmo en pseudocódigo.

D. **Codificación:** escriba el programa en Python.

E. **Validación:** proponga un caso de prueba cuyo resultado pueda calcular a mano.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Densidad del líquido (kg/m³): <b>1000</b>
Altura del líquido (m): <b>10</b>
Presión en el fondo: 98.1 kPa
Presión en el fondo: 0.9681717246484086 atm
</pre>

<pre>
Densidad del líquido (kg/m³): <b>850</b>
Altura del líquido (m): <b>6.4</b>
Presión en el fondo: 53.3664 kPa
Presión en el fondo: 0.5266854182087343 atm
</pre>

**Solución**

**A. Comprensión**

* **Entradas:** densidad del líquido y altura del líquido.
* **Salidas:** presión en kPa y en atmósferas.
* **Constante:** $g = 9.81$.

**B. Descomposición**

1. Leer la densidad y la altura.
2. Calcular la presión en Pa.
3. Convertir a kPa y a atm.
4. Mostrar los resultados.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER densidad
    LEER altura
    presion_pa <- densidad * 9.81 * altura
    presion_kpa <- presion_pa / 1000
    presion_atm <- presion_pa / 101325
    MOSTRAR presion_kpa, presion_atm
FIN
```

**E. Validación:** agua ($\rho = 1000$) a 10 m de altura: $P = 1000 \cdot 9.81 \cdot 10 = 98\,100$ Pa = **98.1 kPa**, casi 1 atm (un dato conocido: cada 10 m de agua agregan aproximadamente una atmósfera).

**D. Codificación**

````python
#@title Solución
GRAVEDAD = 9.81

densidad = float(input("Densidad del líquido (kg/m³): "))
altura = float(input("Altura del líquido (m): "))

presion_pa = densidad * GRAVEDAD * altura
presion_kpa = presion_pa / 1000
presion_atm = presion_pa / 101325

print("Presión en el fondo:", presion_kpa, "kPa")
print("Presión en el fondo:", presion_atm, "atm")
````

## Ejercicio 1.4 — Calentador eléctrico

Un técnico revisa un **calentador eléctrico** de agua. Conoce el voltaje de alimentación y mide la resistencia del elemento calefactor. Con la **ley de Ohm** y la fórmula de potencia:

$$I = \frac{V}{R} \qquad P = V \cdot I$$

Además, quiere saber cuánto cuesta usar el calentador durante un tiempo dado (en horas) si la electricidad cuesta **₡95 por kWh**. La energía consumida es $E = \frac{P \cdot t}{1000}$ kWh.

Aplique la metodología de pensamiento computacional:

B. **Descomposición:** divida el problema en subproblemas.

E. **Validación:** proponga un caso de prueba cuyo resultado pueda calcular a mano.

</pre>

Voltaje (V): <b>100</b>

</pre>

**Solución**

**A. Comprensión**

* **Entradas:** voltaje (V), resistencia (Ω), horas de uso.
* **Salidas:** corriente (A), potencia (W), energía (kWh), costo (₡).

**B. Descomposición**

1. Leer los datos eléctricos y el tiempo de uso.
2. Calcular la corriente.
3. Calcular la potencia.
4. Calcular la energía y el costo.
5. Mostrar los resultados.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER voltaje, resistencia, horas
    corriente <- voltaje / resistencia
    potencia <- voltaje * corriente
    energia <- potencia * horas / 1000
    costo <- energia * 95
    MOSTRAR corriente, potencia, energia, costo
FIN
```

**E. Validación:** con 100 V, 10 Ω y 10 h: $I = 10$ A, $P = 1000$ W, $E = 10$ kWh y costo = **₡950**.

**D. Codificación**

````python
#@title Solución
TARIFA = 95

voltaje = float(input("Voltaje (V): "))
resistencia = float(input("Resistencia (Ω): "))
horas = float(input("Horas de uso: "))

corriente = voltaje / resistencia
potencia = voltaje * corriente
energia = potencia * horas / 1000
costo = energia * TARIFA

print("Corriente:", corriente, "A")
print("Potencia:", potencia, "W")
print("Energía:", energia, "kWh")
print("Costo:", costo, "colones")
````

Además, quiere estimar los litros necesarios para una ruta de cierta longitud.

Aplique la metodología de pensamiento computacional:

**B. Descomposición**

1. Leer el consumo y la longitud de la ruta.

```text
INICIO

    MOSTRAR rendimiento, litros
FIN
```

## Ejercicio 1.6 — Concreto para una losa

Un ingeniero civil debe pedir **concreto premezclado** para colar una losa rectangular. El volumen de la losa es largo × ancho × espesor, pero siempre se pide un **5 % adicional** por desperdicio. El concreto llega en **camiones mezcladores de 7 m³**.

El espesor de la losa se indica en los planos en **centímetros**, mientras que el largo y el ancho están en metros.

Aplique la metodología de pensamiento computacional:

B. **Descomposición:** divida el problema en subproblemas.

C. **Especificación del algoritmo:** escriba el algoritmo en pseudocódigo.

**Solución**

* **Unidades:** el espesor debe pasarse a metros (÷ 100).

**B. Descomposición**

1. Leer las dimensiones.
2. Convertir el espesor a metros.
3. Calcular el volumen de la losa.
4. Agregar el 5 % de desperdicio.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER largo, ancho, espesor_cm
    espesor_m <- espesor_cm / 100
    volumen <- largo * ancho * espesor_m
    volumen_pedido <- volumen * 1.05
    camiones <- volumen_pedido / 7

```

## Ejercicio 1.10 — Ciclo de desarrollo: caída de tensión en un cable

Una empresa de instalaciones eléctricas le pide un programa para calcular la **caída de tensión** en un cable de cobre. $$\Delta V = \frac{2 \cdot L \cdot I \cdot \rho}{S} \qquad \rho_{cobre} = 0.0172 \ \Omega \cdot \text{mm}^2/\text{m}$$

La caída se expresa también como porcentaje del voltaje de alimentación, que es de **120 V**.

Desarrolle la solución siguiendo **las cuatro fases del ciclo de desarrollo de software**:

A. **Análisis:** ¿qué debe hacer el programa? (requerimientos, entradas y salidas).

* El programa debe calcular la caída de tensión en voltios y en porcentaje.
* **Entradas:** corriente $I$, longitud $L$ y sección $S$.
* **Salidas:** caída de tensión (V) y caída porcentual respecto a 120 V.
* **Constantes:** $\rho = 0.0172$, voltaje de alimentación = 120 V.

**B. Diseño (pseudocódigo)**

```text
INICIO
    LEER corriente, longitud, seccion
    caida <- 2 * longitud * corriente * 0.0172 / seccion
    caida_porcentual <- caida / 120 * 100
```

Si la prueba hubiera fallado, se volvería a la fase de implementación (o incluso de diseño): el ciclo de desarrollo es **iterativo**.

````python
#@title Solución
RESISTIVIDAD_COBRE = 0.0172
VOLTAJE_ALIMENTACION = 120

corriente = float(input("Corriente (A): "))
longitud = float(input("Longitud del cable (m): "))
seccion = float(input("Sección del conductor (mm²): "))

caida = 2 * longitud * corriente * RESISTIVIDAD_COBRE / seccion
caida_porcentual = caida / VOLTAJE_ALIMENTACION * 100

print("Caída de tensión:", caida, "V")
print("Caída porcentual:", caida_porcentual, "%")
````

## Ejercicio 1.11 — Esfuerzo en el cable de un ascensor

Un ingeniero mecánico revisa el **cable de acero** que sostiene la cabina de un ascensor. El esfuerzo de tracción en el cable es:

$$\sigma = \frac{F}{A} \qquad F = m \cdot g \qquad A = \frac{\pi d^2}{4}$$

donde $m$ es la masa total (cabina + carga, en kg), $g = 9.81$ m/s², $d$ el diámetro del cable (mm) y $\pi = 3.1416$. Si la fuerza está en newtons y el área en mm², el esfuerzo queda en **MPa**.

Además se quiere conocer el **factor de seguridad**: el esfuerzo de ruptura del cable (**1770 MPa**) dividido entre el esfuerzo calculado.

Aplique la metodología de pensamiento computacional:

A. **Comprensión:** identifique las entradas, las salidas y las fórmulas necesarias.

B. **Descomposición:** divida el problema en subproblemas.

C. **Especificación del algoritmo:** escriba el algoritmo en pseudocódigo.

D. **Codificación:** escriba el programa en Python.

E. **Validación:** proponga un caso de prueba cuyo resultado pueda calcular a mano.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Masa de la cabina (kg): <b>600</b>
Masa de la carga (kg): <b>400</b>
Diámetro del cable (mm): <b>20</b>
Fuerza en el cable: 9810.0 N
Área del cable: 314.16 mm²
Esfuerzo: 31.226126814362107 MPa
Factor de seguridad: 56.68330275229358
</pre>

**Solución**

**A. Comprensión**

* **Entradas:** masa de la cabina (kg), masa de la carga (kg), diámetro del cable (mm).
* **Salidas:** fuerza (N), área (mm²), esfuerzo (MPa) y factor de seguridad.
* **Constantes:** $g = 9.81$, $\pi = 3.1416$, esfuerzo de ruptura = 1770 MPa.

**B. Descomposición**

1. Leer los datos.
2. Calcular la masa total y la fuerza.
3. Calcular el área del cable.
4. Calcular el esfuerzo.
5. Calcular el factor de seguridad.
6. Mostrar los resultados.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER masa_cabina, masa_carga, diametro
    fuerza <- (masa_cabina + masa_carga) * 9.81
    area <- 3.1416 * diametro * diametro / 4
    esfuerzo <- fuerza / area
    factor_seguridad <- 1770 / esfuerzo
    MOSTRAR fuerza, area, esfuerzo, factor_seguridad
FIN
```

**E. Validación:** es difícil elegir datos "redondos" aquí, así que se verifica por partes: con 1000 kg en total, $F = 9810$ N; un cable de 20 mm tiene $A = 314.16$ mm²; $\sigma = 9810 / 314.16 \approx 31.2$ MPa; $FS \approx 56.7$.

**D. Codificación**

````python
#@title Solución
GRAVEDAD = 9.81
PI = 3.1416
ESFUERZO_RUPTURA = 1770

masa_cabina = float(input("Masa de la cabina (kg): "))
masa_carga = float(input("Masa de la carga (kg): "))
diametro = float(input("Diámetro del cable (mm): "))

fuerza = (masa_cabina + masa_carga) * GRAVEDAD
area = PI * diametro * diametro / 4
esfuerzo = fuerza / area
factor_seguridad = ESFUERZO_RUPTURA / esfuerzo

print("Fuerza en el cable:", fuerza, "N")
print("Área del cable:", area, "mm²")
print("Esfuerzo:", esfuerzo, "MPa")
print("Factor de seguridad:", factor_seguridad)
````

La masa se mide en **gramos** y el volumen se obtiene por desplazamiento de agua en una probeta, en **mililitros** (1 mL = 1 cm³). Recuerde que 1 g/cm³ = 1000 kg/m³.

Aplique la metodología de pensamiento computacional:

En el pseudocódigo, use la instrucción `SI ... ENTONCES ... SINO ... FIN SI` para la decisión.

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

</pre>

* **Entradas:** masa (g) y volumen desplazado (mL).
* **Salidas:** densidad en kg/m³ y el mensaje "flota" o "se hunde".
* **Conversión:** $\rho$ (g/cm³) × 1000 = $\rho$ (kg/m³).

**B. Descomposición**

1. Leer la masa y el volumen.
2. Calcular la densidad en g/cm³ y convertirla a kg/m³.
3. Comparar con la densidad del agua y decidir.
4. Mostrar los resultados.

**C. Algoritmo (pseudocódigo)**

```text
INICIO
    LEER masa_g, volumen_ml
    densidad <- masa_g / volumen_ml * 1000
    MOSTRAR densidad
    SI densidad < 1000 ENTONCES
        MOSTRAR "La muestra flota"
    SINO
        MOSTRAR "La muestra se hunde"
    FIN SI
```
