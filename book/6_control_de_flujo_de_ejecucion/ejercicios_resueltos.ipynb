## Ejercicio 6.1 — Clasificación de sismos

La Red Sismológica Nacional clasifica cada sismo según su **magnitud** para decidir qué protocolo activar:

| Magnitud | Clasificación | Protocolo |
| :-: | :-- | :-- |
| Menor que 3.0 | Micro | Solo registro |
| De 3.0 a menos de 5.0 | Leve | Informe a la población |
| De 5.0 a menos de 6.0 | Moderado | Revisión de infraestructura crítica |
| De 6.0 a menos de 7.0 | Fuerte | Activación de comités de emergencia |
| 7.0 o más | Mayor | Alerta nacional |

Escriba un programa en Python que:

A. Solicite la **magnitud** del sismo y su **profundidad** (km).

B. Muestre la clasificación y el protocolo correspondientes.

C. Indique además si el sismo es **superficial** (profundidad menor que 70 km), lo cual aumenta los daños.

D. Si alguno de los datos no es numérico, o si la magnitud o la profundidad son **negativas**, muestre un mensaje de error y no clasifique el sismo.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Magnitud: <b>6.4</b>
Profundidad (km): <b>22</b>
Clasificación: Fuerte
Protocolo: Activación de comités de emergencia
Sismo superficial: mayor potencial de daño.
</pre>

<pre>
Magnitud: <b>4.1</b>
Profundidad (km): <b>130</b>
Clasificación: Leve
Protocolo: Informe a la población
</pre>

<pre>
Magnitud: <b>-2</b>
Profundidad (km): <b>10</b>
Error: la magnitud y la profundidad no pueden ser negativas.
</pre>

**Pregunta de respuesta corta**

¿Por qué en el segundo `elif` basta con escribir `magnitud < 5.0` y no `3.0 <= magnitud < 5.0`?

````python
#@title Solución
try:
    magnitud = float(input("Magnitud: "))
    profundidad_km = float(input("Profundidad (km): "))

    if magnitud < 0 or profundidad_km < 0:
        print("Error: la magnitud y la profundidad no pueden ser negativas.")
    else:
        if magnitud < 3.0:
            clasificacion = "Micro"
            protocolo = "Solo registro"
        elif magnitud < 5.0:
            clasificacion = "Leve"
            protocolo = "Informe a la población"
        elif magnitud < 6.0:
            clasificacion = "Moderado"
            protocolo = "Revisión de infraestructura crítica"
        elif magnitud < 7.0:
            clasificacion = "Fuerte"
            protocolo = "Activación de comités de emergencia"
        else:
            clasificacion = "Mayor"
            protocolo = "Alerta nacional"

        print(f"Clasificación: {clasificacion}")
        print(f"Protocolo: {protocolo}")
        if profundidad_km < 70:
            print("Sismo superficial: mayor potencial de daño.")
except ValueError:
    print("Error: los datos deben ser numéricos.")
````

**Explicación**

**Respuesta corta:** porque un `elif` solo se evalúa si **todas** las condiciones anteriores fueron falsas. Si el programa llega a `magnitud < 5.0`, ya se sabe que la magnitud **no** es menor que 3.0.

## Ejercicio 6.2 — Índice de calidad del aire

Una estación de monitoreo ambiental mide la concentración de **partículas PM2.5** (µg/m³) y reporta un nivel de alerta:

| PM2.5 (µg/m³) | Nivel |
| :-: | :-- |
| 0 a 12 | Bueno |
| Más de 12 a 35.4 | Moderado |
| Más de 35.4 a 55.4 | Dañino para grupos sensibles |
| Más de 55.4 a 150.4 | Dañino |
| Más de 150.4 | Muy dañino |

Escriba un programa en Python que:

A. Solicite la concentración de PM2.5 y si hay un **evento escolar al aire libre** programado (`s`/`n`).

B. Muestre el nivel de calidad del aire.

C. Muestre la recomendación `Suspender el evento al aire libre.` si hay evento **y** el nivel es "Dañino para grupos sensibles" o peor; de lo contrario, `El evento puede realizarse.` (solo si hay evento).

D. Si la concentración es negativa, muestre `Error: lectura inválida.`

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
PM2.5 (µg/m³): <b>41.7</b>
¿Hay evento al aire libre? (s/n): <b>s</b>
Calidad del aire: Dañino para grupos sensibles
Suspender el evento al aire libre.
</pre>

<pre>
PM2.5 (µg/m³): <b>18</b>
¿Hay evento al aire libre? (s/n): <b>S</b>
Calidad del aire: Moderado
El evento puede realizarse.
</pre>

<pre>
PM2.5 (µg/m³): <b>8.2</b>
¿Hay evento al aire libre? (s/n): <b>n</b>
Calidad del aire: Bueno
</pre>

````python
#@title Solución
concentracion = float(input("PM2.5 (µg/m³): "))
hay_evento = input("¿Hay evento al aire libre? (s/n): ").lower() == "s"

if concentracion < 0:
    print("Error: lectura inválida.")
else:
    if concentracion <= 12:
        nivel = "Bueno"
    elif concentracion <= 35.4:
        nivel = "Moderado"
    elif concentracion <= 55.4:
        nivel = "Dañino para grupos sensibles"
    elif concentracion <= 150.4:
        nivel = "Dañino"
    else:
        nivel = "Muy dañino"

    print(f"Calidad del aire: {nivel}")

    if hay_evento and concentracion > 35.4:
        print("Suspender el evento al aire libre.")
    elif hay_evento:
        print("El evento puede realizarse.")
````

**Explicación**

La condición de suspensión se expresa con el **valor** de la concentración (`> 35.4`) en lugar de comparar el texto del nivel con varias opciones, lo que la hace más corta y menos propensa a errores de escritura.

## Ejercicio 6.3 — Selección de material con `match`

Un taller de manufactura fabrica barras cilíndricas de distintos materiales. La masa de una barra es $m = \rho \cdot V$, con $V = \frac{\pi d^2}{4} L$ (use $\pi = 3.1416$). Las densidades son:

| Código | Material | Densidad (kg/m³) |
| :-: | :-- | :-: |
| `A` | Acero | 7850 |
| `L` | Aluminio | 2700 |
| `T` | Titanio | 4500 |
| `C` | Cobre | 8960 |

Escriba un programa en Python que:

A. Solicite el **código** del material (acepte mayúsculas o minúsculas), el **diámetro** (mm) y la **longitud** (m) de la barra.

B. Use una estructura **`match`** para obtener el nombre y la densidad del material.

C. Calcule y muestre el volumen (cm³, 1 decimal) y la masa (kg, 3 decimales).

D. Si el código no existe, muestre `Material desconocido: <código>` y no haga cálculos.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Código del material (A/L/T/C): <b>t</b>
Diámetro (mm): <b>25</b>
Longitud (m): <b>1.2</b>
Material: Titanio
Volumen: 589.1 cm³
Masa: 2.651 kg
</pre>

<pre>
Código del material (A/L/T/C): <b>X</b>
Diámetro (mm): <b>25</b>
Longitud (m): <b>1.2</b>
Material desconocido: X
</pre>

````python
#@title Solución
PI = 3.1416

codigo = input("Código del material (A/L/T/C): ").upper()
diametro_mm = float(input("Diámetro (mm): "))
longitud_m = float(input("Longitud (m): "))

material_valido = True
match codigo:
    case "A":
        material = "Acero"
        densidad = 7850
    case "L":
        material = "Aluminio"
        densidad = 2700
    case "T":
        material = "Titanio"
        densidad = 4500
    case "C":
        material = "Cobre"
        densidad = 8960
    case _:
        material_valido = False

if material_valido:
    diametro_m = diametro_mm / 1000
    volumen_m3 = PI * diametro_m ** 2 / 4 * longitud_m
    masa_kg = densidad * volumen_m3
    print(f"Material: {material}")
    print(f"Volumen: {volumen_m3 * 1e6:.1f} cm³")
    print(f"Masa: {masa_kg:.3f} kg")
else:
    print(f"Material desconocido: {codigo}")
````

**Explicación**

El caso `case _:` funciona como el `else` de un `if`: atrapa cualquier valor que no coincidió con los casos anteriores. La bandera `material_valido` evita hacer los cálculos con una densidad que nunca se asignó (lo que produciría un `NameError`).

## Ejercicio 6.4 — Tarifa eléctrica por bloques

La compañía de electricidad cobra el consumo **residencial** por bloques: los primeros **200 kWh** a ₡80/kWh, los siguientes **100 kWh** (de 201 a 300) a ₡122/kWh y **cada kWh por encima de 300** a ₡156/kWh. Además, se cobra un cargo fijo de ₡1500 por **alumbrado público** y el **IVA** (13 %) solo si el consumo **supera** 280 kWh.

Escriba un programa en Python que:

A. Solicite el consumo del mes (kWh, entero). Si es negativo, muestre un error.

B. Calcule cuántos kWh caen en **cada bloque** y el costo de cada uno.

C. Calcule el subtotal (bloques + alumbrado), el IVA (0 si no aplica) y el total.

D. Imprima el desglose con los montos alineados a la derecha en 12 caracteres y 2 decimales.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Consumo del mes (kWh): <b>347</b>

Bloque 1 (200 kWh)           16000.00
Bloque 2 (100 kWh)           12200.00
Bloque 3 (47 kWh)             7332.00
Alumbrado público             1500.00
IVA                           4814.16
TOTAL                       ₡41846.16
</pre>

<pre>
Consumo del mes (kWh): <b>180</b>

Bloque 1 (180 kWh)           14400.00
Bloque 2 (0 kWh)                 0.00
Bloque 3 (0 kWh)                 0.00
Alumbrado público             1500.00
IVA                              0.00
TOTAL                       ₡15900.00
</pre>

**Pregunta de respuesta corta**

Para un consumo de exactamente **300 kWh**, ¿cuánto se paga en cada bloque? ¿Se cobra IVA?

````python
#@title Solución
PRECIO_BLOQUE_1 = 80
PRECIO_BLOQUE_2 = 122
PRECIO_BLOQUE_3 = 156
CARGO_ALUMBRADO = 1500
LIMITE_IVA = 280
PORCENTAJE_IVA = 0.13

consumo = int(input("Consumo del mes (kWh): "))

if consumo < 0:
    print("Error: el consumo no puede ser negativo.")
else:
    if consumo <= 200:
        kwh_bloque_1 = consumo
        kwh_bloque_2 = 0
        kwh_bloque_3 = 0
    elif consumo <= 300:
        kwh_bloque_1 = 200
        kwh_bloque_2 = consumo - 200
        kwh_bloque_3 = 0
    else:
        kwh_bloque_1 = 200
        kwh_bloque_2 = 100
        kwh_bloque_3 = consumo - 300

    costo_bloque_1 = kwh_bloque_1 * PRECIO_BLOQUE_1
    costo_bloque_2 = kwh_bloque_2 * PRECIO_BLOQUE_2
    costo_bloque_3 = kwh_bloque_3 * PRECIO_BLOQUE_3
    subtotal = costo_bloque_1 + costo_bloque_2 + costo_bloque_3 + CARGO_ALUMBRADO

    if consumo > LIMITE_IVA:
        iva = subtotal * PORCENTAJE_IVA
    else:
        iva = 0

    total = subtotal + iva

    print()
    print(f"Bloque 1 ({kwh_bloque_1} kWh)".ljust(25) + f"{costo_bloque_1:.2f}".rjust(12))
    print(f"Bloque 2 ({kwh_bloque_2} kWh)".ljust(25) + f"{costo_bloque_2:.2f}".rjust(12))
    print(f"Bloque 3 ({kwh_bloque_3} kWh)".ljust(25) + f"{costo_bloque_3:.2f}".rjust(12))
    print("Alumbrado público".ljust(25) + f"{CARGO_ALUMBRADO:.2f}".rjust(12))
    print("IVA".ljust(25) + f"{iva:.2f}".rjust(12))
    print("TOTAL".ljust(25) + f"₡{total:.2f}".rjust(12))
````

**Explicación**

**Respuesta corta:** 200 kWh × ₡80 = ₡16 000 en el bloque 1, 100 kWh × ₡122 = ₡12 200 en el bloque 2 y ₡0 en el bloque 3. Sí se cobra IVA, porque 300 > 280.

## Ejercicio 6.5 — Lectura validada de un manómetro

Un operador digita a mano la lectura del **manómetro** de una caldera. El instrumento mide entre **0 y 250 psi**; cualquier valor fuera de ese rango, o que no sea un número, se debe rechazar y **volver a solicitar**.

Una vez obtenida una lectura válida, la caldera se clasifica así:

* Menor que 80 psi: `Presión baja: revisar el quemador.`
* De 80 a 180 psi: `Operación normal.`
* Más de 180 psi: `¡Sobrepresión! Abrir la válvula de alivio.`

Escriba un programa en Python que solicite la lectura con un ciclo **`while`** (sin `break` ni `while True`), cuente cuántos **intentos inválidos** hubo, y al final muestre la lectura aceptada, su clasificación y la cantidad de intentos inválidos.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Lectura del manómetro (psi): <b>19O</b>
Error: ingrese un número.
Lectura del manómetro (psi): <b>-5</b>
Fuera de rango: debe estar entre 0 y 250 psi.
Lectura del manómetro (psi): <b>310</b>
Fuera de rango: debe estar entre 0 y 250 psi.
Lectura del manómetro (psi): <b>196.5</b>

Lectura aceptada: 196.5 psi
¡Sobrepresión! Abrir la válvula de alivio.
Intentos inválidos: 3
</pre>

````python
#@title Solución
PRESION_MINIMA = 0
PRESION_MAXIMA = 250

lectura_valida = False
intentos_invalidos = 0

while not lectura_valida:
    try:
        presion = float(input("Lectura del manómetro (psi): "))
        if PRESION_MINIMA <= presion <= PRESION_MAXIMA:
            lectura_valida = True
        else:
            intentos_invalidos += 1
            print(f"Fuera de rango: debe estar entre {PRESION_MINIMA} y {PRESION_MAXIMA} psi.")
    except ValueError:
        intentos_invalidos += 1
        print("Error: ingrese un número.")

if presion < 80:
    mensaje = "Presión baja: revisar el quemador."
elif presion <= 180:
    mensaje = "Operación normal."
else:
    mensaje = "¡Sobrepresión! Abrir la válvula de alivio."

print(f"\nLectura aceptada: {presion} psi")
print(mensaje)
print(f"Intentos inválidos: {intentos_invalidos}")
````

**Explicación**

La bandera `lectura_valida` empieza en `False` para que el ciclo se ejecute al menos una vez, y solo cambia a `True` cuando el dato pasa **todas** las validaciones. Observe que `19O` (con la letra O en lugar de cero) es un error de digitación real y común.

## Ejercicio 6.6 — Tabla de dilatación térmica de un riel

Los rieles de un tren se instalan dejando una **junta de dilatación**, porque el acero se alarga con el calor:

$$\Delta L = \alpha \cdot L_0 \cdot \Delta T$$

con $\alpha = 1.2 \times 10^{-5}$ 1/°C para el acero, $L_0$ la longitud del riel (m) y $\Delta T$ el cambio de temperatura respecto a la temperatura de instalación.

Escriba un programa en Python que:

A. Solicite la **longitud** del riel (m), la **temperatura de instalación** (°C, entero) y el **ancho de la junta** (mm).

B. Con un ciclo **`for`** y `range()`, recorra las temperaturas desde 0 °C hasta 60 °C **de 10 en 10**, e imprima una tabla con la temperatura, el cambio de longitud en **mm** (2 decimales) y el estado: `OK` si el alargamiento **no supera** el ancho de la junta, o `PANDEO` si lo supera. Los valores negativos indican que el riel se contrae.

C. Al final, muestre cuántas temperaturas de la tabla provocarían pandeo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Longitud del riel (m): <b>25</b>
Temperatura de instalación (°C): <b>22</b>
Ancho de la junta (mm): <b>8</b>

  T (°C)     ΔL (mm)    Estado
       0       -6.60        OK
      10       -3.60        OK
      20       -0.60        OK
      30        2.40        OK
      40        5.40        OK
      50        8.40    PANDEO
      60       11.40    PANDEO

Temperaturas con riesgo de pandeo: 2
</pre>

````python
#@title Solución
COEFICIENTE_ACERO = 1.2e-5

longitud_m = float(input("Longitud del riel (m): "))
temperatura_instalacion = int(input("Temperatura de instalación (°C): "))
ancho_junta_mm = float(input("Ancho de la junta (mm): "))

temperaturas_con_pandeo = 0

print("\n" + "T (°C)".rjust(8) + "ΔL (mm)".rjust(12) + "Estado".rjust(10))
for temperatura in range(0, 61, 10):
    delta_t = temperatura - temperatura_instalacion
    delta_l_mm = COEFICIENTE_ACERO * longitud_m * delta_t * 1000

    if delta_l_mm > ancho_junta_mm:
        estado = "PANDEO"
        temperaturas_con_pandeo += 1
    else:
        estado = "OK"

    print(f"{temperatura:8d}{delta_l_mm:12.2f}{estado:>10}")

print(f"\nTemperaturas con riesgo de pandeo: {temperaturas_con_pandeo}")
````

**Explicación**

`range(0, 61, 10)` genera 0, 10, 20, ..., 60: el límite superior (61) **no** se incluye, por eso se usa 61 y no 60. El formato `{valor:8d}` reserva 8 espacios para un entero y `{texto:>10}` alinea un texto a la derecha en 10 caracteres, igual que `.rjust(10)`.

## Ejercicio 6.7 — Descarga de un capacitor

En un circuito RC, un capacitor cargado a un voltaje $V_0$ se descarga a través de una resistencia. Una forma sencilla de **simular** la descarga es avanzar en pequeños pasos de tiempo $\Delta t$ y actualizar el voltaje con:

$$V_{nuevo} = V - V \cdot \frac{\Delta t}{R \cdot C}$$

Escriba un programa en Python que:

A. Solicite $V_0$ (V), $R$ (Ω), $C$ (en **microfaradios**, 1 µF = 10⁻⁶ F) y el paso de tiempo $\Delta t$ (en **milisegundos**).

B. Con un ciclo **`while`**, simule la descarga **mientras** el voltaje sea mayor o igual al **5 %** de $V_0$, contando los pasos.

C. Muestre la constante de tiempo $\tau = R \cdot C$ (ms), la cantidad de pasos, el tiempo total simulado (ms, 1 decimal) y cuántas constantes de tiempo transcurrieron (2 decimales).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Resistencia (Ω): <b>4700</b>
Capacitancia (µF): <b>100</b>
Paso de tiempo (ms): <b>1</b>

Constante de tiempo: 470.0 ms
Pasos simulados: 1407
Tiempo hasta el 5%: 1407.0 ms
Constantes de tiempo transcurridas: 2.99
</pre>

**Pregunta de respuesta corta**

La teoría indica que un capacitor se descarga al 5 % en aproximadamente **3 constantes de tiempo**. ¿Qué cree que pasaría con la precisión de la simulación y con la cantidad de iteraciones si se usa un paso de tiempo **mucho más grande**, por ejemplo 100 ms?

````python
#@title Solución
voltaje_inicial = float(input("Voltaje inicial (V): "))
resistencia = float(input("Resistencia (Ω): "))
capacitancia_uf = float(input("Capacitancia (µF): "))
paso_ms = float(input("Paso de tiempo (ms): "))

capacitancia = capacitancia_uf * 1e-6
paso_s = paso_ms / 1000
tau_s = resistencia * capacitancia
voltaje_final = 0.05 * voltaje_inicial

voltaje = voltaje_inicial
pasos = 0
while voltaje >= voltaje_final:
    voltaje = voltaje - voltaje * paso_s / tau_s
    pasos += 1

tiempo_total_ms = pasos * paso_ms

print(f"\nConstante de tiempo: {tau_s * 1000:.1f} ms")
print(f"Pasos simulados: {pasos}")
print(f"Tiempo hasta el 5%: {tiempo_total_ms:.1f} ms")
print(f"Constantes de tiempo transcurridas: {tiempo_total_ms / (tau_s * 1000):.2f}")
````

**Explicación**

        Con un paso de 1 ms, la simulación da 2.99 constantes de tiempo, muy cerca del valor teórico ($\ln 20 \approx 3.00$).

**Respuesta corta:** la simulación es una **aproximación** que supone que el voltaje es constante durante cada paso. Con un paso grande ese supuesto deja de ser razonable, el error crece y el resultado se aleja del valor teórico, aunque el ciclo haga **muchas menos iteraciones**. Elegir el paso es un compromiso entre precisión y tiempo de cómputo.

## Ejercicio 6.9 — Contenido GC de una secuencia de ADN

En bioingeniería, el **contenido GC** de una secuencia de ADN (el porcentaje de bases que son guanina `G` o citosina `C`) es importante para diseñar cebadores de PCR: se recomienda que esté **entre 40 % y 60 %**.

Escriba un programa en Python que:

A. Solicite una secuencia de ADN y conviértala a mayúsculas.

B. Recorra la secuencia con un ciclo **`for`** sobre sus caracteres y cuente las bases `A`, `T`, `G` y `C`. Si encuentra un carácter que **no** es una base válida (por ejemplo, un espacio o un error de digitación), cuéntelo como inválido y use **`continue`** para pasar al siguiente.

C. Muestre la cantidad de cada base, la cantidad de caracteres inválidos, el contenido GC (sobre las bases **válidas**, 1 decimal) y si la secuencia es **apta** como cebador.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Secuencia de ADN: <b>atgcgt acggtaxcatg</b>

A: 4  T: 4  G: 5  C: 3
Caracteres inválidos: 2
Contenido GC: 50.0%
Secuencia apta como cebador.
</pre>

````python
#@title Solución
secuencia = input("Secuencia de ADN: ").upper()

cantidad_a = 0
cantidad_t = 0
cantidad_g = 0
cantidad_c = 0
invalidos = 0

for base in secuencia:
    if base not in "ATGC":
        invalidos += 1
        continue

    if base == "A":
        cantidad_a += 1
    elif base == "T":
        cantidad_t += 1
    elif base == "G":
        cantidad_g += 1
    else:
        cantidad_c += 1

bases_validas = cantidad_a + cantidad_t + cantidad_g + cantidad_c

print(f"\nA: {cantidad_a}  T: {cantidad_t}  G: {cantidad_g}  C: {cantidad_c}")
print(f"Caracteres inválidos: {invalidos}")

if bases_validas > 0:
    contenido_gc = (cantidad_g + cantidad_c) / bases_validas * 100
    print(f"Contenido GC: {contenido_gc:.1f}%")
    if 40 <= contenido_gc <= 60:
        print("Secuencia apta como cebador.")
    else:
        print("Secuencia NO apta como cebador.")
else:
    print("La secuencia no contiene bases válidas.")
````

**Explicación**

`continue` salta **el resto del cuerpo del ciclo** para ese carácter y pasa directamente al siguiente; así, los caracteres inválidos no llegan a la cadena de `if`/`elif`. El `if bases_validas > 0` evita una división entre cero si la persona usuaria no ingresó ninguna base.

El departamento financiero de una empresa constructora compara dos métodos para **depreciar** una excavadora a lo largo de su vida útil:

* **Línea recta:** cada año se deprecia lo mismo: $\frac{\text{costo} - \text{valor residual}}{\text{vida útil}}$.
* **Saldo decreciente doble:** cada año se deprecia $\frac{2}{\text{vida útil}}$ del valor en libros **al inicio de ese año**, pero sin bajar nunca del valor residual.

Escriba un programa en Python que:

A. Solicite el **costo** de la máquina, su **valor residual** y su **vida útil** en años (entero).

B. Con un ciclo **`for`**, imprima una tabla año por año con el **valor en libros al final del año** según cada método (separador de miles, 0 decimales).

C. En el método de saldo decreciente, si la depreciación del año haría que el valor quede por debajo del residual, deprecie solo hasta llegar exactamente al residual.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Valor residual ($): <b>20000</b>

 Año     Línea recta   Saldo decreciente
   1         157,500             123,333
   2         130,000              82,222
   3         102,500              54,815
   4          75,000              36,543
   5          47,500              24,362
   6          20,000              20,000

**Pregunta de respuesta corta**

¿Cuántas veces se ejecuta el cuerpo del ciclo si la vida útil es 6? ¿Por qué se usa `range(1, vida_util + 1)` y no `range(vida_util)`?

````python
#@title Solución
costo = float(input("Costo de la máquina ($): "))
valor_residual = float(input("Valor residual ($): "))
vida_util = int(input("Vida útil (años): "))

depreciacion_lineal = (costo - valor_residual) / vida_util
tasa_decreciente = 2 / vida_util

valor_lineal = costo
valor_decreciente = costo

print("\n" + "Año".rjust(4) + "Línea recta".rjust(16) + "Saldo decreciente".rjust(20))
for anio in range(1, vida_util + 1):
    valor_lineal -= depreciacion_lineal

    depreciacion_anio = valor_decreciente * tasa_decreciente
    if valor_decreciente - depreciacion_anio < valor_residual:
        depreciacion_anio = valor_decreciente - valor_residual
    valor_decreciente -= depreciacion_anio

    print(f"{anio:4d}{valor_lineal:16,.0f}{valor_decreciente:20,.0f}")
````

Escriba un programa en Python que:

C. Cada **30 días**, imprima el día y el volumen (2 decimales).

Día 330: 52.61 hm³

````python
#@title Solución
SEGUNDOS_POR_DIA = 24 * 3600
M3_POR_HM3 = 1e6
TASA_EVAPORACION = 0.002
DIAS_MAXIMOS = 365

volumen = float(input("Volumen inicial (hm³): "))
capacidad = float(input("Capacidad (hm³): "))
caudal_rio = float(input("Caudal del río (m³/s): "))
turbinado_diario = float(input("Volumen turbinado por día (hm³): "))

entrada_diaria = caudal_rio * SEGUNDOS_POR_DIA / M3_POR_HM3

dia = 0
while volumen < capacidad and dia < DIAS_MAXIMOS:
    dia += 1
    evaporacion = volumen * TASA_EVAPORACION
    volumen = volumen + entrada_diaria - evaporacion - turbinado_diario
    if dia % 30 == 0:
        print(f"Día {dia}: {volumen:.2f} hm³")

if volumen >= capacidad:
    print(f"\nEl embalse se llenó en {dia} días.")
else:
    print("\nEl embalse no se llena en un año.")
````

## Ejercicio 6.12 — Código de colores de una resistencia

Las resistencias de un circuito indican su valor con **bandas de colores**. Con 3 bandas, las dos primeras son dígitos y la tercera es el **multiplicador** (una potencia de 10):

| Color | Valor | Color | Valor |
| :-- | :-: | :-- | :-: |
| negro | 0 | verde | 5 |
| cafe | 1 | azul | 6 |
| rojo | 2 | violeta | 7 |
| naranja | 3 | gris | 8 |
| amarillo | 4 | blanco | 9 |

Por ejemplo, `amarillo violeta rojo` es $47 \times 10^2 = 4700 \ \Omega$.

Escriba un programa en Python que:

A. Solicite los **tres colores** en una sola línea, separados por espacios, y sepárelos con `split()`.

B. Recorra los colores con un ciclo **`for`** y use **`match`** para traducir cada uno a su valor numérico. Si algún color no existe, márquelo como inválido.

C. Si todos los colores son válidos, muestre el valor en ohmios y, si es de 1000 Ω o más, también en **kΩ** (2 decimales).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Colores de las bandas: <b>Amarillo violeta rojo</b>
Resistencia: 4700 Ω
Resistencia: 4.70 kΩ
</pre>

<pre>
Colores de las bandas: <b>cafe negro naranja</b>
Resistencia: 10000 Ω
Resistencia: 10.00 kΩ
</pre>

<pre>
Colores de las bandas: <b>rojo morado cafe</b>
Color inválido: morado
No se puede calcular la resistencia.
</pre>

````python
#@title Solución
colores = input("Colores de las bandas: ").lower().split()

valores = []
colores_validos = len(colores) == 3

for color in colores:
    match color:
        case "negro":
            valor = 0
        case "cafe":
            valor = 1
        case "rojo":
            valor = 2
        case "naranja":
            valor = 3
        case "amarillo":
            valor = 4
        case "verde":
            valor = 5
        case "azul":
            valor = 6
        case "violeta":
            valor = 7
        case "gris":
            valor = 8
        case "blanco":
            valor = 9
        case _:
            valor = -1
            colores_validos = False
            print(f"Color inválido: {color}")
    valores.append(valor)

if colores_validos:
    resistencia = (valores[0] * 10 + valores[1]) * 10 ** valores[2]
    print(f"Resistencia: {resistencia} Ω")
    if resistencia >= 1000:
        print(f"Resistencia: {resistencia / 1000:.2f} kΩ")
else:
    print("No se puede calcular la resistencia.")
````

**Explicación**

La bandera `colores_validos` empieza verificando que haya **exactamente tres** colores, y se vuelve `False` si alguno no se reconoce. El valor `-1` se agrega igualmente a la lista para mantener la posición de cada banda, pero nunca se usa en el cálculo.

B. Detenga la toma de lecturas en cuanto una supere 95 °C.

C. Al final, muestre la **temperatura máxima** registrada, el **promedio** de las lecturas tomadas y, según el caso, `DESCONECTAR EL TRANSFORMADOR (lectura N)` o `Transformador en operación normal.`

<pre>
Lectura 1 (minuto 0) °C: <b>78.2</b>

Lectura 1 (minuto 0) °C: <b>70</b>

Lectura 4 (minuto 30) °C: <b>72.8</b>

````python
#@title Solución
ESFUERZO_ADMISIBLE = 10
NMM_POR_KNM = 1e6

momento_diseno_knm = float(input("Momento flector de diseño (kN·m): "))

encontro_seccion = False
area_minima = 0
mejor_ancho = 0
mejor_altura = 0

print("\nSecciones que resisten:")
for ancho in range(50, 151, 25):
    for altura in range(100, 301, 50):
        modulo_seccion = ancho * altura ** 2 / 6
        momento_resistente_knm = ESFUERZO_ADMISIBLE * modulo_seccion / NMM_POR_KNM

        if momento_resistente_knm >= momento_diseno_knm:
            area = ancho * altura
            print(f"  {ancho} x {altura} mm: área {area} mm², M = {momento_resistente_knm:.1f} kN·m")
            if not encontro_seccion or area < area_minima:
                encontro_seccion = True
                area_minima = area
                mejor_ancho = ancho
                mejor_altura = altura

if encontro_seccion:
    print(f"\nSección más liviana: {mejor_ancho} x {mejor_altura} mm ({area_minima} mm²)")
else:
    print("\nNinguna sección comercial resiste el momento.")
````

**Explicación**

La condición `not encontro_seccion or area < area_minima` permite guardar la **primera** sección válida sin compararla (todavía no hay mínimo), y a partir de ahí solo reemplazarla si aparece una más liviana.

**Respuesta corta:** el ciclo externo tiene 5 anchos (50, 75, 100, 125, 150) y el interno 5 alturas (100 a 300), así que la línea se ejecuta **5 × 5 = 25 veces**.

## Ejercicio 6.15 — Velocidad a partir de un *encoder*

Un **encoder** óptico montado en el eje de un motor produce una señal digital: un `1` cuando la luz pasa por una ranura del disco y un `0` cuando no. El sistema de adquisición entrega las muestras como una cadena, por ejemplo `0011100111001110`.

Cada **flanco de subida** (un `0` seguido de un `1`) corresponde a una ranura. Si el disco tiene $N$ ranuras por vuelta y la ventana de muestreo dura $t$ segundos:

$$\text{rpm} = \frac{\text{flancos}}{N} \cdot \frac{60}{t}$$

Escriba un programa en Python que:

A. Solicite la cadena de muestras, la cantidad de **ranuras por vuelta** y la **duración** de la ventana (ms).

B. Verifique que la cadena solo contenga `0` y `1`; si no, muestre un error.

C. Recorra la cadena con `for i in range(1, len(muestras))` y cuente los **flancos de subida** comparando cada muestra con la anterior.

D. Muestre la cantidad de flancos y la velocidad en rpm (1 decimal).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

</pre>

<pre>

Ranuras por vuelta: <b>20</b>
Duración de la ventana (ms): <b>5</b>
Error: la señal solo puede contener 0 y 1.
</pre>

````python
#@title Solución
muestras = input("Muestras del encoder: ")
ranuras_por_vuelta = int(input("Ranuras por vuelta: "))
ventana_ms = float(input("Duración de la ventana (ms): "))

senal_valida = True
for muestra in muestras:
    if muestra != "0" and muestra != "1":
        senal_valida = False

if not senal_valida:
    print("Error: la señal solo puede contener 0 y 1.")
else:
    flancos = 0
    for i in range(1, len(muestras)):
        if muestras[i - 1] == "0" and muestras[i] == "1":
            flancos += 1

    ventana_s = ventana_ms / 1000
    rpm = flancos / ranuras_por_vuelta * 60 / ventana_s

    print(f"Flancos de subida: {flancos}")
    print(f"Velocidad: {rpm:.1f} rpm")
````

El ciclo empieza en 1 (y no en 0) porque cada iteración compara la muestra `i` con la **anterior** (`i - 1`); si empezara en 0, `muestras[-1]` tomaría el **último** carácter, lo cual sería un error lógico.

````python
#@title Solución
mes_inicial = int(input("Mes inicial (1-12): "))
mes_final = int(input("Mes final (1-12): "))

if not (1 <= mes_inicial <= mes_final <= 12):
    print("Error: los meses deben estar entre 1 y 12, y el inicial no puede ser mayor que el final.")
else:
    total_tareas = 0
    for mes in range(mes_inicial, mes_final + 1):
        match mes:
            case 1: nombre = "Enero"
            case 2: nombre = "Febrero"
            case 3: nombre = "Marzo"
            case 4: nombre = "Abril"
            case 5: nombre = "Mayo"
            case 6: nombre = "Junio"
            case 7: nombre = "Julio"
            case 8: nombre = "Agosto"
            case 9: nombre = "Septiembre"
            case 10: nombre = "Octubre"
            case 11: nombre = "Noviembre"
            case _: nombre = "Diciembre"

        print(f"\n{nombre}:")
        print("  - Limpieza de rejillas")
        total_tareas += 1
        if mes % 3 == 0:
            print("  - Calibración de sensores")
            total_tareas += 1
        if mes % 6 == 0:
            print("  - Cambio de aceite de sopladores")
            total_tareas += 1
        if mes == 12:
            print("  - Revisión general del sistema")
            total_tareas += 1
        if 5 <= mes <= 11:
            print("  - Inspección de canales pluviales")
            total_tareas += 1

    print(f"\nTotal de tareas programadas: {total_tareas}")
````

## Ejercicio 6.17 — Diámetro de una tubería por el método de Newton

Para transportar un caudal $Q$ (m³/s) a una velocidad $v$ (m/s), una tubería necesita un área $A = Q / v$ y, por tanto, un diámetro $d = \sqrt{4A/\pi}$.

Suponga que **no** se puede usar `** 0.5` ni ninguna función de raíz cuadrada. El **método de Newton** calcula $\sqrt{x}$ partiendo de una estimación inicial $r = x$ y mejorándola repetidamente con:

$$r_{nuevo} = \frac{r + x / r}{2}$$

hasta que la diferencia entre dos estimaciones consecutivas sea menor que una **tolerancia**.

Escriba un programa en Python que:

A. Solicite el caudal y la velocidad de diseño (use $\pi = 3.1416$).

B. Calcule $x = 4A/\pi$ y aplique el método de Newton con un ciclo **`while`** y una tolerancia de $10^{-10}$, imprimiendo cada estimación.

C. Muestre el diámetro en **milímetros** (2 decimales) y la cantidad de iteraciones.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Caudal (m³/s): <b>0.085</b>
Velocidad de diseño (m/s): <b>1.5</b>
Iteración 1: d = 0.5360750361 m
Iteración 2: d = 0.3353322690 m
Iteración 3: d = 0.2752461064 m
Iteración 4: d = 0.2686877086 m
Iteración 5: d = 0.2686076666 m
Iteración 6: d = 0.2686076547 m
Iteración 7: d = 0.2686076547 m

Diámetro requerido: 268.61 mm
Iteraciones: 7
</pre>

````python
#@title Solución
PI = 3.1416
TOLERANCIA = 1e-10

caudal = float(input("Caudal (m³/s): "))
velocidad = float(input("Velocidad de diseño (m/s): "))

area = caudal / velocidad
x = 4 * area / PI

estimacion = x
diferencia = TOLERANCIA + 1
iteraciones = 0

while diferencia >= TOLERANCIA:
    nueva_estimacion = (estimacion + x / estimacion) / 2
    diferencia = abs(nueva_estimacion - estimacion)
    estimacion = nueva_estimacion
    iteraciones += 1
    print(f"Iteración {iteraciones}: d = {estimacion:.10f} m")

print(f"\nDiámetro requerido: {estimacion * 1000:.2f} mm")
print(f"Iteraciones: {iteraciones}")
````

**Explicación**

`diferencia` se inicializa con un valor **mayor** que la tolerancia para garantizar que el ciclo se ejecute al menos una vez. El método de Newton converge muy rápido: en pocas iteraciones las estimaciones dejan de cambiar incluso en el décimo decimal.

Un **montacargas** de una bodega tiene una capacidad máxima de **1200 kg**. Su sistema de control funciona con un menú:

```text

```

* **Cargar:** solicita el peso del bulto; si con él se supera la capacidad, lo rechaza con `Rechazado: excede la capacidad en X kg.`
* **Descargar:** solicita el peso del bulto a descargar; no se puede descargar más peso del que hay.
* **Ver estado:** muestra la carga actual, la cantidad de bultos y el **porcentaje de capacidad** usado (1 decimal).
* **Apagar:** solo se permite si el montacargas está **vacío**; si no, muestra `No se puede apagar con carga.`

Escriba un programa en Python que implemente este control con un ciclo **`while`** (sin `break`) y una estructura **`match`** para las opciones. Los pesos que no sean numéricos o que sean menores o iguales a cero deben rechazarse con un mensaje, y cualquier opción desconocida debe mostrar `Opción inválida.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

Opción: <b>1</b>
Peso del bulto (kg): <b>700</b>
Cargado. Carga actual: 1150.0 kg

1. Cargar bulto  2. Descargar bulto  3. Ver estado  4. Apagar
Opción: <b>3</b>

1. Cargar bulto  2. Descargar bulto  3. Ver estado  4. Apagar
Opción: <b>4</b>

</pre>

**Pregunta de respuesta corta**

¿Qué hace el patrón `case "1" | "2":`? ¿Cómo se escribiría la misma condición con un `if`?

````python
#@title Solución
CAPACIDAD_KG = 1200

carga_actual = 0
cantidad_bultos = 0
encendido = True

while encendido:
    print("\n1. Cargar bulto  2. Descargar bulto  3. Ver estado  4. Apagar")
    opcion = input("Opción: ")

    match opcion:
        case "1" | "2":
            try:
                peso = float(input("Peso del bulto (kg): "))
                if peso <= 0:
                    print("Error: el peso debe ser mayor que cero.")
                elif opcion == "1" and carga_actual + peso > CAPACIDAD_KG:
                    exceso = carga_actual + peso - CAPACIDAD_KG
                    print(f"Rechazado: excede la capacidad en {exceso:.1f} kg.")
                elif opcion == "1":
                    carga_actual += peso
                    cantidad_bultos += 1
                    print(f"Cargado. Carga actual: {carga_actual:.1f} kg")
                elif peso > carga_actual:
                    print("Error: no hay tanto peso en el montacargas.")
                else:
                    carga_actual -= peso
                    cantidad_bultos -= 1
                    print(f"Descargado. Carga actual: {carga_actual:.1f} kg")
            except ValueError:
                print("Error: el peso debe ser numérico.")
        case "3":
            porcentaje = carga_actual / CAPACIDAD_KG * 100
            print(f"Carga: {carga_actual:.1f} kg en {cantidad_bultos} bultos ({porcentaje:.1f}% de capacidad)")
        case "4":
            if carga_actual == 0:
                encendido = False
                print("Montacargas apagado.")
            else:
                print("No se puede apagar con carga.")
        case _:
            print("Opción inválida.")
````

**Explicación**

**Respuesta corta:** el operador `|` dentro de un `case` significa "**o**": el caso se ejecuta si la opción es `"1"` o `"2"`. Con un `if` se escribiría `if opcion == "1" or opcion == "2":`. Se agruparon porque cargar y descargar comparten la lectura y validación del peso.
