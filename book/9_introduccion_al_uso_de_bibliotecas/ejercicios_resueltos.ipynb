## Ejercicio 9.1 — Caja sobre una rampa de carga

En un centro de distribución, las cajas bajan por una **rampa inclinada**. Para una caja de masa $m$ sobre una rampa con ángulo $\theta$:

$$F_{\text{paralela}} = m g \sin\theta \qquad N = m g \cos\theta \qquad F_{\text{fricción}} = \mu N$$

con $g = 9.81$ m/s². La caja **se desliza sola** si la fuerza paralela supera la fricción.

Escriba un programa en Python que:

A. Importe **solo** las funciones `sin`, `cos` y `radians` de la biblioteca `math` (con `from math import ...`).

B. Defina `calcular_fuerzas(masa: float, angulo_grados: float, friccion: float) -> tuple[float, float, float]`, que retorna la fuerza paralela, la normal y la de fricción. Recuerde que las funciones trigonométricas de `math` trabajan en **radianes**.

C. En `main()`, solicite la masa (kg), el ángulo (°) y el coeficiente de fricción, muestre las tres fuerzas (N, 2 decimales) y si la caja se desliza sola o necesita un empujón.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Ángulo de la rampa (°): <b>20</b>

Fuerza normal: 230.46 N
Fricción: 69.14 N
La caja se desliza sola.
</pre>

<pre>
Masa de la caja (kg): <b>25</b>
Ángulo de la rampa (°): <b>12</b>
Coeficiente de fricción: <b>0.3</b>

Fuerza paralela: 50.99 N
Fuerza normal: 239.89 N
Fricción: 71.97 N
La caja necesita un empujón.
</pre>

**Pregunta de respuesta corta**

¿Qué valor daría `sin(30)` y por qué **no** es 0.5?

````python
#@title Solución
from math import sin, cos, radians

GRAVEDAD = 9.81


def calcular_fuerzas(masa: float, angulo_grados: float, friccion: float) -> tuple[float, float, float]:
    """Retorna la fuerza paralela a la rampa, la normal y la de fricción (N)."""
    angulo = radians(angulo_grados)
    peso = masa * GRAVEDAD
    fuerza_paralela = peso * sin(angulo)
    normal = peso * cos(angulo)
    return fuerza_paralela, normal, friccion * normal


def main() -> None:
    masa = float(input("Masa de la caja (kg): "))
    angulo = float(input("Ángulo de la rampa (°): "))
    friccion = float(input("Coeficiente de fricción: "))

    paralela, normal, fuerza_friccion = calcular_fuerzas(masa, angulo, friccion)

    print(f"\nFuerza paralela: {paralela:.2f} N")
    print(f"Fuerza normal: {normal:.2f} N")
    print(f"Fricción: {fuerza_friccion:.2f} N")
    if paralela > fuerza_friccion:
        print("La caja se desliza sola.")
    else:
        print("La caja necesita un empujón.")


main()
````

**Explicación**

**Respuesta corta:** `sin(30)` calcula el seno de 30 **radianes**, que es aproximadamente −0.988. Para obtener el seno de 30 grados hay que convertir primero: `sin(radians(30))` ≈ 0.5. Este detalle se descubre leyendo la documentación: `help(math.sin)` dice *"Return the sine of x (measured in radians)"*.

## Ejercicio 9.2 — Alcance del chorro de una manguera

Un ingeniero del Cuerpo de Bomberos estudia el **alcance** del chorro de agua de una manguera. Si se desprecia la resistencia del aire, un chorro que sale a velocidad $v$ (m/s) con un ángulo $\theta$ desde una altura $h_0$ (m) alcanza una distancia horizontal:

$$d = \frac{v \cos\theta}{g} \left( v \sin\theta + \sqrt{(v \sin\theta)^2 + 2 g h_0} \right)$$

La velocidad de salida depende del caudal $Q$ (L/min) y del diámetro de la boquilla $D$ (mm): $v = \frac{Q / 60000}{\pi D^2 / 4}$, con $D$ en metros.

Escriba un programa en Python que importe la biblioteca **completa** (`import math`) y use `math.pi`, `math.sqrt`, `math.sin`, `math.cos` y `math.radians`, con las funciones:

A. `calcular_velocidad(caudal_lpm: float, diametro_mm: float) -> float`

B. `calcular_alcance(velocidad: float, angulo_grados: float, altura: float) -> float`

C. `main() -> None`: solicita el caudal, el diámetro y la altura de la boquilla, y muestra la velocidad de salida y el alcance para los ángulos **15°, 30°, 45° y 60°**, indicando cuál da el **mayor** alcance.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Caudal (L/min): <b>475</b>
Diámetro de la boquilla (mm): <b>19</b>
Altura de la boquilla (m): <b>1.2</b>

Velocidad de salida: 27.92 m/s
  15°: 43.80 m
  30°: 70.85 m
  45°: 80.66 m
  60°: 69.51 m
Mayor alcance con 45°: 80.66 m
</pre>

````python
#@title Solución
import math

GRAVEDAD = 9.81


def calcular_velocidad(caudal_lpm: float, diametro_mm: float) -> float:
    """Calcula la velocidad de salida (m/s) del agua en la boquilla."""
    caudal_m3s = caudal_lpm / 60000
    diametro_m = diametro_mm / 1000
    area = math.pi * diametro_m ** 2 / 4
    return caudal_m3s / area


def calcular_alcance(velocidad: float, angulo_grados: float, altura: float) -> float:
    """Calcula el alcance horizontal (m) de un chorro lanzado desde una altura."""
    angulo = math.radians(angulo_grados)
    vertical = velocidad * math.sin(angulo)
    horizontal = velocidad * math.cos(angulo)
    return horizontal / GRAVEDAD * (vertical + math.sqrt(vertical ** 2 + 2 * GRAVEDAD * altura))


def main() -> None:
    caudal = float(input("Caudal (L/min): "))
    diametro = float(input("Diámetro de la boquilla (mm): "))
    altura = float(input("Altura de la boquilla (m): "))

    velocidad = calcular_velocidad(caudal, diametro)
    print(f"\nVelocidad de salida: {velocidad:.2f} m/s")

    mejor_angulo = 0
    mejor_alcance = 0
    for angulo in [15, 30, 45, 60]:
        alcance = calcular_alcance(velocidad, angulo, altura)
        print(f"  {angulo}°: {alcance:.2f} m")
        if alcance > mejor_alcance:
            mejor_angulo = angulo
            mejor_alcance = alcance

    print(f"Mayor alcance con {mejor_angulo}°: {mejor_alcance:.2f} m")


main()
````

**Explicación**

Con `import math` cada función se escribe con el prefijo `math.`: es un poco más largo, pero deja claro de **dónde** viene cada función y evita confusiones con variables o funciones propias que tengan el mismo nombre.

## Ejercicio 9.3 — Fuerza resultante en un anclaje

Un anclaje de una torre de telecomunicaciones recibe la tensión de **varios cables**. Cada cable tira con una fuerza (kN) en una dirección dada por su ángulo (°), medido desde el eje x. La fuerza resultante se obtiene sumando las componentes:

$$R_x = \sum F_i \cos\theta_i \qquad R_y = \sum F_i \sin\theta_i \qquad R = \sqrt{R_x^2 + R_y^2} \qquad \alpha = \text{atan2}(R_y, R_x)$$

Escriba un programa en Python que importe `math` con el **alias** `m` (`import math as m`) y:

A. Defina `sumar_fuerzas(fuerzas: list[tuple[float, float]]) -> tuple[float, float]`, que recibe una lista de tuplas `(magnitud, angulo_grados)` y retorna $R_x$ y $R_y$.

B. En `main()`, solicite la cantidad de cables y la fuerza y el ángulo de cada uno; muestre $R_x$, $R_y$, la magnitud de la resultante (use `m.hypot`) y su dirección en grados (use `m.atan2` y `m.degrees`), con 2 decimales.

C. Si la resultante supera la **capacidad del anclaje, 25 kN**, muestre `Anclaje sobrecargado.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de cables: <b>3</b>
Cable 1 - fuerza (kN): <b>12</b>
Cable 1 - ángulo (°): <b>30</b>
Cable 2 - fuerza (kN): <b>15</b>
Cable 2 - ángulo (°): <b>150</b>
Cable 3 - fuerza (kN): <b>9</b>
Cable 3 - ángulo (°): <b>270</b>

Rx = -2.60 kN, Ry = 4.50 kN
Resultante: 5.20 kN a 120.00°
</pre>

**Pregunta de respuesta corta**

¿Por qué se usa `atan2(Ry, Rx)` y no `atan(Ry / Rx)` para calcular la dirección?

````python
#@title Solución
import math as m

CAPACIDAD_ANCLAJE_KN = 25


def sumar_fuerzas(fuerzas: list[tuple[float, float]]) -> tuple[float, float]:
    """Retorna la suma de componentes x e y de una lista de fuerzas (magnitud, ángulo)."""
    suma_x = 0
    suma_y = 0
    for magnitud, angulo_grados in fuerzas:
        angulo = m.radians(angulo_grados)
        suma_x += magnitud * m.cos(angulo)
        suma_y += magnitud * m.sin(angulo)
    return suma_x, suma_y


def main() -> None:
    cantidad = int(input("Cantidad de cables: "))
    fuerzas = []
    for numero in range(1, cantidad + 1):
        magnitud = float(input(f"Cable {numero} - fuerza (kN): "))
        angulo = float(input(f"Cable {numero} - ángulo (°): "))
        fuerzas.append((magnitud, angulo))

    resultante_x, resultante_y = sumar_fuerzas(fuerzas)
    magnitud_resultante = m.hypot(resultante_x, resultante_y)
    direccion = m.degrees(m.atan2(resultante_y, resultante_x))

    print(f"\nRx = {resultante_x:.2f} kN, Ry = {resultante_y:.2f} kN")
    print(f"Resultante: {magnitud_resultante:.2f} kN a {direccion:.2f}°")
    if magnitud_resultante > CAPACIDAD_ANCLAJE_KN:
        print("Anclaje sobrecargado.")


main()
````

**Explicación**

**Respuesta corta:** `atan(Ry / Rx)` solo retorna ángulos entre −90° y 90°, así que no distingue, por ejemplo, una fuerza hacia el noreste de una hacia el suroeste (el cociente es el mismo), y además falla con `ZeroDivisionError` si `Rx = 0`. `atan2` recibe **ambas** componentes por separado, determina el cuadrante correcto y funciona aunque `Rx` sea cero.

## Ejercicio 9.4 — Decaimiento de un radioisótopo médico

Un hospital usa **tecnecio-99m** para estudios de imagen. Su actividad disminuye con el tiempo:

$$A(t) = A_0 \cdot e^{-\lambda t} \qquad \lambda = \frac{\ln 2}{T_{1/2}}$$

donde $T_{1/2} = 6.01$ horas es la vida media. Los residuos solo pueden desecharse como basura común cuando su actividad baja de un **límite** dado. Despejando, el tiempo necesario es $t = \frac{\ln(A_0 / A_{\text{límite}})}{\lambda}$.

Escriba un programa en Python con las funciones:

A. `calcular_actividad(actividad_inicial: float, horas: float) -> float` (use `math.exp` y `math.log`).

B. `calcular_horas_de_espera(actividad_inicial: float, actividad_limite: float) -> int`: retorna las horas de espera **redondeadas hacia arriba** con `math.ceil`.

C. `main() -> None`: solicita la actividad inicial y el límite (MBq), muestra la actividad a las 6, 12, 24 y 48 horas (2 decimales), y las horas y **días** (1 decimal) que deben almacenarse los residuos.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Actividad inicial (MBq): <b>740</b>
Límite para desecho (MBq): <b>0.5</b>

Actividad a las 6 h: 370.43 MBq
Actividad a las 12 h: 185.43 MBq
Actividad a las 24 h: 46.46 MBq
Actividad a las 48 h: 2.92 MBq

Almacenar durante 64 h (2.7 días)
</pre>

````python
#@title Solución
import math

VIDA_MEDIA_HORAS = 6.01


def calcular_actividad(actividad_inicial: float, horas: float) -> float:
    """Calcula la actividad (MBq) después de las horas indicadas."""
    constante = math.log(2) / VIDA_MEDIA_HORAS
    return actividad_inicial * math.exp(-constante * horas)


def calcular_horas_de_espera(actividad_inicial: float, actividad_limite: float) -> int:
    """Calcula las horas enteras necesarias para bajar del límite de actividad."""
    constante = math.log(2) / VIDA_MEDIA_HORAS
    return math.ceil(math.log(actividad_inicial / actividad_limite) / constante)


def main() -> None:
    actividad_inicial = float(input("Actividad inicial (MBq): "))
    actividad_limite = float(input("Límite para desecho (MBq): "))

    print()
    for horas in [6, 12, 24, 48]:
        print(f"Actividad a las {horas} h: {calcular_actividad(actividad_inicial, horas):.2f} MBq")

    espera = calcular_horas_de_espera(actividad_inicial, actividad_limite)
    print(f"\nAlmacenar durante {espera} h ({espera / 24:.1f} días)")


main()
````

**Explicación**

A las 6 horas (casi una vida media) la actividad es aproximadamente la mitad, y a las 12 horas, la cuarta parte: comparar con este comportamiento conocido es una forma rápida de **validar** el programa. `math.log` es el logaritmo **natural** ($\ln$); para base 10 existe `math.log10`.

## Ejercicio 9.5 — Ruido combinado en una planta

Un ingeniero de salud ocupacional mide el **nivel sonoro** (dB) de cada máquina de un taller por separado. Los decibeles **no** se suman directamente: el nivel combinado de varias fuentes es

$$L_{total} = 10 \log_{10} \left( \sum_i 10^{L_i / 10} \right)$$

La normativa permite una exposición de **8 horas** a 85 dB, y por cada 3 dB adicionales el tiempo permitido se reduce a la **mitad**: $t = \frac{8}{2^{(L - 85)/3}}$ horas.

Escriba un programa en Python con las funciones:

A. `combinar_niveles(niveles: list[float]) -> float` (use `math.log10` y `math.pow`).

B. `calcular_tiempo_permitido(nivel: float) -> float`

C. `main() -> None`: solicita los niveles de las máquinas en una sola línea (separados por espacios), muestra el nivel combinado (1 decimal) y el tiempo de exposición permitido en horas y minutos. Si el tiempo es menor que la jornada de 8 horas, muestra `Uso obligatorio de protección auditiva.`

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Niveles de las máquinas (dB): <b>82 85 79</b>

Nivel combinado: 87.4 dB
Exposición permitida: 4 h 33 min
Uso obligatorio de protección auditiva.
</pre>

<pre>
Niveles de las máquinas (dB): <b>80 80</b>

Nivel combinado: 83.0 dB
Exposición permitida: 12 h 40 min
</pre>

**Pregunta de respuesta corta**

Dos máquinas de 80 dB cada una, ¿producen 160 dB juntas? Justifique con el segundo ejemplo.

````python
#@title Solución
import math

NIVEL_REFERENCIA = 85
HORAS_REFERENCIA = 8


def combinar_niveles(niveles: list[float]) -> float:
    """Combina varios niveles sonoros (dB) en un nivel total."""
    suma_intensidades = 0
    for nivel in niveles:
        suma_intensidades += math.pow(10, nivel / 10)
    return 10 * math.log10(suma_intensidades)


def calcular_tiempo_permitido(nivel: float) -> float:
    """Calcula las horas de exposición permitidas para un nivel sonoro."""
    return HORAS_REFERENCIA / math.pow(2, (nivel - NIVEL_REFERENCIA) / 3)


def main() -> None:
    niveles = []
    for texto in input("Niveles de las máquinas (dB): ").split():
        niveles.append(float(texto))

    nivel_total = combinar_niveles(niveles)
    tiempo = calcular_tiempo_permitido(nivel_total)
    horas = int(tiempo)
    minutos = round((tiempo - horas) * 60)

    print(f"\nNivel combinado: {nivel_total:.1f} dB")
    print(f"Exposición permitida: {horas} h {minutos} min")
    if tiempo < HORAS_REFERENCIA:
        print("Uso obligatorio de protección auditiva.")


main()
````

**Explicación**

**Respuesta corta:** no. Los decibeles son una escala **logarítmica**: dos fuentes iguales suman solo unos **3 dB** más (80 + 80 → 83 dB), porque duplicar la intensidad sonora equivale a sumar $10 \log_{10} 2 \approx 3$ dB.

## Ejercicio 9.6 — Desgaste uniforme de engranajes

En una caja de engranajes, si los números de dientes del piñón y de la rueda tienen **factores comunes**, los mismos pares de dientes se encuentran una y otra vez y se desgastan de forma desigual. Por eso los diseñadores prefieren combinaciones cuyo **máximo común divisor (MCD) sea 1** (llamadas *hunting tooth*).

Además, el número de vueltas del piñón hasta que **un mismo par de dientes** vuelve a encontrarse es $\frac{\text{mcm}(z_1, z_2)}{z_1}$.

Escriba un programa en Python que importe `gcd` y `lcm` de `math` y:

A. Defina `analizar_par(dientes_pinon: int, dientes_rueda: int) -> tuple[int, int, float]`, que retorna el MCD, las vueltas del piñón hasta repetir el mismo par y la relación de transmisión $z_2 / z_1$.

B. En `main()`, solicite los dientes del piñón y la cantidad de **ruedas candidatas** con sus dientes. Muestre para cada candidata la relación (2 decimales), el MCD, las vueltas hasta repetir y si es `Desgaste uniforme` (MCD = 1) o `Desgaste desigual`.

**Ejemplo de ejecución**

Cantidad de ruedas candidatas: <b>3</b>

  Relación 3.11:1, MCD 2, se repite cada 28 vueltas -&gt; Desgaste desigual
</pre>

````python
#@title Solución
from math import gcd, lcm


def analizar_par(dientes_pinon: int, dientes_rueda: int) -> tuple[int, int, float]:
    """Retorna el MCD, las vueltas hasta repetir el mismo par de dientes y la relación."""
    divisor = gcd(dientes_pinon, dientes_rueda)
    vueltas = lcm(dientes_pinon, dientes_rueda) // dientes_pinon
    return divisor, vueltas, dientes_rueda / dientes_pinon


def main() -> None:
    dientes_pinon = int(input("Dientes del piñón: "))
    cantidad = int(input("Cantidad de ruedas candidatas: "))

    for numero in range(1, cantidad + 1):
        dientes_rueda = int(input(f"\nDientes de la rueda {numero}: "))
        divisor, vueltas, relacion = analizar_par(dientes_pinon, dientes_rueda)
        if divisor == 1:
            tipo = "Desgaste uniforme"
        else:
            tipo = "Desgaste desigual"
        print(f"  Relación {relacion:.2f}:1, MCD {divisor}, se repite cada {vueltas} vueltas -> {tipo}")


main()
````

**Explicación**

Con 18 y 54 dientes (MCD = 18), cada diente del piñón toca siempre los **mismos 3** dientes de la rueda: si uno tiene un defecto, dañará a esos mismos dientes en cada vuelta. Con 55 dientes (MCD = 1), cada diente del piñón termina encontrándose con **todos** los de la rueda. `math.lcm` existe desde Python 3.9: revisar la documentación de la versión instalada es importante.

## Ejercicio 9.7 — Muestreo de aceptación de un lote

Un cliente recibe un lote de $N$ piezas y, en lugar de revisarlas todas, toma una **muestra** de $n$ piezas al azar. Si en la muestra **no** aparece ninguna defectuosa, acepta el lote. Si el lote tiene $D$ piezas defectuosas, la probabilidad de **no** encontrar ninguna en la muestra es:

$$P_{aceptar} = \frac{\binom{N - D}{n}}{\binom{N}{n}}$$

donde $\binom{a}{b}$ es el número de combinaciones, que en Python se calcula con `math.comb(a, b)`.

Escriba un programa en Python con las funciones:

A. `probabilidad_de_aceptar(tamano_lote: int, defectuosas: int, tamano_muestra: int) -> float`

B. `main() -> None`: solicita el tamaño del lote y el de la muestra, y muestra una tabla con la probabilidad de aceptar el lote (como porcentaje con 1 decimal) cuando tiene **1, 2, 5, 10 y 20** piezas defectuosas. Al final, indica cuántas defectuosas se necesitan, como mínimo, para que el lote se rechace con una probabilidad de al menos 95 % (es decir, $P_{aceptar} \le 5\%$), buscando con un ciclo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Tamaño del lote: <b>500</b>
Tamaño de la muestra: <b>50</b>

Defectuosas   P(aceptar)
          1       90.0%
          2       81.0%
          5       58.9%
         10       34.5%
         20       11.6%

Con 28 o más defectuosas, el lote se rechaza con al menos 95% de probabilidad.
</pre>

````python
#@title Solución
import math

PROBABILIDAD_OBJETIVO = 0.05


def probabilidad_de_aceptar(tamano_lote: int, defectuosas: int, tamano_muestra: int) -> float:
    """Probabilidad de que la muestra no contenga piezas defectuosas."""
    return math.comb(tamano_lote - defectuosas, tamano_muestra) / math.comb(tamano_lote, tamano_muestra)


def main() -> None:
    tamano_lote = int(input("Tamaño del lote: "))
    tamano_muestra = int(input("Tamaño de la muestra: "))

    print("\nDefectuosas   P(aceptar)")
    for defectuosas in [1, 2, 5, 10, 20]:
        probabilidad = probabilidad_de_aceptar(tamano_lote, defectuosas, tamano_muestra)
        print(f"{defectuosas:11d}   {probabilidad:9.1%}")

    defectuosas = 0
    probabilidad = 1
    while probabilidad > PROBABILIDAD_OBJETIVO and defectuosas < tamano_lote:
        defectuosas += 1
        probabilidad = probabilidad_de_aceptar(tamano_lote, defectuosas, tamano_muestra)

    print(f"\nCon {defectuosas} o más defectuosas, el lote se rechaza con al menos 95% de probabilidad.")


main()
````

**Explicación**

La tabla muestra una limitación importante del muestreo: con un lote de 500 piezas y una muestra de 50, un lote con 5 piezas defectuosas se **acepta** más de la mitad de las veces. `math.comb` evita tener que programar factoriales enormes: `math.comb(500, 50)` es un número de 70 cifras.

## Ejercicio 9.8 — Selección aleatoria para auditoría

El departamento de calidad de una planta de ensamble debe auditar al azar algunos **números de serie** de la producción del día, para que ningún operario pueda anticipar qué piezas se revisarán. Además, cada pieza auditada se asigna al azar a uno de los inspectores disponibles.

Escriba un programa en Python que importe `random` y:

A. Defina `generar_series(prefijo: str, primera: int, cantidad: int) -> list[str]`, que retorna los números de serie con el formato `PREFIJO-00001` (5 dígitos).

B. Defina `seleccionar_auditoria(series: list[str], cantidad: int) -> list[str]`, que usa `random.sample` para elegir piezas **sin repetir** y las retorna **ordenadas**.

C. En `main()`, fije la semilla con `random.seed(...)` usando el número de **lote** que ingresa la persona usuaria, solicite el prefijo, el primer número de serie, la producción del día y el porcentaje a auditar. Muestre las series seleccionadas y, para cada una, el inspector asignado con `random.choice(["Ana", "Luis", "Marta"])`.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

  VX-01971 -&gt; Luis
  VX-02005 -&gt; Marta
  VX-02060 -&gt; Marta
</pre>

**Pregunta de respuesta corta**

¿Qué pasaría si se usara `random.choice(series)` varias veces en lugar de `random.sample`? ¿Por qué es importante fijar la semilla en una auditoría?

````python
#@title Solución
import random

INSPECTORES = ["Ana", "Luis", "Marta"]


def generar_series(prefijo: str, primera: int, cantidad: int) -> list[str]:
    """Genera los números de serie de la producción."""
    series = []
    for numero in range(primera, primera + cantidad):
        series.append(f"{prefijo}-{numero:05d}")
    return series


def seleccionar_auditoria(series: list[str], cantidad: int) -> list[str]:
    """Selecciona al azar, sin repetir, las series que se auditarán."""
    return sorted(random.sample(series, cantidad))


def main() -> None:
    lote = int(input("Número de lote (semilla): "))
    random.seed(lote)

    prefijo = input("Prefijo de la serie: ").upper()
    primera = int(input("Primer número de serie: "))
    produccion = int(input("Producción del día: "))
    porcentaje = float(input("Porcentaje a auditar: "))

    series = generar_series(prefijo, primera, produccion)
    cantidad = max(1, round(produccion * porcentaje / 100))
    seleccionadas = seleccionar_auditoria(series, cantidad)

    print(f"\nPiezas a auditar ({cantidad} de {produccion}):")
    for serie in seleccionadas:
        print(f"  {serie} -> {random.choice(INSPECTORES)}")


main()
````

**Explicación**

**Respuesta corta:** `random.choice` puede elegir la **misma** pieza más de una vez, mientras que `random.sample` garantiza elementos distintos. Fijar la semilla hace que la selección sea **reproducible**: con el número de lote, un auditor externo puede verificar qué piezas debieron revisarse, sin que eso impida que la selección sea impredecible de antemano para los operarios.

<span id="ejercicio-9-9-ruido-en-las-lecturas-de-un-sensor"></span>

## Ejercicio 9.9 — Ruido en las lecturas de un sensor

Antes de instalar un sensor de temperatura en un reactor, un ingeniero quiere **simular** cómo se verían sus lecturas. El fabricante indica que el sensor tiene un **ruido** con distribución normal de media 0 y desviación estándar $\sigma$ (°C). Cada lectura simulada es la temperatura real más un valor aleatorio `random.gauss(0, sigma)`.

Escriba un programa en Python que importe `random` y `statistics` y:

A. Defina `simular_lecturas(temperatura_real: float, sigma: float, cantidad: int) -> list[float]`.

B. Defina `contar_falsas_alarmas(lecturas: list[float], limite: float) -> int`: cuántas lecturas superan el límite de alarma, aunque la temperatura real **no** lo supere.

C. En `main()`, use `random.seed(42)`, solicite la temperatura real, $\sigma$, la cantidad de lecturas y el límite de alarma. Muestre las 5 primeras lecturas (2 decimales), la media y la desviación estándar muestrales (con `statistics.mean` y `statistics.stdev`, 3 decimales), el mínimo y el máximo, y el **porcentaje** de falsas alarmas.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Temperatura real (°C): <b>180</b>
Desviación estándar del ruido (°C): <b>1.5</b>
Cantidad de lecturas: <b>1000</b>
Límite de alarma (°C): <b>183</b>

Primeras lecturas: [179.78, 179.74, 179.83, 181.05, 179.81]
Media: 179.973 °C
Desviación estándar: 1.529 °C
Mínimo: 175.27 °C  Máximo: 185.63 °C
Falsas alarmas: 26 (2.6%)
</pre>

````python
#@title Solución
import random
import statistics


def simular_lecturas(temperatura_real: float, sigma: float, cantidad: int) -> list[float]:
    """Simula lecturas de un sensor con ruido normal."""
    lecturas = []
    for _ in range(cantidad):
        lecturas.append(temperatura_real + random.gauss(0, sigma))
    return lecturas


def contar_falsas_alarmas(lecturas: list[float], limite: float) -> int:
    """Cuenta las lecturas que superan el límite de alarma."""
    falsas = 0
    for lectura in lecturas:
        if lectura > limite:
            falsas += 1
    return falsas


def main() -> None:
    random.seed(42)

    temperatura_real = float(input("Temperatura real (°C): "))
    sigma = float(input("Desviación estándar del ruido (°C): "))
    cantidad = int(input("Cantidad de lecturas: "))
    limite = float(input("Límite de alarma (°C): "))

    lecturas = simular_lecturas(temperatura_real, sigma, cantidad)
    falsas = contar_falsas_alarmas(lecturas, limite)

    primeras = []
    for lectura in lecturas[:5]:
        primeras.append(round(lectura, 2))

    print(f"\nPrimeras lecturas: {primeras}")
    print(f"Media: {statistics.mean(lecturas):.3f} °C")
    print(f"Desviación estándar: {statistics.stdev(lecturas):.3f} °C")
    print(f"Mínimo: {min(lecturas):.2f} °C  Máximo: {max(lecturas):.2f} °C")
    print(f"Falsas alarmas: {falsas} ({falsas / cantidad:.1%})")


main()
````

**Explicación**

La media y la desviación estándar de la muestra quedan muy cerca de los valores del modelo (180 °C y 1.5 °C), como se espera con 1000 lecturas. Un límite de alarma a solo **2σ** de la temperatura normal produce falsas alarmas en alrededor del 2 % al 3 % de las lecturas: la simulación ayuda a elegir un límite razonable **antes** de instalar el equipo.

Los instrumentos de un laboratorio acreditado deben **recalibrarse** periódicamente. Cada tipo de instrumento tiene su propio intervalo:

C. En `main()`, solicite la **fecha de la auditoría** y, para cada instrumento (hasta que se ingrese un código vacío), su tipo y la fecha de su última calibración. Muestre la fecha de vencimiento (formato `DD/MM/AAAA` con `strftime`) y los días que faltan, o `VENCIDO hace N días`. Al final, muestre cuántos instrumentos están vencidos.

<pre>
Fecha de la auditoría: <b>15/03/2026</b>

````python
#@title Solución
from datetime import date, datetime, timedelta

INTERVALOS_DIAS = {"balanza": 365, "termometro": 180, "manometro": 730}
FORMATO = "%d/%m/%Y"


def calcular_vencimiento(fecha_calibracion: date, tipo: str) -> date:
    """Retorna la fecha en que vence la calibración de un instrumento."""
    return fecha_calibracion + timedelta(days=INTERVALOS_DIAS[tipo])


def leer_fecha(mensaje: str) -> date:
    """Lee una fecha DD/MM/AAAA, repitiendo mientras sea inválida."""
    fecha = None
    while fecha is None:
        try:
            fecha = datetime.strptime(input(mensaje), FORMATO).date()
        except ValueError:
            print("  Formato inválido, use DD/MM/AAAA.")
    return fecha


def main() -> None:
    fecha_auditoria = leer_fecha("Fecha de la auditoría: ")
    vencidos = 0

    codigo = input("\nCódigo del instrumento (Enter para terminar): ").upper()
    while codigo != "":
        tipo = input("  Tipo (balanza/termometro/manometro): ").lower()
        if tipo not in INTERVALOS_DIAS:
            print("  Tipo desconocido.")
        else:
            calibracion = leer_fecha("  Última calibración: ")
            vencimiento = calcular_vencimiento(calibracion, tipo)
            dias = (vencimiento - fecha_auditoria).days
            if dias >= 0:
                print(f"  Vence el {vencimiento.strftime(FORMATO)} (faltan {dias} días)")
            else:
                vencidos += 1
                print(f"  Venció el {vencimiento.strftime(FORMATO)}: VENCIDO hace {-dias} días")
        codigo = input("\nCódigo del instrumento (Enter para terminar): ").upper()

    print(f"\nInstrumentos vencidos: {vencidos}")


main()
````

**Explicación**

Restar dos objetos `date` produce un `timedelta`, cuyo atributo `.days` es la diferencia en días; sumar un `timedelta` a una fecha "salta" los días correctos, considerando la cantidad de días de cada mes y los años bisiestos. Hacer estos cálculos a mano sería una fuente segura de errores.

Una línea de **extrusión de tubería de PVC** registra la fecha y hora de inicio y fin de cada corrida de producción, que puede terminar al día siguiente. La línea produce a una velocidad constante (m/min), pero durante la corrida hay **paros** (minutos) por cambios de rollo y limpieza.

Escriba un programa en Python que importe `datetime` de la biblioteca `datetime` y:

B. Defina `calcular_produccion(inicio: datetime, fin: datetime, paros_min: float, velocidad: float) -> tuple[float, float, float]`, que retorna la duración total en horas, la **disponibilidad** (tiempo productivo / tiempo total) y los metros producidos. Use `(fin - inicio).total_seconds()`.

C. En `main()`, solicite los datos, valide que el fin sea posterior al inicio, y muestre la duración (2 decimales), la disponibilidad (1 decimal, `%`), los metros producidos y la cantidad de **tubos de 6 m** completos.

<pre>

Disponibilidad: 86.5%

````python
#@title Solución
from datetime import datetime

FORMATO = "%d/%m/%Y %H:%M"
LARGO_TUBO_M = 6


def leer_fecha_hora(mensaje: str) -> datetime:
    """Lee una fecha y hora en formato DD/MM/AAAA HH:MM."""
    return datetime.strptime(input(mensaje), FORMATO)


def calcular_produccion(inicio: datetime, fin: datetime, paros_min: float,
                        velocidad: float) -> tuple[float, float, float]:
    """Retorna la duración (h), la disponibilidad y los metros producidos."""
    minutos_totales = (fin - inicio).total_seconds() / 60
    minutos_productivos = minutos_totales - paros_min
    disponibilidad = minutos_productivos / minutos_totales
    return minutos_totales / 60, disponibilidad, minutos_productivos * velocidad


def main() -> None:
    inicio = leer_fecha_hora("Inicio de la corrida (DD/MM/AAAA HH:MM): ")
    fin = leer_fecha_hora("Fin de la corrida (DD/MM/AAAA HH:MM): ")
    paros = float(input("Minutos de paros: "))
    velocidad = float(input("Velocidad de extrusión (m/min): "))

    if fin <= inicio:
        print("Error: el fin debe ser posterior al inicio.")
    else:
        horas, disponibilidad, metros = calcular_produccion(inicio, fin, paros, velocidad)
        print(f"\nDuración: {horas:.2f} h")
        print(f"Disponibilidad: {disponibilidad:.1%}")
        print(f"Producción: {metros:,.1f} m ({int(metros // LARGO_TUBO_M)} tubos de {LARGO_TUBO_M} m)")


main()
````

**Explicación**

## Ejercicio 9.12 — Próximo mantenimiento de un compresor

Un compresor industrial debe recibir mantenimiento cada **2000 horas de operación**. El compresor trabaja una cantidad fija de horas por día, **de lunes a viernes** (los fines de semana está apagado).

Defina `calcular_fecha_mantenimiento(fecha_actual: date, horas_restantes: float, horas_por_dia: float) -> date`: avanza **día por día** con `timedelta(days=1)`, descontando las horas solo en los días laborales (use `fecha.weekday()`, que retorna 0 para lunes y 6 para domingo), hasta que se cumplan las horas.

B. En `main()`, solicite la fecha actual (`DD/MM/AAAA`), el **horómetro** del compresor (horas acumuladas) y las horas de operación por día laboral. Calcule las horas **restantes** hasta el siguiente múltiplo de 2000 (use `%`). Muestre la fecha estimada del mantenimiento con el **nombre del día en español** (use una lista de nombres indexada con `weekday()`) y cuántos días calendario faltan.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Fecha actual (DD/MM/AAAA): <b>22/09/2026</b>
Horómetro (h): <b>13415</b>
Horas de operación por día laboral: <b>16</b>

Horas restantes para el mantenimiento: 585 h
Fecha estimada: jueves 12/11/2026
Faltan 51 días calendario.
</pre>

````python
#@title Solución
from datetime import date, datetime, timedelta

INTERVALO_HORAS = 2000
DIAS_SEMANA = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]


def calcular_fecha_mantenimiento(fecha_actual: date, horas_restantes: float,
                                 horas_por_dia: float) -> date:
    """Estima la fecha en que se cumplirán las horas restantes de operación."""
    fecha = fecha_actual
    while horas_restantes > 0:
        fecha = fecha + timedelta(days=1)
        if fecha.weekday() < 5:
            horas_restantes -= horas_por_dia
    return fecha


def main() -> None:
    fecha_actual = datetime.strptime(input("Fecha actual (DD/MM/AAAA): "), "%d/%m/%Y").date()
    horometro = float(input("Horómetro (h): "))
    horas_por_dia = float(input("Horas de operación por día laboral: "))

    horas_restantes = INTERVALO_HORAS - horometro % INTERVALO_HORAS
    fecha = calcular_fecha_mantenimiento(fecha_actual, horas_restantes, horas_por_dia)
    dias = (fecha - fecha_actual).days

    print(f"\nHoras restantes para el mantenimiento: {horas_restantes:.0f} h")
    print(f"Fecha estimada: {DIAS_SEMANA[fecha.weekday()]} {fecha.strftime('%d/%m/%Y')}")
    print(f"Faltan {dias} días calendario.")


main()
````

**Explicación**

Se usa una lista propia de nombres porque `strftime("%A")` retorna el nombre del día en el **idioma del sistema** (normalmente inglés), lo que haría que el programa se comporte distinto en cada computadora. `weekday()` retorna un número que no depende del idioma.

C. En `main()`, use `random.seed(7)`, genere y muestre **4** códigos completos; luego solicite códigos para verificar **hasta** que se ingrese `fin`, e indique si cada uno es `VÁLIDO` o `INVÁLIDO`.

**Pregunta de respuesta corta**

El cuarto código de prueba tiene dos caracteres **intercambiados** (`JD` → `DJ`) y, sin embargo, el programa lo acepta como válido. ¿Por qué este carácter de control **no** detecta ese tipo de error?

````python
#@title Solución
import random
import string

ALFABETO = string.digits + string.ascii_uppercase
LONGITUD = 8


def generar_codigo(longitud: int) -> str:
    """Genera un código aleatorio de letras mayúsculas y dígitos."""
    return "".join(random.choices(ALFABETO, k=longitud))


def calcular_control(codigo: str) -> str:
    """Calcula el carácter de control de un código."""
    suma = 0
    for caracter in codigo:
        suma += ALFABETO.index(caracter)
    return ALFABETO[suma % len(ALFABETO)]


def verificar(codigo_completo: str) -> bool:
    """Indica si el último carácter coincide con el control del resto del código."""
    valido = len(codigo_completo) == LONGITUD + 1
    for caracter in codigo_completo:
        if caracter not in ALFABETO:
            valido = False
    if valido:
        valido = calcular_control(codigo_completo[:-1]) == codigo_completo[-1]
    return valido


def main() -> None:
    random.seed(7)

    print("Códigos generados:")
    for _ in range(4):
        codigo = generar_codigo(LONGITUD)
        print(f"  {codigo}{calcular_control(codigo)}")

    entrada = input("\nCódigo a verificar (o 'fin'): ").upper()
    while entrada != "FIN":
        if verificar(entrada):
            print("  VÁLIDO")
        else:
            print("  INVÁLIDO")
        entrada = input("Código a verificar (o 'fin'): ").upper()


main()
````

**Explicación**

La biblioteca `string` evita escribir a mano el alfabeto: `string.ascii_uppercase` es `'ABCDEFGHIJKLMNOPQRSTUVWXYZ'` y `string.digits` es `'0123456789'`.

**Respuesta corta:** porque la suma **no depende del orden** de los caracteres: intercambiar dos caracteres produce la misma suma y, por tanto, el mismo control. Los dígitos verificadores reales (como el de un código de barras) multiplican cada carácter por un **peso** según su posición, justamente para detectar las transposiciones.

## Ejercicio 9.14 — Ducto de ventilación equivalente

En sistemas de aire acondicionado, a veces un ducto **circular** debe reemplazarse por uno **rectangular** que cabe mejor sobre el cielo raso. El diámetro equivalente de un ducto rectangular de lados $a$ y $b$ (misma pérdida de presión) es:

$$D_{eq} = 1.30 \cdot \frac{(a \cdot b)^{0.625}}{(a + b)^{0.25}}$$

Escriba un programa en Python que use **las cuatro formas de importar** estudiadas:

* `import math` para usar `math.pi`,

* `from math import pow as potencia` para las potencias fraccionarias.

A. Defina `calcular_diametro_equivalente(lado_a: float, lado_b: float) -> float`.

B. En `main()`, solicite el diámetro del ducto circular original (mm) y el **alto** disponible sobre el cielo raso (mm). Pruebe anchos desde el alto hasta 4 veces el alto, de **50 en 50 mm**, y muestre el **primer** ancho cuyo diámetro equivalente sea **mayor o igual** que el original. Muestre también las áreas de ambos ductos (cm²) y el promedio de los diámetros equivalentes probados.

**Ejemplo de ejecución**

**Pregunta de respuesta corta**

¿Qué problema podría causar escribir `from math import *` en un programa que también tiene una variable llamada `e` o una función propia llamada `pow`?

````python
#@title Solución
import math
from math import sqrt
import statistics as est
from math import pow as potencia


def calcular_diametro_equivalente(lado_a: float, lado_b: float) -> float:
    """Diámetro equivalente (mm) de un ducto rectangular."""
    return 1.30 * potencia(lado_a * lado_b, 0.625) / potencia(lado_a + lado_b, 0.25)


def main() -> None:
    diametro_original = float(input("Diámetro del ducto circular (mm): "))
    alto = int(input("Alto disponible (mm): "))

    ancho_elegido = 0
    diametros_probados = []
    ancho = alto
    while ancho <= 4 * alto and ancho_elegido == 0:
        diametro = calcular_diametro_equivalente(ancho, alto)
        diametros_probados.append(diametro)
        if diametro >= diametro_original:
            ancho_elegido = ancho
        ancho += 50

    if ancho_elegido == 0:
        print("No hay un ducto rectangular adecuado con ese alto.")
    else:
        area_circular = math.pi * diametro_original ** 2 / 4 / 100
        area_rectangular = ancho_elegido * alto / 100
        print(f"\nDucto rectangular: {ancho_elegido} x {alto} mm")
        print(f"Diámetro equivalente: {diametros_probados[-1]:.1f} mm")
        print(f"Área circular: {area_circular:.1f} cm², área rectangular: {area_rectangular:.1f} cm²")
        print(f"Lado de un ducto cuadrado de igual área que el circular: {sqrt(area_circular * 100):.1f} mm")
        print(f"Promedio de diámetros probados: {est.mean(diametros_probados):.1f} mm")


main()
````

**Explicación**

**Respuesta corta:** `from math import *` trae **todos** los nombres de `math` al programa, incluidos `e`, `pow`, `log`, etc. Si el programa tiene sus propias variables o funciones con esos nombres, unos **reemplazan** a los otros según el orden en que se definan, lo que produce errores difíciles de encontrar. Por eso se prefiere importar solo lo necesario o usar el prefijo `math.`.

## Ejercicio 9.15 — Comparación de mediciones con `math.isclose`

Un laboratorio de metrología compara las mediciones de **dos instrumentos** sobre las mismas piezas. Como los flotantes no deben compararse con `==`, se decide usar la función `math.isclose`.

Lea la documentación con `help(math.isclose)` y responda: ¿qué significan los parámetros `rel_tol` y `abs_tol`? Luego escriba un programa en Python que:

A. Solicite las mediciones del instrumento A y del B, cada grupo en una sola línea.

B. Para cada par de mediciones, use `math.isclose` con una tolerancia **relativa** de 0.5 % y muestre si concuerdan, junto con la diferencia relativa (%).

C. Repita la comparación con una tolerancia **absoluta** de 0.02 mm (`rel_tol=0`), y muestre cuántos pares concuerdan con cada criterio.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Mediciones del instrumento A (mm): <b>2.015 12.480 25.310 98.40 0.512</b>
Mediciones del instrumento B (mm): <b>2.030 12.495 25.390 98.62 0.509</b>

   A (mm)    B (mm)  Dif. rel.  Relativa  Absoluta
    2.015     2.030     0.74%     False      True
   12.480    12.495     0.12%      True      True
   25.310    25.390     0.32%      True     False
   98.400    98.620     0.22%      True     False
    0.512     0.509     0.59%     False      True

Concuerdan con tolerancia relativa del 0.5%: 3
Concuerdan con tolerancia absoluta de 0.02 mm: 3
</pre>

````python
#@title Solución
import math

TOLERANCIA_RELATIVA = 0.005
TOLERANCIA_ABSOLUTA = 0.02


def leer_mediciones(mensaje: str) -> list[float]:
    """Lee una línea de mediciones separadas por espacios."""
    mediciones = []
    for texto in input(mensaje).split():
        mediciones.append(float(texto))
    return mediciones


def main() -> None:
    instrumento_a = leer_mediciones("Mediciones del instrumento A (mm): ")
    instrumento_b = leer_mediciones("Mediciones del instrumento B (mm): ")

    concuerdan_relativa = 0
    concuerdan_absoluta = 0
    print("\n   A (mm)    B (mm)  Dif. rel.  Relativa  Absoluta")
    for medida_a, medida_b in zip(instrumento_a, instrumento_b):
        relativa = math.isclose(medida_a, medida_b, rel_tol=TOLERANCIA_RELATIVA)
        absoluta = math.isclose(medida_a, medida_b, rel_tol=0, abs_tol=TOLERANCIA_ABSOLUTA)
        diferencia = abs(medida_a - medida_b) / max(abs(medida_a), abs(medida_b))
        concuerdan_relativa += relativa
        concuerdan_absoluta += absoluta
        print(f"{medida_a:9.3f} {medida_b:9.3f} {diferencia:9.2%}  {str(relativa):>8}  {str(absoluta):>8}")

    print(f"\nConcuerdan con tolerancia relativa del 0.5%: {concuerdan_relativa}")
    print(f"Concuerdan con tolerancia absoluta de 0.02 mm: {concuerdan_absoluta}")


main()
````

**Explicación**

**Respuesta:** según la documentación, `isclose(a, b)` es verdadero si `abs(a - b) <= max(rel_tol * max(abs(a), abs(b)), abs_tol)`. `rel_tol` es una tolerancia **proporcional** al tamaño de los números (útil cuando las piezas tienen dimensiones muy diferentes) y `abs_tol` es una tolerancia **fija** en las unidades de la medición (útil cerca de cero, donde una tolerancia relativa sería casi nula). En el ejemplo, la pieza de 98.4 mm concuerda con el criterio relativo pero no con el absoluto, y la de 2.015 mm al revés.

Sumar booleanos (`concuerdan_relativa += relativa`) funciona porque `True` vale 1 y `False` vale 0.

````python
#@title Solución
import csv

NIVEL_CRITICO_M = 35


def leer_bitacora(nombre_archivo: str) -> list[dict]:
    """Lee la bitácora del pozo y convierte los valores numéricos."""
    registros = []
    with open(nombre_archivo, encoding="utf-8") as archivo:
        for fila in csv.DictReader(archivo):
            fila["nivel_m"] = float(fila["nivel_m"])
            fila["volumen_m3"] = float(fila["volumen_m3"])
            registros.append(fila)
    return registros


def resumir(registros: list[dict]) -> tuple[float, float, str]:
    """Retorna el volumen total, el nivel más profundo y su fecha."""
    volumen_total = 0
    nivel_maximo = registros[0]["nivel_m"]
    fecha_maximo = registros[0]["fecha"]
    for registro in registros:
        volumen_total += registro["volumen_m3"]
        if registro["nivel_m"] > nivel_maximo:
            nivel_maximo = registro["nivel_m"]
            fecha_maximo = registro["fecha"]
    return volumen_total, nivel_maximo, fecha_maximo


def main() -> None:
    registros = leer_bitacora("pozo.csv")
    volumen_total, nivel_maximo, fecha_maximo = resumir(registros)

    fechas_criticas = []
    for registro in registros:
        if registro["nivel_m"] > NIVEL_CRITICO_M:
            fechas_criticas.append(registro["fecha"])

    print(f"Registros leídos: {len(registros)}")
    print(f"Volumen total: {volumen_total:.0f} m³ (promedio {volumen_total / len(registros):.1f} m³/día)")
    print(f"Nivel más profundo: {nivel_maximo} m el {fecha_maximo}")
    print(f"Días con nivel crítico (> {NIVEL_CRITICO_M} m): {fechas_criticas}")


main()
````

Así el código es más legible y no se rompe si alguien cambia el orden de las columnas. Todos los valores se leen como **texto**, por eso hay que convertirlos.

## Ejercicio 9.17 — Gráfico de control de un proceso

En control estadístico de procesos, una **carta de control** define un límite superior e inferior a partir de mediciones de referencia del proceso "sano":

$$LSC = \bar{x} + 3\sigma \qquad LIC = \bar{x} - 3\sigma$$

Cualquier medición nueva fuera de esos límites indica que el proceso está **fuera de control**.

Una embotelladora controla el **volumen de llenado** (mL) de sus botellas. Escriba un programa en Python que importe `mean`, `stdev` y `median` de la biblioteca `statistics` y:

A. Defina `calcular_limites(referencia: list[float]) -> tuple[float, float, float]`, que retorna la media, el LIC y el LSC.

B. Defina `buscar_fuera_de_control(mediciones: list[float], lic: float, lsc: float) -> list[tuple[int, float]]`: retorna tuplas `(número de medición, valor)` de las mediciones fuera de los límites (numeradas desde 1).

C. En `main()`, solicite las mediciones de **referencia** y las mediciones **nuevas** (cada grupo en una sola línea), muestre la media, la mediana y la desviación estándar de la referencia, los límites (2 decimales) y las mediciones fuera de control, o `Proceso bajo control.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Mediciones de referencia (mL): <b>500.2 499.8 500.5 500.1 499.6 500.3 500.0 499.9 500.4 499.7</b>
Mediciones nuevas (mL): <b>500.1 499.5 501.4 500.2 498.7 500.0</b>

Media: 500.05 mL  Mediana: 500.05 mL  σ: 0.303 mL
Límites de control: [499.14, 500.96] mL
  Medición 3: 501.4 mL FUERA DE CONTROL
  Medición 5: 498.7 mL FUERA DE CONTROL
</pre>

````python
#@title Solución
from statistics import mean, stdev, median


def leer_valores(mensaje: str) -> list[float]:
    """Lee una línea de valores separados por espacios."""
    valores = []
    for texto in input(mensaje).split():
        valores.append(float(texto))
    return valores


def calcular_limites(referencia: list[float]) -> tuple[float, float, float]:
    """Retorna la media y los límites de control inferior y superior."""
    promedio = mean(referencia)
    desviacion = stdev(referencia)
    return promedio, promedio - 3 * desviacion, promedio + 3 * desviacion


def buscar_fuera_de_control(mediciones: list[float], lic: float, lsc: float) -> list[tuple[int, float]]:
    """Retorna las mediciones (número, valor) fuera de los límites de control."""
    fuera = []
    for numero, valor in enumerate(mediciones, start=1):
        if valor < lic or valor > lsc:
            fuera.append((numero, valor))
    return fuera


def main() -> None:
    referencia = leer_valores("Mediciones de referencia (mL): ")
    nuevas = leer_valores("Mediciones nuevas (mL): ")

    promedio, lic, lsc = calcular_limites(referencia)
    fuera = buscar_fuera_de_control(nuevas, lic, lsc)

    print(f"\nMedia: {promedio:.2f} mL  Mediana: {median(referencia):.2f} mL  σ: {stdev(referencia):.3f} mL")
    print(f"Límites de control: [{lic:.2f}, {lsc:.2f}] mL")
    if len(fuera) == 0:
        print("Proceso bajo control.")
    else:
        for numero, valor in fuera:
            print(f"  Medición {numero}: {valor} mL FUERA DE CONTROL")


main()
````

**Explicación**

Las mediciones de referencia se usan **solo** para calcular los límites; las nuevas se comparan con ellos. Si se calcularan los límites incluyendo las mediciones anómalas, éstas "ensancharían" los límites y podrían pasar desapercibidas.

<span id="ejercicio-9-18-materiales-para-enchapar-un-ba-o"></span>

## Ejercicio 9.18 — Materiales para enchapar un baño

Un maestro de obras debe **enchapar** con cerámica las paredes de un baño rectangular hasta cierta altura. La cerámica se vende en **cajas** que cubren un área fija, y se compra un **10 % adicional** por cortes y roturas. El pegamento se vende en **sacos de 20 kg** y rinde **4 m² por saco**. Se descuenta el área de la puerta.

Escriba un programa en Python que importe `ceil` y `floor` de `math` y:

A. Defina `calcular_area_enchape(largo: float, ancho: float, altura: float, area_puerta: float) -> float`.

B. Defina `calcular_materiales(area: float, cobertura_caja: float) -> tuple[int, int]`, que retorna las cajas de cerámica y los sacos de pegamento, **redondeados hacia arriba** con `ceil`.

C. En `main()`, solicite las dimensiones, el área de la puerta, la cobertura de cada caja (m²) y el precio de la caja y del saco. Muestre el área, las cajas y los sacos, el costo total, y cuántos m² de cerámica **sobrarán** aproximadamente. Si se vendieran cajas **sueltas por m² completo**, muestre también cuántos m² completos sobrarían (use `floor`).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Ancho del baño (m): <b>1.8</b>

Área de la puerta (m²): <b>1.6</b>

Sobrante aproximado: 2.68 m² (2 m² completos)
</pre>

**Pregunta de respuesta corta**

¿Qué diferencia hay entre `math.floor(-2.5)`, `int(-2.5)` y `math.ceil(-2.5)`?

````python
#@title Solución
from math import ceil, floor

DESPERDICIO = 0.10
RENDIMIENTO_SACO_M2 = 4


def calcular_area_enchape(largo: float, ancho: float, altura: float, area_puerta: float) -> float:
    """Calcula el área de pared a enchapar (m²)."""
    perimetro = 2 * (largo + ancho)
    return perimetro * altura - area_puerta


def calcular_materiales(area: float, cobertura_caja: float) -> tuple[int, int]:
    """Retorna las cajas de cerámica y los sacos de pegamento necesarios."""
    cajas = ceil(area * (1 + DESPERDICIO) / cobertura_caja)
    sacos = ceil(area / RENDIMIENTO_SACO_M2)
    return cajas, sacos


def main() -> None:
    largo = float(input("Largo del baño (m): "))
    ancho = float(input("Ancho del baño (m): "))
    altura = float(input("Altura del enchape (m): "))
    area_puerta = float(input("Área de la puerta (m²): "))
    cobertura_caja = float(input("Cobertura por caja (m²): "))
    precio_caja = float(input("Precio por caja (₡): "))
    precio_saco = float(input("Precio por saco de pegamento (₡): "))

    area = calcular_area_enchape(largo, ancho, altura, area_puerta)
    cajas, sacos = calcular_materiales(area, cobertura_caja)
    costo = cajas * precio_caja + sacos * precio_saco
    sobrante = cajas * cobertura_caja - area

    print(f"\nÁrea a enchapar: {area:.2f} m²")
    print(f"Cajas de cerámica: {cajas}")
    print(f"Sacos de pegamento: {sacos}")
    print(f"Costo total: ₡{costo:,.0f}")
    print(f"Sobrante aproximado: {sobrante:.2f} m² ({floor(sobrante)} m² completos)")


main()
````

**Explicación**

**Respuesta corta:** `math.floor(-2.5)` es **−3** (redondea hacia menos infinito), `int(-2.5)` es **−2** (trunca, es decir, redondea hacia cero) y `math.ceil(-2.5)` es **−2** (redondea hacia más infinito). Con números positivos `floor` e `int` coinciden, pero con negativos no, por eso conviene conocer ambas.

## Ejercicio 9.19 — Orden aleatorio de un diseño de experimentos

Un ingeniero químico estudia cómo la **temperatura** y la **concentración de catalizador** afectan el rendimiento de una reacción. Diseña un experimento con todas las combinaciones de ambos factores, y cada combinación se repite varias veces (**réplicas**). Para evitar que efectos externos (como el desgaste del equipo a lo largo del día) sesguen los resultados, las corridas deben realizarse en **orden aleatorio**.

Escriba un programa en Python que importe `random` y:

A. Defina `crear_corridas(temperaturas: list[int], concentraciones: list[float], replicas: int) -> list[tuple[int, float, int]]`, que genera todas las corridas `(temperatura, concentración, número de réplica)` con ciclos anidados.

B. Defina `aleatorizar(corridas: list) -> list`, que retorna una **copia** barajada con `random.shuffle` (sin modificar la lista original).

C. En `main()`, use `random.seed(2026)`, solicite las temperaturas y las concentraciones (cada grupo en una sola línea) y la cantidad de réplicas, y muestre el total de corridas y la **hoja de trabajo** con el orden de ejecución, numerada con `enumerate`.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Temperaturas (°C): <b>60 80</b>
Concentraciones de catalizador (%): <b>0.5 1.0 1.5</b>
Réplicas por combinación: <b>2</b>

Total de corridas: 12
Orden  Temp (°C)  Catalizador (%)  Réplica
    1         60              0.5        1
    2         80              0.5        1
    3         60              1.0        1
    4         60              1.0        2
    5         80              1.5        1
    6         60              1.5        1
    7         80              0.5        2
    8         80              1.5        2
    9         80              1.0        2
   10         80              1.0        1
   11         60              1.5        2
   12         60              0.5        2
</pre>

````python
#@title Solución
import random


def crear_corridas(temperaturas: list[int], concentraciones: list[float],
                   replicas: int) -> list[tuple[int, float, int]]:
    """Genera todas las combinaciones de factores con sus réplicas."""
    corridas = []
    for temperatura in temperaturas:
        for concentracion in concentraciones:
            for replica in range(1, replicas + 1):
                corridas.append((temperatura, concentracion, replica))
    return corridas


def aleatorizar(corridas: list) -> list:
    """Retorna una copia de las corridas en orden aleatorio."""
    copia = list(corridas)
    random.shuffle(copia)
    return copia


def main() -> None:
    random.seed(2026)

    temperaturas = []
    for texto in input("Temperaturas (°C): ").split():
        temperaturas.append(int(texto))
    concentraciones = []
    for texto in input("Concentraciones de catalizador (%): ").split():
        concentraciones.append(float(texto))
    replicas = int(input("Réplicas por combinación: "))

    corridas = crear_corridas(temperaturas, concentraciones, replicas)
    orden = aleatorizar(corridas)

    print(f"\nTotal de corridas: {len(corridas)}")
    print("Orden  Temp (°C)  Catalizador (%)  Réplica")
    for numero, (temperatura, concentracion, replica) in enumerate(orden, start=1):
        print(f"{numero:5d}  {temperatura:9d}  {concentracion:15.1f}  {replica:7d}")


main()
````

**Explicación**

`random.shuffle` **modifica** la lista que recibe y retorna `None`; por eso se baraja una copia (`list(corridas)`) y se retorna la copia. Un error común es escribir `orden = random.shuffle(corridas)`, que deja `orden` con el valor `None`.

## Ejercicio 9.20 — Recorrido de un dron de inspección

Un dron inspecciona una línea de transmisión eléctrica y registra puntos GPS con la hora de cada uno. La distancia entre dos puntos de la superficie terrestre se calcula con la **fórmula de haversine**:

$$a = \sin^2\left(\frac{\Delta\varphi}{2}\right) + \cos\varphi_1 \cos\varphi_2 \sin^2\left(\frac{\Delta\lambda}{2}\right) \qquad d = 2R \cdot \text{asin}(\sqrt{a})$$

donde $\varphi$ es la latitud, $\lambda$ la longitud (ambas en **radianes**) y $R = 6371$ km.

Escriba un programa en Python que importe `math` y `datetime` (de la biblioteca `datetime`) y:

A. Defina `calcular_distancia_km(punto_a: tuple[float, float], punto_b: tuple[float, float]) -> float`.

B. Defina `leer_registro() -> tuple[datetime, tuple[float, float]]`, que lee una línea con el formato `HH:MM:SS lat lon`.

C. En `main()`, solicite la **fecha** del vuelo y la cantidad de registros, lea los registros y muestre, para cada tramo, la distancia (m) y la velocidad (km/h). Al final muestre la distancia total (km, 3 decimales), la duración del vuelo (`HH:MM:SS`, usando la resta de `datetime`), la velocidad promedio y la velocidad **máxima**. Si algún tramo supera los **60 km/h**, advierta que se excedió la velocidad de inspección.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Fecha del vuelo (DD/MM/AAAA): <b>12/06/2026</b>
Cantidad de registros: <b>4</b>
Registro (HH:MM:SS lat lon): <b>09:15:00 9.93420 -84.08750</b>
Registro (HH:MM:SS lat lon): <b>09:16:10 9.93890 -84.08110</b>
Registro (HH:MM:SS lat lon): <b>09:17:05 9.94510 -84.07560</b>
Registro (HH:MM:SS lat lon): <b>09:17:50 9.94980 -84.06890</b>

Tramo 1: 874.4 m a 45.0 km/h
Tramo 2: 915.5 m a 59.9 km/h
Tramo 3: 900.9 m a 72.1 km/h

Distancia total: 2.691 km
Duración del vuelo: 0:02:50
Velocidad promedio: 57.0 km/h (máxima 72.1 km/h)
Advertencia: se excedió la velocidad de inspección.
</pre>

````python
#@title Solución
import math
from datetime import datetime

RADIO_TIERRA_KM = 6371
VELOCIDAD_MAXIMA_KMH = 60


def calcular_distancia_km(punto_a: tuple[float, float], punto_b: tuple[float, float]) -> float:
    """Distancia (km) entre dos puntos GPS con la fórmula de haversine."""
    latitud_a, longitud_a = map(math.radians, punto_a)
    latitud_b, longitud_b = map(math.radians, punto_b)
    delta_latitud = latitud_b - latitud_a
    delta_longitud = longitud_b - longitud_a
    a = (math.sin(delta_latitud / 2) ** 2
         + math.cos(latitud_a) * math.cos(latitud_b) * math.sin(delta_longitud / 2) ** 2)
    return 2 * RADIO_TIERRA_KM * math.asin(math.sqrt(a))


def leer_registro(fecha: str) -> tuple[datetime, tuple[float, float]]:
    """Lee un registro 'HH:MM:SS lat lon' y lo combina con la fecha del vuelo."""
    hora, latitud, longitud = input("Registro (HH:MM:SS lat lon): ").split()
    momento = datetime.strptime(f"{fecha} {hora}", "%d/%m/%Y %H:%M:%S")
    return momento, (float(latitud), float(longitud))


def main() -> None:
    fecha = input("Fecha del vuelo (DD/MM/AAAA): ")
    cantidad = int(input("Cantidad de registros: "))

    registros = []
    for _ in range(cantidad):
        registros.append(leer_registro(fecha))

    distancia_total = 0
    velocidad_maxima = 0
    print()
    for i in range(1, len(registros)):
        momento_anterior, punto_anterior = registros[i - 1]
        momento, punto = registros[i]
        distancia = calcular_distancia_km(punto_anterior, punto)
        horas = (momento - momento_anterior).total_seconds() / 3600
        velocidad = distancia / horas
        distancia_total += distancia
        velocidad_maxima = max(velocidad_maxima, velocidad)
        print(f"Tramo {i}: {distancia * 1000:.1f} m a {velocidad:.1f} km/h")

    duracion = registros[-1][0] - registros[0][0]
    velocidad_promedio = distancia_total / (duracion.total_seconds() / 3600)

    print(f"\nDistancia total: {distancia_total:.3f} km")
    print(f"Duración del vuelo: {duracion}")
    print(f"Velocidad promedio: {velocidad_promedio:.1f} km/h (máxima {velocidad_maxima:.1f} km/h)")
    if velocidad_maxima > VELOCIDAD_MAXIMA_KMH:
        print("Advertencia: se excedió la velocidad de inspección.")


main()
````

**Explicación**

`map(math.radians, punto_a)` aplica la conversión a **cada** elemento de la tupla, y el resultado se desempaqueta en dos variables. Restar dos `datetime` da un `timedelta` que, al imprimirse, ya aparece con el formato `H:MM:SS`. Este programa combina dos bibliotecas estándar para resolver un problema real sin programar trigonometría ni aritmética de fechas desde cero.
