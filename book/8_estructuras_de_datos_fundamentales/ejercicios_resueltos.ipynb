## Ejercicio 8.1 — Perfil de elevación de una carretera

Un equipo de topografía mide la **elevación** (m) de una carretera en estaciones separadas cada **20 m**. La pendiente de cada tramo entre dos estaciones consecutivas es:

$$\text{pendiente}_i = \frac{\text{elevación}_{i+1} - \text{elevación}_i}{20} \cdot 100 \ (\%)$$

Escriba un programa en Python con las siguientes funciones:

A. `leer_elevaciones() -> list[float]`: solicita las elevaciones **en una sola línea**, separadas por espacios, y retorna una lista de `float`.

B. `calcular_pendientes(elevaciones: list[float], separacion: float) -> list[float]`: retorna una lista **nueva** con la pendiente de cada tramo (tendrá un elemento menos que las elevaciones).

C. `tramos_empinados(pendientes: list[float], limite: float) -> list[int]`: retorna los **números de tramo** (empezando en 1) cuya pendiente, en valor absoluto, supera el límite.

D. `main() -> None`: muestra las pendientes (redondeadas a 1 decimal), la máxima subida, la máxima bajada, el desnivel total y los tramos que superan el **10 %**. Si alguna elevación no es un número, muestra `Error: las elevaciones deben ser números separados por espacios.` y **finaliza** el programa (no las vuelve a solicitar); si se ingresan menos de dos elevaciones, muestra `Error: se requieren al menos dos elevaciones.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Elevaciones (m): <b>1150.0 1151.2 1153.8 1156.5 1155.9 1153.4 1152.8</b>

Pendientes (%): [6.0, 13.0, 13.5, -3.0, -12.5, -3.0]
Máxima subida: 13.5%
Máxima bajada: -12.5%
Desnivel total: 2.80 m
Tramos con más de 10%: [2, 3, 5]
</pre>

````python
#@title Solución
SEPARACION_M = 20
PENDIENTE_LIMITE = 10


def leer_elevaciones() -> list[float]:
    """Lee las elevaciones separadas por espacios y las retorna como flotantes."""
    textos = input("Elevaciones (m): ").split()
    elevaciones = []
    for texto in textos:
        elevaciones.append(float(texto))
    return elevaciones


def calcular_pendientes(elevaciones: list[float], separacion: float) -> list[float]:
    """Calcula la pendiente (%) de cada tramo entre estaciones consecutivas."""
    pendientes = []
    for i in range(len(elevaciones) - 1):
        pendientes.append((elevaciones[i + 1] - elevaciones[i]) / separacion * 100)
    return pendientes


def tramos_empinados(pendientes: list[float], limite: float) -> list[int]:
    """Retorna los números de tramo (desde 1) cuya pendiente supera el límite."""
    tramos = []
    for numero, pendiente in enumerate(pendientes, start=1):
        if abs(pendiente) > limite:
            tramos.append(numero)
    return tramos


def main() -> None:
    try:
        elevaciones = leer_elevaciones()

        if len(elevaciones) < 2:
            print("Error: se requieren al menos dos elevaciones.")
        else:
            pendientes = calcular_pendientes(elevaciones, SEPARACION_M)

            pendientes_redondeadas = []
            for pendiente in pendientes:
                pendientes_redondeadas.append(round(pendiente, 1))

            print(f"\nPendientes (%): {pendientes_redondeadas}")
            print(f"Máxima subida: {max(pendientes):.1f}%")
            print(f"Máxima bajada: {min(pendientes):.1f}%")
            print(f"Desnivel total: {elevaciones[-1] - elevaciones[0]:.2f} m")
            print(f"Tramos con más de {PENDIENTE_LIMITE}%: {tramos_empinados(pendientes, PENDIENTE_LIMITE)}")
    except ValueError:
        print("Error: las elevaciones deben ser números separados por espacios.")


main()
````

**Explicación**

El rango del ciclo en `calcular_pendientes` es `len(elevaciones) - 1` porque cada tramo usa la estación `i` **y la siguiente**; si se recorriera hasta el final, `elevaciones[i + 1]` produciría un `IndexError`. `enumerate(..., start=1)` numera los tramos desde 1, que es como los reportaría un ingeniero. Como todas las elevaciones se digitan en una sola línea, lo razonable es **terminar** el programa con un mensaje si alguna no es un número: volver a pedir la línea completa sería lo mismo que ejecutar el programa otra vez. La revisión de `len(elevaciones) < 2` evita además un `ValueError` distinto, el de `max()` sobre una lista vacía.

## Ejercicio 8.2 — Caudales mensuales de un río

Una empresa hidroeléctrica analiza los **caudales promedio mensuales** (m³/s) de un río durante un año, de enero a diciembre. En Costa Rica, la **época lluviosa** va de **mayo a noviembre**. Por normativa ambiental, el río debe conservar un **caudal ecológico** mínimo que no se puede turbinar.

Escriba un programa en Python con las siguientes funciones:

A. `promedio(datos: list[float]) -> float`

B. `meses_criticos(caudales: list[float], meses: list[str], caudal_ecologico: float) -> list[str]`: retorna los **nombres** de los meses cuyo caudal es **menor** que el caudal ecológico.

C. `main() -> None`: usa la lista `MESES = ["Ene", "Feb", ..., "Dic"]`, solicita los 12 caudales en una sola línea y el caudal ecológico, y muestra:
   1. El promedio anual y el promedio de la época lluviosa (obtenga los meses de mayo a noviembre con **slicing**).
   2. El mes de caudal **máximo** y el de caudal **mínimo** (use `max()`, `min()` y el método `index()`).
   3. Los meses críticos.

Si no se ingresan exactamente 12 caudales, muestre un error. Si alguno de los datos no es un número, muestre `Error: los caudales y el caudal ecológico deben ser números.` y **finalice** el programa (no los vuelva a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Caudales de enero a diciembre (m³/s): <b>8.2 6.9 5.4 5.1 12.6 24.3 21.8 26.5 38.7 45.2 30.4 14.1</b>
Caudal ecológico (m³/s): <b>6</b>

Promedio anual: 19.93 m³/s
Promedio época lluviosa: 28.50 m³/s
Caudal máximo: 45.2 m³/s en Oct
Caudal mínimo: 5.1 m³/s en Abr
Meses críticos: [&#x27;Mar&#x27;, &#x27;Abr&#x27;]
</pre>

<pre>
Caudales de enero a diciembre (m³/s): <b>8.2 6.9 5.4</b>
Error: se esperaban 12 caudales y se recibieron 3.
</pre>

**Pregunta de respuesta corta**

¿Por qué el *slice* de mayo a noviembre es `caudales[4:11]` y no `caudales[5:11]`?

````python
#@title Solución
MESES = ["Ene", "Feb", "Mar", "Abr", "May", "Jun", "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]


def promedio(datos: list[float]) -> float:
    """Calcula el promedio de una lista de números."""
    return sum(datos) / len(datos)


def meses_criticos(caudales: list[float], meses: list[str], caudal_ecologico: float) -> list[str]:
    """Retorna los meses cuyo caudal es menor que el caudal ecológico."""
    criticos = []
    for mes, caudal in zip(meses, caudales):
        if caudal < caudal_ecologico:
            criticos.append(mes)
    return criticos


def main() -> None:
    try:
        caudales = []
        for texto in input("Caudales de enero a diciembre (m³/s): ").split():
            caudales.append(float(texto))

        if len(caudales) != len(MESES):
            print(f"Error: se esperaban 12 caudales y se recibieron {len(caudales)}.")
        else:
            caudal_ecologico = float(input("Caudal ecológico (m³/s): "))
            epoca_lluviosa = caudales[4:11]

            mes_maximo = MESES[caudales.index(max(caudales))]
            mes_minimo = MESES[caudales.index(min(caudales))]

            print(f"\nPromedio anual: {promedio(caudales):.2f} m³/s")
            print(f"Promedio época lluviosa: {promedio(epoca_lluviosa):.2f} m³/s")
            print(f"Caudal máximo: {max(caudales)} m³/s en {mes_maximo}")
            print(f"Caudal mínimo: {min(caudales)} m³/s en {mes_minimo}")
            print(f"Meses críticos: {meses_criticos(caudales, MESES, caudal_ecologico)}")
    except ValueError:
        print("Error: los caudales y el caudal ecológico deben ser números.")


main()
````

**Explicación**

**Respuesta corta:** los índices empiezan en 0, así que mayo (el mes 5) está en el índice **4**; el límite superior del *slice* **no se incluye**, así que `11` hace que el último elemento sea el índice 10, que corresponde a noviembre.

## Ejercicio 8.3 — Filtro de media móvil para vibraciones

Un acelerómetro instalado en un motor mide la **amplitud de vibración** (mm/s) una vez por segundo. Las lecturas tienen ruido, así que se suavizan con una **media móvil de 3 puntos**: cada valor suavizado es el promedio de una lectura y sus dos siguientes.

Escriba un programa en Python con las siguientes funciones:

A. `media_movil(datos: list[float], ventana: int) -> list[float]`: retorna una lista nueva donde el elemento `i` es el promedio de `datos[i:i + ventana]` (use *slicing* y `sum()`). La lista resultante tiene `len(datos) - ventana + 1` elementos.

B. `encontrar_picos(datos: list[float]) -> list[int]`: retorna los **índices** de los picos, es decir, los valores estrictamente mayores que su vecino izquierdo **y** que su vecino derecho (el primero y el último no pueden ser picos).

C. `main() -> None`: solicita las lecturas en una sola línea y el **límite de alarma** (mm/s). Muestra la señal suavizada (1 decimal), los índices de los picos de la señal suavizada y si **algún** valor suavizado supera el límite (en ese caso, indica el segundo en que ocurrió por primera vez). Si alguna lectura o el límite no son números, muestra `Error: las lecturas y el límite deben ser números.` y **finaliza** el programa (no los vuelve a solicitar); si hay menos de 3 lecturas, muestra `Error: se requieren al menos 3 lecturas.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Lecturas de vibración (mm/s): <b>2.1 2.8 2.3 3.9 4.6 4.2 5.8 7.1 6.4 4.9 3.8</b>
Límite de alarma (mm/s): <b>5.5</b>

Señal suavizada: [2.4, 3.0, 3.6, 4.2, 4.9, 5.7, 6.4, 6.1, 5.0]
Picos en los índices: [6]
ALARMA: se superó el límite por primera vez en el segundo 5.
</pre>

````python
#@title Solución
VENTANA = 3


def media_movil(datos: list[float], ventana: int) -> list[float]:
    """Suaviza una señal con una media móvil del tamaño de la ventana."""
    suavizados = []
    for i in range(len(datos) - ventana + 1):
        suavizados.append(sum(datos[i:i + ventana]) / ventana)
    return suavizados


def encontrar_picos(datos: list[float]) -> list[int]:
    """Retorna los índices de los valores mayores que sus dos vecinos."""
    picos = []
    for i in range(1, len(datos) - 1):
        if datos[i] > datos[i - 1] and datos[i] > datos[i + 1]:
            picos.append(i)
    return picos


def main() -> None:
    try:
        lecturas = []
        for texto in input("Lecturas de vibración (mm/s): ").split():
            lecturas.append(float(texto))

        if len(lecturas) < VENTANA:
            print(f"Error: se requieren al menos {VENTANA} lecturas.")
        else:
            limite = float(input("Límite de alarma (mm/s): "))

            suavizada = media_movil(lecturas, VENTANA)

            suavizada_redondeada = []
            for valor in suavizada:
                suavizada_redondeada.append(round(valor, 1))

            print(f"\nSeñal suavizada: {suavizada_redondeada}")
            print(f"Picos en los índices: {encontrar_picos(suavizada)}")

            segundo_alarma = -1
            indice = 0
            while indice < len(suavizada) and segundo_alarma == -1:
                if suavizada[indice] > limite:
                    segundo_alarma = indice
                indice += 1

            if segundo_alarma >= 0:
                print(f"ALARMA: se superó el límite por primera vez en el segundo {segundo_alarma}.")
            else:
                print("Vibración dentro del límite.")
    except ValueError:
        print("Error: las lecturas y el límite deben ser números.")


main()
````

**Explicación**

La búsqueda de la primera alarma usa un `while` con dos condiciones en lugar de `break`: el ciclo termina cuando se recorren todos los valores **o** cuando ya se encontró uno que supera el límite. Observe que el filtro reduce el ruido: la señal original tiene más "picos" que la suavizada.

````python
#@title Solución
CAPACIDAD_PILA = 5
PESO_MINIMO = 0.1
PESO_MAXIMO = 40


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def apilar(pila: list[tuple[str, float]], codigo: str, peso: float) -> bool:
    """Coloca una bobina arriba de la pila si hay espacio."""
    apilada = len(pila) < CAPACIDAD_PILA
    if apilada:
        pila.append((codigo, peso))
    return apilada


def retirar(pila: list[tuple[str, float]]) -> tuple[str, float]:
    """Quita y retorna la bobina de arriba de la pila."""
    return pila.pop()


def bobinas_encima(pila: list[tuple[str, float]], codigo: str) -> int:
    """Cuenta cuántas bobinas hay encima de la bobina buscada, o -1 si no está."""
    codigos = []
    for codigo_bobina, peso in pila:
        codigos.append(codigo_bobina)

    encima = -1
    if codigo in codigos:
        encima = len(codigos) - 1 - codigos.index(codigo)
    return encima


def mostrar_pila(pila: list[tuple[str, float]]) -> None:
    """Muestra los códigos de la pila (de abajo hacia arriba) y su peso total."""
    peso_total = 0
    codigos = []
    for codigo, peso in pila:
        codigos.append(codigo)
        peso_total += peso
    print(f"Pila (abajo -> arriba): {codigos} | Peso total: {peso_total:.1f} t")


def main() -> None:
    pila = []
    opcion = ""
    while opcion != "4":
        opcion = input("\n1. Apilar  2. Retirar  3. Buscar  4. Salir\nOpción: ")
        if opcion == "1":
            codigo = input("Código de la bobina: ").upper()
            peso = pedir_numero_seguro("Peso (t): ", PESO_MINIMO, PESO_MAXIMO)
            if not apilar(pila, codigo, peso):
                print("Pila llena: no se puede apilar.")
            mostrar_pila(pila)
        elif opcion == "2":
            if len(pila) == 0:
                print("La pila está vacía.")
            else:
                codigo, peso = retirar(pila)
                print(f"Se retiró la bobina {codigo} ({peso} t).")
            mostrar_pila(pila)
        elif opcion == "3":
            codigo = input("Código a buscar: ").upper()
            encima = bobinas_encima(pila, codigo)
            if encima == -1:
                print(f"La bobina {codigo} no está en la pila.")
            else:
                print(f"Hay que retirar {encima} bobina(s) antes de sacar {codigo}.")
        elif opcion != "4":
            print("Opción inválida.")


main()
````

**Explicación**

Una **pila** (*stack*) sigue la regla "el último en entrar es el primero en salir". Con una lista, `append()` agrega al final y `pop()` sin argumentos quita del final, así que ambas operaciones trabajan sobre la "parte de arriba". Se usan **tuplas** para cada bobina porque su código y su peso no deberían cambiar mientras está almacenada.

## Ejercicio 8.5 — Levantamiento topográfico de un lote

Un topógrafo mide las **coordenadas** $(x, y)$ en metros de los vértices de un lote, recorriéndolos en orden. El **área** del polígono se calcula con la fórmula de Gauss (del zapato):

$$A = \frac{1}{2} \left| \sum_{i=0}^{n-1} (x_i \cdot y_{i+1} - x_{i+1} \cdot y_i) \right|$$

donde el vértice siguiente al último es el **primero** (el polígono se cierra). El **perímetro** es la suma de las distancias entre vértices consecutivos, incluyendo del último al primero.

Escriba un programa en Python con las siguientes funciones:

A. `pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int`: muestra el mensaje y lee un número entero; si no lo es o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar**.

B. `leer_vertices() -> list[tuple[float, float]]`: solicita con `pedir_entero_seguro` la cantidad de vértices (de 3 a 50) y luego cada vértice en el formato `x,y`, **volviendo a solicitar** el vértice mientras el formato sea inválido (use `try`/`except`).

C. `calcular_distancia(punto_a: tuple[float, float], punto_b: tuple[float, float]) -> float`: **desempaque** cada tupla en sus coordenadas.

D. `calcular_perimetro(vertices: list[tuple[float, float]]) -> float` y `calcular_area(vertices: list[tuple[float, float]]) -> float`: use `(i + 1) % len(vertices)` para obtener el vértice siguiente.

E. `main() -> None`: muestra el perímetro (m, 2 decimales), el área en m² y en **hectáreas** (1 ha = 10 000 m², 4 decimales).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de vértices: <b>5</b>
Vértice 1 (x,y): <b>0 0</b>
Error: escriba las coordenadas en el formato x,y (por ejemplo, 12.5,8).
Vértice 1 (x,y): <b>0,0</b>
Vértice 2 (x,y): <b>42.5,0</b>
Vértice 3 (x,y): <b>48,31.2</b>
Vértice 4 (x,y): <b>20.3,45.8</b>
Vértice 5 (x,y): <b>-4.6,28</b>

Vértices: [(0.0, 0.0), (42.5, 0.0), (48.0, 31.2), (20.3, 45.8), (-4.6, 28.0)]
Perímetro: 164.48 m
Área: 1835.06 m² (0.1835 ha)
</pre>

**Pregunta de respuesta corta**

¿Qué ocurriría si se intentara corregir un vértice con `vertices[0][1] = 2.5`? ¿Cómo se haría correctamente?

````python
#@title Solución
M2_POR_HECTAREA = 10000


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def leer_vertices() -> list[tuple[float, float]]:
    """Solicita los vértices del lote en el formato x,y."""
    cantidad = pedir_entero_seguro("Cantidad de vértices: ", 3, 50)

    vertices = []
    for numero in range(1, cantidad + 1):
        vertice_valido = False
        while not vertice_valido:
            try:
                x_texto, y_texto = input(f"Vértice {numero} (x,y): ").split(",")
                vertices.append((float(x_texto), float(y_texto)))
                vertice_valido = True
            except ValueError:
                print("Error: escriba las coordenadas en el formato x,y (por ejemplo, 12.5,8).")
    return vertices


def calcular_distancia(punto_a: tuple[float, float], punto_b: tuple[float, float]) -> float:
    """Calcula la distancia euclidiana entre dos puntos."""
    x_a, y_a = punto_a
    x_b, y_b = punto_b
    return ((x_b - x_a) ** 2 + (y_b - y_a) ** 2) ** 0.5


def calcular_perimetro(vertices: list[tuple[float, float]]) -> float:
    """Suma las distancias entre vértices consecutivos, cerrando el polígono."""
    perimetro = 0
    for i in range(len(vertices)):
        siguiente = (i + 1) % len(vertices)
        perimetro += calcular_distancia(vertices[i], vertices[siguiente])
    return perimetro


def calcular_area(vertices: list[tuple[float, float]]) -> float:
    """Calcula el área del polígono con la fórmula de Gauss."""
    suma = 0
    for i in range(len(vertices)):
        x_actual, y_actual = vertices[i]
        x_siguiente, y_siguiente = vertices[(i + 1) % len(vertices)]
        suma += x_actual * y_siguiente - x_siguiente * y_actual
    return abs(suma) / 2


def main() -> None:
    vertices = leer_vertices()
    area = calcular_area(vertices)

    print(f"\nVértices: {vertices}")
    print(f"Perímetro: {calcular_perimetro(vertices):.2f} m")
    print(f"Área: {area:.2f} m² ({area / M2_POR_HECTAREA:.4f} ha)")


main()
````

**Explicación**

El operador `%` hace que, para el último vértice, el "siguiente" sea el índice 0, cerrando el polígono sin escribir un caso especial.

**Respuesta corta:** se produciría un **`TypeError`**, porque las tuplas son **inmutables**: no se puede cambiar uno de sus elementos. Hay que reemplazar la tupla completa por una nueva: `vertices[0] = (vertices[0][0], 2.5)` (la **lista** sí es mutable).

El vértice mal escrito se **vuelve a solicitar** en lugar de terminar el programa: si un levantamiento de 20 vértices se cayera en el último, habría que digitarlos todos otra vez.

## Ejercicio 8.6 — Resumen estadístico de un sensor

Una estación de monitoreo de **calidad del agua** registra el **pH** de un río cada hora durante un día. La normativa exige que el pH se mantenga entre **6.5 y 8.5**.

Escriba un programa en Python con las siguientes funciones:

A. `calcular_resumen(datos: list[float]) -> tuple[float, float, float, float]`: retorna, **empaquetados en una tupla**, el mínimo, el máximo, el promedio y la **desviación estándar** $\sigma = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n}}$.

B. `contar_fuera_de_rango(datos: list[float], minimo: float, maximo: float) -> tuple[int, int]`: retorna cuántas lecturas están **por debajo** del mínimo y cuántas **por encima** del máximo.

C. `main() -> None`: solicita las lecturas en una sola línea, **desempaqueta** los resultados de ambas funciones en variables con nombres descriptivos y muestra el reporte (2 decimales) y el **porcentaje de cumplimiento** (lecturas dentro de la norma). Si alguna lectura no es un número, muestra `Error: las lecturas de pH deben ser números.` y **finaliza** el programa (no las vuelve a solicitar); si no se ingresa ninguna lectura, muestra `Error: no se ingresaron lecturas.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Lecturas de pH: <b>7.2 7.4 7.1 6.9 6.4 6.2 6.8 7.5 8.1 8.7 8.9 8.3</b>

Lecturas: 12
pH mínimo: 6.20  pH máximo: 8.90
pH promedio: 7.46 ± 0.84
Lecturas ácidas (&lt; 6.5): 2
Lecturas alcalinas (&gt; 8.5): 2
Cumplimiento de la norma: 66.7%
</pre>

````python
#@title Solución
PH_MINIMO = 6.5
PH_MAXIMO = 8.5


def calcular_resumen(datos: list[float]) -> tuple[float, float, float, float]:
    """Retorna el mínimo, el máximo, el promedio y la desviación estándar."""
    promedio = sum(datos) / len(datos)
    suma_cuadrados = 0
    for dato in datos:
        suma_cuadrados += (dato - promedio) ** 2
    desviacion = (suma_cuadrados / len(datos)) ** 0.5
    return min(datos), max(datos), promedio, desviacion


def contar_fuera_de_rango(datos: list[float], minimo: float, maximo: float) -> tuple[int, int]:
    """Cuenta las lecturas por debajo del mínimo y por encima del máximo."""
    por_debajo = 0
    por_encima = 0
    for dato in datos:
        if dato < minimo:
            por_debajo += 1
        elif dato > maximo:
            por_encima += 1
    return por_debajo, por_encima


def main() -> None:
    try:
        lecturas = []
        for texto in input("Lecturas de pH: ").split():
            lecturas.append(float(texto))

        if len(lecturas) == 0:
            print("Error: no se ingresaron lecturas.")
        else:
            ph_minimo, ph_maximo, ph_promedio, desviacion = calcular_resumen(lecturas)
            acidas, alcalinas = contar_fuera_de_rango(lecturas, PH_MINIMO, PH_MAXIMO)
            cumplimiento = (len(lecturas) - acidas - alcalinas) / len(lecturas) * 100

            print(f"\nLecturas: {len(lecturas)}")
            print(f"pH mínimo: {ph_minimo:.2f}  pH máximo: {ph_maximo:.2f}")
            print(f"pH promedio: {ph_promedio:.2f} ± {desviacion:.2f}")
            print(f"Lecturas ácidas (< {PH_MINIMO}): {acidas}")
            print(f"Lecturas alcalinas (> {PH_MAXIMO}): {alcalinas}")
            print(f"Cumplimiento de la norma: {cumplimiento:.1f}%")
    except ValueError:
        print("Error: las lecturas de pH deben ser números.")


main()
````

**Explicación**

`return min(datos), max(datos), promedio, desviacion` **empaqueta** automáticamente los cuatro valores en una tupla, y `ph_minimo, ph_maximo, ph_promedio, desviacion = ...` la **desempaqueta**. La cantidad de variables a la izquierda debe coincidir exactamente con la cantidad de elementos; si no, se produce un `ValueError`.

<span id="ejercicio-8-7-verificaci-n-de-un-lote-de-resistencias"></span>

## Ejercicio 8.7 — Verificación de un lote de resistencias

Un técnico de control de calidad mide con un multímetro las resistencias de un lote y las compara con sus valores **nominales**. Cada resistencia tiene su propia **tolerancia** (%). Los datos se registran en tres listas paralelas:

```python
nominales = [100, 220, 470, 1000, 2200, 4700]
tolerancias = [5, 5, 1, 1, 5, 10]
```

Escriba un programa en Python con las siguientes funciones:

A. `calcular_error(medido: float, nominal: float) -> float`: error porcentual con signo.

B. `verificar_lote(nominales: list[float], tolerancias: list[float], medidos: list[float]) -> list[int]`: recorre las tres listas **a la vez con `zip()`**, imprime una tabla con el número de resistencia (con `enumerate()`), el nominal, el medido, el error (2 decimales, con signo) y `OK` o `RECHAZO`, y retorna los números de las resistencias rechazadas.

C. `main() -> None`: solicita los valores **medidos** en una sola línea (deben ser tantos como nominales), verifica el lote y muestra cuántas se rechazaron. Si el rechazo supera el **20 %**, muestra `Lote rechazado: devolver al proveedor.` Si algún valor medido no es un número, muestra `Error: los valores medidos deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Valores medidos (Ω): <b>98.7 231.5 468.2 1012.4 2176 5020</b>

  #   Nominal   Medido     Error  Estado
  1     100.0     98.7    -1.30%  OK
  2     220.0    231.5    +5.23%  RECHAZO
  3     470.0    468.2    -0.38%  OK
  4    1000.0   1012.4    +1.24%  RECHAZO
  5    2200.0   2176.0    -1.09%  OK
  6    4700.0   5020.0    +6.81%  OK

Rechazadas: 2 [2, 4] (33.3%)
Lote rechazado: devolver al proveedor.
</pre>

````python
#@title Solución
NOMINALES = [100, 220, 470, 1000, 2200, 4700]
TOLERANCIAS = [5, 5, 1, 1, 5, 10]
RECHAZO_MAXIMO = 20


def calcular_error(medido: float, nominal: float) -> float:
    """Calcula el error porcentual de la medición respecto al nominal."""
    return (medido - nominal) / nominal * 100


def verificar_lote(nominales: list[float], tolerancias: list[float], medidos: list[float]) -> list[int]:
    """Imprime la verificación de cada resistencia y retorna las rechazadas."""
    rechazadas = []
    print("\n  #   Nominal   Medido     Error  Estado")
    for numero, (nominal, tolerancia, medido) in enumerate(zip(nominales, tolerancias, medidos), start=1):
        error = calcular_error(medido, nominal)
        if abs(error) <= tolerancia:
            estado = "OK"
        else:
            estado = "RECHAZO"
            rechazadas.append(numero)
        print(f"{numero:3d} {nominal:9.1f} {medido:8.1f} {error:+8.2f}%  {estado}")
    return rechazadas


def main() -> None:
    try:
        medidos = []
        for texto in input("Valores medidos (Ω): ").split():
            medidos.append(float(texto))

        if len(medidos) != len(NOMINALES):
            print(f"Error: se esperaban {len(NOMINALES)} mediciones.")
        else:
            rechazadas = verificar_lote(NOMINALES, TOLERANCIAS, medidos)
            porcentaje_rechazo = len(rechazadas) / len(NOMINALES) * 100

            print(f"\nRechazadas: {len(rechazadas)} {rechazadas} ({porcentaje_rechazo:.1f}%)")
            if porcentaje_rechazo > RECHAZO_MAXIMO:
                print("Lote rechazado: devolver al proveedor.")
            else:
                print("Lote aceptado.")
    except ValueError:
        print("Error: los valores medidos deben ser números.")


main()
````

**Explicación**

`zip()` agrupa los elementos que están en la **misma posición** de cada lista en una tupla, y `enumerate()` agrega el número de posición; por eso el `for` desempaqueta `numero, (nominal, tolerancia, medido)`. Usar listas paralelas requiere cuidado: si una lista tiene un elemento de menos, `zip()` simplemente se detiene en la más corta, sin avisar.

<span id="ejercicio-8-8-cat-logo-de-materiales-estructurales"></span>

## Ejercicio 8.8 — Catálogo de materiales estructurales

Un ingeniero mecánico consulta las propiedades de varios materiales para diseñar un **eje** macizo sometido a tracción. Las propiedades se guardan en un diccionario cuya clave es el nombre del material y cuyo valor es una tupla `(densidad kg/m³, módulo de elasticidad GPa, esfuerzo de fluencia MPa)`:

```python
MATERIALES = {
    "acero": (7850, 200, 250),
    "aluminio": (2700, 69, 95),
    "titanio": (4430, 114, 880),
    "laton": (8500, 100, 200),
}
```

Para una barra de diámetro $d$ (mm), longitud $L$ (m) y carga axial $F$ (kN): $A = \frac{\pi d^2}{4}$, $\sigma = \frac{F}{A}$, $\delta = \frac{F \cdot L}{A \cdot E}$ y el **factor de seguridad** es $FS = \sigma_{fluencia} / \sigma$.

Escriba un programa en Python con las funciones:

A. `analizar_barra(material: str, diametro_mm: float, longitud_m: float, carga_kn: float) -> tuple[float, float, float]`: retorna la masa (kg), el alargamiento (mm) y el factor de seguridad.

B. `main() -> None`: muestra los materiales disponibles (use `", ".join(MATERIALES.keys())`), solicita los datos de la barra y un material, y muestra los resultados. Si el material no está en el diccionario, muestra un error. Si $FS < 2$, muestra `Diseño inseguro: aumente el diámetro o cambie de material.` Si el diámetro, la longitud o la carga no son números, muestra `Error: el diámetro, la longitud y la carga deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Materiales disponibles: acero, aluminio, titanio, laton
Material: <b>Aluminio</b>
Diámetro (mm): <b>30</b>
Longitud (m): <b>1.5</b>
Carga axial (kN): <b>40</b>

Masa: 2.86 kg
Alargamiento: 1.230 mm
Factor de seguridad: 1.68
Diseño inseguro: aumente el diámetro o cambie de material.
</pre>

<pre>
Materiales disponibles: acero, aluminio, titanio, laton
Material: <b>Titanio</b>
Diámetro (mm): <b>30</b>
Longitud (m): <b>1.5</b>
Carga axial (kN): <b>40</b>

Masa: 4.70 kg
Alargamiento: 0.745 mm
Factor de seguridad: 15.55
</pre>

<pre>
Materiales disponibles: acero, aluminio, titanio, laton
Material: <b>cobre</b>
Error: el material &#x27;cobre&#x27; no está en el catálogo.
</pre>

````python
#@title Solución
PI = 3.1416
FACTOR_SEGURIDAD_MINIMO = 2

MATERIALES = {
    "acero": (7850, 200, 250),
    "aluminio": (2700, 69, 95),
    "titanio": (4430, 114, 880),
    "laton": (8500, 100, 200),
}


def analizar_barra(material: str, diametro_mm: float, longitud_m: float,
                   carga_kn: float) -> tuple[float, float, float]:
    """Retorna la masa (kg), el alargamiento (mm) y el factor de seguridad de la barra."""
    densidad, modulo_gpa, fluencia_mpa = MATERIALES[material]

    area_mm2 = PI * diametro_mm ** 2 / 4
    carga_n = carga_kn * 1000
    esfuerzo_mpa = carga_n / area_mm2

    masa_kg = densidad * area_mm2 / 1e6 * longitud_m
    alargamiento_mm = carga_n * longitud_m * 1000 / (area_mm2 * modulo_gpa * 1000)
    factor_seguridad = fluencia_mpa / esfuerzo_mpa
    return masa_kg, alargamiento_mm, factor_seguridad


def main() -> None:
    print(f"Materiales disponibles: {', '.join(MATERIALES.keys())}")
    material = input("Material: ").lower()

    if material not in MATERIALES:
        print(f"Error: el material '{material}' no está en el catálogo.")
    else:
        try:
            diametro = float(input("Diámetro (mm): "))
            longitud = float(input("Longitud (m): "))
            carga = float(input("Carga axial (kN): "))

            masa, alargamiento, factor_seguridad = analizar_barra(material, diametro, longitud, carga)

            print(f"\nMasa: {masa:.2f} kg")
            print(f"Alargamiento: {alargamiento:.3f} mm")
            print(f"Factor de seguridad: {factor_seguridad:.2f}")
            if factor_seguridad < FACTOR_SEGURIDAD_MINIMO:
                print("Diseño inseguro: aumente el diámetro o cambie de material.")
        except ValueError:
            print("Error: el diámetro, la longitud y la carga deben ser números.")


main()
````

**Explicación**

El diccionario permite buscar las propiedades **por nombre** en un solo paso (`MATERIALES[material]`), sin recorrer listas. Antes de acceder, se verifica con `in` que la clave exista: acceder a una clave inexistente produce un **`KeyError`**. Las unidades se convierten con cuidado: N/mm² equivale a MPa y 1 GPa = 1000 MPa.

Un ingeniero realiza una **auditoría energética** en un taller de manufactura para identificar qué equipos consumen más energía. Para cada equipo registra su potencia (kW) y sus horas de uso al mes.

Escriba un programa en Python con las siguientes funciones:

A. `pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float`: muestra el mensaje y lee un número; si no es numérico o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar** (use `while` y `try`/`except`, sin `break`).

B. `registrar_equipos() -> dict[str, float]`: solicita equipos **hasta** que se ingrese un nombre vacío; para cada uno solicita, con `pedir_numero_seguro`, la potencia (de 0.01 a 1000 kW) y las horas de uso al mes (de 0 a 744), de modo que un dato mal digitado se **vuelva a solicitar** sin perder los equipos ya registrados, y guarda en el diccionario la **energía mensual** (kWh) con el nombre como clave. Si el equipo ya estaba registrado, **suma** la energía nueva a la existente.

C. `equipo_de_mayor_consumo(consumos: dict[str, float]) -> str`: recorre el diccionario con `items()` y retorna el nombre del equipo que más consume.

D. `main() -> None`: muestra, para cada equipo, su energía mensual y su **porcentaje** del total; luego el consumo total, el **costo** mensual (₡112 por kWh, separador de miles) y el equipo de mayor consumo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

  Potencia (kW): <b>2.4</b>
  Horas de uso al mes: <b>260</b>
Equipo (Enter para terminar): <b>compresor</b>
  Potencia (kW): <b>15</b>
  Horas de uso al mes: <b>40</b>

````python
#@title Solución
TARIFA_KWH = 112


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def registrar_equipos() -> dict[str, float]:
    """Registra los equipos y su energía mensual (kWh), acumulando repetidos."""
    consumos = {}
    nombre = input("Equipo (Enter para terminar): ").lower()
    while nombre != "":
        potencia = pedir_numero_seguro("  Potencia (kW): ", 0.01, 1000)
        horas = pedir_numero_seguro("  Horas de uso al mes: ", 0, 744)
        energia = potencia * horas
        if nombre in consumos:
            consumos[nombre] += energia
        else:
            consumos[nombre] = energia
        nombre = input("Equipo (Enter para terminar): ").lower()
    return consumos


def equipo_de_mayor_consumo(consumos: dict[str, float]) -> str:
    """Retorna el nombre del equipo con mayor energía mensual."""
    mayor_nombre = ""
    mayor_energia = -1
    for nombre, energia in consumos.items():
        if energia > mayor_energia:
            mayor_nombre = nombre
            mayor_energia = energia
    return mayor_nombre


def main() -> None:
    consumos = registrar_equipos()

    if len(consumos) == 0:
        print("No se registraron equipos.")
    else:
        total = sum(consumos.values())
        print("\n--- Auditoría energética ---")
        for nombre, energia in consumos.items():
            print(f"{nombre.ljust(20)}{energia:10.1f} kWh{energia / total:8.1%}")
        print(f"\nConsumo total: {total:,.1f} kWh")
        print(f"Costo mensual: ₡{total * TARIFA_KWH:,.0f}")
        print(f"Mayor consumidor: {equipo_de_mayor_consumo(consumos)}")


main()
````

**Explicación**

Como el compresor se registró dos veces, su energía se **acumula** en la misma clave (15 × 180 + 15 × 40 = 3300 kWh). Esto funciona porque los nombres se convierten a minúsculas antes de usarlos como clave: para un diccionario, `"Compresor"` y `"compresor"` serían claves distintas.

Se conoce la descripción de cada código:

```python
DESCRIPCIONES = {
    "P-LOW": "Presión baja en la línea",
    "T-HIGH": "Temperatura alta en el motor",
    "V-FAIL": "Falla de válvula",
    "COM-LOST": "Pérdida de comunicación",
}
```

Escriba un programa en Python con las siguientes funciones:

A. `contar_alarmas(bitacora: str) -> dict[str, int]`: separa los códigos (sin espacios extra y en mayúsculas) y cuenta cuántas veces aparece cada uno, usando `dict.get(clave, 0)`.

B. `main() -> None`: solicita la bitácora y muestra cada código con su descripción (o `Código desconocido` si no está en el diccionario) y su frecuencia, **ordenados de mayor a menor frecuencia**. Muestre también el total de alarmas y cuántos códigos **distintos** hubo. Si la bitácora está vacía, muestre `Error: la bitácora está vacía.` y **finalice** el programa (no la vuelva a solicitar).

*Pista:* para ordenar, cree una lista de tuplas `(frecuencia, codigo)` y use `sorted(..., reverse=True)`.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

  2 x T-HIGH    Temperatura alta en el motor

  1 x COM-LOST  Pérdida de comunicación

Total de alarmas: 9

</pre>

````python
#@title Solución
DESCRIPCIONES = {
    "P-LOW": "Presión baja en la línea",
    "T-HIGH": "Temperatura alta en el motor",
    "V-FAIL": "Falla de válvula",
    "COM-LOST": "Pérdida de comunicación",
}


def contar_alarmas(bitacora: str) -> dict[str, int]:
    """Cuenta cuántas veces aparece cada código de alarma en la bitácora."""
    frecuencias = {}
    for codigo in bitacora.split(","):
        codigo = codigo.strip().upper()
        frecuencias[codigo] = frecuencias.get(codigo, 0) + 1
    return frecuencias


def main() -> None:
    bitacora = input("Bitácora de alarmas: ")

    if bitacora.strip() == "":
        print("Error: la bitácora está vacía.")
    else:
        frecuencias = contar_alarmas(bitacora)

        pares = []
        for codigo, frecuencia in frecuencias.items():
            pares.append((frecuencia, codigo))

        print("\n--- Resumen de alarmas ---")
        for frecuencia, codigo in sorted(pares, reverse=True):
            descripcion = DESCRIPCIONES.get(codigo, "Código desconocido")
            print(f"{frecuencia:3d} x {codigo.ljust(9)} {descripcion}")

        print(f"\nTotal de alarmas: {sum(frecuencias.values())}")
        print(f"Códigos distintos: {len(frecuencias)}")


main()
````

**Explicación**

`frecuencias.get(codigo, 0)` retorna la cuenta actual del código o `0` si todavía no existe, lo que evita escribir un `if` para el primer caso. Las tuplas se ordenan comparando **primero** su primer elemento: por eso se guarda `(frecuencia, codigo)` y no `(codigo, frecuencia)`.

Escriba un programa en Python con las siguientes funciones:

C. `main() -> None`: solicita una nueva lectura (código y lluvia) y la agrega; luego muestra una tabla con cada estación, su provincia, el total, los días secos y el máximo. Finalmente muestra `Alerta de sequía en <provincia>` para cada estación con **6 o más** días secos. ````python
#@title Solución
ESTACIONES = {
    "ST-01": {"provincia": "Limón", "altitud": 15, "lluvia": [45.2, 60.1, 12.0, 0.0, 88.4, 30.5, 22.1]},
    "ST-02": {"provincia": "San José", "altitud": 1150, "lluvia": [0.0, 5.2, 18.6, 22.0, 0.0, 0.0, 3.1]},
    "ST-03": {"provincia": "Guanacaste", "altitud": 120, "lluvia": [0.0, 0.0, 0.0, 2.5, 0.0, 0.0, 0.0]},
}
DIAS_SECOS_ALERTA = 6


def resumir_estacion(datos: dict) -> tuple[float, int, float]:
    """Retorna la lluvia total, los días secos y la lluvia máxima de la semana."""
    lluvias = datos["lluvia"]
    return sum(lluvias), lluvias.count(0.0), max(lluvias)


def agregar_lectura(estaciones: dict, codigo: str, lluvia: float) -> bool:
    """Agrega la lectura más reciente y descarta la más antigua."""
    existe = codigo in estaciones
    if existe:
        estaciones[codigo]["lluvia"].append(lluvia)
        estaciones[codigo]["lluvia"].pop(0)
    return existe


def main() -> None:
    try:
        codigo = input("Estación de la nueva lectura: ").upper()
        lluvia = float(input("Lluvia del día (mm): "))
        if not agregar_lectura(ESTACIONES, codigo, lluvia):
            print(f"La estación {codigo} no existe; no se agregó la lectura.")

        print("\nEstación  Provincia      Total (mm)  Días secos  Máx (mm)")
        for codigo_estacion, datos in ESTACIONES.items():
            total, dias_secos, maximo = resumir_estacion(datos)
            print(f"{codigo_estacion:<10}{datos['provincia']:<15}{total:>10.1f}{dias_secos:>12}{maximo:>10.1f}")

        for datos in ESTACIONES.values():
            if resumir_estacion(datos)[1] >= DIAS_SECOS_ALERTA:
                print(f"Alerta de sequía en {datos['provincia']}")
    except ValueError:
        print("Error: la lluvia debe ser un número.")


main()
````

Para llegar a la lista de lluvias se encadenan dos accesos: `estaciones[codigo]` obtiene el diccionario de la estación y `["lluvia"]` la lista dentro de él. FACTORES_A_METRO = {"mm": 0.001, "cm": 0.01, "m": 1, "km": 1000,

````python
#@title Solución
FACTORES_A_METRO = {"mm": 0.001, "cm": 0.01, "m": 1, "km": 1000,
                    "in": 0.0254, "ft": 0.3048, "yd": 0.9144, "mi": 1609.344}


def convertir(valor: float, origen: str, destino: str) -> float:
    """Convierte un valor entre dos unidades de longitud del diccionario."""
    en_metros = valor * FACTORES_A_METRO[origen]
    return en_metros / FACTORES_A_METRO[destino]


def interpretar_medida(texto: str) -> tuple[float, str]:
    """Separa un texto como '12.5 ft' en su valor numérico y su unidad."""
    valor_texto, unidad = texto.split()
    return float(valor_texto), unidad.lower()


def main() -> None:
    entrada = input("Medida (por ejemplo, 12.5 ft) o 'fin': ")
    while entrada.lower() != "fin":
        try:
            valor, unidad = interpretar_medida(entrada)
            if unidad in FACTORES_A_METRO:
                for destino in FACTORES_A_METRO:
                    print(f"  {convertir(valor, unidad, destino):.4g} {destino}")
            else:
                print(f"  Unidad desconocida. Disponibles: {', '.join(FACTORES_A_METRO)}")
        except ValueError:
            print("  Formato inválido: escriba un número, un espacio y la unidad.")
        entrada = input("Medida (por ejemplo, 12.5 ft) o 'fin': ")


main()
````

**Explicación**

Un integrador de sistemas de **Internet de las Cosas (IoT)** debe verificar qué **protocolos de comunicación** comparten los dispositivos de una planta con la **pasarela** (*gateway*) que los conecta a la nube. Los protocolos de cada equipo son:

PASARELA = {"MQTT", "Modbus", "OPC-UA", "HTTP"}
DISPOSITIVOS = {

Escriba un programa en Python con las siguientes funciones:

Indique si la pasarela ya soporta **todos** los protocolos de la planta (use `issubset()`).

Dispositivos incompatibles: [&#x27;Medidor de energía&#x27;]

````python
#@title Solución
PASARELA = {"MQTT", "Modbus", "OPC-UA", "HTTP"}
DISPOSITIVOS = {
    "Sensor de nivel": {"Modbus", "HART"},
    "Variador": {"Modbus", "Profinet", "OPC-UA"},
    "Cámara térmica": {"RTSP", "HTTP"},
    "Medidor de energía": {"BACnet", "Zigbee"},
}


def protocolos_comunes(dispositivo: set[str], pasarela: set[str]) -> set[str]:
    """Retorna los protocolos que comparten el dispositivo y la pasarela."""
    return dispositivo & pasarela


def analizar_red(dispositivos: dict[str, set[str]], pasarela: set[str]) -> tuple[list[str], set[str]]:
    """Muestra la compatibilidad de cada dispositivo y retorna los incompatibles y todos los protocolos."""
    incompatibles = []
    todos_los_protocolos = set()
    for nombre, protocolos in dispositivos.items():
        todos_los_protocolos = todos_los_protocolos | protocolos
        comunes = protocolos_comunes(protocolos, pasarela)
        if len(comunes) > 0:
            print(f"  {nombre}: {sorted(comunes)}")
        else:
            print(f"  {nombre}: INCOMPATIBLE")
            incompatibles.append(nombre)
    return incompatibles, todos_los_protocolos


def main() -> None:
    pasarela = set(PASARELA)
    expansion = input("Protocolos del módulo de expansión (separados por comas, Enter si no hay): ")
    for protocolo in expansion.split(","):
        if protocolo.strip() != "":
            pasarela.add(protocolo.strip())

    print(f"\nPasarela: {sorted(pasarela)}")
    print("Compatibilidad:")
    incompatibles, todos_los_protocolos = analizar_red(DISPOSITIVOS, pasarela)

    print(f"\nDispositivos incompatibles: {incompatibles}")
    print(f"Protocolos no soportados: {sorted(todos_los_protocolos - pasarela)}")
    if todos_los_protocolos.issubset(pasarela):
        print("La pasarela soporta todos los protocolos de la planta.")


main()
````

**Explicación**

Los conjuntos son la estructura natural para este problema: no importa el orden de los protocolos, no tiene sentido que se repitan y las operaciones que se necesitan (`&` intersección, `|` unión, `-` diferencia) son exactamente las de la teoría de conjuntos. Se trabaja sobre una **copia** (`set(PASARELA)`) para no modificar la configuración original.

Escriba un programa en Python con las siguientes funciones:

A. `leer_bom(nombre_producto: str) -> list[str]`: solicita la lista de materiales y retorna una lista con los códigos en mayúsculas y sin espacios.

B. `comparar_boms(bom_a: list[str], bom_b: list[str]) -> tuple[set[str], set[str], set[str]]`: convierte las listas a conjuntos y retorna los componentes **comunes**, los **exclusivos** del producto A y los exclusivos del B.

C. `main() -> None`: lee las dos listas y muestra, para cada producto, la cantidad total de componentes y la cantidad de componentes **distintos**; luego los comunes (con cuántas unidades de cada uno usa **en total** entre ambos productos, usando `count()`), los exclusivos y el **porcentaje de similitud** $\frac{|A \cap B|}{|A \cup B|} \cdot 100$.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
BOM de la licuadora: <b>mot-120, TOR-M4, TOR-M4, CAP-10u, SW-01, VAS-15, TOR-M4</b>
BOM de la batidora: <b>MOT-90, TOR-M4, TOR-M4, CAP-10u, SW-01, BAT-02</b>

Licuadora: 7 componentes, 5 distintos
Batidora: 6 componentes, 5 distintos

Componentes comunes (unidades totales):
  CAP-10U: 2
  SW-01: 2
  TOR-M4: 5
Solo licuadora: [&#x27;MOT-120&#x27;, &#x27;VAS-15&#x27;]
Solo batidora: [&#x27;BAT-02&#x27;, &#x27;MOT-90&#x27;]
Similitud: 42.9%
</pre>

````python
#@title Solución
def leer_bom(nombre_producto: str) -> list[str]:
    """Lee la lista de materiales de un producto."""
    codigos = []
    for codigo in input(f"BOM de {nombre_producto}: ").split(","):
        codigos.append(codigo.strip().upper())
    return codigos


def comparar_boms(bom_a: list[str], bom_b: list[str]) -> tuple[set[str], set[str], set[str]]:
    """Retorna los componentes comunes y los exclusivos de cada producto."""
    componentes_a = set(bom_a)
    componentes_b = set(bom_b)
    return componentes_a & componentes_b, componentes_a - componentes_b, componentes_b - componentes_a


def main() -> None:
    bom_licuadora = leer_bom("la licuadora")
    bom_batidora = leer_bom("la batidora")

    comunes, solo_licuadora, solo_batidora = comparar_boms(bom_licuadora, bom_batidora)
    union = set(bom_licuadora) | set(bom_batidora)
    similitud = len(comunes) / len(union) * 100

    print(f"\nLicuadora: {len(bom_licuadora)} componentes, {len(set(bom_licuadora))} distintos")
    print(f"Batidora: {len(bom_batidora)} componentes, {len(set(bom_batidora))} distintos")
    print("\nComponentes comunes (unidades totales):")
    for codigo in sorted(comunes):
        unidades = bom_licuadora.count(codigo) + bom_batidora.count(codigo)
        print(f"  {codigo}: {unidades}")
    print(f"Solo licuadora: {sorted(solo_licuadora)}")
    print(f"Solo batidora: {sorted(solo_batidora)}")
    print(f"Similitud: {similitud:.1f}%")


main()
````

**Explicación**

La lista conserva los **repetidos** (necesarios para saber cuántas unidades se compran) y el conjunto los **elimina** (necesario para saber qué componentes distintos hay). Por eso el programa usa ambas estructuras a partir de los mismos datos: elegir la estructura depende de la **pregunta** que se quiere responder.

## Ejercicio 8.15 — Cobertura de detectores de humo

Un ingeniero de protección contra incendios revisa si los **detectores de humo** de un edificio cubren todas sus zonas. Cada detector cubre un conjunto de zonas (identificadas con números del 1 al 12). La norma exige que **todas** las zonas estén cubiertas, y se recomienda que las zonas críticas (cuarto eléctrico: zona **4**; bodega de químicos: zona **9**) estén cubiertas por **al menos dos** detectores.

Escriba un programa en Python con las siguientes funciones:

A. `pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int`: muestra el mensaje y lee un número entero; si no lo es o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar**.

B. `pedir_zonas_seguras(codigo: str) -> set[int]`: lee en una sola línea los números de zona que cubre un detector; si la línea queda vacía o contiene algo que no es un número entero, muestra un error y la **vuelve a solicitar**.

C. `leer_detectores() -> dict[str, set[int]]`: solicita la cantidad de detectores (de 1 a 20) y, para cada uno, su código y las zonas que cubre, **reutilizando** las dos funciones anteriores.

D. `zonas_cubiertas(detectores: dict[str, set[int]]) -> set[int]`: unión de todas las zonas.

E. `contar_cobertura(detectores: dict[str, set[int]], zona: int) -> int`: cuántos detectores cubren una zona.

F. `main() -> None`: muestra las zonas **sin cobertura** (diferencia con el conjunto de zonas 1 a 12, creado con `set(range(1, 13))`), la cantidad de detectores que cubren cada zona crítica y un dictamen final: `CUMPLE`, o `NO CUMPLE` con los motivos.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de detectores: <b>4</b>
Código del detector 1: <b>dh-1</b>
  Zonas que cubre DH-1: <b>1 2 tres 4</b>
  Error: escriba los números de zona separados por espacios.
  Zonas que cubre DH-1: <b>1 2 3 4</b>
Código del detector 2: <b>DH-2</b>
  Zonas que cubre DH-2: <b>3 4 5 6</b>
Código del detector 3: <b>DH-3</b>
  Zonas que cubre DH-3: <b>7 8 9</b>
Código del detector 4: <b>DH-4</b>
  Zonas que cubre DH-4: <b>9 10 12</b>

Zonas sin cobertura: [11]
Zona crítica 4: 2 detector(es)
Zona crítica 9: 2 detector(es)

Dictamen: NO CUMPLE
</pre>

````python
#@title Solución
ZONAS_EDIFICIO = set(range(1, 13))
ZONAS_CRITICAS = {4, 9}
COBERTURA_MINIMA_CRITICA = 2


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def pedir_zonas_seguras(codigo: str) -> set[int]:
    """Pide las zonas que cubre un detector, repitiendo mientras la línea sea inválida."""
    zonas = set()
    while len(zonas) == 0:
        try:
            for texto in input(f"  Zonas que cubre {codigo}: ").split():
                zonas.add(int(texto))
            if len(zonas) == 0:
                print("  Error: indique al menos una zona.")
        except ValueError:
            zonas = set()
            print("  Error: escriba los números de zona separados por espacios.")
    return zonas


def leer_detectores() -> dict[str, set[int]]:
    """Solicita los detectores y el conjunto de zonas que cubre cada uno."""
    detectores = {}
    cantidad = pedir_entero_seguro("Cantidad de detectores: ", 1, 20)
    for numero in range(1, cantidad + 1):
        codigo = input(f"Código del detector {numero}: ").upper()
        detectores[codigo] = pedir_zonas_seguras(codigo)
    return detectores


def zonas_cubiertas(detectores: dict[str, set[int]]) -> set[int]:
    """Retorna la unión de las zonas cubiertas por todos los detectores."""
    cubiertas = set()
    for zonas in detectores.values():
        cubiertas = cubiertas | zonas
    return cubiertas


def contar_cobertura(detectores: dict[str, set[int]], zona: int) -> int:
    """Cuenta cuántos detectores cubren una zona."""
    cantidad = 0
    for zonas in detectores.values():
        if zona in zonas:
            cantidad += 1
    return cantidad


def main() -> None:
    detectores = leer_detectores()

    sin_cobertura = ZONAS_EDIFICIO - zonas_cubiertas(detectores)
    cumple = len(sin_cobertura) == 0

    print(f"\nZonas sin cobertura: {sorted(sin_cobertura)}")
    for zona in sorted(ZONAS_CRITICAS):
        cantidad = contar_cobertura(detectores, zona)
        print(f"Zona crítica {zona}: {cantidad} detector(es)")
        if cantidad < COBERTURA_MINIMA_CRITICA:
            cumple = False
            print(f"  -> Se requieren al menos {COBERTURA_MINIMA_CRITICA} detectores.")

    if cumple:
        print("\nDictamen: CUMPLE")
    else:
        print("\nDictamen: NO CUMPLE")


main()
````

**Explicación**

`set(range(1, 13))` crea el conjunto de **todas** las zonas del edificio, y la diferencia con las cubiertas da, en una sola operación, las zonas que faltan. Hacer lo mismo con listas requeriría un ciclo y una búsqueda con `in` por cada zona. En `pedir_zonas_seguras`, el conjunto se **vacía** antes de volver a preguntar: si no, las zonas que alcanzaron a leerse antes del error quedarían mezcladas con las de la nueva línea.

<span id="ejercicio-8-16-red-de-distribuci-n-entre-plantas"></span>

## Ejercicio 8.16 — Red de distribución entre plantas

Una empresa de alimentos tiene **4 plantas** y conoce la **distancia por carretera** (km) entre cada par, guardada en una **matriz** (lista de listas), donde `DISTANCIAS[i][j]` es la distancia de la planta `i` a la planta `j`:

```python
PLANTAS = ["Cartago", "Heredia", "Liberia", "Puntarenas"]
DISTANCIAS = [
    [0, 45, 250, 135],
    [45, 0, 205, 95],
    [250, 205, 0, 135],
    [135, 95, 135, 0],
]
```

Escriba un programa en Python con las siguientes funciones:

A. `par_mas_cercano(distancias: list[list[int]]) -> tuple[int, int]`: con dos ciclos anidados, retorna los índices del par de plantas **distintas** más cercanas (recorra solo `j > i` para no repetir pares).

B. `distancia_de_ruta(distancias: list[list[int]], ruta: list[int]) -> int`: suma las distancias entre plantas consecutivas de la ruta.

C. `main() -> None`: muestra el par más cercano; luego solicita una **ruta** de reparto como números de planta (1 a 4) separados por espacios, la convierte a índices, verifica que todos sean válidos y muestra los nombres de la ruta unidos con `" -> "` (use `join`), la distancia total y el **costo** de combustible (₡950 por km). Si la ruta contiene algo que no es un número de planta válido, muestra `Ruta inválida: use al menos dos plantas numeradas del 1 al 4.` y **finaliza** el programa (no la vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Plantas más cercanas: Cartago y Heredia (45 km)

Ruta (números de planta 1-4): <b>1 2 4 3 1</b>
Ruta: Cartago -&gt; Heredia -&gt; Puntarenas -&gt; Liberia -&gt; Cartago
Distancia total: 525 km
Costo de combustible: ₡498,750
</pre>

<pre>
Plantas más cercanas: Cartago y Heredia (45 km)

Ruta (números de planta 1-4): <b>1 5</b>
Ruta inválida: use al menos dos plantas numeradas del 1 al 4.
</pre>

````python
#@title Solución
PLANTAS = ["Cartago", "Heredia", "Liberia", "Puntarenas"]
DISTANCIAS = [
    [0, 45, 250, 135],
    [45, 0, 205, 95],
    [250, 205, 0, 135],
    [135, 95, 135, 0],
]
COSTO_POR_KM = 950


def par_mas_cercano(distancias: list[list[int]]) -> tuple[int, int]:
    """Retorna los índices del par de plantas distintas más cercanas."""
    mejor_i, mejor_j = 0, 1
    for i in range(len(distancias)):
        for j in range(i + 1, len(distancias)):
            if distancias[i][j] < distancias[mejor_i][mejor_j]:
                mejor_i, mejor_j = i, j
    return mejor_i, mejor_j


def distancia_de_ruta(distancias: list[list[int]], ruta: list[int]) -> int:
    """Suma las distancias entre plantas consecutivas de la ruta."""
    total = 0
    for k in range(len(ruta) - 1):
        total += distancias[ruta[k]][ruta[k + 1]]
    return total


def main() -> None:
    i, j = par_mas_cercano(DISTANCIAS)
    print(f"Plantas más cercanas: {PLANTAS[i]} y {PLANTAS[j]} ({DISTANCIAS[i][j]} km)")

    ruta = []
    ruta_valida = True
    try:
        for texto in input("\nRuta (números de planta 1-4): ").split():
            indice = int(texto) - 1
            if 0 <= indice < len(PLANTAS):
                ruta.append(indice)
            else:
                ruta_valida = False
    except ValueError:
        ruta_valida = False

    if not ruta_valida or len(ruta) < 2:
        print("Ruta inválida: use al menos dos plantas numeradas del 1 al 4.")
    else:
        nombres = []
        for indice in ruta:
            nombres.append(PLANTAS[indice])
        distancia = distancia_de_ruta(DISTANCIAS, ruta)
        print(f"Ruta: {' -> '.join(nombres)}")
        print(f"Distancia total: {distancia} km")
        print(f"Costo de combustible: ₡{distancia * COSTO_POR_KM:,}")


main()
````

**Explicación**

El ciclo interno empieza en `i + 1` porque la matriz es **simétrica** (la distancia de A a B es la misma que de B a A) y la diagonal es 0: así se revisa cada par una sola vez y se evita comparar una planta consigo misma. El `try`/`except` alrededor del ciclo hace que un texto como `1 dos 3` se trate igual que un número de planta fuera de rango: en ambos casos la ruta es inválida y el programa termina con un solo mensaje.

Escriba un programa en Python con las siguientes funciones:

Si una línea no contiene exactamente tres números, muestra `  Error: escriba tres números: voltaje, corriente y rpm.` y **vuelve a solicitar** la medición (el ciclo continúa; no termina el programa).

Construya cada línea con `",".join(...)` (recuerde que `join` solo acepta cadenas).

````python
#@title Solución
ENCABEZADO = ["punto", "voltaje_V", "corriente_A", "rpm", "potencia_W"]


def leer_mediciones() -> list[tuple[float, float, int]]:
    """Lee mediciones 'voltaje corriente rpm' hasta una línea vacía."""
    mediciones = []
    linea = input("Medición (V A rpm, Enter para terminar): ")
    while linea != "":
        try:
            voltaje, corriente, rpm = linea.split()
            mediciones.append((float(voltaje), float(corriente), int(rpm)))
        except ValueError:
            print("  Error: escriba tres números: voltaje, corriente y rpm.")
        linea = input("Medición (V A rpm, Enter para terminar): ")
    return mediciones


def convertir_a_csv(mediciones: list[tuple[float, float, int]]) -> list[str]:
    """Convierte las mediciones en líneas de texto en formato CSV."""
    lineas = [",".join(ENCABEZADO)]
    for punto, (voltaje, corriente, rpm) in enumerate(mediciones, start=1):
        potencia = round(voltaje * corriente, 1)
        campos = [str(punto), str(voltaje), str(corriente), str(rpm), str(potencia)]
        lineas.append(",".join(campos))
    return lineas


def main() -> None:
    mediciones = leer_mediciones()
    lineas = convertir_a_csv(mediciones)

    print("\n--- mediciones.csv ---")
    print("\n".join(lineas))
    print(f"--- {len(lineas) - 1} filas de datos ---")


main()
````

<span id="ejercicio-8-18-mapa-t-rmico-de-una-placa-electr-nica"></span>

## Ejercicio 8.18 — Mapa térmico de una placa electrónica

Una cámara térmica toma la temperatura (°C) de una **placa electrónica** en una cuadrícula. Cada **fila** de la cuadrícula se ingresa en una línea, con las temperaturas separadas por espacios; todas las filas tienen la misma cantidad de columnas. Un **punto caliente** es una celda que supera los **85 °C**.

Escriba un programa en Python con las siguientes funciones:

A. `leer_matriz(filas: int) -> list[list[float]]`

B. `promedios_por_fila(matriz: list[list[float]]) -> list[float]` y `promedios_por_columna(matriz: list[list[float]]) -> list[float]`

C. `puntos_calientes(matriz: list[list[float]], limite: float) -> list[tuple[int, int]]`: retorna las posiciones `(fila, columna)` de los puntos calientes.

D. `main() -> None`: solicita la cantidad de filas y la matriz, y muestra los promedios por fila y por columna (1 decimal), la temperatura máxima con su posición, y la lista de puntos calientes. Si hay más de **2** puntos calientes, muestra `Rediseñar la disipación térmica.` Si la cantidad de filas o alguna temperatura no son números, muestra `Error: la cantidad de filas y las temperaturas deben ser números.` y **finaliza** el programa (no los vuelve a solicitar); si la cantidad de filas no es mayor que cero, muestra `Error: la cantidad de filas debe ser mayor que cero.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de filas: <b>4</b>
Fila 1: <b>42.1 45.3 51.0 48.2 44.0</b>
Fila 2: <b>47.5 68.9 88.2 71.4 49.3</b>
Fila 3: <b>49.0 79.6 96.5 86.1 52.8</b>
Fila 4: <b>44.2 52.3 60.7 55.1 46.9</b>

Promedios por fila: [46.1, 65.1, 72.8, 51.8]
Promedios por columna: [45.7, 61.5, 74.1, 65.2, 48.2]
Temperatura máxima: 96.5 °C en (2, 2)
Puntos calientes (&gt; 85 °C): [(1, 2), (2, 2), (2, 3)]
Rediseñar la disipación térmica.
</pre>

````python
#@title Solución
LIMITE_CALIENTE = 85
PUNTOS_CALIENTES_MAXIMOS = 2


def leer_matriz(filas: int) -> list[list[float]]:
    """Lee una matriz de temperaturas, una fila por línea."""
    matriz = []
    for numero in range(1, filas + 1):
        fila = []
        for texto in input(f"Fila {numero}: ").split():
            fila.append(float(texto))
        matriz.append(fila)
    return matriz


def promedios_por_fila(matriz: list[list[float]]) -> list[float]:
    """Calcula el promedio de cada fila."""
    promedios = []
    for fila in matriz:
        promedios.append(round(sum(fila) / len(fila), 1))
    return promedios


def promedios_por_columna(matriz: list[list[float]]) -> list[float]:
    """Calcula el promedio de cada columna."""
    promedios = []
    for columna in range(len(matriz[0])):
        suma = 0
        for fila in matriz:
            suma += fila[columna]
        promedios.append(round(suma / len(matriz), 1))
    return promedios


def puntos_calientes(matriz: list[list[float]], limite: float) -> list[tuple[int, int]]:
    """Retorna las posiciones (fila, columna) cuya temperatura supera el límite."""
    posiciones = []
    for i, fila in enumerate(matriz):
        for j, temperatura in enumerate(fila):
            if temperatura > limite:
                posiciones.append((i, j))
    return posiciones


def main() -> None:
    try:
        filas = int(input("Cantidad de filas: "))

        if filas < 1:
            print("Error: la cantidad de filas debe ser mayor que cero.")
        else:
            matriz = leer_matriz(filas)

            maxima = matriz[0][0]
            posicion_maxima = (0, 0)
            for i, fila in enumerate(matriz):
                for j, temperatura in enumerate(fila):
                    if temperatura > maxima:
                        maxima = temperatura
                        posicion_maxima = (i, j)

            calientes = puntos_calientes(matriz, LIMITE_CALIENTE)

            print(f"\nPromedios por fila: {promedios_por_fila(matriz)}")
            print(f"Promedios por columna: {promedios_por_columna(matriz)}")
            print(f"Temperatura máxima: {maxima} °C en {posicion_maxima}")
            print(f"Puntos calientes (> {LIMITE_CALIENTE} °C): {calientes}")
            if len(calientes) > PUNTOS_CALIENTES_MAXIMOS:
                print("Rediseñar la disipación térmica.")
    except ValueError:
        print("Error: la cantidad de filas y las temperaturas deben ser números.")


main()
````

**Explicación**

En una matriz, `matriz[i]` es la **fila** `i` (una lista) y `matriz[i][j]` el elemento de esa fila en la columna `j`. Recorrer por filas es directo (`for fila in matriz`), pero para recorrer por **columnas** hay que fijar el índice de columna y visitar ese elemento en cada fila.

## Ejercicio 8.19 — Trama de datos de una red de sensores

Un concentrador de una red inalámbrica de sensores en un **invernadero** envía cada minuto una **trama** de texto con las lecturas de humedad del suelo (%) de cada sensor, con el formato `ID:valor` separados por punto y coma:

```text
S01:34.5;S02:41.2;S03:18.7;S04:ERR;S05:27.9;S06:52.3
```

Un sensor que reporta `ERR` está dañado.

Escriba un programa en Python con las siguientes funciones:

A. `procesar_trama(trama: str) -> tuple[dict[str, float], list[str]]`: retorna un diccionario `{id: humedad}` con las lecturas válidas y una lista con los ID de los sensores dañados. Use `split(";")` y `split(":")`.

B. `sensores_bajo_umbral(lecturas: dict[str, float], umbral: float) -> list[str]`: retorna los ID cuya humedad es menor que el umbral, **ordenados** alfabéticamente.

C. `main() -> None`: solicita la trama y el umbral de riego, y muestra: la cantidad de lecturas válidas, los sensores dañados, la humedad promedio, los sensores que necesitan **riego** y, si la mitad o más de los sensores válidos necesitan riego, `Activar riego general.`; si no, `Riego localizado.`

Si el umbral no es un número o la trama no respeta el formato `ID:valor`, muestre `Error: revise el formato de la trama y el umbral.` y **finalice** el programa (no vuelva a solicitar los datos); si la trama no trae ninguna lectura válida, muestre `Error: la trama no contiene lecturas válidas.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Trama: <b>S05:27.9;S01:34.5;S02:41.2;S03:18.7;S04:ERR;S06:52.3;S07:22.4</b>
Umbral de riego (%): <b>30</b>

Lecturas válidas: 6
Sensores dañados: [&#x27;S04&#x27;]
Humedad promedio: 32.8%
Necesitan riego: [&#x27;S03&#x27;, &#x27;S05&#x27;, &#x27;S07&#x27;]
Activar riego general.
</pre>

````python
#@title Solución
def procesar_trama(trama: str) -> tuple[dict[str, float], list[str]]:
    """Separa la trama en lecturas válidas y sensores dañados."""
    lecturas = {}
    danados = []
    for registro in trama.split(";"):
        identificador, valor = registro.split(":")
        if valor == "ERR":
            danados.append(identificador)
        else:
            lecturas[identificador] = float(valor)
    return lecturas, danados


def sensores_bajo_umbral(lecturas: dict[str, float], umbral: float) -> list[str]:
    """Retorna, ordenados, los sensores con humedad menor que el umbral."""
    secos = []
    for identificador, humedad in lecturas.items():
        if humedad < umbral:
            secos.append(identificador)
    return sorted(secos)


def main() -> None:
    try:
        trama = input("Trama: ")
        umbral = float(input("Umbral de riego (%): "))

        lecturas, danados = procesar_trama(trama)

        if len(lecturas) == 0:
            print("Error: la trama no contiene lecturas válidas.")
        else:
            necesitan_riego = sensores_bajo_umbral(lecturas, umbral)
            promedio = sum(lecturas.values()) / len(lecturas)

            print(f"\nLecturas válidas: {len(lecturas)}")
            print(f"Sensores dañados: {danados}")
            print(f"Humedad promedio: {promedio:.1f}%")
            print(f"Necesitan riego: {necesitan_riego}")
            if len(necesitan_riego) >= len(lecturas) / 2:
                print("Activar riego general.")
            else:
                print("Riego localizado.")
    except ValueError:
        print("Error: revise el formato de la trama y el umbral.")


main()
````

**Explicación**

La trama se procesa en dos niveles: `split(";")` separa los registros y `split(":")` separa cada registro en ID y valor. Convertir el texto a un **diccionario** permite después consultar la humedad de cualquier sensor por su ID, sin volver a procesar el texto.

INVENTARIO = {

La opción 2 pide dos códigos y dice si pueden almacenarse en el mismo estante. **Pregunta de respuesta corta**

¿Por qué las parejas incompatibles se guardan como **tuplas** dentro de un **conjunto**, y no como listas dentro de una lista?

````python
#@title Solución
INVENTARIO = {
    "HCL": {"nombre": "Ácido clorhídrico", "clase": "acido", "litros": 12.0},
    "NAOH": {"nombre": "Hidróxido de sodio", "clase": "base", "litros": 8.5},
    "ETOH": {"nombre": "Etanol", "clase": "inflamable", "litros": 20.0},
    "H2O2": {"nombre": "Peróxido de hidrógeno", "clase": "oxidante", "litros": 5.0},
}
INCOMPATIBLES = {("acido", "base"), ("inflamable", "oxidante"), ("acido", "oxidante")}
MINIMO_LITROS = 6


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def son_compatibles(clase_a: str, clase_b: str) -> bool:
    """Indica si dos clases de riesgo pueden almacenarse juntas."""
    return (clase_a, clase_b) not in INCOMPATIBLES and (clase_b, clase_a) not in INCOMPATIBLES


def retirar(inventario: dict, codigo: str, litros: float) -> str:
    """Descuenta litros de un reactivo y retorna un mensaje con el resultado."""
    if codigo not in inventario:
        mensaje = f"El reactivo {codigo} no existe."
    elif inventario[codigo]["litros"] < litros:
        mensaje = f"Cantidad insuficiente: solo hay {inventario[codigo]['litros']} L."
    else:
        inventario[codigo]["litros"] -= litros
        mensaje = f"Retiro exitoso. Quedan {inventario[codigo]['litros']} L de {codigo}."
    return mensaje


def reactivos_bajo_minimo(inventario: dict, minimo: float) -> list[str]:
    """Retorna los códigos de los reactivos con menos litros que el mínimo."""
    bajos = []
    for codigo, datos in inventario.items():
        if datos["litros"] < minimo:
            bajos.append(codigo)
    return bajos


def mostrar_reporte(inventario: dict) -> None:
    """Muestra el inventario, las clases presentes y los reactivos bajo el mínimo."""
    clases = set()
    for codigo, datos in inventario.items():
        clases.add(datos["clase"])
        print(f"  {codigo:<6}{datos['nombre']:<24}{datos['clase']:<12}{datos['litros']:>6.1f} L")
    print(f"  Clases presentes: {sorted(clases)}")
    print(f"  Bajo {MINIMO_LITROS} L: {reactivos_bajo_minimo(inventario, MINIMO_LITROS)}")


def main() -> None:
    opcion = ""
    while opcion != "4":
        opcion = input("\n1. Retirar  2. Verificar almacenamiento  3. Reporte  4. Salir\nOpción: ")
        if opcion == "1":
            codigo = input("Código: ").upper()
            litros = pedir_numero_seguro("Litros a retirar: ", 0.1, 200)
            print(retirar(INVENTARIO, codigo, litros))
        elif opcion == "2":
            codigo_a = input("Primer reactivo: ").upper()
            codigo_b = input("Segundo reactivo: ").upper()
            if codigo_a in INVENTARIO and codigo_b in INVENTARIO:
                clase_a = INVENTARIO[codigo_a]["clase"]
                clase_b = INVENTARIO[codigo_b]["clase"]
                if son_compatibles(clase_a, clase_b):
                    print(f"Compatibles: {codigo_a} y {codigo_b} pueden compartir estante.")
                else:
                    print(f"¡PELIGRO! {clase_a} y {clase_b} son incompatibles.")
            else:
                print("Alguno de los reactivos no existe.")
        elif opcion == "3":
            mostrar_reporte(INVENTARIO)
        elif opcion != "4":
            print("Opción inválida.")


main()
````

**Explicación**

**Respuesta corta:** el conjunto permite verificar con `in` si una pareja es incompatible **de forma directa y muy rápida**, y no admite parejas repetidas. Sus elementos tienen que ser **inmutables**, por eso cada pareja es una tupla: una lista no puede ser elemento de un conjunto (produciría un `TypeError: unhashable type: 'list'`).
