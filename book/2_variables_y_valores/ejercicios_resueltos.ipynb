## Ejercicio 2.1 — Registro de una muestra de laboratorio

Un laboratorio de química registra cada muestra que recibe. Escriba un programa que:

1. Solicite el **código** de la muestra (texto), la **masa** en gramos (número decimal), la **cantidad de réplicas** (número entero) y si la muestra **requiere refrigeración** (la persona usuaria escribe `s` o `n`).
2. Convierta cada dato al tipo adecuado. Para el dato de refrigeración, obtenga un **booleano** comparando la respuesta con `"s"`.
3. Calcule la **masa por réplica** (masa entre cantidad de réplicas).
4. Imprima cada dato junto con el **nombre de su tipo** (use `type(variable).__name__`).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Código de la muestra: <b>QM-118</b>
Masa (g): <b>12.6</b>
Cantidad de réplicas: <b>4</b>
¿Requiere refrigeración? (s/n): <b>s</b>

--- Registro de muestra ---
Código: QM-118 (str)
Masa: 12.6 g (float)
Réplicas: 4 (int)
Refrigeración: True (bool)
Masa por réplica: 3.15 g (float)
</pre>

````python
#@title Solución
codigo_muestra = input("Código de la muestra: ")
masa_gramos = float(input("Masa (g): "))
cantidad_replicas = int(input("Cantidad de réplicas: "))
requiere_refrigeracion = input("¿Requiere refrigeración? (s/n): ") == "s"

masa_por_replica = masa_gramos / cantidad_replicas

print("\n--- Registro de muestra ---")
print(f"Código: {codigo_muestra} ({type(codigo_muestra).__name__})")
print(f"Masa: {masa_gramos} g ({type(masa_gramos).__name__})")
print(f"Réplicas: {cantidad_replicas} ({type(cantidad_replicas).__name__})")
print(f"Refrigeración: {requiere_refrigeracion} ({type(requiere_refrigeracion).__name__})")
print(f"Masa por réplica: {masa_por_replica} g ({type(masa_por_replica).__name__})")
````

**Explicación**

Observe que `masa_por_replica` es un `float` aunque `cantidad_replicas` sea un `int`: el operador `/` **siempre** produce un flotante. Además, `input()` siempre devuelve `str`, por eso es necesario convertir con `float()` e `int()`.

## Ejercicio 2.2 — Conversor de presión

Un sensor de una estación meteorológica reporta la presión atmosférica en **kilopascales (kPa)**. Escriba un programa que solicite la presión en kPa y la convierta a:

* **atmósferas**: $1 \text{ atm} = 101.325 \text{ kPa}$
* **milímetros de mercurio**: $1 \text{ kPa} = 7.50062 \text{ mmHg}$
* **libras por pulgada cuadrada**: $1 \text{ kPa} = 0.145038 \text{ psi}$

Muestre los resultados con **3 decimales**.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Presión medida (kPa): <b>86.4</b>
86.400 kPa equivalen a:
  0.853 atm
  648.054 mmHg
  12.531 psi
</pre>

````python
#@title Solución
KPA_POR_ATM = 101.325
MMHG_POR_KPA = 7.50062
PSI_POR_KPA = 0.145038

presion_kpa = float(input("Presión medida (kPa): "))

presion_atm = presion_kpa / KPA_POR_ATM
presion_mmhg = presion_kpa * MMHG_POR_KPA
presion_psi = presion_kpa * PSI_POR_KPA

print(f"{presion_kpa:.3f} kPa equivalen a:")
print(f"  {presion_atm:.3f} atm")
print(f"  {presion_mmhg:.3f} mmHg")
print(f"  {presion_psi:.3f} psi")
````

**Explicación**

Guardar los factores de conversión en variables con nombres en MAYÚSCULAS indica que son **constantes** y evita "números mágicos" dentro de las fórmulas.

## Ejercicio 2.3 — Código de lote

Una planta farmacéutica identifica cada lote con un código de exactamente **16 caracteres** con el formato `PP-AAAA-LOTE-NNN`, por ejemplo `CR-2026-LOTE-457`, donde:

* `PP` son dos letras del país de fabricación,
* `AAAA` es el año,
* `NNN` es el número de lote (tres dígitos).

Escriba un programa que solicite el código y, usando **indexación y *slicing***, extraiga el país, el año y el número de lote. Convierta el año y el número a `int` y muestre:

1. El país, el año y el número de lote.
2. El **número del siguiente lote** (el número actual más 1).
3. Los **años transcurridos** desde la fabricación hasta 2026.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Código del lote: <b>CR-2023-LOTE-457</b>
País: CR
Año de fabricación: 2023
Número de lote: 457
Siguiente lote: 458
Años transcurridos: 3
</pre>

````python
#@title Solución
ANIO_ACTUAL = 2026

codigo_lote = input("Código del lote: ")

pais = codigo_lote[0:2]
anio_fabricacion = int(codigo_lote[3:7])
numero_lote = int(codigo_lote[-3:])

siguiente_lote = numero_lote + 1
anios_transcurridos = ANIO_ACTUAL - anio_fabricacion

print(f"País: {pais}")
print(f"Año de fabricación: {anio_fabricacion}")
print(f"Número de lote: {numero_lote}")
print(f"Siguiente lote: {siguiente_lote}")
print(f"Años transcurridos: {anios_transcurridos}")
````

**Explicación**

El índice negativo `-3:` toma los **últimos tres** caracteres sin importar la longitud total del código. Si el número fuera `"007"`, `int("007")` produciría `7`, lo cual es correcto para hacer cálculos.

## Ejercicio 2.4 — Parte entera y parte decimal

Una balanza digital muestra la masa de una muestra en **kilogramos** con decimales, por ejemplo `3.745`. Escriba un programa que solicite esa lectura y, **sin usar condicionales ni funciones de redondeo**, calcule:

1. Los **kilogramos completos** (parte entera, usando `int()`).
2. Los **gramos restantes** (la parte decimal convertida a gramos y luego a entero).
3. La masa total expresada en **gramos enteros**.

*Pista:* la parte decimal de `x` es `x - int(x)`.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Lectura de la balanza (kg): <b>3.745</b>
Kilogramos completos: 3 kg
Gramos restantes: 745 g
Masa total: 3745 g
</pre>

````python
#@title Solución
masa_kg = float(input("Lectura de la balanza (kg): "))

kilogramos_completos = int(masa_kg)
parte_decimal = masa_kg - kilogramos_completos
gramos_restantes = int(parte_decimal * 1000 + 0.5)
masa_total_gramos = kilogramos_completos * 1000 + gramos_restantes

print(f"Kilogramos completos: {kilogramos_completos} kg")
print(f"Gramos restantes: {gramos_restantes} g")
print(f"Masa total: {masa_total_gramos} g")
````

**Explicación**

`int()` **trunca** (corta) la parte decimal, no redondea. Por la representación binaria de los flotantes, `0.745 * 1000` puede dar `744.9999...`; sumar `0.5` antes de truncar lo convierte en un redondeo al entero más cercano.

## Ejercicio 2.5 — Rotación de bombas

En una planta de tratamiento de agua hay tres bombas centrífugas identificadas por su código. Para repartir el desgaste, cada semana los roles **rotan**: la bomba que estaba en **operación** pasa a **respaldo**, la de respaldo pasa a **mantenimiento** y la de mantenimiento pasa a **operación**.

Escriba un programa que solicite el código de la bomba asignada a cada rol, realice la rotación **usando una sola variable temporal** (sin usar la asignación múltiple `a, b = b, a`) e imprima la asignación antes y después.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Bomba en operación: <b>B-101</b>
Bomba en respaldo: <b>B-102</b>
Bomba en mantenimiento: <b>B-103</b>

Antes -&gt; operación: B-101, respaldo: B-102, mantenimiento: B-103
Después -&gt; operación: B-103, respaldo: B-101, mantenimiento: B-102
</pre>

````python
#@title Solución
bomba_operacion = input("Bomba en operación: ")
bomba_respaldo = input("Bomba en respaldo: ")
bomba_mantenimiento = input("Bomba en mantenimiento: ")

print(f"\nAntes -> operación: {bomba_operacion}, respaldo: {bomba_respaldo}, "
      f"mantenimiento: {bomba_mantenimiento}")

temporal = bomba_mantenimiento
bomba_mantenimiento = bomba_respaldo
bomba_respaldo = bomba_operacion
bomba_operacion = temporal

print(f"Después -> operación: {bomba_operacion}, respaldo: {bomba_respaldo}, "
      f"mantenimiento: {bomba_mantenimiento}")
````

**Explicación**

El orden de las asignaciones importa: se guarda primero el valor que se **sobrescribe primero** (`bomba_mantenimiento`). Si se hiciera `bomba_operacion = bomba_mantenimiento` al inicio sin guardar nada, se perdería el código de la bomba que estaba en operación.

## Ejercicio 2.6 — Configuración de un variador de frecuencia

Un ingeniero de control configura un **variador de frecuencia** para un motor de inducción. La velocidad síncrona del motor (en rpm) se calcula con:

$$n_s = \frac{120 \cdot f}{p}$$

donde $f$ es la frecuencia (Hz) y $p$ la cantidad de polos del motor.

Escriba un programa en Python que:

A. Solicite, en este orden: el **modelo** del motor (texto), la **frecuencia** de la red (decimal), la **cantidad de polos** (entero) y el **voltaje nominal** (decimal).

B. Calcule la **velocidad síncrona** y la **velocidad síncrona entera** (convertida con `int()`).

C. Calcule un booleano `alta_velocidad` que sea `True` si la velocidad síncrona **supera** las 3000 rpm.

D. Imprima cada valor junto con el **nombre de su tipo** (`type(valor).__name__`).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Modelo del motor: <b>WEG-W22</b>
Frecuencia (Hz): <b>60</b>
Cantidad de polos: <b>4</b>
Voltaje nominal (V): <b>460</b>

--- Configuración del variador ---
Modelo: WEG-W22 (str)
Frecuencia: 60.0 Hz (float)
Polos: 4 (int)
Voltaje: 460.0 V (float)
Velocidad síncrona: 1800.0 rpm (float)
Velocidad entera: 1800 rpm (int)
¿Alta velocidad?: False (bool)
</pre>

**Pregunta de respuesta corta**

Sin ejecutar el código, indique el **valor** y el **tipo** de cada variable:

```python
a = 120 / 4
b = "60" * 2
c = float("60") + True
d = str(4.0)
```

````python
#@title Solución
modelo = input("Modelo del motor: ")
frecuencia_hz = float(input("Frecuencia (Hz): "))
cantidad_polos = int(input("Cantidad de polos: "))
voltaje_nominal = float(input("Voltaje nominal (V): "))

velocidad_sincrona = 120 * frecuencia_hz / cantidad_polos
velocidad_sincrona_entera = int(velocidad_sincrona)
alta_velocidad = velocidad_sincrona > 3000

print("\n--- Configuración del variador ---")
print(f"Modelo: {modelo} ({type(modelo).__name__})")
print(f"Frecuencia: {frecuencia_hz} Hz ({type(frecuencia_hz).__name__})")
print(f"Polos: {cantidad_polos} ({type(cantidad_polos).__name__})")
print(f"Voltaje: {voltaje_nominal} V ({type(voltaje_nominal).__name__})")
print(f"Velocidad síncrona: {velocidad_sincrona} rpm ({type(velocidad_sincrona).__name__})")
print(f"Velocidad entera: {velocidad_sincrona_entera} rpm ({type(velocidad_sincrona_entera).__name__})")
print(f"¿Alta velocidad?: {alta_velocidad} ({type(alta_velocidad).__name__})")
````

**Explicación**

La velocidad síncrona es `float` aunque los polos sean `int`, porque `/` siempre produce un flotante.

**Respuesta corta:** `a = 30.0` (`float`, por la división `/`); `b = '6060'` (`str`, repetición de cadenas); `c = 61.0` (`float`, `True` vale 1); `d = '4.0'` (`str`).

## Ejercicio 2.7 — Texto que parece número

Un sistema recibe desde un sensor la cantidad de **alarmas activas** como texto. Escriba un programa que solicite ese texto y muestre:

1. El resultado de `bool()` aplicado **directamente al texto**.
2. El resultado de `bool()` aplicado al texto **convertido a entero**.
3. El resultado de `bool()` aplicado al texto **convertido a flotante**.

Ejecútelo con la entrada `0` y responda: ¿por qué el primer resultado es distinto de los otros dos?

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de alarmas activas: <b>0</b>
bool(texto): True
bool(entero): False
bool(flotante): False
</pre>

<pre>
Cantidad de alarmas activas: <b>3</b>
bool(texto): True
bool(entero): True
bool(flotante): True
</pre>

````python
#@title Solución
alarmas_texto = input("Cantidad de alarmas activas: ")

alarmas_entero = int(alarmas_texto)
alarmas_flotante = float(alarmas_texto)

print(f"bool(texto): {bool(alarmas_texto)}")
print(f"bool(entero): {bool(alarmas_entero)}")
print(f"bool(flotante): {bool(alarmas_flotante)}")
````

**Explicación**

Para las cadenas, `bool()` solo revisa si están **vacías**: `""` es `False` y **cualquier otro texto**, incluido `"0"` o `"False"`, es `True`. Para los números, `bool()` devuelve `False` únicamente cuando el valor es cero. Por eso es importante **convertir primero** el texto al tipo numérico antes de interpretarlo.

## Ejercicio 2.8 — Costo de combustible de un camión de carga

Una empresa de logística necesita calcular el **costo de combustible por tonelada** transportada por un camión de carga. Un compañero escribió el siguiente programa, pero **no se puede ejecutar** porque varios nombres de variables son inválidos, y otros, aunque válidos, no son descriptivos:

```python
1distancia = float(input("Distancia de la ruta (km): "))
rendimiento-km-litro = float(input("Rendimiento del camión (km/L): "))
Precio del litro = float(input("Precio del diésel (₡/L): "))
global = float(input("Carga transportada (t): "))
x = 1distancia / rendimiento-km-litro
y = x * Precio del litro
z = y / global
print(z)
```

Reescriba el programa en Python con nombres **válidos y descriptivos** en `snake_case`, de modo que:

A. Solicite la distancia, el rendimiento, el precio del diésel y las toneladas de carga.

B. Calcule los **litros de diésel**, el **costo total** y el **costo por tonelada**.

C. Imprima un reporte con los tres valores con **2 decimales** y sus unidades.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Distancia de la ruta (km): <b>145.5</b>
Rendimiento del camión (km/L): <b>3.8</b>
Precio del diésel (₡/L): <b>745</b>
Carga transportada (t): <b>18</b>

--- Costo de la ruta ---
Diésel necesario: 38.29 L
Costo total: ₡28525.66
Costo por tonelada: ₡1584.76
</pre>

**Pregunta de respuesta corta**

Para cada uno de los nombres `1distancia`, `rendimiento-km-litro`, `Precio del litro` y `global`, explique **por qué** es inválido.

````python
#@title Solución
distancia_km = float(input("Distancia de la ruta (km): "))
rendimiento_km_por_litro = float(input("Rendimiento del camión (km/L): "))
precio_por_litro = float(input("Precio del diésel (₡/L): "))
toneladas_de_carga = float(input("Carga transportada (t): "))

litros_de_diesel = distancia_km / rendimiento_km_por_litro
costo_total = litros_de_diesel * precio_por_litro
costo_por_tonelada = costo_total / toneladas_de_carga

print("\n--- Costo de la ruta ---")
print(f"Diésel necesario: {litros_de_diesel:.2f} L")
print(f"Costo total: ₡{costo_total:.2f}")
print(f"Costo por tonelada: ₡{costo_por_tonelada:.2f}")
````

**Explicación**

**Respuesta corta:**

* `1distancia`: un nombre **no puede iniciar con un dígito**.
* `rendimiento-km-litro`: el guion `-` se interpreta como el operador de **resta**.
* `Precio del litro`: los nombres **no pueden tener espacios**.
* `global`: es una **palabra reservada** de Python.

`x`, `y` y `z` son válidos, pero **no describen** lo que guardan.

En un ensayo de transferencia de calor, una estudiante registra la temperatura de una placa en tres instantes y, antes de agregar una **cuarta medición**, quiere guardar un **respaldo** de las tres primeras.

Escriba un programa en Python que:

A. Solicite las **tres primeras temperaturas** (°C) y las guarde en una lista llamada `mediciones`.

B. Cree una lista `respaldo` que sea una **copia independiente** de `mediciones` (use `list(mediciones)`).

C. Solicite una **cuarta temperatura** y agréguela **solo** a `mediciones` con `append()`.

D. Imprima ambas listas, la **cantidad** de mediciones de cada una (con `len()`) y el **aumento de temperatura** entre la primera y la última medición.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Temperatura 1 (°C): <b>21.5</b>
Temperatura 2 (°C): <b>22.0</b>
Temperatura 3 (°C): <b>22.4</b>
Temperatura 4 (°C): <b>23.1</b>

Mediciones: [21.5, 22.0, 22.4, 23.1] (4 valores)

¿Qué imprimiría el programa si en el inciso B se escribiera `respaldo = mediciones`? Explique por qué, y por qué ese problema **no** ocurre con una variable de tipo `float`.
