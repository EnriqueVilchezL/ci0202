## Ejercicio 11.1 — Bitácora de mantenimiento de un generador

Los técnicos de una planta eléctrica de respaldo registran cada intervención al **generador** en un archivo de texto, `bitacora.txt`, con una línea por intervención en el formato `fecha | técnico | horómetro | descripción`.

Escriba un programa en Python con las funciones:

A. `registrar(nombre_archivo: str, fecha: str, tecnico: str, horometro: float, descripcion: str) -> None`: **agrega** una línea al final del archivo sin borrar lo que ya tiene (modo `"a"`).

B. `leer_bitacora(nombre_archivo: str) -> list[list[str]]`: lee el archivo línea por línea y retorna una lista con los campos de cada línea (use `strip()` y `split(" | ")`).

C. `main() -> None`: solicita los datos de una nueva intervención, la registra, vuelve a leer la bitácora completa y muestra cada intervención numerada, la cantidad total de intervenciones y las **horas de operación** transcurridas entre la primera y la última.

**Archivo `bitacora.txt`**

```text
03/02/2026 | L. Mora | 1250.5 | Cambio de aceite y filtros
17/03/2026 | A. Rojas | 1498.0 | Revisión del sistema de enfriamiento
28/04/2026 | L. Mora | 1745.2 | Cambio de baterías de arranque
```

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Fecha (DD/MM/AAAA): <b>09/06/2026</b>
Técnico: <b>M. Solís</b>
Horómetro (h): <b>2010.8</b>
Descripción: <b>Ajuste del regulador de voltaje</b>

--- Bitácora del generador ---
1. 03/02/2026 (1250.5 h) - L. Mora: Cambio de aceite y filtros
2. 17/03/2026 (1498.0 h) - A. Rojas: Revisión del sistema de enfriamiento
3. 28/04/2026 (1745.2 h) - L. Mora: Cambio de baterías de arranque
4. 09/06/2026 (2010.8 h) - M. Solís: Ajuste del regulador de voltaje

Intervenciones registradas: 4
Horas de operación entre la primera y la última: 760.3 h
</pre>

Contenido del archivo generado `bitacora.txt`:

```text
03/02/2026 | L. Mora | 1250.5 | Cambio de aceite y filtros
17/03/2026 | A. Rojas | 1498.0 | Revisión del sistema de enfriamiento
28/04/2026 | L. Mora | 1745.2 | Cambio de baterías de arranque
09/06/2026 | M. Solís | 2010.8 | Ajuste del regulador de voltaje
```

**Pregunta de respuesta corta**

¿Qué habría pasado con las intervenciones anteriores si el archivo se hubiera abierto con el modo `"w"` en lugar de `"a"`?

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('bitacora.txt', "w", encoding="utf-8") as archivo:
    archivo.write('03/02/2026 | L. Mora | 1250.5 | Cambio de aceite y filtros\n17/03/2026 | A. Rojas | 1498.0 | Revisión del sistema de enfriamiento\n28/04/2026 | L. Mora | 1745.2 | Cambio de baterías de arranque\n')
````

````python
#@title Solución
def registrar(nombre_archivo: str, fecha: str, tecnico: str, horometro: float, descripcion: str) -> None:
    """Agrega una intervención al final de la bitácora."""
    with open(nombre_archivo, "a", encoding="utf-8") as archivo:
        archivo.write(f"{fecha} | {tecnico} | {horometro} | {descripcion}\n")


def leer_bitacora(nombre_archivo: str) -> list[list[str]]:
    """Lee la bitácora y retorna los campos de cada intervención."""
    intervenciones = []
    with open(nombre_archivo, encoding="utf-8") as archivo:
        for linea in archivo:
            if linea.strip() != "":
                intervenciones.append(linea.strip().split(" | "))
    return intervenciones


def main() -> None:
    fecha = input("Fecha (DD/MM/AAAA): ")
    tecnico = input("Técnico: ")
    horometro = float(input("Horómetro (h): "))
    descripcion = input("Descripción: ")

    registrar("bitacora.txt", fecha, tecnico, horometro, descripcion)
    intervenciones = leer_bitacora("bitacora.txt")

    print("\n--- Bitácora del generador ---")
    for numero, (fecha, tecnico, horometro, descripcion) in enumerate(intervenciones, start=1):
        print(f"{numero}. {fecha} ({horometro} h) - {tecnico}: {descripcion}")

    horas = float(intervenciones[-1][2]) - float(intervenciones[0][2])
    print(f"\nIntervenciones registradas: {len(intervenciones)}")
    print(f"Horas de operación entre la primera y la última: {horas:.1f} h")


main()
````

**Explicación**

**Respuesta corta:** el modo `"w"` (*write*) **borra** todo el contenido del archivo al abrirlo, así que la bitácora quedaría solo con la intervención nueva. El modo `"a"` (*append*) conserva el contenido y escribe al final. El bloque `with` cierra el archivo automáticamente al terminar, incluso si ocurre un error.

## Ejercicio 11.2 — Registro de un datalogger con comentarios

Un **datalogger** instalado en una cámara de refrigeración de vacunas guarda la temperatura cada 15 minutos en un archivo de texto. El archivo tiene **líneas de comentario** que empiezan con `#`, y algunas lecturas quedaron marcadas como `ERROR`. Las vacunas deben conservarse **entre 2 °C y 8 °C**.

Escriba un programa en Python con las funciones:

A. `leer_temperaturas(nombre_archivo: str) -> tuple[list[float], int]`: ignora las líneas vacías y las de comentario, y retorna la lista de temperaturas válidas y la cantidad de lecturas con `ERROR`.

B. `escribir_reporte(nombre_archivo: str, temperaturas: list[float], errores: int) -> None`: escribe en un archivo nuevo un reporte con la cantidad de lecturas, el mínimo, el máximo, el promedio, las lecturas fuera de rango y el dictamen `CONFORME` o `NO CONFORME`.

C. `main() -> None`: lee `camara_frio.txt`, escribe `reporte_camara.txt` y muestra un mensaje indicando que el reporte fue generado.

**Archivo `camara_frio.txt`**

```text
# Datalogger DL-220 - Cámara de vacunas 3
# Fecha: 11/05/2026 - Intervalo: 15 min
4.8
5.1
5.4
ERROR
6.2
7.9
8.6
# Puerta abierta 10 minutos
8.3
6.5
5.0
ERROR
4.7
```

**Ejemplo de ejecución**

<pre>
Reporte generado: reporte_camara.txt
</pre>

Contenido del archivo generado `reporte_camara.txt`:

```text
REPORTE DE CÁMARA DE FRÍO
Lecturas válidas: 10 (con error: 2)
Mínima: 4.7 °C
Máxima: 8.6 °C
Promedio: 6.25 °C
Fuera de rango: [8.6, 8.3]
Dictamen: NO CONFORME
```

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('camara_frio.txt', "w", encoding="utf-8") as archivo:
    archivo.write('# Datalogger DL-220 - Cámara de vacunas 3\n# Fecha: 11/05/2026 - Intervalo: 15 min\n4.8\n5.1\n5.4\nERROR\n6.2\n7.9\n8.6\n# Puerta abierta 10 minutos\n8.3\n6.5\n5.0\nERROR\n4.7\n')
````

````python
#@title Solución
TEMPERATURA_MINIMA = 2
TEMPERATURA_MAXIMA = 8


def leer_temperaturas(nombre_archivo: str) -> tuple[list[float], int]:
    """Lee las temperaturas válidas y cuenta las lecturas con error."""
    temperaturas = []
    errores = 0
    with open(nombre_archivo, encoding="utf-8") as archivo:
        for linea in archivo:
            linea = linea.strip()
            if linea == "" or linea.startswith("#"):
                continue
            if linea == "ERROR":
                errores += 1
            else:
                temperaturas.append(float(linea))
    return temperaturas, errores


def escribir_reporte(nombre_archivo: str, temperaturas: list[float], errores: int) -> None:
    """Escribe el reporte de conformidad de la cámara."""
    fuera_de_rango = []
    for temperatura in temperaturas:
        if not TEMPERATURA_MINIMA <= temperatura <= TEMPERATURA_MAXIMA:
            fuera_de_rango.append(temperatura)

    dictamen = "CONFORME" if len(fuera_de_rango) == 0 else "NO CONFORME"
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write("REPORTE DE CÁMARA DE FRÍO\n")
        archivo.write(f"Lecturas válidas: {len(temperaturas)} (con error: {errores})\n")
        archivo.write(f"Mínima: {min(temperaturas):.1f} °C\n")
        archivo.write(f"Máxima: {max(temperaturas):.1f} °C\n")
        archivo.write(f"Promedio: {sum(temperaturas) / len(temperaturas):.2f} °C\n")
        archivo.write(f"Fuera de rango: {fuera_de_rango}\n")
        archivo.write(f"Dictamen: {dictamen}\n")


def main() -> None:
    temperaturas, errores = leer_temperaturas("camara_frio.txt")
    escribir_reporte("reporte_camara.txt", temperaturas, errores)
    print("Reporte generado: reporte_camara.txt")


main()
````

**Explicación**

Al recorrer un archivo con `for linea in archivo`, cada línea incluye el salto de línea final (`\n`); por eso se usa `strip()` antes de compararla o convertirla. Aquí `continue` salta las líneas que no son datos, lo que evita anidar el resto del cuerpo dentro de otro `if`.

## Ejercicio 11.3 — Conversión de un formato propietario a CSV

Un **analizador de calidad de energía** exporta sus mediciones en un formato propio, con una medición por línea y los campos como pares `clave=valor` separados por punto y coma:

```text
t=00:00;V=229.8;I=12.4;fp=0.91
```

El ingeniero quiere abrir los datos en una hoja de cálculo, así que necesita convertirlos a **CSV**. Además, quiere agregar una columna con la **potencia activa** $P = V \cdot I \cdot fp$ (W).

Escriba un programa en Python con las funciones:

A. `interpretar_linea(linea: str) -> dict[str, str]`: convierte una línea en un diccionario `{"t": "00:00", "V": "229.8", ...}`.

B. `convertir_a_csv(archivo_entrada: str, archivo_salida: str) -> int`: lee el archivo de entrada, escribe el CSV con el encabezado `hora,voltaje_V,corriente_A,fp,potencia_W` y retorna la cantidad de filas escritas.

C. `main() -> None`: realiza la conversión de `analizador.txt` a `analizador.csv` y muestra cuántas filas se convirtieron.

**Archivo `analizador.txt`**

```text
t=00:00;V=229.8;I=12.4;fp=0.91
t=00:15;V=228.5;I=18.9;fp=0.88
t=00:30;V=227.1;I=25.3;fp=0.85
t=00:45;V=228.9;I=21.7;fp=0.87
t=01:00;V=230.2;I=14.0;fp=0.92
```

**Ejemplo de ejecución**

<pre>
Filas convertidas: 5
</pre>

Contenido del archivo generado `analizador.csv`:

```text
hora,voltaje_V,corriente_A,fp,potencia_W
00:00,229.8,12.4,0.91,2593.1
00:15,228.5,18.9,0.88,3800.4
00:30,227.1,25.3,0.85,4883.8
00:45,228.9,21.7,0.87,4321.4
01:00,230.2,14.0,0.92,2965.0
```

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('analizador.txt', "w", encoding="utf-8") as archivo:
    archivo.write('t=00:00;V=229.8;I=12.4;fp=0.91\nt=00:15;V=228.5;I=18.9;fp=0.88\nt=00:30;V=227.1;I=25.3;fp=0.85\nt=00:45;V=228.9;I=21.7;fp=0.87\nt=01:00;V=230.2;I=14.0;fp=0.92\n')
````

````python
#@title Solución
def interpretar_linea(linea: str) -> dict[str, str]:
    """Convierte una línea 'clave=valor;...' en un diccionario."""
    campos = {}
    for par in linea.strip().split(";"):
        clave, valor = par.split("=")
        campos[clave] = valor
    return campos


def convertir_a_csv(archivo_entrada: str, archivo_salida: str) -> int:
    """Convierte el archivo del analizador a CSV y retorna la cantidad de filas."""
    filas = 0
    with open(archivo_entrada, encoding="utf-8") as entrada, \
            open(archivo_salida, "w", encoding="utf-8") as salida:
        salida.write("hora,voltaje_V,corriente_A,fp,potencia_W\n")
        for linea in entrada:
            if linea.strip() != "":
                campos = interpretar_linea(linea)
                potencia = float(campos["V"]) * float(campos["I"]) * float(campos["fp"])
                salida.write(f"{campos['t']},{campos['V']},{campos['I']},{campos['fp']},{potencia:.1f}\n")
                filas += 1
    return filas


def main() -> None:
    filas = convertir_a_csv("analizador.txt", "analizador.csv")
    print(f"Filas convertidas: {filas}")


main()
````

**Explicación**

Se pueden abrir **dos** archivos en el mismo `with`, separados por coma: uno para leer y otro para escribir. Convertir cada línea en un diccionario permite acceder a los campos por su **nombre**, así que el programa seguiría funcionando si el analizador cambiara el orden de los campos.

## Ejercicio 11.4 — Inventario de transformadores con pandas

Una cooperativa eléctrica lleva el inventario de sus **transformadores de distribución**. Para cada uno conoce su código, su potencia nominal (kVA), la carga medida en la hora pico (kVA) y su año de instalación.

Escriba un programa en Python que use pandas y:

A. Defina `crear_inventario() -> pd.DataFrame`, que construye el `DataFrame` a partir de un diccionario con estos datos:

   | codigo | potencia_kva | carga_kva | anio |
   | :-- | :-: | :-: | :-: |
   | TR-101 | 50 | 38.5 | 2008 |
   | TR-102 | 75 | 81.2 | 2012 |
   | TR-103 | 50 | 22.0 | 2019 |
   | TR-104 | 100 | 94.7 | 2005 |
   | TR-105 | 37.5 | 40.1 | 2015 |
   | TR-106 | 75 | 51.3 | 2021 |

B. Agregue las columnas `carga_pct` (carga / potencia × 100, redondeada a 1 decimal) y `antiguedad` (2026 − año).

C. En `main()`, muestre el inventario completo, los transformadores **sobrecargados** (carga mayor al 100 %), los que tienen **más de 15 años** y una carga mayor al 80 % (candidatos a reemplazo prioritario), y exporte el inventario completo a `inventario_transformadores.csv` **sin** el índice.

**Ejemplo de ejecución**

<pre>
   codigo  potencia_kva  carga_kva  anio  carga_pct  antiguedad
0  TR-101          50.0       38.5  2008       77.0          18
1  TR-102          75.0       81.2  2012      108.3          14
2  TR-103          50.0       22.0  2019       44.0           7
3  TR-104         100.0       94.7  2005       94.7          21
4  TR-105          37.5       40.1  2015      106.9          11
5  TR-106          75.0       51.3  2021       68.4           5

Sobrecargados:
   codigo  carga_pct

Reemplazo prioritario:
   codigo  antiguedad  carga_pct
3  TR-104          21       94.7

Inventario exportado a inventario_transformadores.csv
</pre>

Contenido del archivo generado `inventario_transformadores.csv`:

```text
codigo,potencia_kva,carga_kva,anio,carga_pct,antiguedad
TR-101,50.0,38.5,2008,77.0,18
TR-102,75.0,81.2,2012,108.3,14
TR-103,50.0,22.0,2019,44.0,7

TR-105,37.5,40.1,2015,106.9,11
TR-106,75.0,51.3,2021,68.4,5
```

````python
#@title Solución
import pandas as pd

ANIO_ACTUAL = 2026


def crear_inventario() -> pd.DataFrame:
    """Construye el inventario de transformadores."""
    datos = {
        "codigo": ["TR-101", "TR-102", "TR-103", "TR-104", "TR-105", "TR-106"],
        "potencia_kva": [50, 75, 50, 100, 37.5, 75],
        "carga_kva": [38.5, 81.2, 22.0, 94.7, 40.1, 51.3],
        "anio": [2008, 2012, 2019, 2005, 2015, 2021],
    }
    inventario = pd.DataFrame(datos)
    inventario["carga_pct"] = (inventario["carga_kva"] / inventario["potencia_kva"] * 100).round(1)
    inventario["antiguedad"] = ANIO_ACTUAL - inventario["anio"]
    return inventario


def main() -> None:
    inventario = crear_inventario()
    sobrecargados = inventario[inventario["carga_pct"] > 100]
    reemplazo = inventario[(inventario["antiguedad"] > 15) & (inventario["carga_pct"] > 80)]

    print(inventario)
    print(f"\nSobrecargados:\n{sobrecargados[['codigo', 'carga_pct']]}")
    print(f"\nReemplazo prioritario:\n{reemplazo[['codigo', 'antiguedad', 'carga_pct']]}")

    inventario.to_csv("inventario_transformadores.csv", index=False)
    print("\nInventario exportado a inventario_transformadores.csv")


main()
````

**Explicación**

Una operación entre columnas (`inventario["carga_kva"] / inventario["potencia_kva"]`) se aplica fila por fila, igual que con los arreglos de NumPy. Para filtrar con varias condiciones se usan `&` (y) y `|` (o), con cada condición entre paréntesis. Con `index=False` el CSV no incluye la columna de números de fila, que no es un dato real.

## Ejercicio 11.5 — Resultados de un ensayo de tracción

Un laboratorio de materiales ensaya a tracción varias **probetas de acero de refuerzo**. La máquina de ensayo exporta un CSV con el código de la probeta, su diámetro (mm), la carga de fluencia (kN) y la carga máxima (kN).

El esfuerzo se calcula como $\sigma = \frac{F \cdot 1000}{A}$ (MPa), con $A = \frac{\pi d^2}{4}$. La norma exige un esfuerzo de fluencia de al menos **420 MPa** y que la relación entre el esfuerzo máximo y el de fluencia sea de al menos **1.25**.

Escriba un programa en Python que use pandas y NumPy y:

A. Defina `calcular_esfuerzos(ensayos: pd.DataFrame) -> pd.DataFrame`, que agrega las columnas `fluencia_mpa`, `maximo_mpa` (1 decimal) y `relacion` (máximo / fluencia, 3 decimales).

B. Defina `evaluar(ensayos: pd.DataFrame) -> pd.DataFrame`, que agrega la columna `cumple` (booleana) según ambos criterios.

C. En `main()`, lea `ensayo_traccion.csv`, muestre la tabla con los resultados, el resumen estadístico de las columnas de esfuerzo con `describe()` (redondeado a 1 decimal), los códigos de las probetas que **no cumplen** y el porcentaje de cumplimiento.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('ensayo_traccion.csv', "w", encoding="utf-8") as archivo:
    archivo.write('probeta,diametro_mm,carga_fluencia_kN,carga_maxima_kN\nP-01,12.7,56.1,72.4\nP-02,12.7,55.8,71.0\nP-03,12.7,51.9,70.3\nP-04,15.9,86.5,110.9\nP-05,15.9,88.2,103.7\nP-06,15.9,87.9,112.5\n')
````

````python
#@title Solución
import numpy as np
import pandas as pd

FLUENCIA_MINIMA_MPA = 420
RELACION_MINIMA = 1.25


def calcular_esfuerzos(ensayos: pd.DataFrame) -> pd.DataFrame:
    """Agrega los esfuerzos de fluencia y máximo, y su relación."""
    area = np.pi * ensayos["diametro_mm"] ** 2 / 4
    ensayos["fluencia_mpa"] = (ensayos["carga_fluencia_kN"] * 1000 / area).round(1)
    ensayos["maximo_mpa"] = (ensayos["carga_maxima_kN"] * 1000 / area).round(1)
    ensayos["relacion"] = (ensayos["maximo_mpa"] / ensayos["fluencia_mpa"]).round(3)
    return ensayos


def evaluar(ensayos: pd.DataFrame) -> pd.DataFrame:
    """Agrega una columna que indica si cada probeta cumple la norma."""
    ensayos["cumple"] = (ensayos["fluencia_mpa"] >= FLUENCIA_MINIMA_MPA) & \
                        (ensayos["relacion"] >= RELACION_MINIMA)
    return ensayos


def main() -> None:
    ensayos = evaluar(calcular_esfuerzos(pd.read_csv("ensayo_traccion.csv")))

    print(ensayos[["probeta", "fluencia_mpa", "maximo_mpa", "relacion", "cumple"]])
    print(f"\nResumen estadístico:\n{ensayos[['fluencia_mpa', 'maximo_mpa']].describe().round(1)}")

    no_cumplen = ensayos.loc[~ensayos["cumple"], "probeta"]
    print(f"\nNo cumplen: {list(no_cumplen)}")
    print(f"Cumplimiento: {ensayos['cumple'].mean():.0%}")


main()
````

**Explicación**

`describe()` calcula en una sola instrucción la cantidad de datos, la media, la desviación estándar, el mínimo, los cuartiles y el máximo de cada columna numérica. El promedio de una columna booleana (`ensayos["cumple"].mean()`) es la **fracción** de valores `True`, porque cada `True` cuenta como 1. La probeta P-03 falla por fluencia y la P-05 por la relación máximo/fluencia.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('pesajes.csv', "w", encoding="utf-8") as archivo:
    archivo.write('placa,tipo,peso_t,empresa\nC-145987,C2,15.2,Transportes del Valle\nC-162344,C3,25.8,Agregados Irazú\nC-103220,T3S2,43.9,Logística Caribe\nC-177651,C2,16.3,Transportes del Valle\nC-150093,T3S2,40.8,Agregados Irazú\nC-181204,C3,26.9,Agregados Irazú\nC-129870,C2,17.4,Logística Caribe\n')
````

````python
#@title Solución
import pandas as pd

PESOS_MAXIMOS = {"C2": 16.0, "C3": 24.0, "T3S2": 41.5}
TOLERANCIA_T = 0.5


def agregar_exceso(pesajes: pd.DataFrame) -> pd.DataFrame:
    """Agrega el peso máximo permitido y el exceso de cada camión."""
    pesajes["maximo_t"] = pesajes["tipo"].map(PESOS_MAXIMOS)
    pesajes["exceso_t"] = (pesajes["peso_t"] - pesajes["maximo_t"]).clip(lower=0).round(1)
    return pesajes


def infractores(pesajes: pd.DataFrame, empresa: str = "") -> pd.DataFrame:
    """Retorna los camiones con exceso sobre la tolerancia, ordenados de mayor a menor."""
    seleccion = pesajes[pesajes["exceso_t"] > TOLERANCIA_T]
    if empresa != "":
        seleccion = seleccion[seleccion["empresa"] == empresa]
    return seleccion.sort_values("exceso_t", ascending=False)


def main() -> None:
    pesajes = agregar_exceso(pd.read_csv("pesajes.csv"))
    columnas = ["placa", "tipo", "peso_t", "exceso_t"]

    print(f"Infractores del día:\n{infractores(pesajes)[columnas]}")

    empresa = input("\nEmpresa a consultar: ")
    de_la_empresa = infractores(pesajes, empresa)
    if len(de_la_empresa) == 0:
        print("Sin infracciones")
    else:
        print(de_la_empresa[columnas])


main()
````

**Explicación**

`Series.map(diccionario)` reemplaza cada valor de la columna por el valor correspondiente del diccionario, lo que evita un ciclo con `if`. `clip(lower=0)` convierte los valores negativos en 0: un camión liviano no tiene "exceso negativo". El camión C-177651 excede el máximo por 0.3 t, pero queda dentro de la tolerancia de la báscula.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('estacion.csv', "w", encoding="utf-8") as archivo:
    archivo.write('hora,temperatura_c,humedad_pct,lluvia_mm\n06:00,18.2,92,0.0\n07:00,19.1,,0.4\n08:00,,85,\n09:00,22.8,80,0.0\n10:00,24.5,,1.2\n11:00,25.9,71,\n12:00,26.4,68,3.8\n')
````

````python
#@title Solución
import pandas as pd


def contar_faltantes(datos: pd.DataFrame) -> pd.Series:
    """Cuenta los valores faltantes de cada columna."""
    return datos.isna().sum()


def completar(datos: pd.DataFrame) -> pd.DataFrame:
    """Retorna una copia con los valores faltantes completados."""
    completos = datos.copy()
    completos["temperatura_c"] = completos["temperatura_c"].interpolate()
    completos["humedad_pct"] = completos["humedad_pct"].interpolate()
    completos["lluvia_mm"] = completos["lluvia_mm"].fillna(0)
    return completos


def main() -> None:
    datos = pd.read_csv("estacion.csv")
    completos = completar(datos)

    print(f"Datos originales:\n{datos}")
    print(f"\nFaltantes por columna:\n{contar_faltantes(datos)}")
    print(f"\nDatos completados:\n{completos.round(1)}")
    print(f"\nLluvia total: {completos['lluvia_mm'].sum():.1f} mm")

    completos.round(1).to_csv("estacion_completa.csv", index=False)


main()
````

**Explicación**

**Respuesta corta:** la temperatura varía de forma **continua**, así que el valor faltante debe estar cerca de sus vecinos; rellenarla con 0 inventaría una helada a las 8 de la mañana y alteraría cualquier promedio. La lluvia, en cambio, ocurre en **eventos** discretos: interpolar entre 0.4 mm y 0.0 mm supondría que llovió 0.2 mm sin evidencia. Elegir cómo completar datos es una decisión de ingeniería, que debe documentarse.

## Ejercicio 11.8 — Consumo de agua por sector de una ciudad

El acueducto de una ciudad tiene los **macromedidores** de sus sectores en un CSV con el sector, el mes, el volumen **entregado** a la red (m³) y el volumen **facturado** a los clientes (m³). La diferencia es el **agua no contabilizada** (fugas, conexiones ilegales, errores de medición).

Escriba un programa en Python que use pandas y:

A. Defina `agregar_perdidas(consumos: pd.DataFrame) -> pd.DataFrame`: agrega las columnas `perdida_m3` y `perdida_pct` (sobre lo entregado, 1 decimal).

B. Defina `resumir_por_sector(consumos: pd.DataFrame) -> pd.DataFrame`: agrupa por sector con `groupby` y suma lo entregado, lo facturado y lo perdido; calcula el porcentaje de pérdida de cada sector y ordena de **mayor a menor** pérdida porcentual.

C. En `main()`, lea `consumo_sectores.csv`, muestre el resumen por sector, el sector **crítico** (el de mayor porcentaje) y el porcentaje de pérdida de toda la ciudad. Si algún sector supera el **30 %**, recomiende una campaña de detección de fugas en ese sector.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('consumo_sectores.csv', "w", encoding="utf-8") as archivo:
    archivo.write('sector,mes,entregado_m3,facturado_m3\nNorte,Ene,48200,36900\nNorte,Feb,46100,35800\nCentro,Ene,61500,51200\nCentro,Feb,59800,50100\nSur,Ene,39400,25300\nSur,Feb,40100,26200\nEste,Ene,27800,21900\nEste,Feb,28300,22400\n')
````

````python
#@title Solución
import pandas as pd

PERDIDA_MAXIMA_PCT = 30


def agregar_perdidas(consumos: pd.DataFrame) -> pd.DataFrame:
    """Agrega la pérdida de agua en m³ y en porcentaje."""
    consumos["perdida_m3"] = consumos["entregado_m3"] - consumos["facturado_m3"]
    consumos["perdida_pct"] = (consumos["perdida_m3"] / consumos["entregado_m3"] * 100).round(1)
    return consumos


def resumir_por_sector(consumos: pd.DataFrame) -> pd.DataFrame:
    """Suma los volúmenes de cada sector y calcula su porcentaje de pérdida."""
    resumen = consumos.groupby("sector")[["entregado_m3", "facturado_m3", "perdida_m3"]].sum()
    resumen["perdida_pct"] = (resumen["perdida_m3"] / resumen["entregado_m3"] * 100).round(1)
    return resumen.sort_values("perdida_pct", ascending=False)


def main() -> None:
    consumos = agregar_perdidas(pd.read_csv("consumo_sectores.csv"))
    resumen = resumir_por_sector(consumos)

    print(resumen)
    print(f"\nSector crítico: {resumen.index[0]} ({resumen['perdida_pct'].iloc[0]}%)")
    perdida_ciudad = consumos["perdida_m3"].sum() / consumos["entregado_m3"].sum() * 100
    print(f"Pérdida de la ciudad: {perdida_ciudad:.1f}%")

    for sector, fila in resumen.iterrows():
        if fila["perdida_pct"] > PERDIDA_MAXIMA_PCT:
            print(f"Recomendación: campaña de detección de fugas en el sector {sector}.")


main()
````

**Explicación**

`groupby("sector")` reúne las filas de cada sector y `sum()` suma sus columnas, dejando el sector como **índice** del resultado (por eso se accede con `resumen.index[0]`). Observe que el porcentaje del sector se calcula **después** de sumar los volúmenes: promediar los porcentajes mensuales daría un resultado ligeramente distinto e incorrecto si los volúmenes de cada mes son diferentes.

````python
#@title Solución
import pandas as pd


def aplicar_factor(lecturas: pd.DataFrame, medidor: str, factor: float) -> None:
    """Multiplica por el factor todas las lecturas de un medidor."""
    filas = lecturas["medidor"] == medidor
    lecturas.loc[filas, "consumo_m3"] = (lecturas.loc[filas, "consumo_m3"] * factor).round(0)


def corregir_lectura(lecturas: pd.DataFrame, medidor: str, mes: str, valor: float) -> bool:
    """Reemplaza la lectura de un medidor en un mes; retorna si existía."""
    fila = (lecturas["medidor"] == medidor) & (lecturas["mes"] == mes)
    existe = bool(fila.any())
    if existe:
        lecturas.loc[fila, "consumo_m3"] = valor
    return existe


def main() -> None:
    lecturas = pd.read_csv("lecturas_gas.csv")
    lecturas["consumo_m3"] = lecturas["consumo_m3"].astype(float)
    print(f"Primeras filas antes:\n{lecturas.iloc[:3]}")

    aplicar_factor(lecturas, "MG-3307", 1.12)

    medidor = input("\nMedidor a corregir: ").upper()
    mes = input("Mes: ").capitalize()
    valor = float(input("Lectura corregida (m³): "))
    if not corregir_lectura(lecturas, medidor, mes, valor):
        print("No existe esa lectura.")

    print(f"\nPrimeras filas después:\n{lecturas.iloc[:3]}")
    print(f"\nTabla corregida:\n{lecturas}")
    print(f"\nConsumo total por medidor:\n{lecturas.groupby('medidor')['consumo_m3'].sum()}")


main()
````

**Explicación**

**Respuesta corta:** `lecturas[filas]` crea un `DataFrame` **nuevo** (una copia filtrada), así que la asignación modificaría esa copia y no la tabla original; pandas lo advierte con un `SettingWithCopyWarning`. Con `loc[filas, columna]` se selecciona y se modifica **en un solo paso** sobre el `DataFrame` original. `iloc` selecciona por **posición** (como una lista) y `loc` por **etiquetas** o condiciones.

C. En `main()`, muestre el resumen (fechas en formato `DD/MM/AAAA`), las fechas con caudal **insuficiente** y las de **creciente**, y la **variación** del caudal de un día al siguiente (`diff()`), indicando el mayor aumento en un solo día.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('caudales_rio.csv', "w", encoding="utf-8") as archivo:
    archivo.write('fecha,caudal_m3s\n2026-10-01,18.4\n2026-10-02,16.9\n2026-10-03,14.2\n2026-10-04,11.8\n2026-10-05,10.9\n2026-10-06,27.5\n2026-10-07,48.3\n2026-10-08,39.6\n2026-10-09,25.1\n2026-10-10,19.7\n')
````

````python
#@title Solución
import pandas as pd

CAUDAL_MINIMO = 12
CAUDAL_CRECIENTE = 45
FORMATO = "%d/%m/%Y"


def leer_caudales(nombre_archivo: str) -> pd.Series:
    """Lee los caudales diarios como una Series indexada por fecha."""
    datos = pd.read_csv(nombre_archivo, index_col="fecha", parse_dates=True)
    return datos["caudal_m3s"]


def resumir(caudales: pd.Series) -> dict:
    """Resume los caudales: promedio, máximo, mínimo y sus fechas."""
    return {
        "promedio": caudales.mean(),
        "maximo": caudales.max(),
        "fecha_maximo": caudales.idxmax(),
        "minimo": caudales.min(),
        "fecha_minimo": caudales.idxmin(),
    }


def main() -> None:
    caudales = leer_caudales("caudales_rio.csv")
    resumen = resumir(caudales)

    print(f"Promedio: {resumen['promedio']:.2f} m³/s")
    print(f"Máximo: {resumen['maximo']} m³/s el {resumen['fecha_maximo'].strftime(FORMATO)}")
    print(f"Mínimo: {resumen['minimo']} m³/s el {resumen['fecha_minimo'].strftime(FORMATO)}")

    insuficientes = caudales[caudales < CAUDAL_MINIMO]
    crecientes = caudales[caudales > CAUDAL_CRECIENTE]
    print(f"\nDías con caudal insuficiente: {list(insuficientes.index.strftime(FORMATO))}")
    print(f"Días de creciente: {list(crecientes.index.strftime(FORMATO))}")

    variacion = caudales.diff()
    print(f"\nVariación diaria (m³/s):\n{variacion.round(1)}")
    print(f"Mayor aumento en un día: {variacion.max():.1f} m³/s el {variacion.idxmax().strftime(FORMATO)}")


main()
````

**Explicación**

Una `Series` es como una columna con **etiquetas**: al filtrarla, se conservan las fechas de los valores seleccionados, y `idxmax()` retorna la **etiqueta** (la fecha) del máximo, no su posición. `diff()` resta a cada valor el anterior; el primer día queda como `NaN` porque no tiene un día anterior con el cual compararse.

Escriba un programa en Python que use pandas y:

C. En `main()`, lea `produccion_semana.csv`, genere `reporte_produccion.xlsx` y, para **verificar**, vuelva a leer la hoja `Resumen` del archivo Excel con `pd.read_excel(..., sheet_name="Resumen")` y muéstrela, junto con la línea con **menor** porcentaje de rechazo.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('produccion_semana.csv', "w", encoding="utf-8") as archivo:
    archivo.write('dia,linea,producidas,rechazadas\nLun,L1,12400,186\nLun,L2,9800,245\nMar,L1,12650,152\nMar,L2,10100,268\nMie,L1,11900,214\nMie,L2,9650,193\nJue,L1,12800,141\nJue,L2,10300,309\n')
````

````python
#@title Solución
import pandas as pd


def resumir_lineas(produccion: pd.DataFrame) -> pd.DataFrame:
    """Totales y porcentaje de rechazo por línea de envasado."""
    resumen = produccion.groupby("linea")[["producidas", "rechazadas"]].sum()
    resumen["rechazo_pct"] = (resumen["rechazadas"] / resumen["producidas"] * 100).round(2)
    return resumen.reset_index()


def exportar_reporte(produccion: pd.DataFrame, resumen: pd.DataFrame, nombre_archivo: str) -> None:
    """Escribe el reporte en un archivo Excel con dos hojas."""
    with pd.ExcelWriter(nombre_archivo) as escritor:
        produccion.to_excel(escritor, sheet_name="Datos", index=False)
        resumen.to_excel(escritor, sheet_name="Resumen", index=False)


def main() -> None:
    produccion = pd.read_csv("produccion_semana.csv")
    resumen = resumir_lineas(produccion)
    exportar_reporte(produccion, resumen, "reporte_produccion.xlsx")
    print("Reporte exportado: reporte_produccion.xlsx")

    verificacion = pd.read_excel("reporte_produccion.xlsx", sheet_name="Resumen")
    print(f"\nHoja 'Resumen' leída desde Excel:\n{verificacion}")
    mejor = verificacion.loc[verificacion["rechazo_pct"].idxmin()]
    print(f"\nLínea con menor rechazo: {mejor['linea']} ({mejor['rechazo_pct']}%)")


main()
````

**Explicación**

Para escribir **varias hojas** en un mismo archivo se usa `pd.ExcelWriter` dentro de un `with`; escribir dos veces con `to_excel("archivo.xlsx")` reemplazaría el archivo y dejaría solo la última hoja. `reset_index()` convierte el índice creado por `groupby` (la línea) en una columna normal, para que aparezca como tal en Excel. Leer el archivo generado es una buena forma de **verificar** la exportación.

````python
#@title Solución
import pandas as pd


def leer_espectrometro(nombre_archivo: str) -> pd.DataFrame:
    """Lee un CSV exportado en formato europeo."""
    return pd.read_csv(nombre_archivo, sep=";", decimal=",")


def verificar_composicion(muestras: pd.DataFrame) -> pd.DataFrame:
    """Agrega la suma de la composición y si es válida."""
    elementos = muestras.columns[1:]
    muestras["total_pct"] = muestras[elementos].sum(axis=1).round(2)
    muestras["valida"] = muestras["total_pct"].between(99.5, 100.5)
    return muestras


def main() -> None:
    incorrecto = pd.read_csv("espectrometro.csv")
    print(f"Sin parámetros: forma {incorrecto.shape}, columnas {list(incorrecto.columns)}")

    muestras = verificar_composicion(leer_espectrometro("espectrometro.csv"))
    print(f"\nLectura correcta:\n{muestras}")

    muestras.to_csv("espectrometro_cr.csv", index=False)
    print("\nExportado en formato estándar: espectrometro_cr.csv")


main()
````

**Explicación**

Sin `sep=";"`, pandas busca comas para separar columnas y encuentra las comas **decimales**, por lo que produce una tabla deformada.
````python
#@title Solución
import numpy as np
import pandas as pd


def clasificar(muestras: pd.DataFrame) -> pd.DataFrame:
    """Agrega la clase según el oxígeno disuelto y si la muestra es apta para riego."""
    condiciones = [muestras["od_mgl"] >= 6, muestras["od_mgl"] >= 4]
    muestras["clase"] = np.select(condiciones, ["Buena", "Regular"], default="Mala")
    muestras["apta_riego"] = (muestras["conductividad_uscm"] < 750) & muestras["ph"].between(6.5, 8.4)
    return muestras


def main() -> None:
    muestras = clasificar(pd.read_csv("calidad_agua.csv"))

    print(muestras)
    print(f"\nMuestras por clase:\n{muestras.groupby('clase').size()}")
    print(f"\nOD promedio por río (mg/L):\n{muestras.groupby('rio')['od_mgl'].mean().round(2)}")
    print(f"\nNo aptas para riego: {list(muestras.loc[~muestras['apta_riego'], 'muestra'])}")

    muestras.to_csv("calidad_agua_clasificada.csv", index=False)


main()
````

**Explicación**

`np.select` evalúa las condiciones **en orden** y asigna el valor de la **primera** que se cumple, igual que una cadena de `if`/`elif`; `default` es el equivalente del `else`. Por eso la segunda condición puede ser simplemente `od >= 4`: las muestras con 6 o más ya quedaron clasificadas como "Buena" por la primera.

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('fallas_linea.csv', "w", encoding="utf-8") as archivo:
    archivo.write('maquina,tipo,inicio,fin\nLlenadora,Atasco,2026-08-03 06:42,2026-08-03 06:58\nTaponadora,Sensor,2026-08-03 07:15,2026-08-03 07:22\nLlenadora,Atasco,2026-08-03 08:30,2026-08-03 08:51\nEtiquetadora,Eléctrica,2026-08-03 09:05,2026-08-03 10:12\nLlenadora,Sensor,2026-08-03 11:20,2026-08-03 11:29\nTaponadora,Atasco,2026-08-03 12:02,2026-08-03 12:14\nLlenadora,Atasco,2026-08-03 13:40,2026-08-03 13:55\n')
````

````python
#@title Solución
import pandas as pd


def calcular_duraciones(fallas: pd.DataFrame) -> pd.DataFrame:
    """Agrega la duración en minutos de cada paro."""
    fallas["inicio"] = pd.to_datetime(fallas["inicio"])
    fallas["fin"] = pd.to_datetime(fallas["fin"])
    fallas["duracion_min"] = (fallas["fin"] - fallas["inicio"]).dt.total_seconds() / 60
    return fallas


def indicadores_por_maquina(fallas: pd.DataFrame, horas_turno: float) -> pd.DataFrame:
    """Cantidad de fallas, minutos de paro y disponibilidad de cada máquina."""
    indicadores = fallas.groupby("maquina").agg(
        fallas=("tipo", "count"),
        paro_min=("duracion_min", "sum"),
    )
    minutos_turno = horas_turno * 60
    indicadores["disponibilidad_pct"] = ((minutos_turno - indicadores["paro_min"])
                                         / minutos_turno * 100).round(1)
    return indicadores.sort_values("disponibilidad_pct")


def main() -> None:
    fallas = calcular_duraciones(pd.read_csv("fallas_linea.csv"))
    horas_turno = float(input("Horas del turno: "))

    print(f"\n{indicadores_por_maquina(fallas, horas_turno)}")

    por_tipo = fallas.groupby("tipo")["duracion_min"]
    frecuencias = por_tipo.count()
    print(f"\nFalla más frecuente: {frecuencias.idxmax()} ({frecuencias.max()} veces)")
    print(f"Tiempo promedio de reparación por tipo (min):\n{por_tipo.mean().round(1)}")


main()
````

**Explicación**

Restar dos columnas de fechas produce una columna de **duraciones**; su accesor `.dt` permite obtener, por ejemplo, los segundos totales. `agg` permite calcular varias agregaciones distintas a la vez y darles nombre: `fallas=("tipo", "count")` significa "crear la columna `fallas` contando los valores de `tipo`". La Etiquetadora tuvo **una sola** falla, pero fue la más larga: contar fallas no basta para decidir dónde intervenir.

## Ejercicio 11.15 — Detección de defectos en una lámina metálica

Escriba un programa en Python que use Pillow (`from PIL import Image`) y NumPy, con las funciones:

C. `guardar_mascara(mascara: np.ndarray, nombre_archivo: str) -> None`: guarda una imagen en blanco y negro donde los defectos aparecen en **blanco** (use `Image.fromarray` con un arreglo `uint8`).

D. `main() -> None`: solicita el umbral, muestra el tamaño de la imagen, su brillo promedio, la cantidad de píxeles defectuosos, el **porcentaje de área** defectuosa, la **región** (filas y columnas mínima y máxima) donde están los defectos, y el dictamen: la lámina se rechaza si más del **0.5 %** del área tiene defectos. Guarde la máscara en `defectos.png`.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

````python
#@title Preparación: crear los archivos de datos del ejercicio
# Crea una imagen sintética de una lámina brillante con un rayón y una mancha de óxido
import numpy as np
from PIL import Image

generador = np.random.default_rng(3)
lamina = 200 + generador.normal(0, 6, (240, 320))
lamina[118:123, 40:210] = 60             # rayón horizontal
filas, columnas = np.ogrid[:240, :320]
mancha = (filas - 70) ** 2 + (columnas - 250) ** 2 < 14 ** 2
lamina[mancha] = 85                      # mancha de óxido
Image.fromarray(np.clip(lamina, 0, 255).astype(np.uint8)).save("lamina.jpg", quality=95)
````

````python
#@title Solución
import numpy as np
from PIL import Image

AREA_MAXIMA_DEFECTOS = 0.005


def cargar_en_grises(nombre_archivo: str) -> np.ndarray:
    """Abre una imagen y la retorna como arreglo en escala de grises."""
    return np.array(Image.open(nombre_archivo).convert("L"))


def detectar_defectos(imagen: np.ndarray, umbral: int) -> np.ndarray:
    """Retorna una máscara con los píxeles más oscuros que el umbral."""
    return imagen < umbral


def guardar_mascara(mascara: np.ndarray, nombre_archivo: str) -> None:
    """Guarda la máscara con los defectos en blanco y el fondo en negro."""
    Image.fromarray((mascara * 255).astype(np.uint8)).save(nombre_archivo)


def main() -> None:
    umbral = int(input("Umbral de detección (0-255): "))

    imagen = cargar_en_grises("lamina.jpg")
    mascara = detectar_defectos(imagen, umbral)
    defectuosos = int(mascara.sum())
    porcentaje = defectuosos / mascara.size

    print(f"\nTamaño de la imagen: {imagen.shape[1]} x {imagen.shape[0]} píxeles")
    print(f"Brillo promedio: {imagen.mean():.1f}")
    print(f"Píxeles defectuosos: {defectuosos} ({porcentaje:.2%} del área)")

    if defectuosos > 0:
        filas, columnas = np.where(mascara)
        print(f"Región con defectos: filas {filas.min()}-{filas.max()}, columnas {columnas.min()}-{columnas.max()}")

    if porcentaje > AREA_MAXIMA_DEFECTOS:
        print("Dictamen: LÁMINA RECHAZADA")
    else:
        print("Dictamen: LÁMINA ACEPTADA")

    guardar_mascara(mascara, "defectos.png")
    print("Máscara guardada en defectos.png")


main()
````

**Explicación**

Una imagen en escala de grises es simplemente una **matriz** de NumPy: cada elemento es el brillo de un píxel. Por eso toda la inspección se hace con operaciones ya conocidas: comparaciones para crear la máscara, `sum()` para contar y `np.where` para encontrar las posiciones. La máscara se guarda en **PNG** y no en JPEG, porque la compresión JPEG altera los valores y "ensuciaría" los bordes de una imagen en blanco y negro.

## Ejercicio 11.16 — Cobertura vegetal desde una imagen de dron

Una empresa de ingeniería agrícola toma fotografías aéreas de un cultivo con un **dron** para estimar el porcentaje del terreno cubierto por plantas. Un índice sencillo es el **exceso de verde**:

$$ExG = 2G - R - B$$

donde $R$, $G$ y $B$ son los valores de los canales rojo, verde y azul de cada píxel. Un píxel se considera **vegetación** si su ExG supera un umbral.

Escriba un programa en Python que use Pillow y NumPy, con las funciones:

A. `calcular_exceso_verde(nombre_archivo: str) -> np.ndarray`: abre `cultivo.jpg` en modo RGB, lo convierte a arreglo (forma `(alto, ancho, 3)`), separa los canales con *slicing* (`imagen[:, :, 0]`, etc.) y calcula el ExG. **Importante:** convierta el arreglo a `int` antes de operar (`astype(int)`), porque el tipo `uint8` no admite negativos y "da la vuelta" al pasar de 255.

B. `cobertura_por_franja(vegetacion: np.ndarray, franjas: int) -> list[float]`: divide la imagen en franjas **verticales** del mismo ancho (columnas) y retorna el porcentaje de vegetación de cada una.

C. `main() -> None`: solicita el umbral y el área real que cubre la foto (m²), y muestra la cobertura total (%), el área con vegetación (m²) y la cobertura de 4 franjas, indicando la franja con **menor** cobertura (posible zona con problemas de riego).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Área real de la foto (m²): <b>1200</b>

Área con vegetación: 580.8 m²
Cobertura por franja (%): [51.0, 51.0, 51.0, 40.6]
Franja con menor cobertura: 4
</pre>

**Pregunta de respuesta corta**

Con arreglos `uint8`, ¿cuánto da `np.array([200], dtype=np.uint8) * 2`? ¿Por qué eso arruinaría el cálculo del ExG?

````python
#@title Preparación: crear los archivos de datos del ejercicio
# Crea una imagen sintética de un cultivo: suelo café con hileras de plantas verdes
import numpy as np
from PIL import Image

generador = np.random.default_rng(8)
alto, ancho = 200, 400
imagen = np.zeros((alto, ancho, 3))
imagen[:, :] = [140, 105, 75]                          # suelo
for fila in range(15, alto, 30):                       # hileras de plantas
    imagen[fila:fila + 14, :] = [60, 150, 55]
imagen[:, 300:400][generador.random((alto, 100)) < 0.55] = [140, 105, 75]  # zona seca
imagen += generador.normal(0, 8, imagen.shape)
Image.fromarray(np.clip(imagen, 0, 255).astype(np.uint8)).save("cultivo.jpg", quality=95)
````

````python
#@title Solución
import numpy as np
from PIL import Image


def calcular_exceso_verde(nombre_archivo: str) -> np.ndarray:
    """Calcula el índice de exceso de verde (2G - R - B) de cada píxel."""
    imagen = np.array(Image.open(nombre_archivo).convert("RGB")).astype(int)
    rojo = imagen[:, :, 0]
    verde = imagen[:, :, 1]
    azul = imagen[:, :, 2]
    return 2 * verde - rojo - azul


def cobertura_por_franja(vegetacion: np.ndarray, franjas: int) -> list[float]:
    """Porcentaje de vegetación en franjas verticales del mismo ancho."""
    ancho_franja = vegetacion.shape[1] // franjas
    coberturas = []
    for franja in range(franjas):
        columnas = vegetacion[:, franja * ancho_franja:(franja + 1) * ancho_franja]
        coberturas.append(round(float(columnas.mean()) * 100, 1))
    return coberturas


def main() -> None:
    umbral = int(input("Umbral de ExG: "))
    area_foto = float(input("Área real de la foto (m²): "))

    exceso_verde = calcular_exceso_verde("cultivo.jpg")
    vegetacion = exceso_verde > umbral
    cobertura = vegetacion.mean()
    franjas = cobertura_por_franja(vegetacion, 4)

    print(f"\nImagen de {vegetacion.shape[1]} x {vegetacion.shape[0]} píxeles")
    print(f"Cobertura vegetal: {cobertura:.1%}")
    print(f"Área con vegetación: {cobertura * area_foto:.1f} m²")
    print(f"Cobertura por franja (%): {franjas}")
    print(f"Franja con menor cobertura: {franjas.index(min(franjas)) + 1}")


main()
````

**Explicación**

`float(...)` convierte el resultado de NumPy (un `np.float64`) en un `float` de Python, para que la lista se imprima de forma legible.

**Respuesta corta:** da **144**, no 400: un `uint8` solo representa enteros de 0 a 255, y al pasarse "da la vuelta" (400 − 256 = 144). Del mismo modo, una resta que debería ser negativa se convertiría en un número grande positivo. El ExG quedaría lleno de valores sin sentido, por eso se convierte primero a `int`.

Escriba un programa en Python que use pandas y:

A. Defina `marcar_anomalos(ciclos: pd.DataFrame) -> pd.DataFrame`: agrega la columna `z` (puntaje estandarizado: $(x - \bar{x}) / \sigma$, 2 decimales) y la columna `anomalo`.

B. Defina `escribir_informe(ciclos: pd.DataFrame, nombre_archivo: str) -> None`: escribe un informe de texto con el resumen estadístico del tiempo de ciclo (`describe()` convertido a texto con `to_string()`), la cantidad de anomalías y una línea por cada pieza anómala.

C. En `main()`, lea `ciclos_robot.csv`, marque los anómalos, escriba `informe_robot.txt` y muestre la cantidad de piezas analizadas, la cantidad de anomalías y el **tiempo de ciclo promedio sin las anomalías**, que es el valor que se usaría para planificar la producción.

````python
#@title Solución
import pandas as pd

LIMITE_Z = 2


def marcar_anomalos(ciclos: pd.DataFrame) -> pd.DataFrame:
    """Agrega el puntaje estandarizado y la marca de anomalía de cada ciclo."""
    media = ciclos["tiempo_s"].mean()
    desviacion = ciclos["tiempo_s"].std()
    ciclos["z"] = ((ciclos["tiempo_s"] - media) / desviacion).round(2)
    ciclos["anomalo"] = ciclos["z"].abs() > LIMITE_Z
    return ciclos


def escribir_informe(ciclos: pd.DataFrame, nombre_archivo: str) -> None:
    """Escribe el informe de tiempos de ciclo en un archivo de texto."""
    anomalos = ciclos[ciclos["anomalo"]]
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write("INFORME DE TIEMPOS DE CICLO - ROBOT DE SOLDADURA\n\n")
        archivo.write(ciclos["tiempo_s"].describe().round(2).to_string() + "\n\n")
        archivo.write(f"Ciclos anómalos: {len(anomalos)}\n")
        for _, fila in anomalos.iterrows():
            archivo.write(f"  Pieza {fila['pieza']}: {fila['tiempo_s']} s (z = {fila['z']})\n")


def main() -> None:
    ciclos = marcar_anomalos(pd.read_csv("ciclos_robot.csv"))
    escribir_informe(ciclos, "informe_robot.txt")

    normales = ciclos.loc[~ciclos["anomalo"], "tiempo_s"]
    print(f"Piezas analizadas: {len(ciclos)}")
    print(f"Anomalías: {ciclos['anomalo'].sum()}")
    print(f"Tiempo de ciclo promedio sin anomalías: {normales.mean():.2f} s")
    print("Informe escrito en informe_robot.txt")


main()
````

**Explicación**

`to_string()` convierte una `Series` o un `DataFrame` en el mismo texto que se vería con `print`, listo para escribirse en un archivo. `iterrows()` recorre las filas de un `DataFrame` como pares `(índice, fila)`; el índice no se necesita, por eso se nombra `_`. Las dos anomalías, una lenta y una rápida, desplazan poco la media, pero **inflan** la desviación estándar.

````python
#@title Solución
import pandas as pd

FACTOR_CO2_KG_KWH = 0.0385
TARIFA_KWH = 98


def agregar_indicadores(edificios: pd.DataFrame) -> pd.DataFrame:
    """Agrega la intensidad energética y las emisiones de CO₂."""
    edificios["kwh_m2"] = (edificios["consumo_kwh"] / edificios["area_m2"]).round(1)
    edificios["co2_t"] = (edificios["consumo_kwh"] * FACTOR_CO2_KG_KWH / 1000).round(1)
    return edificios


def ranking(edificios: pd.DataFrame, columna: str, cantidad: int) -> pd.DataFrame:
    """Los edificios con mayor valor en la columna indicada."""
    return edificios.sort_values(columna, ascending=False).head(cantidad)


def main() -> None:
    edificios = agregar_indicadores(pd.read_csv("edificios_campus.csv"))

    print(f"Mayor consumo total:\n{ranking(edificios, 'consumo_kwh', 3)[['edificio', 'consumo_kwh']]}")
    print(f"\nMayor intensidad:\n{ranking(edificios, 'kwh_m2', 3)[['edificio', 'kwh_m2']]}")

    promedio = edificios["kwh_m2"].mean()
    ineficientes = edificios[edificios["kwh_m2"] > promedio]
    print(f"\nIntensidad promedio: {promedio:.1f} kWh/m²")
    print(f"Sobre el promedio: {list(ineficientes['edificio'])}")

    reduccion = float(input("\nReducción esperada (%): ")) / 100
    ahorro_kwh = ineficientes["consumo_kwh"].sum() * reduccion
    print(f"Ahorro anual: {ahorro_kwh:,.0f} kWh = ₡{ahorro_kwh * TARIFA_KWH:,.0f}")
    print(f"Emisiones evitadas: {ahorro_kwh * FACTOR_CO2_KG_KWH / 1000:.1f} t de CO₂")


main()
````

**Explicación**

Los dos rankings cuentan historias distintas: Ingeniería es el edificio que **más** consume, pero el Centro de Cómputo, con un área mucho menor, es el más **intenso**. Normalizar por el área (o por la cantidad de usuarios, o de horas) es clave para comparar de forma justa, y es una operación de una línea con pandas.

````python
#@title Solución
import pandas as pd

ESPESOR_MINIMO = 48
ESPESOR_MAXIMO = 52


def leer_turnos(nombres_archivo: dict[str, str]) -> pd.DataFrame:
    """Lee los archivos de cada turno y los une en un solo DataFrame."""
    tablas = []
    for turno, nombre_archivo in nombres_archivo.items():
        tabla = pd.read_csv(nombre_archivo)
        tabla["turno"] = turno
        tablas.append(tabla)
    return pd.concat(tablas, ignore_index=True)


def resumir_turnos(produccion: pd.DataFrame) -> pd.DataFrame:
    """Indicadores de producción y calidad de cada turno."""
    produccion["aceptable"] = produccion["espesor_um"].between(ESPESOR_MINIMO, ESPESOR_MAXIMO)
    resumen = produccion.groupby("turno").agg(
        rollos=("rollo", "count"),
        kg=("peso_kg", "sum"),
        espesor_prom=("espesor_um", "mean"),
        aceptables_pct=("aceptable", "mean"),
    )
    resumen["espesor_prom"] = resumen["espesor_prom"].round(2)
    resumen["aceptables_pct"] = (resumen["aceptables_pct"] * 100).round(1)
    return resumen.sort_values("aceptables_pct", ascending=False)


def main() -> None:
    archivos = {"Mañana": "turno_manana.csv", "Tarde": "turno_tarde.csv", "Noche": "turno_noche.csv"}
    produccion = leer_turnos(archivos)
    resumen = resumir_turnos(produccion)

    print(f"Rollos del día: {len(produccion)} ({produccion['peso_kg'].sum():.1f} kg)")
    print(f"\n{resumen}")

    produccion.to_csv("produccion_dia.csv", index=False)
    print("\nConsolidado exportado a produccion_dia.csv")


main()
````

**Explicación**

`pd.concat` "apila" varios `DataFrame` con las mismas columnas, uno debajo del otro. Sin `ignore_index=True`, el resultado tendría índices repetidos (0, 1, 2, 0, 1, 2, 3, ...), porque cada tabla conserva los suyos. Agregar la columna `turno` **antes** de unir las tablas es lo que permite saber después de dónde vino cada rollo.

Un laboratorio de metrología calibró los **manómetros** de una planta química comparando cada uno con un manómetro **patrón** en varios puntos de presión. El archivo de resultados tiene algunas lecturas faltantes (el instrumento no respondió), y los técnicos escribieron los códigos de los manómetros sin un formato uniforme.

Un manómetro **aprueba** si, en **todos** sus puntos, el error absoluto no supera el **1 %** del alcance máximo de su escala (su *fondo de escala*).

Escriba un programa en Python que use pandas y:

A. Defina `limpiar(datos: pd.DataFrame) -> pd.DataFrame`: normaliza los códigos (mayúsculas y sin espacios, con `str.upper()` y `str.strip()`) y **elimina** las filas con lecturas faltantes (`dropna`), informando cuántas se eliminaron.

B. Defina `evaluar(datos: pd.DataFrame) -> pd.DataFrame`: agrega el error (lectura − patrón), el error como porcentaje del fondo de escala y si ese punto cumple.

C. Defina `resumir(datos: pd.DataFrame) -> pd.DataFrame`: agrupa por manómetro y calcula la cantidad de puntos, el error máximo en valor absoluto (%) y si **aprueba** (todos sus puntos cumplen: agregación `"all"`).

D. En `main()`, lea `calibracion_manometros.csv`, muestre el resumen, exporte el detalle a `detalle_calibracion.csv` y escriba `certificados.txt` con una línea por manómetro: `APROBADO` o `RECHAZADO` y su error máximo.

**Pregunta de respuesta corta**

¿Qué habría pasado con el agrupamiento si **no** se hubieran normalizado los códigos de los manómetros?

````python
#@title Preparación: crear los archivos de datos del ejercicio
with open('calibracion_manometros.csv', "w", encoding="utf-8") as archivo:
    archivo.write('manometro,fondo_escala_bar,patron_bar,lectura_bar\npi-101,10,2.0,2.04\nPI-101 ,10,5.0,5.06\nPI-101,10,8.0,8.07\npi-102,16,4.0,4.21\nPI-102,16,8.0,\nPI-102,16,12.0,12.19\nPI-103,25,5.0,5.10\npi-103,25,15.0,15.18\nPI-103,25,20.0,20.35\n')
````

````python
#@title Solución
import pandas as pd

ERROR_MAXIMO_PCT = 1


def limpiar(datos: pd.DataFrame) -> pd.DataFrame:
    """Normaliza los códigos y elimina las filas con lecturas faltantes."""
    datos["manometro"] = datos["manometro"].str.upper().str.strip()
    limpios = datos.dropna()
    print(f"Filas eliminadas por datos faltantes: {len(datos) - len(limpios)}")
    return limpios.copy()


def evaluar(datos: pd.DataFrame) -> pd.DataFrame:
    """Agrega el error de cada punto y si cumple la tolerancia."""
    datos["error_bar"] = (datos["lectura_bar"] - datos["patron_bar"]).round(3)
    datos["error_pct_fe"] = (datos["error_bar"] / datos["fondo_escala_bar"] * 100).round(2)
    datos["cumple"] = datos["error_pct_fe"].abs() <= ERROR_MAXIMO_PCT
    return datos


def resumir(datos: pd.DataFrame) -> pd.DataFrame:
    """Resultado de la calibración de cada manómetro."""
    datos["error_abs_pct"] = datos["error_pct_fe"].abs()
    return datos.groupby("manometro").agg(
        puntos=("cumple", "count"),
        error_max_pct=("error_abs_pct", "max"),
        aprueba=("cumple", "all"),
    )


def main() -> None:
    datos = evaluar(limpiar(pd.read_csv("calibracion_manometros.csv")))
    resumen = resumir(datos)

    print(f"\n{resumen}")
    datos.drop(columns="error_abs_pct").to_csv("detalle_calibracion.csv", index=False)

    with open("certificados.txt", "w", encoding="utf-8") as archivo:
        for manometro, fila in resumen.iterrows():
            estado = "APROBADO" if fila["aprueba"] else "RECHAZADO"
            archivo.write(f"{manometro}: {estado} (error máximo {fila['error_max_pct']}% FE)\n")
    print("\nArchivos generados: detalle_calibracion.csv, certificados.txt")


main()
````

**Explicación**

**Respuesta corta:** `groupby` habría tratado `"pi-101"`, `"PI-101 "` y `"PI-101"` como **tres manómetros distintos**, cada uno con un solo punto, y el resumen tendría grupos falsos. La limpieza de datos (formatos, espacios, mayúsculas, faltantes) casi siempre es el primer paso de un análisis real. El accesor `.str` aplica los métodos de cadenas a toda una columna.
