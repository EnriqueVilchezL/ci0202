## Ejercicio 3.1 — Empaque de tornillos

Una fábrica de tornillos empaca su producción en cajas de capacidad fija. Escriba un programa que solicite la **cantidad de tornillos producidos** y la **capacidad de cada caja** (ambos enteros) y calcule, usando **solo operadores aritméticos**:

1. La cantidad de **cajas llenas**.
2. La cantidad de **tornillos sobrantes** (que no completan una caja).
3. La cantidad de **cajas necesarias** para empacar **todo**, incluyendo una caja incompleta si hay sobrantes.
4. Un booleano `empaque_exacto` que sea `True` si no sobran tornillos.

*Pista:* para dividir `a` entre `b` redondeando hacia arriba, puede usar `(a + b - 1) // b`.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Tornillos producidos: <b>12530</b>
Capacidad de cada caja: <b>250</b>
Cajas llenas: 50
Tornillos sobrantes: 30
Cajas necesarias: 51
¿Empaque exacto?: False
</pre>

<pre>
Tornillos producidos: <b>5000</b>
Capacidad de cada caja: <b>250</b>
Cajas llenas: 20
Tornillos sobrantes: 0
Cajas necesarias: 20
¿Empaque exacto?: True
</pre>

````python
#@title Solución
tornillos_producidos = int(input("Tornillos producidos: "))
capacidad_caja = int(input("Capacidad de cada caja: "))

cajas_llenas = tornillos_producidos // capacidad_caja
tornillos_sobrantes = tornillos_producidos % capacidad_caja
cajas_necesarias = (tornillos_producidos + capacidad_caja - 1) // capacidad_caja
empaque_exacto = tornillos_sobrantes == 0

print(f"Cajas llenas: {cajas_llenas}")
print(f"Tornillos sobrantes: {tornillos_sobrantes}")
print(f"Cajas necesarias: {cajas_necesarias}")
print(f"¿Empaque exacto?: {empaque_exacto}")
````

**Explicación**

Sumar `capacidad_caja - 1` antes de la división entera hace que cualquier sobrante, por pequeño que sea, "empuje" el resultado a la siguiente caja; si no hay sobrante, la suma no alcanza para cambiar el cociente.

## Ejercicio 3.2 — Dígito verificador de una pieza

Una empresa aeroespacial identifica cada pieza con un número de **4 dígitos** y le agrega un **dígito verificador** para detectar errores de digitación. El dígito se calcula así:

1. Se multiplica cada dígito por un peso: el de los **millares** por 5, las **centenas** por 4, las **decenas** por 3 y las **unidades** por 2.
2. Se suman los productos.
3. El dígito verificador es el **residuo** de dividir esa suma entre 10.

Escriba un programa que solicite el número de la pieza (entero), extraiga cada dígito **usando `//` y `%`** (sin convertir a texto), calcule el dígito verificador y muestre el **código completo** (el número multiplicado por 10 más el dígito verificador).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Número de pieza (4 dígitos): <b>4817</b>
Dígitos: 4 8 1 7
Suma ponderada: 69
Dígito verificador: 9
Código completo: 48179
</pre>

````python
#@title Solución
numero_pieza = int(input("Número de pieza (4 dígitos): "))

millares = numero_pieza // 1000
centenas = numero_pieza // 100 % 10
decenas = numero_pieza // 10 % 10
unidades = numero_pieza % 10

suma_ponderada = millares * 5 + centenas * 4 + decenas * 3 + unidades * 2
digito_verificador = suma_ponderada % 10
codigo_completo = numero_pieza * 10 + digito_verificador

print(f"Dígitos: {millares} {centenas} {decenas} {unidades}")
print(f"Suma ponderada: {suma_ponderada}")
print(f"Dígito verificador: {digito_verificador}")
print(f"Código completo: {codigo_completo}")
````

**Explicación**

`numero // 100 % 10` se evalúa de **izquierda a derecha** porque `//` y `%` tienen la misma precedencia: primero se eliminan las dos últimas cifras y luego se toma la última cifra que queda.

## Ejercicio 3.3 — Cohete de agua

Un equipo de estudiantes de ingeniería lanza un cohete de agua **verticalmente**. Si se desprecia la resistencia del aire, con una velocidad inicial $v_0$ (m/s) y $g = 9.81 \text{ m/s}^2$:

$$h_{max} = \frac{v_0^2}{2g} \qquad t_{subida} = \frac{v_0}{g} \qquad t_{vuelo} = 2 \cdot t_{subida}$$

Escriba un programa que solicite $v_0$ y la **altura objetivo** de la competencia, y muestre la altura máxima, el tiempo de subida y el tiempo total de vuelo (2 decimales), además de un booleano que indique si el cohete **alcanza o supera** la altura objetivo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Velocidad inicial (m/s): <b>24</b>
Altura objetivo (m): <b>30</b>
Altura máxima: 29.36 m
Tiempo de subida: 2.45 s
Tiempo de vuelo: 4.89 s
¿Alcanza el objetivo?: False
</pre>

````python
#@title Solución
GRAVEDAD = 9.81

velocidad_inicial = float(input("Velocidad inicial (m/s): "))
altura_objetivo = float(input("Altura objetivo (m): "))

altura_maxima = velocidad_inicial ** 2 / (2 * GRAVEDAD)
tiempo_subida = velocidad_inicial / GRAVEDAD
tiempo_vuelo = 2 * tiempo_subida
alcanza_objetivo = altura_maxima >= altura_objetivo

print(f"Altura máxima: {altura_maxima:.2f} m")
print(f"Tiempo de subida: {tiempo_subida:.2f} s")
print(f"Tiempo de vuelo: {tiempo_vuelo:.2f} s")
print(f"¿Alcanza el objetivo?: {alcanza_objetivo}")
````

**Explicación**

Los paréntesis en `(2 * GRAVEDAD)` son **indispensables**: sin ellos, `v ** 2 / 2 * g` se evaluaría como `(v ** 2 / 2) * g`, porque `/` y `*` tienen la misma precedencia y se asocian de izquierda a derecha.

## Ejercicio 3.4 — Rango de operación de un sensor

Un transductor de presión de una tubería de gas funciona correctamente entre un **mínimo** y un **máximo** de operación. Escriba un programa que solicite la lectura, el mínimo y el máximo (decimales) y calcule:

1. `en_rango`: `True` si la lectura está entre el mínimo y el máximo, **inclusive** (use una comparación encadenada).
2. `fuera_de_rango`: la negación del anterior.
3. `lectura_invalida`: `True` si la lectura es **negativa** (un transductor de presión absoluta no puede medir valores negativos).
4. `activar_alarma`: `True` si la lectura está fuera de rango **o** es inválida.
5. El **porcentaje de uso** del rango: $\frac{\text{lectura} - \text{mínimo}}{\text{máximo} - \text{mínimo}} \cdot 100$.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Lectura (bar): <b>7.8</b>
Presión mínima (bar): <b>2</b>
Presión máxima (bar): <b>8</b>
¿En rango?: True
¿Fuera de rango?: False
¿Lectura inválida?: False
¿Activar alarma?: False
Uso del rango: 96.7%
</pre>

<pre>
Lectura (bar): <b>9.1</b>
Presión mínima (bar): <b>2</b>
Presión máxima (bar): <b>8</b>
¿En rango?: False
¿Fuera de rango?: True
¿Lectura inválida?: False
¿Activar alarma?: True
Uso del rango: 118.3%
</pre>

````python
#@title Solución
lectura = float(input("Lectura (bar): "))
presion_minima = float(input("Presión mínima (bar): "))
presion_maxima = float(input("Presión máxima (bar): "))

en_rango = presion_minima <= lectura <= presion_maxima
fuera_de_rango = not en_rango
lectura_invalida = lectura < 0
activar_alarma = fuera_de_rango or lectura_invalida
porcentaje_uso = (lectura - presion_minima) / (presion_maxima - presion_minima) * 100

print(f"¿En rango?: {en_rango}")
print(f"¿Fuera de rango?: {fuera_de_rango}")
print(f"¿Lectura inválida?: {lectura_invalida}")
print(f"¿Activar alarma?: {activar_alarma}")
print(f"Uso del rango: {porcentaje_uso:.1f}%")
````

**Explicación**

`a <= x <= b` es equivalente a `a <= x and x <= b`. Un porcentaje de uso mayor que 100 % indica que la lectura sobrepasó el máximo.

## Ejercicio 3.5 — Esfuerzo en una viga de madera

Un ingeniero estructural revisa una viga de madera **simplemente apoyada** de sección rectangular, con una carga distribuida uniforme. Las fórmulas son:

$$M = \frac{w \cdot L^2}{8} \qquad I = \frac{b \cdot h^3}{12} \qquad c = \frac{h}{2} \qquad \sigma = \frac{M \cdot c}{I}$$

donde $w$ es la carga (N/m), $L$ la luz de la viga (m), $b$ y $h$ el ancho y la altura de la sección (m), $M$ el momento máximo (N·m), $I$ el momento de inercia (m⁴) y $\sigma$ el esfuerzo máximo (Pa).

Escriba un programa en Python que:

A. Solicite $w$, $L$, $b$ y $h$.

B. Calcule $M$, $I$, $c$ y $\sigma$, escribiendo **cada fórmula en una sola línea** y usando paréntesis **solo donde sean necesarios**.

C. Muestre el esfuerzo en **megapascales** (1 MPa = 10⁶ Pa) con 2 decimales y un booleano `viga_segura` que sea `True` si el esfuerzo **no supera** el esfuerzo admisible de la madera, **10 MPa**.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Carga distribuida w (N/m): <b>2500</b>
Luz de la viga L (m): <b>4</b>
Ancho de la sección b (m): <b>0.1</b>
Altura de la sección h (m): <b>0.25</b>

Momento máximo: 5000.00 N·m
Momento de inercia: 0.00013021 m⁴
Esfuerzo máximo: 4.80 MPa
¿Viga segura?: True
</pre>

<pre>
Carga distribuida w (N/m): <b>4000</b>
Luz de la viga L (m): <b>5</b>
Ancho de la sección b (m): <b>0.1</b>
Altura de la sección h (m): <b>0.2</b>

Momento máximo: 12500.00 N·m
Momento de inercia: 0.00006667 m⁴
Esfuerzo máximo: 18.75 MPa
¿Viga segura?: False
</pre>

**Pregunta de respuesta corta**

Con `w = 8`, `L = 2` y `h = 3`, indique el valor de cada expresión y el orden en que se evalúan sus operadores:

```python
e1 = w * L ** 2 / 8
e2 = w * L ** 2 / 2 * 4
e3 = h ** L ** 2
e4 = -h ** 2
```

````python
#@title Solución
ESFUERZO_ADMISIBLE_MPA = 10
PA_POR_MPA = 1e6

carga_n_m = float(input("Carga distribuida w (N/m): "))
luz_m = float(input("Luz de la viga L (m): "))
ancho_m = float(input("Ancho de la sección b (m): "))
altura_m = float(input("Altura de la sección h (m): "))

momento_maximo = carga_n_m * luz_m ** 2 / 8
momento_inercia = ancho_m * altura_m ** 3 / 12
distancia_fibra = altura_m / 2
esfuerzo_pa = momento_maximo * distancia_fibra / momento_inercia

esfuerzo_mpa = esfuerzo_pa / PA_POR_MPA
viga_segura = esfuerzo_mpa <= ESFUERZO_ADMISIBLE_MPA

print(f"\nMomento máximo: {momento_maximo:.2f} N·m")
print(f"Momento de inercia: {momento_inercia:.8f} m⁴")
print(f"Esfuerzo máximo: {esfuerzo_mpa:.2f} MPa")
print(f"¿Viga segura?: {viga_segura}")
````

**Explicación**

`luz_m ** 2` se evalúa **antes** que `*` y `/` porque `**` tiene mayor precedencia, así que no hacen falta paréntesis en ninguna fórmula.

**Respuesta corta:**

* `e1 = 4.0`: `2 ** 2 = 4`, `8 * 4 = 32`, `32 / 8 = 4.0`.
* `e2 = 64.0`: `*` y `/` se evalúan de izquierda a derecha: `32 / 2 = 16.0`, `16.0 * 4 = 64.0` (no es `32 / 8`).
* `e3 = 81`: `**` asocia de **derecha a izquierda**: `2 ** 2 = 4`, `3 ** 4 = 81`.
* `e4 = -9`: `**` se evalúa antes que el signo: `-(3 ** 2)`.

## Ejercicio 3.6 — Ventilación de un invernadero

* la temperatura es **alta** y **no** está lloviendo, **o**

Escriba un programa que pregunte por cada sensor (la persona usuaria responde `s` o `n`), obtenga un booleano por sensor comparando la respuesta con `"s"`, y calcule `abrir_ventanas`.

## Ejercicio 3.7 — Balance diario de un tanque

Un acueducto lleva el **balance diario** de un tanque de almacenamiento. Durante el día ocurren, en orden, estos eventos:

1. **Llenado nocturno:** entra un volumen de agua (m³).
2. **Consumo de la mañana:** cada bomba activa extrae 30 m³.
3. **Hora pico:** se **duplica** la cantidad de bombas activas.
4. **Consumo de la tarde:** cada bomba activa extrae 12.5 m³.
5. **Pérdidas por fugas:** se pierde un porcentaje del volumen restante.

Escriba un programa en Python que:

A. Solicite el **nivel inicial** (m³), el **volumen del llenado nocturno** (m³), la cantidad de **bombas activas** al inicio del día (entero) y el **porcentaje de fugas**.

B. Actualice las variables **usando únicamente operadores de asignación compuesta** (`+=`, `-=`, `*=`) para cada evento, e imprima el nivel del tanque después de cada uno.

C. Al final, imprima un booleano `requiere_llenado_extra` que sea `True` si el nivel quedó **por debajo** del 25 % del nivel inicial.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Nivel inicial (m³): <b>120</b>
Llenado nocturno (m³): <b>45</b>
Bombas activas: <b>2</b>
Fugas (%): <b>4</b>

Después del llenado: 165.00 m³
Después del consumo de la mañana: 105.00 m³
Bombas activas en hora pico: 4
Después del consumo de la tarde: 55.00 m³
Después de las fugas: 52.80 m³
¿Requiere llenado extra?: False
</pre>

**Pregunta de respuesta corta**

Sin ejecutar el código, indique el valor final de `x`:

```python
x = 10
x += 2 * 3
x //= 5
x **= 2
```

````python
#@title Solución
CONSUMO_MANANA_POR_BOMBA = 30
CONSUMO_TARDE_POR_BOMBA = 12.5

nivel_m3 = float(input("Nivel inicial (m³): "))
llenado_nocturno = float(input("Llenado nocturno (m³): "))
bombas_activas = int(input("Bombas activas: "))
porcentaje_fugas = float(input("Fugas (%): "))

nivel_inicial = nivel_m3

nivel_m3 += llenado_nocturno
print(f"\nDespués del llenado: {nivel_m3:.2f} m³")

nivel_m3 -= CONSUMO_MANANA_POR_BOMBA * bombas_activas
print(f"Después del consumo de la mañana: {nivel_m3:.2f} m³")

bombas_activas *= 2
print(f"Bombas activas en hora pico: {bombas_activas}")

nivel_m3 -= CONSUMO_TARDE_POR_BOMBA * bombas_activas
print(f"Después del consumo de la tarde: {nivel_m3:.2f} m³")

nivel_m3 *= 1 - porcentaje_fugas / 100
print(f"Después de las fugas: {nivel_m3:.2f} m³")

requiere_llenado_extra = nivel_m3 < 0.25 * nivel_inicial
print(f"¿Requiere llenado extra?: {requiere_llenado_extra}")
````

**Explicación**

En `x -= expresión`, **primero** se evalúa toda la expresión de la derecha y luego se resta. Observe que `nivel_inicial` guarda una copia del valor inicial **antes** de modificarlo, para poder compararlo al final.

**Respuesta corta:** `x = 10` → `x += 6` → `16` → `x //= 5` → `3` → `x **= 2` → **`9`**.

## Ejercicio 3.8 — Transporte de materiales peligrosos

Una empresa de transporte verifica si puede aceptar una carga química. Se conocen:

```python
codigos_prohibidos = ["UN1203", "UN1017", "UN2814"]
clases_permitidas = "3689"
```

<span id="ejercicio-3-17-resistencias-en-serie-y-en-paralelo"></span>

## Ejercicio 3.17 — Resistencias en serie y en paralelo

Un técnico escribió este código para calcular la resistencia equivalente de **tres resistencias en paralelo**, pero el resultado no coincide con el que mide en el laboratorio:

```python
resistencia_paralelo = 1 / 1 / r1 + 1 / r2 + 1 / r3
```

La fórmula correcta es $R_p = \dfrac{1}{\frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_3}}$ y, en serie, $R_s = R_1 + R_2 + R_3$.

1. Explique cómo evalúa Python la línea del técnico.
2. Escriba un programa que solicite las tres resistencias (Ω) y muestre la resistencia en serie y en paralelo (2 decimales). Muestre además un booleano que verifique una propiedad física: la resistencia en paralelo siempre es **menor** que la menor de las tres.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
R1 (Ω): <b>220</b>
R2 (Ω): <b>330</b>
R3 (Ω): <b>470</b>
Resistencia en serie: 1020.00 Ω
Resistencia en paralelo: 103.06 Ω
¿Rp es menor que cada resistencia?: True
</pre>

````python
#@title Solución
r1 = float(input("R1 (Ω): "))
r2 = float(input("R2 (Ω): "))
r3 = float(input("R3 (Ω): "))

resistencia_serie = r1 + r2 + r3
resistencia_paralelo = 1 / (1 / r1 + 1 / r2 + 1 / r3)
propiedad_cumplida = resistencia_paralelo < r1 and resistencia_paralelo < r2 and resistencia_paralelo < r3

print(f"Resistencia en serie: {resistencia_serie:.2f} Ω")
print(f"Resistencia en paralelo: {resistencia_paralelo:.2f} Ω")
print(f"¿Rp es menor que cada resistencia?: {propiedad_cumplida}")
````

**Explicación**

La línea del técnico se evalúa como `((1 / 1) / r1) + (1 / r2) + (1 / r3)`: las divisiones se hacen de izquierda a derecha y **antes** que las sumas. El resultado es la suma de conductancias, no la resistencia equivalente. Los paréntesis alrededor de la suma del denominador corrigen el error lógico.

## Ejercicio 3.19 — Caudal con evaluación en cortocircuito

Un sistema de riego calcula el caudal $Q = V / t$ (litros por minuto) de un aspersor, y marca una alerta si el caudal **supera** un límite. Cuando el sensor falla, reporta un tiempo de `0`.

Escriba un programa que solicite el volumen (L), el tiempo (min) y el caudal límite, y calcule `alerta_caudal` **sin que el programa se detenga con un error** cuando el tiempo sea 0 y **sin usar condicionales**. Calcule también `sensor_con_falla`.

*Pista:* `and` no evalúa su lado derecho si el izquierdo es `False`.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Volumen medido (L): <b>90</b>
Tiempo medido (min): <b>4</b>
Caudal límite (L/min): <b>20</b>
¿Sensor con falla?: False
¿Alerta de caudal?: True
</pre>

<pre>
Volumen medido (L): <b>90</b>
Tiempo medido (min): <b>0</b>
Caudal límite (L/min): <b>20</b>
¿Sensor con falla?: True
¿Alerta de caudal?: False
</pre>

````python
#@title Solución
volumen_litros = float(input("Volumen medido (L): "))
tiempo_minutos = float(input("Tiempo medido (min): "))
caudal_limite = float(input("Caudal límite (L/min): "))

sensor_con_falla = tiempo_minutos == 0
alerta_caudal = not sensor_con_falla and volumen_litros / tiempo_minutos > caudal_limite

print(f"¿Sensor con falla?: {sensor_con_falla}")
print(f"¿Alerta de caudal?: {alerta_caudal}")
````

**Explicación**

Cuando `tiempo_minutos` es 0, `not sensor_con_falla` es `False` y Python **ya sabe** que toda la expresión `and` será `False`, así que **no evalúa** la división y no se produce el `ZeroDivisionError`. Si se invierte el orden (`volumen / tiempo > limite and not sensor_con_falla`), el programa falla, porque la división se evalúa primero.

## Ejercicio 3.20 — Cloración de un tanque de agua potable

Un acueducto rural desinfecta un tanque de agua con **hipoclorito de sodio** líquido. Se conocen:

* El **volumen** del tanque en m³ (1 m³ = 1000 L).
* La **dosis** de cloro deseada en mg/L.
* La **concentración** del hipoclorito en % (un hipoclorito al 12 % contiene 120 g de cloro por litro).

Escriba un programa que calcule:

1. Los **gramos de cloro** necesarios: $\frac{volumen\_L \cdot dosis}{1000}$.
2. Los **litros de hipoclorito** necesarios.
3. La cantidad de **bidones de 20 L** que se deben comprar (redondeando hacia arriba, usando solo operadores enteros sobre los **mililitros** de hipoclorito convertidos a `int`).
4. Un booleano `dosis_segura`, que sea `True` si la dosis está entre **0.3 y 1.5 mg/L** (inclusive), el rango de la norma de agua potable.
5. Un booleano `requiere_supervision`, que sea `True` si la dosis no es segura **o** se necesitan más de 3 bidones.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Volumen del tanque (m³): <b>850</b>
Dosis de cloro (mg/L): <b>1.2</b>
Concentración del hipoclorito (%): <b>12</b>

--- Plan de cloración ---
Cloro necesario: 1020.00 g
Hipoclorito necesario: 8.50 L
Bidones de 20 L: 1
¿Dosis segura?: True
¿Requiere supervisión?: False
</pre>

````python
#@title Solución
LITROS_POR_M3 = 1000
MILILITROS_POR_BIDON = 20 * 1000

volumen_m3 = float(input("Volumen del tanque (m³): "))
dosis_mg_l = float(input("Dosis de cloro (mg/L): "))
concentracion_porcentaje = float(input("Concentración del hipoclorito (%): "))

volumen_litros = volumen_m3 * LITROS_POR_M3
gramos_cloro = volumen_litros * dosis_mg_l / 1000
gramos_por_litro_hipoclorito = concentracion_porcentaje * 10
litros_hipoclorito = gramos_cloro / gramos_por_litro_hipoclorito
mililitros_hipoclorito = int(litros_hipoclorito * 1000)
bidones = (mililitros_hipoclorito + MILILITROS_POR_BIDON - 1) // MILILITROS_POR_BIDON

dosis_segura = 0.3 <= dosis_mg_l <= 1.5
requiere_supervision = not dosis_segura or bidones > 3

print("\n--- Plan de cloración ---")
print(f"Cloro necesario: {gramos_cloro:.2f} g")
print(f"Hipoclorito necesario: {litros_hipoclorito:.2f} L")
print(f"Bidones de 20 L: {bidones}")
print(f"¿Dosis segura?: {dosis_segura}")
print(f"¿Requiere supervisión?: {requiere_supervision}")
````
