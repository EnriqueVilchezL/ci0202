### Entrada Estándar con `input()`

La forma más común de "escuchar" al usuario es a través del teclado. En Python, la función `input()` pausa el programa y espera a que el usuario escriba algo y presione "Enter".

🧠 **Analogía:** La función `input()` es como un recepcionista que te pregunta "¿Cuál es tu nombre?" y espera pacientemente tu respuesta para anotarla.

**¡Importante!** 💡 Todo lo que `input()` recibe es siempre una cadena de texto (`str`). Si necesitas un número, ¡hay que convertirlo!

---

````python
# Pedimos el nombre al usuario y lo saludamos
print("Por favor, introduce tu nombre:")
nombre = input()
print(f"¡Hola, {nombre}! Bienvenido a la programación.")

# Ahora pedimos un número, lo convertimos y lo usamos
print("\nIntroduce un número para multiplicarlo por 2:")
numero_texto = input()
numero_entero = int(numero_texto) # Convertimos el texto a un número entero
resultado = numero_entero * 2
print(f"El resultado es: {resultado}")
````

## 2. Salida estándar y salida formateada

---

La "salida" es cómo el programa comunica información al exterior, generalmente mostrándola en la pantalla. La función `print()` es la principal herramienta para esta tarea.

`print()` puede mostrar texto, números y variables. Pero a menudo se quiere que la salida sea más legible y estructurada. A esto se le llama "formatear la salida".

🚀 **La mejor manera: f-strings (Cadenas f)**

Las "f-strings" son la forma más moderna, fácil y legible de formatear cadenas en Python. Simplemente se pone una `f` antes de las comillas y se escriben los nombres de las variables entre llaves `{}`.

`nombre = "Alan"`  
`edad = 41`  
`print(f"El famoso científico {nombre} tenía {edad} años.")`

---

````python
# Ejemplo de diferentes formas de salida
nombre_lenguaje = "Python"
version = 3.12

# Salida básica
print("El lenguaje es", nombre_lenguaje)

# Salida con f-string (la forma recomendada)
print(f"Estamos aprendiendo {nombre_lenguaje} en su versión {version}.")

# También podemos incluir expresiones dentro de las f-strings
a = 10
b = 5
print(f"La suma de {a} + {b} es {a + b}.")

# Formateando números: podemos especificar decimales
pi = 3.1415926535
print(f"El valor de PI con solo dos decimales es {pi:.2f}")
````

### Formateo avanzado de la salida en Python

Además de las f-strings, Python ofrece varias formas de mejorar la presentación de la información en pantalla.

#### 1. Alinear texto con `.ljust()`, `.rjust()` y `.center()`

- `.ljust(ancho)`: Alinea a la izquierda y rellena con espacios a la derecha.
- `.rjust(ancho)`: Alinea a la derecha y rellena con espacios a la izquierda.
- `.center(ancho)`: Centra el texto y rellena con espacios a ambos lados.

---

````python
titulo = "Factura"
print(titulo.center(30, '='))  # Centra el texto y rellena con '='
producto = "Café"
print(producto.ljust(15), "$", str(7.5).rjust(10))
````

#### 2. Tabuladores (`\t`) y saltos de línea (`\n`)

- `\t` agrega una tabulación (espacio grande, como en una tabla).
- `\n` agrega un salto de línea.

---

````python
print("Producto\tCantidad\tPrecio")
print("Café\t\t2\t\t$7.50")
print("Leche\t\t1\t\t$1.20")
print("\n¡Gracias por su compra!\n")
````

#### 3. Formateo de números y decimales

- Se puede controlar la cantidad de decimales con f-strings: `{valor:.2f}`

---

````python
pi = 3.1415926535
print(f"El valor de PI es {pi:.3f}")  # Muestra 3 decimales
````

#### Ejemplo combinado

````python
print("Producto".ljust(15), "Cantidad".center(10), "Precio".rjust(10))
print("Café".ljust(15), str(2).center(10), f"${7.5:.2f}".rjust(10))
print("Leche".ljust(15), str(1).center(10), f"${1.2:.2f}".rjust(10))
````

## 3. Manejo Básico de Errores en la Entrada

---

¿Qué pasa si se le pide al usuario un número y escribe "hola"? ¡El programa se romperá con un `ValueError`! Un buen programa debe ser robusto y anticipar estos problemas.

Se usa un bloque `try...except` para "intentar" hacer algo que podría fallar. Si falla, en lugar de detenerse, el programa ejecuta el código dentro del bloque `except`.

🧠 **Analogía:** Es como intentar abrir una puerta (`try`). Si la llave funciona, se puede entrar. Si no funciona (`except`), en lugar de quedarse atascado, se tiene un plan B, como tocar el timbre.

`try -> except`  
`Intentar código riesgoso -> Manejar el error si ocurre`

---

````python
dividendo_str = input("Introduce el dividendo (el número a dividir): ")
dividendo = float(dividendo_str)
divisor_str = input("Introduce el divisor (el número por el que dividir): ")
divisor = float(divisor_str)
resultado = dividendo / divisor
print("El resultado de", dividendo, "/", divisor, "es:", resultado)
````

#### ➗ Ejercicio: División Segura

Crear un programa que pida al usuario dos números (dividendo y divisor). El programa debe intentar realizar la división y mostrar el resultado. Se tienen que manejar dos posibles errores:

1.  `ValueError`: Si el usuario introduce algo que no es un número.
2.  `ZeroDivisionError`: Si el usuario introduce 0 como divisor.

Muestra mensajes claros para cada tipo de error.

---

````python
"""
Pide dos números y realiza una división, manejando los errores
ValueError y ZeroDivisionError.
"""
try:
    dividendo_str = input("Introduce el dividendo (el número a dividir): ")
    dividendo = float(dividendo_str) # Usamos float para permitir decimales
    divisor_str = input("Introduce el divisor (el número por el que dividir): ")
    divisor = float(divisor_str)
    resultado = dividendo / divisor
    print(f"El resultado de {dividendo} / {divisor} es: {resultado:.4f}")
except ValueError:
    print("Error de Valor: Por favor, introduce solo números.")
except ZeroDivisionError:
    print("Error de División: ¡No se puede dividir por cero!")
finally:
    # El bloque 'finally' se ejecuta siempre, haya error o no.
    print("\n--- Fin del intento de división. ---")
````

**Ejercicio 5.1 - Manguera**

Se tiene una manguera de longitud L y un carrete de radio R. Escribir un programa que calcule el número de vueltas enteras que la manguera podría enrollarse en el carrete y la longitud o trozo de manguera restante.

Pida la longitud L y el radio R al usuario.

Recuerde: La fórmula de la circunferencia de un círculo es 2*pi*r.

Su salida debería verse similar a esta:

```txt
=================================
Enrollando una manguera
=================================
Longitud de la manguera:     5m
Radio del carrete:           4m
---------------------------------
VUELTAS COMPLETAS:           5
=================================
```

````python
# Su código acá
````

````python
#@title Pruebas automáticas (solo ejecutar, no modificar)
from notebook_cell_tester.tester import ColabTestFramework, TestCase, TestSection

tester = ColabTestFramework()

tests = [
    TestSection("Sección 1 — Correctitud", [
        TestCase(
            name="Vueltas completas (L=20, R=1)",
            test_type="regex_output",
            stdin_input="20\n1\n",
            pattern=r"\b3\b",
            success_message="Calcula correctamente el número de vueltas completas.",
            error_message="Con L=20 y R=1, la circunferencia es 2*pi*1 ≈ 6.28, por lo que se esperaban 3 vueltas completas (20 // 6.28). Revise el cálculo de la circunferencia y el uso de división entera.",
        ),
        TestCase(
            name="Longitud restante (L=20, R=1)",
            test_type="regex_output",
            stdin_input="20\n1\n",
            pattern=r"1\.15(?!\d)",
            success_message="Calcula correctamente la longitud de manguera restante, con 2 decimales.",
            error_message="Con L=20 y R=1, después de dar 3 vueltas completas debería sobrar aproximadamente 1.15m de manguera. Revise el cálculo del resto y que se muestre con 2 decimales.",
        ),
    ]),
    TestSection("Sección 2 — Casos límite", [
        TestCase(
            name="La manguera no alcanza para ninguna vuelta completa (L=1, R=1)",
            test_type="regex_output",
            stdin_input="1\n1\n",
            pattern=r"\b0\b",
            success_message="Maneja correctamente el caso en que no se completa ninguna vuelta.",
            error_message="Con L=1 y R=1, la circunferencia (≈6.28) es mayor que la longitud de la manguera, por lo que se esperaban 0 vueltas completas.",
        ),
    ]),
    TestSection("Sección 3 — Manejo de errores", [
        TestCase(
            name="Entrada no numérica",
            test_type="regex_output",
            stdin_input="abc\n1\n",
            pattern=r"(?i)(error|inv[aá]lid|num[eé]ric)",
            success_message="Se maneja correctamente una entrada que no es un número.",
            error_message="Al ingresar un texto que no es un número para L se esperaba un mensaje de error, sin que el programa se detenga con una excepción.",
        ),
        TestCase(
            name="Radio del carrete en cero",
            test_type="regex_output",
            stdin_input="10\n0\n",
            pattern=r"(?i)(error|cero|radio)",
            success_message="Se maneja correctamente el caso de un radio igual a cero (que provocaría una división por cero).",
            error_message="Un radio de 0 hace que la circunferencia sea 0, lo que provocaría una división por cero al calcular las vueltas. Se esperaba un mensaje de error en vez de que el programa se detenga.",
        ),
    ]),
    TestSection("Sección 4 — Estilo de código", [
        TestCase(
            name="Se utiliza un bloque try/except",
            test_type="regex",
            pattern=r"\btry\b",
            success_message="Se utiliza un bloque try para manejar los errores posibles.",
            error_message="No se detectó un bloque 'try'. Este ejercicio pide un programa resistente a errores de tipo de dato y de cálculo; use try/except.",
        ),
    ]),
]

tester.run_sections(tests)
tester.display_results()
````

````python
#@title Posible solución (ejecutar para ver una solución de referencia)
import math

longitud_str = input("Ingrese la longitud de la manguera (L): ")
radio_str = input("Ingrese el radio del carrete (R): ")

try:
    longitud = float(longitud_str)
    radio = float(radio_str)
    circunferencia = 2 * math.pi * radio
    vueltas_completas = int(longitud // circunferencia)
    resto = longitud - vueltas_completas * circunferencia

    print("=" * 33)
    print("Enrollando una manguera")
    print("=" * 33)
    print(f"Longitud de la manguera:     {longitud}m")
    print(f"Radio del carrete:           {radio}m")
    print("-" * 33)
    print(f"VUELTAS COMPLETAS:           {vueltas_completas}")
    print(f"MANGUERA RESTANTE:           {resto:.2f}m")
    print("=" * 33)
except ValueError:
    print("Error: por favor ingrese valores numéricos válidos para L y R.")
except ZeroDivisionError:
    print("Error: el radio del carrete no puede ser cero.")
````
