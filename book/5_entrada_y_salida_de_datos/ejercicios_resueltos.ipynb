## Ejercicio 5.1 — Hoja de datos de un resistor

Un técnico de un laboratorio de electrónica necesita imprimir la **hoja de datos** de cada resistor que recibe. Un resistor tiene un valor **nominal** (Ω) y una **tolerancia** (%), de modo que su valor real puede estar entre:

$$R_{min} = R \cdot \left(1 - \frac{t}{100}\right) \qquad R_{max} = R \cdot \left(1 + \frac{t}{100}\right)$$

Escriba un programa en Python que:

A. Solicite el **código** del resistor, su valor **nominal** y su **tolerancia**.

B. Calcule los valores **mínimo** y **máximo** admisibles.

C. Imprima una hoja de datos de **30 caracteres de ancho**, con un encabezado centrado entre signos `=`, las etiquetas alineadas a la **izquierda** en 18 caracteres y los valores alineados a la **derecha** en 12 caracteres, con **2 decimales**, tal como en el ejemplo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Código del resistor: <b>R-4K7</b>
Valor nominal (Ω): <b>4700</b>
Tolerancia (%): <b>5</b>

=========== R-4K7 ============
Nominal (Ω):           4700.00
Tolerancia (%):           5.00
Mínimo (Ω):            4465.00
Máximo (Ω):            4935.00
==============================
</pre>

**Pregunta de respuesta corta**

¿Qué tipo de dato devuelve `f"{nominal:.2f}"`? ¿Por qué se le puede aplicar el método `.rjust()`?

````python
#@title Solución
ANCHO = 30

codigo = input("Código del resistor: ")
nominal = float(input("Valor nominal (Ω): "))
tolerancia = float(input("Tolerancia (%): "))

resistencia_minima = nominal * (1 - tolerancia / 100)
resistencia_maxima = nominal * (1 + tolerancia / 100)

print()
print(f" {codigo} ".center(ANCHO, "="))
print("Nominal (Ω):".ljust(18) + f"{nominal:.2f}".rjust(12))
print("Tolerancia (%):".ljust(18) + f"{tolerancia:.2f}".rjust(12))
print("Mínimo (Ω):".ljust(18) + f"{resistencia_minima:.2f}".rjust(12))
print("Máximo (Ω):".ljust(18) + f"{resistencia_maxima:.2f}".rjust(12))
print("=" * ANCHO)
````

**Explicación**

**Respuesta corta:** una *f-string* siempre produce un `str`, por eso admite métodos de cadenas como `.rjust()`. En cambio, `nominal.rjust(12)` produciría un `AttributeError`, porque un `float` no tiene ese método.

## Ejercicio 5.2 — Placa de identificación de un generador

Una empresa que instala **plantas eléctricas de respaldo** imprime una placa de identificación para cada generador. La corriente nominal de un generador trifásico es:

$$I = \frac{S \cdot 1000}{\sqrt{3} \cdot V}$$

donde $S$ es la potencia aparente (kVA) y $V$ el voltaje de línea (V).

Escriba un programa en Python que:

A. Solicite el **modelo** del generador, su **potencia** (kVA), el **voltaje** (V) y la **frecuencia** (Hz).

B. Calcule la **corriente nominal**.

C. Imprima una placa de **36 caracteres** de ancho, con bordes hechos con `*`, el modelo en **mayúsculas y centrado**, y cada dato en su propia línea, como en el ejemplo. La corriente debe mostrarse con **1 decimal**.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Modelo: <b>cummins c150</b>
Potencia (kVA): <b>150</b>
Voltaje (V): <b>480</b>
Frecuencia (Hz): <b>60</b>

************************************
*           CUMMINS C150           *
************************************
* Potencia: 150 kVA                *
* Voltaje: 480 V                   *
* Frecuencia: 60 Hz                *
* Corriente: 180.4 A               *
************************************
</pre>

````python
#@title Solución
ANCHO = 36

modelo = input("Modelo: ")
potencia_kva = float(input("Potencia (kVA): "))
voltaje = float(input("Voltaje (V): "))
frecuencia = int(input("Frecuencia (Hz): "))

corriente_nominal = potencia_kva * 1000 / (3 ** 0.5 * voltaje)

print()
print("*" * ANCHO)
print("*" + modelo.upper().center(ANCHO - 2) + "*")
print("*" * ANCHO)
print("*" + f" Potencia: {potencia_kva:.0f} kVA".ljust(ANCHO - 2) + "*")
print("*" + f" Voltaje: {voltaje:.0f} V".ljust(ANCHO - 2) + "*")
print("*" + f" Frecuencia: {frecuencia} Hz".ljust(ANCHO - 2) + "*")
print("*" + f" Corriente: {corriente_nominal:.1f} A".ljust(ANCHO - 2) + "*")
print("*" * ANCHO)
````

**Explicación**

El texto interior se ajusta a `ANCHO - 2` caracteres porque los dos bordes `*` ocupan un carácter cada uno. Guardar el ancho en una constante permite cambiar el tamaño de la placa modificando una sola línea.

## Ejercicio 5.3 — Conversor de caudal por línea de comandos

Los operadores de una planta potabilizadora quieren un programa que se ejecute **desde la terminal** y convierta un caudal dado en **litros por segundo (L/s)** a otras unidades:

* $1 \text{ L/s} = 3.6 \text{ m}^3\text{/h}$
* $1 \text{ L/s} = 15.8503 \text{ gal/min}$ (galones estadounidenses)

Escriba un programa en Python, `caudal.py`, que:

A. Reciba el caudal como **argumento de línea de comandos** (`sys.argv[1]`).

B. Muestre las tres unidades con **2 decimales**.

C. Si no se pasó ningún argumento (`IndexError`), muestre `Uso: python caudal.py <caudal_en_L/s>`. Si el argumento no es un número (`ValueError`), muestre `Error: el caudal debe ser un número.`

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
$ python caudal.py <b>12.5</b>
Caudal: 12.50 L/s
Caudal: 45.00 m³/h
Caudal: 198.13 gal/min
</pre>

<pre>
$ python caudal.py <b></b>
Uso: python caudal.py &lt;caudal_en_L/s&gt;
</pre>

<pre>
$ python caudal.py <b>doce</b>
Error: el caudal debe ser un número.
</pre>

**Pregunta de respuesta corta**

¿Qué contiene `sys.argv[0]`? ¿De qué tipo son **todos** los elementos de `sys.argv`?

````python
#@title Solución
import sys

# En un notebook, simule el argumento con: sys.argv = ["caudal.py", "12.5"]

M3H_POR_LS = 3.6
GPM_POR_LS = 15.8503

try:
    caudal_ls = float(sys.argv[1])
    print(f"Caudal: {caudal_ls:.2f} L/s")
    print(f"Caudal: {caudal_ls * M3H_POR_LS:.2f} m³/h")
    print(f"Caudal: {caudal_ls * GPM_POR_LS:.2f} gal/min")
except IndexError:
    print("Uso: python caudal.py <caudal_en_L/s>")
except ValueError:
    print("Error: el caudal debe ser un número.")
````

**Explicación**

**Respuesta corta:** `sys.argv[0]` es el **nombre del programa** (`caudal.py`); los argumentos que escribe la persona usuaria empiezan en el índice 1. Todos los elementos de `sys.argv` son **cadenas (`str`)**, por eso hay que convertirlos, igual que con `input()`.

## Ejercicio 5.4 — Pendiente de una carretera

Un ingeniero de caminos revisa la **pendiente** de tramos de carretera. Para cada tramo conoce la **distancia horizontal** (m) y el **desnivel** (m) entre sus extremos:

$$\text{pendiente} = \frac{\text{desnivel}}{\text{distancia horizontal}} \qquad \text{relación } 1:n \text{ con } n = \frac{\text{distancia}}{\text{desnivel}}$$

Escriba un programa en Python, `pendiente.py`, que:

A. Reciba el **nombre del tramo**, la **distancia horizontal** y el **desnivel** como argumentos de línea de comandos, en ese orden.

B. Calcule la pendiente y muéstrela como **porcentaje con 2 decimales** usando el formato `:.2%` (que multiplica por 100 y agrega el símbolo `%`).

C. Muestre la relación $1:n$ con $n$ redondeado a **1 decimal** y un booleano que indique si la pendiente **supera el 12 %**, el máximo permitido para carreteras de montaña.

D. Maneje los errores de argumentos faltantes, argumentos no numéricos y desnivel igual a cero (`ZeroDivisionError`), con mensajes claros.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
$ python pendiente.py <b>Tramo-3 850 76.5</b>
Tramo: Tramo-3
Pendiente: 9.00%
Relación: 1:11.1
¿Excede el 12%?: False
</pre>

<pre>
$ python pendiente.py <b>Tramo-7 400 58</b>
Tramo: Tramo-7
Pendiente: 14.50%
Relación: 1:6.9
¿Excede el 12%?: True
</pre>

<pre>
$ python pendiente.py <b>Tramo-9 300 0</b>
Error: el tramo es plano (desnivel 0) o tiene distancia 0.
</pre>

````python
#@title Solución
import sys

# En un notebook: sys.argv = ["pendiente.py", "Tramo-3", "850", "76.5"]

PENDIENTE_MAXIMA = 0.12

try:
    nombre_tramo = sys.argv[1]
    distancia_horizontal = float(sys.argv[2])
    desnivel = float(sys.argv[3])

    pendiente = desnivel / distancia_horizontal
    relacion_n = distancia_horizontal / desnivel
    excede_maximo = pendiente > PENDIENTE_MAXIMA

    print(f"Tramo: {nombre_tramo}")
    print(f"Pendiente: {pendiente:.2%}")
    print(f"Relación: 1:{relacion_n:.1f}")
    print(f"¿Excede el 12%?: {excede_maximo}")
except IndexError:
    print("Uso: python pendiente.py <tramo> <distancia_m> <desnivel_m>")
except ValueError:
    print("Error: la distancia y el desnivel deben ser números.")
except ZeroDivisionError:
    print("Error: el tramo es plano (desnivel 0) o tiene distancia 0.")
````

**Explicación**

El formato `:.2%` evita multiplicar por 100 a mano: `f"{0.09:.2%}"` produce `'9.00%'`. Observe que con desnivel 0 la **pendiente** sí se puede calcular (es 0), pero la **relación** $1:n$ no, y el `ZeroDivisionError` ocurre en esa línea.

## Ejercicio 5.5 — Presupuesto de cableado eléctrico

Un ingeniero eléctrico prepara el presupuesto del cableado de un edificio, que usa tres calibres de cable. Escriba un programa en Python que:

A. Solicite, para cada uno de los calibres **12 AWG**, **10 AWG** y **8 AWG**, la cantidad de **metros** (decimal) y el **precio por metro** en colones.

B. Calcule el **subtotal** de cada calibre, el **subtotal general**, el **IVA** (13 %) y el **total**.

C. Imprima una tabla con columnas de 10, 10, 12 y 14 caracteres (`Calibre`, `Metros`, `₡/m`, `Subtotal`); los números deben alinearse a la derecha, con **2 decimales** y **separador de miles** (formato `:,.2f`).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Metros de cable 12 AWG: <b>1250</b>
Precio por metro 12 AWG (₡): <b>385</b>
Metros de cable 10 AWG: <b>640</b>
Precio por metro 10 AWG (₡): <b>590</b>
Metros de cable 8 AWG: <b>180.5</b>
Precio por metro 8 AWG (₡): <b>940</b>

Calibre       Metros         ₡/m      Subtotal
----------------------------------------------

10 AWG        640.00      590.00    377,600.00
8 AWG         180.50      940.00    169,670.00
----------------------------------------------
Subtotal                          1,028,520.00
IVA (13%)                           133,707.60

</pre>

````python
#@title Solución
PORCENTAJE_IVA = 0.13

metros_12 = float(input("Metros de cable 12 AWG: "))
precio_12 = float(input("Precio por metro 12 AWG (₡): "))
metros_10 = float(input("Metros de cable 10 AWG: "))
precio_10 = float(input("Precio por metro 10 AWG (₡): "))
metros_8 = float(input("Metros de cable 8 AWG: "))
precio_8 = float(input("Precio por metro 8 AWG (₡): "))

subtotal_12 = metros_12 * precio_12
subtotal_10 = metros_10 * precio_10
subtotal_8 = metros_8 * precio_8
subtotal_general = subtotal_12 + subtotal_10 + subtotal_8
iva = subtotal_general * PORCENTAJE_IVA
total = subtotal_general + iva

print()
print("Calibre".ljust(10) + "Metros".rjust(10) + "₡/m".rjust(12) + "Subtotal".rjust(14))
print("-" * 46)
print("12 AWG".ljust(10) + f"{metros_12:,.2f}".rjust(10) + f"{precio_12:,.2f}".rjust(12)
      + f"{subtotal_12:,.2f}".rjust(14))
print("10 AWG".ljust(10) + f"{metros_10:,.2f}".rjust(10) + f"{precio_10:,.2f}".rjust(12)
      + f"{subtotal_10:,.2f}".rjust(14))
print("8 AWG".ljust(10) + f"{metros_8:,.2f}".rjust(10) + f"{precio_8:,.2f}".rjust(12)
      + f"{subtotal_8:,.2f}".rjust(14))
print("-" * 46)
print("Subtotal".ljust(32) + f"{subtotal_general:,.2f}".rjust(14))
print("IVA (13%)".ljust(32) + f"{iva:,.2f}".rjust(14))
print("TOTAL".ljust(32) + f"₡{total:,.2f}".rjust(14))
````

**Explicación**

La etiqueta de las filas de totales ocupa `10 + 10 + 12 = 32` caracteres para que los montos queden alineados bajo la columna `Subtotal`. El formato `:,` agrega comas como separador de miles, lo que hace mucho más legibles montos como `970,370.00`.

## Ejercicio 5.6 — Ensayo de tracción

En un laboratorio de materiales se ensaya una **probeta de acero** a tracción. Con la carga aplicada $F$ (N), el diámetro inicial $d$ (mm), la longitud inicial $L_0$ (mm) y la longitud final $L_f$ (mm) se calcula:

$$A = \frac{\pi d^2}{4} \qquad \sigma = \frac{F}{A} \qquad \varepsilon = \frac{L_f - L_0}{L_0} \qquad E = \frac{\sigma}{\varepsilon}$$

donde $\sigma$ es el esfuerzo (MPa, pues N/mm² = MPa), $\varepsilon$ la deformación unitaria y $E$ el módulo de elasticidad (MPa). Use $\pi = 3.1416$.

Escriba un programa en Python que:

A. Solicite $F$, $d$, $L_0$ y $L_f$.

B. Calcule $A$, $\sigma$, $\varepsilon$ y $E$.

C. Imprima un reporte donde: el área tenga **2 decimales**, el esfuerzo **1 decimal**, la deformación se muestre como **porcentaje con 3 decimales** (`:.3%`) y el módulo de elasticidad en **notación científica con 3 decimales** (`:.3e`).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Carga aplicada (N): <b>35000</b>
Diámetro inicial (mm): <b>12.5</b>
Longitud inicial (mm): <b>50</b>
Longitud final (mm): <b>50.071</b>

--- Resultados del ensayo ---
Área de la sección: 122.72 mm²
Esfuerzo: 285.2 MPa
Deformación unitaria: 0.142%
Módulo de elasticidad: 2.008e+05 MPa
</pre>

**Pregunta de respuesta corta**

El módulo de elasticidad del acero es de aproximadamente 200 GPa. ¿Es coherente el resultado del ejemplo? Explique cómo se lee el valor `2.008e+05`.

````python
#@title Solución
PI = 3.1416

carga_n = float(input("Carga aplicada (N): "))
diametro_mm = float(input("Diámetro inicial (mm): "))
longitud_inicial = float(input("Longitud inicial (mm): "))
longitud_final = float(input("Longitud final (mm): "))

area_mm2 = PI * diametro_mm ** 2 / 4
esfuerzo_mpa = carga_n / area_mm2
deformacion = (longitud_final - longitud_inicial) / longitud_inicial
modulo_elasticidad = esfuerzo_mpa / deformacion

print("\n--- Resultados del ensayo ---")
print(f"Área de la sección: {area_mm2:.2f} mm²")
print(f"Esfuerzo: {esfuerzo_mpa:.1f} MPa")
print(f"Deformación unitaria: {deformacion:.3%}")
print(f"Módulo de elasticidad: {modulo_elasticidad:.3e} MPa")
````

**Explicación**

**Respuesta corta:** sí. `2.008e+05` significa $2.008 \times 10^{5}$ MPa = 200 800 MPa ≈ **201 GPa**, muy cercano al valor teórico del acero. Comparar el resultado con un valor de referencia conocido es una forma sencilla de **validar** un programa.

## Ejercicio 5.7 — Rendimiento de un panel solar

Una empresa de energía solar evalúa paneles fotovoltaicos. La **eficiencia** de un panel es la fracción de la energía solar que recibe y que logra convertir en electricidad:

$$\eta = \frac{P}{G \cdot A} \cdot 100$$

C. Maneje los errores con `try`/`except`: si algún dato no es numérico, muestre `Error: todos los datos deben ser numéricos.`; si la irradiancia o el área son cero, muestre `Error: la irradiancia y el área deben ser distintas de cero.`

Potencia medida (W): <b>385</b>

## Ejercicio 5.9 — Tanque de gas a presión

Un ingeniero químico verifica la presión dentro de un tanque de **nitrógeno** usando la ley de los gases ideales:

$$P = \frac{n \cdot R \cdot T}{V} \qquad n = \frac{m}{M}$$

con $R = 8.314 \text{ J/(mol}\cdot\text{K)}$, $m$ la masa de gas (g), $M = 28.014 \text{ g/mol}$ la masa molar del nitrógeno, $T$ la temperatura (°C, que debe convertirse a K sumando 273.15) y $V$ el volumen (m³). La presión resulta en pascales.

Escriba un programa en Python que:

A. Solicite la masa, la temperatura en °C y el volumen del tanque en **litros** (1 m³ = 1000 L).

B. Calcule la presión y muéstrela en **kPa** y en **bar** (1 bar = 100 kPa) con **2 decimales**, además de un booleano que indique si supera la presión de diseño del tanque, **200 bar**.

C. Si algún dato no es numérico o el volumen es cero, muestre un mensaje de error apropiado y **no** muestre resultados.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Masa de nitrógeno (g): <b>2100</b>
Temperatura (°C): <b>25</b>
Volumen del tanque (L): <b>10</b>

Presión: 18581.85 kPa
Presión: 185.82 bar
¿Supera la presión de diseño?: False
</pre>

<pre>
Masa de nitrógeno (g): <b>2100</b>
Temperatura (°C): <b>25</b>
Volumen del tanque (L): <b>0</b>
Error: el volumen del tanque no puede ser cero.
</pre>

````python
#@title Solución
R = 8.314
MASA_MOLAR_N2 = 28.014
PRESION_DISENO_BAR = 200

try:
    masa_g = float(input("Masa de nitrógeno (g): "))
    temperatura_c = float(input("Temperatura (°C): "))
    volumen_l = float(input("Volumen del tanque (L): "))

    moles = masa_g / MASA_MOLAR_N2
    temperatura_k = temperatura_c + 273.15
    volumen_m3 = volumen_l / 1000
    presion_pa = moles * R * temperatura_k / volumen_m3

    presion_kpa = presion_pa / 1000
    presion_bar = presion_kpa / 100
    supera_diseno = presion_bar > PRESION_DISENO_BAR

    print(f"\nPresión: {presion_kpa:.2f} kPa")
    print(f"Presión: {presion_bar:.2f} bar")
    print(f"¿Supera la presión de diseño?: {supera_diseno}")
except ValueError:
    print("Error: los datos deben ser numéricos (use punto decimal).")
except ZeroDivisionError:
    print("Error: el volumen del tanque no puede ser cero.")
````

## Ejercicio 5.11 — Resistencia de un conductor

Al diseñar una instalación eléctrica hay que estimar la resistencia de los conductores largos para calcular la **caída de tensión**:

$$R = \frac{\rho \cdot L}{A} \qquad \Delta V = I \cdot R$$

donde $\rho$ es la resistividad del material (Ω·m), $L$ la longitud (m), $A$ el área de la sección (m²) e $I$ la corriente (A). El área se conoce en **mm²** (1 mm² = 10⁻⁶ m²).

Escriba un programa en Python, `conductor.py`, que:

A. Reciba por **línea de comandos**: la resistividad (puede venir en notación científica, como `1.68e-8`), la longitud, el área en mm² y la corriente.

B. Muestre la resistencia en **ohmios con 4 decimales**, la caída de tensión en voltios con **2 decimales**, y la caída como **porcentaje** de un voltaje de alimentación de **240 V** (`:.2%`).

C. Muestre `Uso: python conductor.py <resistividad> <longitud_m> <area_mm2> <corriente_A>` si falta algún argumento, y un mensaje de error si alguno no es numérico.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
$ python conductor.py <b>1.68e-8 85 6 32</b>
Resistencia del conductor: 0.2380 Ω
Caída de tensión: 7.62 V
Caída porcentual: 3.17%
</pre>

<pre>
$ python conductor.py <b>1.68e-8 85</b>
Uso: python conductor.py &lt;resistividad&gt; &lt;longitud_m&gt; &lt;area_mm2&gt; &lt;corriente_A&gt;
</pre>

````python
#@title Solución
import sys

# En un notebook: sys.argv = ["conductor.py", "1.68e-8", "85", "6", "32"]

VOLTAJE_ALIMENTACION = 240
M2_POR_MM2 = 1e-6

try:
    resistividad = float(sys.argv[1])
    longitud_m = float(sys.argv[2])
    area_mm2 = float(sys.argv[3])
    corriente_a = float(sys.argv[4])

    resistencia = resistividad * longitud_m / (area_mm2 * M2_POR_MM2)
    caida_tension = corriente_a * resistencia
    caida_porcentual = caida_tension / VOLTAJE_ALIMENTACION

    print(f"Resistencia del conductor: {resistencia:.4f} Ω")
    print(f"Caída de tensión: {caida_tension:.2f} V")
    print(f"Caída porcentual: {caida_porcentual:.2%}")
except IndexError:
    print("Uso: python conductor.py <resistividad> <longitud_m> <area_mm2> <corriente_A>")
except ValueError:
    print("Error: todos los argumentos deben ser numéricos.")
````

**Explicación**

`float()` acepta notación científica, así que `"1.68e-8"` se convierte sin problema a `0.0000000168`. Una caída de tensión menor al 3 % suele considerarse aceptable en instalaciones residenciales.

## Ejercicio 5.19 — Transmisión de datos de un satélite

Una estación terrena descarga imágenes de un **satélite de observación**. El satélite solo es visible durante una **ventana de contacto** de duración limitada, y hay que saber si la descarga cabe en esa ventana.

* El tamaño de las imágenes se da en **megabytes (MB)** y la velocidad del enlace en **megabits por segundo (Mbps)**; 1 byte = 8 bits.

Escriba un programa en Python que:

A. Solicite el tamaño de los datos (MB), la velocidad del enlace (Mbps) y la duración de la ventana de contacto (**minutos**).

B. Calcule el tiempo de descarga en segundos y muéstrelo en formato `HH:MM:SS` (cada parte con dos dígitos).

C. Muestre qué **porcentaje** de la ventana ocupa la descarga (1 decimal, `:.1%`) y un booleano `cabe_en_ventana`.

D. Maneje los datos no numéricos y una velocidad igual a cero.

**Ejemplos de ejecución**

## Ejercicio 5.20 — Mezcla de dos corrientes en un tanque

En una planta química, dos corrientes de una solución de **salmuera** se mezclan en un tanque. Por balance de masa, la concentración de la mezcla es:

$$C_m = \frac{Q_1 \cdot C_1 + Q_2 \cdot C_2}{Q_1 + Q_2}$$

donde $Q_1, Q_2$ son los caudales (L/min) y $C_1, C_2$ las concentraciones (g/L).

Escriba un programa en Python, `mezcla.py`, que:

A. Reciba por **línea de comandos** el **código del lote** y la **concentración objetivo** (g/L).

B. Solicite con `input()` el caudal y la concentración de cada corriente.

C. Calcule el **caudal total**, la **concentración de la mezcla**, la **masa de sal** que entra por minuto (g/min) y la **desviación** de la concentración respecto al objetivo, en porcentaje.

D. Imprima un reporte con el código del lote centrado en 36 caracteres entre signos `=`, y los resultados alineados (etiquetas en 24 caracteres, valores en 12, con 2 decimales), además de un booleano `dentro_de_especificacion` (desviación absoluta de **5 %** o menos).

E. Maneje los errores de argumentos faltantes, datos no numéricos y caudal total igual a cero.

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
$ python mezcla.py <b>SAL-0931 85</b>

========== Lote SAL-0931 ===========

</pre>

<pre>
$ python mezcla.py <b>SAL-0931</b>
Uso: python mezcla.py &lt;codigo_lote&gt; &lt;concentracion_objetivo&gt;
</pre>

**Pregunta de respuesta corta**

Explique por qué la concentración de la mezcla **no** es simplemente el promedio $\frac{C_1 + C_2}{2}$.

````python
#@title Solución
import sys

# En un notebook: sys.argv = ["mezcla.py", "SAL-0931", "85"]

ANCHO = 36
DESVIACION_MAXIMA = 5

try:
    codigo_lote = sys.argv[1]
    concentracion_objetivo = float(sys.argv[2])

    caudal_1 = float(input("Caudal corriente 1 (L/min): "))
    concentracion_1 = float(input("Concentración corriente 1 (g/L): "))
    caudal_2 = float(input("Caudal corriente 2 (L/min): "))
    concentracion_2 = float(input("Concentración corriente 2 (g/L): "))

    caudal_total = caudal_1 + caudal_2
    masa_sal = caudal_1 * concentracion_1 + caudal_2 * concentracion_2
    concentracion_mezcla = masa_sal / caudal_total
    desviacion = (concentracion_mezcla - concentracion_objetivo) / concentracion_objetivo * 100
    dentro_de_especificacion = abs(desviacion) <= DESVIACION_MAXIMA

    print()
    print(f" Lote {codigo_lote} ".center(ANCHO, "="))
    print("Caudal total (L/min)".ljust(24) + f"{caudal_total:.2f}".rjust(12))
    print("Masa de sal (g/min)".ljust(24) + f"{masa_sal:.2f}".rjust(12))
    print("Concentración (g/L)".ljust(24) + f"{concentracion_mezcla:.2f}".rjust(12))
    print("Desviación (%)".ljust(24) + f"{desviacion:.2f}".rjust(12))
    print("=" * ANCHO)
    print(f"¿Dentro de especificación?: {dentro_de_especificacion}")
except IndexError:
    print("Uso: python mezcla.py <codigo_lote> <concentracion_objetivo>")
except ValueError:
    print("Error: los caudales y concentraciones deben ser numéricos.")
except ZeroDivisionError:
    print("Error: el caudal total no puede ser cero.")
````

**Explicación**

**Respuesta corta:** porque las corrientes tienen **caudales distintos**: la que aporta más litros por minuto "pesa" más en la mezcla. La fórmula es un **promedio ponderado** por los caudales; solo coincide con el promedio simple cuando $Q_1 = Q_2$.
