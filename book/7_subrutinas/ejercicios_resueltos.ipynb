## Ejercicio 7.1 — Conversor de torque

En un taller automotriz, los manuales de distintos fabricantes especifican el **torque** de apriete de los pernos en unidades diferentes. Las equivalencias son:

* $1 \text{ N}\cdot\text{m} = 0.737562 \text{ lb}\cdot\text{ft}$
* $1 \text{ kgf}\cdot\text{m} = 9.80665 \text{ N}\cdot\text{m}$

Escriba un programa en Python con las siguientes funciones:

A. `newton_metro_a_libra_pie(torque: float) -> float`: convierte de N·m a lb·ft.

B. `kilogramo_metro_a_newton_metro(torque: float) -> float`: convierte de kgf·m a N·m.

C. `main() -> None`: solicita un torque en **kgf·m**, lo convierte a N·m y luego a lb·ft **reutilizando** las funciones anteriores, y muestra los tres valores con 2 decimales. Si el torque en N·m **supera** 120, muestra `Use llave de torque de rango alto.`

Si lo digitado no es un número, muestre `Error: el torque debe ser un número.` y **finalice** el programa (no lo vuelva a solicitar).

Las funciones de conversión deben **retornar** el resultado, no imprimirlo.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Torque de apriete (kgf·m): <b>14.5</b>
14.50 kgf·m = 142.20 N·m = 104.88 lb·ft
Use llave de torque de rango alto.
</pre>

**Pregunta de respuesta corta**

¿Qué imprimiría `print(newton_metro_a_libra_pie(10))` si, por error, la función tuviera `print(torque * LBFT_POR_NM)` en lugar de `return torque * LBFT_POR_NM`?

````python
#@title Solución
NM_POR_KGFM = 9.80665
LBFT_POR_NM = 0.737562


def newton_metro_a_libra_pie(torque: float) -> float:
    """Convierte un torque de N·m a lb·ft."""
    return torque * LBFT_POR_NM


def kilogramo_metro_a_newton_metro(torque: float) -> float:
    """Convierte un torque de kgf·m a N·m."""
    return torque * NM_POR_KGFM


def main() -> None:
    try:
        torque_kgfm = float(input("Torque de apriete (kgf·m): "))

        torque_nm = kilogramo_metro_a_newton_metro(torque_kgfm)
        torque_lbft = newton_metro_a_libra_pie(torque_nm)

        print(f"{torque_kgfm:.2f} kgf·m = {torque_nm:.2f} N·m = {torque_lbft:.2f} lb·ft")
        if torque_nm > 120:
            print("Use llave de torque de rango alto.")
    except ValueError:
        print("Error: el torque debe ser un número.")


main()
````

**Explicación**

**Respuesta corta:** imprimiría primero `7.37562` (desde dentro de la función) y luego **`None`**, porque una función sin `return` retorna `None` automáticamente. Además, en `main()` no se podría usar el resultado para seguir calculando.

El `try`/`except` que envuelve el cuerpo de `main()` evita que el programa se caiga con un `ValueError` cuando la entrada no es numérica: en su lugar muestra un mensaje claro y finaliza de forma ordenada. Es la estrategia adecuada en un programa de un solo cálculo como este.

## Ejercicio 7.2 — Número de Reynolds

En mecánica de fluidos, el **número de Reynolds** indica si el flujo en una tubería es laminar o turbulento:

$$Re = \frac{\rho \cdot v \cdot D}{\mu}$$

donde $\rho$ es la densidad (kg/m³), $v$ la velocidad (m/s), $D$ el diámetro (m) y $\mu$ la viscosidad dinámica (Pa·s). El flujo es **laminar** si $Re < 2300$, **de transición** si $2300 \le Re \le 4000$ y **turbulento** si $Re > 4000$.

Escriba un programa en Python con las siguientes funciones:

A. `calcular_reynolds(densidad: float, velocidad: float, diametro: float, viscosidad: float) -> float`

B. `clasificar_flujo(reynolds: float) -> str`: retorna `"laminar"`, `"de transición"` o `"turbulento"`.

C. `main() -> None`: solicita los datos (el diámetro en **milímetros**), invoca ambas funciones y muestra el número de Reynolds (0 decimales) y el tipo de flujo. Si alguno de los datos no es un número, muestre `Error: todos los datos deben ser numéricos.` y **finalice** el programa (no lo vuelva a solicitar); si la viscosidad no es mayor que cero, muestre `Error: la viscosidad debe ser mayor que cero.`

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Densidad (kg/m³): <b>998</b>
Velocidad (m/s): <b>0.8</b>
Diámetro (mm): <b>25</b>
Viscosidad dinámica (Pa·s): <b>0.001</b>
Número de Reynolds: 19960
Flujo turbulento
</pre>

<pre>
Densidad (kg/m³): <b>870</b>
Velocidad (m/s): <b>0.5</b>
Diámetro (mm): <b>20</b>
Viscosidad dinámica (Pa·s): <b>0.08</b>
Número de Reynolds: 109
Flujo laminar
</pre>

````python
#@title Solución
def calcular_reynolds(densidad: float, velocidad: float, diametro: float, viscosidad: float) -> float:
    """Calcula el número de Reynolds de un flujo en una tubería."""
    return densidad * velocidad * diametro / viscosidad


def clasificar_flujo(reynolds: float) -> str:
    """Clasifica el régimen de flujo según el número de Reynolds."""
    if reynolds < 2300:
        regimen = "laminar"
    elif reynolds <= 4000:
        regimen = "de transición"
    else:
        regimen = "turbulento"
    return regimen


def main() -> None:
    try:
        densidad = float(input("Densidad (kg/m³): "))
        velocidad = float(input("Velocidad (m/s): "))
        diametro_mm = float(input("Diámetro (mm): "))
        viscosidad = float(input("Viscosidad dinámica (Pa·s): "))

        if viscosidad <= 0:
            print("Error: la viscosidad debe ser mayor que cero.")
        else:
            reynolds = calcular_reynolds(densidad, velocidad, diametro_mm / 1000, viscosidad)
            regimen = clasificar_flujo(reynolds)

            print(f"Número de Reynolds: {reynolds:.0f}")
            print(f"Flujo {regimen}")
    except ValueError:
        print("Error: todos los datos deben ser numéricos.")


main()
````

**Explicación**

La conversión de milímetros a metros se hace **al invocar** la función (`diametro_mm / 1000`): así `calcular_reynolds` trabaja siempre en unidades del Sistema Internacional y se puede reutilizar en otros programas sin sorpresas. En el segundo ejemplo, el aceite es mucho más viscoso que el agua y el flujo resulta laminar. La viscosidad se revisa aparte porque un cero no produce un `ValueError` sino un `ZeroDivisionError`: son dos problemas distintos y cada uno merece su propio mensaje.

## Ejercicio 7.3 — Distancia de frenado

Un ingeniero de tránsito estima la **distancia total de detención** de un vehículo, que es la suma de la distancia recorrida durante el **tiempo de reacción** del conductor y la **distancia de frenado**:

$$d = v \cdot t_r + \frac{v^2}{2 \cdot \mu \cdot g}$$

donde $v$ es la velocidad (m/s), $t_r$ el tiempo de reacción (s), $\mu$ el coeficiente de fricción entre llanta y pavimento y $g = 9.81 \text{ m/s}^2$.

Escriba un programa en Python con las siguientes funciones:

A. `kmh_a_ms(velocidad: float) -> float`

B. `calcular_distancia_detencion(velocidad: float, friccion: float = 0.7, tiempo_reaccion: float = 1.5) -> float`: la fricción por defecto corresponde a pavimento **seco**.

C. `main() -> None`: solicita la velocidad en km/h y muestra la distancia de detención (1 decimal) en tres escenarios, invocando la función de tres formas distintas:
   1. Pavimento seco (usando los **valores por defecto**).
   2. Pavimento mojado ($\mu = 0.4$), pasando la fricción **por nombre**.
   3. Pavimento mojado y conductor distraído ($t_r = 2.5$ s), pasando ambos **por nombre**.

Si la velocidad digitada no es un número, muestre `Error: la velocidad debe ser un número.` y **finalice** el programa (no la vuelva a solicitar).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Velocidad (km/h): <b>90</b>
Pavimento seco: 83.0 m
Pavimento mojado: 117.1 m
Mojado y conductor distraído: 142.1 m
</pre>

**Pregunta de respuesta corta**

¿Qué valor tomaría cada parámetro en la invocación `calcular_distancia_detencion(25, 2.5)`? ¿Por qué es un error lógico?

````python
#@title Solución
GRAVEDAD = 9.81


def kmh_a_ms(velocidad: float) -> float:
    """Convierte una velocidad de km/h a m/s."""
    return velocidad / 3.6


def calcular_distancia_detencion(velocidad: float, friccion: float = 0.7,
                                 tiempo_reaccion: float = 1.5) -> float:
    """Calcula la distancia total de detención (m) para una velocidad en m/s."""
    distancia_reaccion = velocidad * tiempo_reaccion
    distancia_frenado = velocidad ** 2 / (2 * friccion * GRAVEDAD)
    return distancia_reaccion + distancia_frenado


def main() -> None:
    try:
        velocidad_kmh = float(input("Velocidad (km/h): "))
        velocidad_ms = kmh_a_ms(velocidad_kmh)

        seco = calcular_distancia_detencion(velocidad_ms)
        mojado = calcular_distancia_detencion(velocidad_ms, friccion=0.4)
        distraido = calcular_distancia_detencion(velocidad_ms, friccion=0.4, tiempo_reaccion=2.5)

        print(f"Pavimento seco: {seco:.1f} m")
        print(f"Pavimento mojado: {mojado:.1f} m")
        print(f"Mojado y conductor distraído: {distraido:.1f} m")
    except ValueError:
        print("Error: la velocidad debe ser un número.")


main()
````

**Explicación**

**Respuesta corta:** los argumentos posicionales se asignan **en orden**: `velocidad = 25`, `friccion = 2.5` y `tiempo_reaccion = 1.5` (por defecto). La intención probablemente era cambiar el tiempo de reacción, pero se cambió la fricción a un valor físicamente absurdo. Pasar los argumentos **por nombre** evita este tipo de confusiones.

## Ejercicio 7.4 — Corte de varillas de refuerzo

En una obra de construcción, las varillas de acero de refuerzo se compran en largos comerciales de **6 m** (600 cm), y se cortan en piezas del largo que indica el plano. El sobrante de cada varilla se considera **desperdicio**.

Escriba un programa en Python con las siguientes funciones:

A. `calcular_cortes(largo_pieza_cm: int, largo_varilla_cm: int = 600) -> tuple[int, int]`: usando la función integrada `divmod`, retorna la cantidad de **piezas por varilla** y el **desperdicio por varilla** (cm).

B. `calcular_varillas_necesarias(piezas_requeridas: int, piezas_por_varilla: int) -> int`: cantidad de varillas a comprar (redondeando hacia arriba).

C. `main() -> None`: solicita el largo de cada pieza (cm) y la cantidad de piezas requeridas. Si la pieza es más larga que la varilla, muestra un error. Si no, muestra las piezas por varilla, las varillas a comprar, el desperdicio total (m, 2 decimales) y el **porcentaje de desperdicio** respecto al total de acero comprado (1 decimal). Si alguno de los dos datos no es un número entero, muestra `Error: el largo y la cantidad deben ser números enteros.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Largo de cada pieza (cm): <b>145</b>
Cantidad de piezas requeridas: <b>38</b>
Piezas por varilla: 4 (sobran 20 cm)
Varillas a comprar: 10
Desperdicio total: 4.90 m
Porcentaje de desperdicio: 8.2%
</pre>

<pre>
Largo de cada pieza (cm): <b>650</b>
Cantidad de piezas requeridas: <b>4</b>
Error: la pieza es más larga que una varilla comercial.
</pre>

````python
#@title Solución
LARGO_VARILLA_CM = 600


def calcular_cortes(largo_pieza_cm: int, largo_varilla_cm: int = LARGO_VARILLA_CM) -> tuple[int, int]:
    """Retorna las piezas que salen de una varilla y el desperdicio (cm) por varilla."""
    piezas, desperdicio = divmod(largo_varilla_cm, largo_pieza_cm)
    return piezas, desperdicio


def calcular_varillas_necesarias(piezas_requeridas: int, piezas_por_varilla: int) -> int:
    """Calcula cuántas varillas comprar, redondeando hacia arriba."""
    return (piezas_requeridas + piezas_por_varilla - 1) // piezas_por_varilla


def main() -> None:
    try:
        largo_pieza = int(input("Largo de cada pieza (cm): "))
        piezas_requeridas = int(input("Cantidad de piezas requeridas: "))

        if largo_pieza > LARGO_VARILLA_CM:
            print("Error: la pieza es más larga que una varilla comercial.")
        else:
            piezas_por_varilla, desperdicio_por_varilla = calcular_cortes(largo_pieza)
            varillas = calcular_varillas_necesarias(piezas_requeridas, piezas_por_varilla)

            acero_comprado_cm = varillas * LARGO_VARILLA_CM
            desperdicio_total_cm = acero_comprado_cm - piezas_requeridas * largo_pieza
            porcentaje_desperdicio = desperdicio_total_cm / acero_comprado_cm * 100

            print(f"Piezas por varilla: {piezas_por_varilla} (sobran {desperdicio_por_varilla} cm)")
            print(f"Varillas a comprar: {varillas}")
            print(f"Desperdicio total: {desperdicio_total_cm / 100:.2f} m")
            print(f"Porcentaje de desperdicio: {porcentaje_desperdicio:.1f}%")
    except ValueError:
        print("Error: el largo y la cantidad deben ser números enteros.")


main()
````

**Explicación**

`divmod(a, b)` retorna **al mismo tiempo** el cociente entero y el residuo, equivalentes a `a // b` y `a % b`. El desperdicio total se calcula como todo el acero comprado menos lo que realmente se usa, lo que incluye también las piezas "de más" de la última varilla.

## Ejercicio 7.5 — Rampa de accesibilidad

La normativa de accesibilidad exige que las rampas para sillas de ruedas tengan una **pendiente máxima de 8.33 %** (1:12). Si el desnivel a salvar es mayor, se requieren **descansos** intermedios cada 9 m de rampa.

Escriba un programa en Python con las siguientes funciones:

A. `pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float`: muestra el mensaje y lee un número; si no es numérico o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar** (use `while` y `try`/`except`, sin `break`). Esta función se **reutiliza** en varios ejercicios posteriores de este glosario, así que conviene aprenderse su estructura.

B. `calcular_largo_minimo(desnivel_cm: float) -> float`: retorna el largo mínimo de la rampa en metros para cumplir la pendiente máxima.

C. `calcular_descansos(largo_m: float) -> int`: cantidad de descansos intermedios (uno por cada 9 m completos de rampa, sin contar el final).

D. `main() -> None`: solicita el desnivel (entre 1 y 150 cm) y el espacio disponible (entre 0.5 y 30 m) **reutilizando** `pedir_numero_seguro`, y muestra el largo mínimo, la cantidad de descansos y si la rampa **cabe** en el espacio disponible.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Desnivel a salvar (cm): <b>noventa</b>
Error: ingrese un número.
Desnivel a salvar (cm): <b>90</b>
Espacio disponible (m): <b>40</b>
Error: el valor debe estar entre 0.5 y 30.
Espacio disponible (m): <b>12</b>

Largo mínimo de la rampa: 10.80 m
Descansos intermedios: 1
La rampa cabe en el espacio disponible.
</pre>

````python
#@title Solución
PENDIENTE_MAXIMA = 0.0833
LARGO_ENTRE_DESCANSOS = 9


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def calcular_largo_minimo(desnivel_cm: float) -> float:
    """Calcula el largo mínimo (m) de la rampa según la pendiente máxima."""
    return desnivel_cm / 100 / PENDIENTE_MAXIMA


def calcular_descansos(largo_m: float) -> int:
    """Calcula los descansos intermedios requeridos por la rampa."""
    return int(largo_m // LARGO_ENTRE_DESCANSOS)


def main() -> None:
    desnivel = pedir_numero_seguro("Desnivel a salvar (cm): ", 1, 150)
    espacio = pedir_numero_seguro("Espacio disponible (m): ", 0.5, 30)

    largo = calcular_largo_minimo(desnivel)
    descansos = calcular_descansos(largo)

    print(f"\nLargo mínimo de la rampa: {largo:.2f} m")
    print(f"Descansos intermedios: {descansos}")
    if largo <= espacio:
        print("La rampa cabe en el espacio disponible.")
    else:
        print(f"No cabe: faltan {largo - espacio:.2f} m. Considere una rampa en zigzag.")


main()
````

**Explicación**

Escribir la validación **una sola vez** dentro de una función, y reutilizarla con distintos mensajes y rangos, evita copiar el mismo ciclo `while` para cada dato. Esa es la principal ventaja de la **modularización**. Cuando un dato se puede volver a pedir, `pedir_numero_seguro` es preferible a terminar el programa con un mensaje de error: la persona usuaria corrige el dato equivocado sin perder los que ya había digitado.

Escriba un programa en Python con las siguientes funciones:

D. `ampliar_incertidumbre(incertidumbre: float, factor: float) -> float`: retorna la incertidumbre multiplicada por el factor.

Luego invoca ambas funciones y muestra la lista corregida, el promedio corregido (use `sum()` y `len()`) y el resultado en la forma `promedio ± incertidumbre ampliada`, redondeando ambos valores a 2 decimales con `round()`.

```python
    incertidumbre = incertidumbre * 2

    ampliar(incertidumbre)
    print(incertidumbre)

```

````python
#@title Solución
FACTOR_COBERTURA = 2


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def corregir_lecturas(lecturas: list[float], offset: float) -> None:
    """Resta el offset a cada lectura, modificando la lista recibida."""
    for indice in range(len(lecturas)):
        lecturas[indice] = lecturas[indice] - offset


def ampliar_incertidumbre(incertidumbre: float, factor: float) -> float:
    """Retorna la incertidumbre ampliada con el factor de cobertura."""
    return incertidumbre * factor


def main() -> None:
    cantidad = pedir_entero_seguro("Cantidad de lecturas: ", 1, 20)
    lecturas = []
    for numero in range(1, cantidad + 1):
        lecturas.append(pedir_numero_seguro(f"Lectura {numero} (°C): ", -50, 200))
    offset = pedir_numero_seguro("Offset del instrumento (°C): ", -10, 10)
    incertidumbre = pedir_numero_seguro("Incertidumbre estándar (°C): ", 0, 5)

    corregir_lecturas(lecturas, offset)
    incertidumbre_ampliada = ampliar_incertidumbre(incertidumbre, FACTOR_COBERTURA)
    promedio = sum(lecturas) / len(lecturas)

    print(f"\nLecturas corregidas: {lecturas}")
    print(f"Resultado: ({round(promedio, 2)} ± {round(incertidumbre_ampliada, 2)}) °C")


main()
````

**Explicación**

La lista corregida muestra valores como `25.099999999999998` por la representación binaria de los flotantes; por eso el resultado final se redondea con `round()`.

**Respuesta corta:** imprime `0.06`. El parámetro `incertidumbre` es una variable **local** de `ampliar`: la asignación crea un valor nuevo (los `float` son **inmutables**) y solo cambia la variable local. En cambio, `corregir_lecturas` recibe una referencia a la **misma lista** (mutable) y modifica sus elementos, por eso el cambio se ve desde `main()`.

Aquí conviene **volver a solicitar** los datos inválidos: si el programa terminara al digitar mal la última lectura, se perderían todas las anteriores.

## Ejercicio 7.7 — Ensayo de fatiga

En un ensayo de **fatiga**, varias probetas idénticas de aluminio se someten a una carga cíclica hasta que se rompen, y se registra cuántos **ciclos** resistió cada una. La vida a fatiga del material se describe con la **mediana** de los ciclos (el valor central de los datos ordenados; si la cantidad es par, el promedio de los dos centrales).

Escriba un programa en Python con las siguientes funciones:

A. `pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int`: muestra el mensaje y lee un número **entero**; si no lo es o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar**. Es la versión con `int()` de la función de lectura segura del ejercicio 7.5.

B. `leer_ciclos(cantidad: int) -> list[int]`: solicita y retorna los ciclos de cada probeta (entre 1 y 10 000 000), **reutilizando** `pedir_entero_seguro`.

C. `calcular_mediana(datos: list[int]) -> float`: usa `sorted()` para ordenar una **copia** de los datos (sin modificar la lista original) y retorna la mediana.

D. `calcular_dispersion(datos: list[int]) -> float`: retorna el rango relativo $\frac{\max - \min}{\text{promedio}} \cdot 100$ usando `max()`, `min()` y `sum()`.

E. `main() -> None`: solicita la cantidad de probetas (entre 2 y 50), lee los ciclos, y muestra los datos originales, los datos ordenados, la mediana, el mínimo, el máximo y la dispersión (redondeada a 1 decimal con `round()`). Si la dispersión supera el **50 %**, muestra `Alta dispersión: revisar el montaje del ensayo.`

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de probetas: <b>seis</b>
Error: ingrese un número entero.
Cantidad de probetas: <b>6</b>
Ciclos de la probeta 1: <b>152000</b>
Ciclos de la probeta 2: <b>98500</b>
Ciclos de la probeta 3: <b>187300</b>
Ciclos de la probeta 4: <b>121000</b>
Ciclos de la probeta 5: <b>143800</b>
Ciclos de la probeta 6: <b>76400</b>

Datos originales: [152000, 98500, 187300, 121000, 143800, 76400]
Datos ordenados: [76400, 98500, 121000, 143800, 152000, 187300]
Vida a fatiga (mediana): 132400.0 ciclos
Mínimo: 76400  Máximo: 187300
Dispersión: 85.4%
Alta dispersión: revisar el montaje del ensayo.
</pre>

````python
#@title Solución
DISPERSION_MAXIMA = 50
CICLOS_MAXIMOS = 10000000


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def leer_ciclos(cantidad: int) -> list[int]:
    """Solicita los ciclos a la falla de cada probeta."""
    ciclos = []
    for numero in range(1, cantidad + 1):
        ciclos.append(pedir_entero_seguro(f"Ciclos de la probeta {numero}: ", 1, CICLOS_MAXIMOS))
    return ciclos


def calcular_mediana(datos: list[int]) -> float:
    """Calcula la mediana sin modificar la lista original."""
    ordenados = sorted(datos)
    centro = len(ordenados) // 2
    if len(ordenados) % 2 == 1:
        mediana = ordenados[centro]
    else:
        mediana = (ordenados[centro - 1] + ordenados[centro]) / 2
    return mediana


def calcular_dispersion(datos: list[int]) -> float:
    """Calcula el rango relativo (%) de los datos respecto a su promedio."""
    promedio = sum(datos) / len(datos)
    return (max(datos) - min(datos)) / promedio * 100


def main() -> None:
    cantidad = pedir_entero_seguro("Cantidad de probetas: ", 2, 50)
    ciclos = leer_ciclos(cantidad)

    mediana = calcular_mediana(ciclos)
    dispersion = calcular_dispersion(ciclos)

    print(f"\nDatos originales: {ciclos}")
    print(f"Datos ordenados: {sorted(ciclos)}")
    print(f"Vida a fatiga (mediana): {mediana} ciclos")
    print(f"Mínimo: {min(ciclos)}  Máximo: {max(ciclos)}")
    print(f"Dispersión: {round(dispersion, 1)}%")
    if dispersion > DISPERSION_MAXIMA:
        print("Alta dispersión: revisar el montaje del ensayo.")


main()
````

**Explicación**

`sorted(datos)` retorna una **lista nueva** ordenada y deja intacta la original; en cambio, el método `datos.sort()` ordenaría la lista **en su lugar**, y el cambio se vería desde `main()` porque las listas son mutables. En fatiga se prefiere la mediana al promedio porque es menos sensible a una probeta con un defecto.

````python
#@title Solución
def cumple_tolerancia(medida: float, nominal: float, tolerancia: float) -> bool:
    """Indica si la medida está dentro de nominal ± tolerancia."""
    return abs(medida - nominal) <= tolerancia


def revisar_dimension(nombre: str, medida: float, nominal: float, tolerancia: float) -> bool:
    """Imprime el resultado de la inspección de una dimensión y retorna si cumple."""
    cumple = cumple_tolerancia(medida, nominal, tolerancia)
    if cumple:
        estado = "OK"
    else:
        estado = "FUERA"
    print(f"{nombre.ljust(20)} {medida:8.2f} mm  desviación {medida - nominal:+.2f}  {estado}")
    return cumple


def main() -> None:
    try:
        largo = float(input("Largo (mm): "))
        ancho = float(input("Ancho (mm): "))
        espesor = float(input("Espesor del cordón (mm): "))

        print()
        fuera_de_tolerancia = 0
        if not revisar_dimension("Largo", largo, 250.0, 0.5):
            fuera_de_tolerancia += 1
        if not revisar_dimension("Ancho", ancho, 80.0, 0.3):
            fuera_de_tolerancia += 1
        if not revisar_dimension("Espesor del cordón", espesor, 6.0, 0.8):
            fuera_de_tolerancia += 1

        if fuera_de_tolerancia == 0:
            print("\nSOPORTE ACEPTADO")
        else:
            print(f"\nSOPORTE RECHAZADO ({fuera_de_tolerancia} dimensiones fuera de tolerancia)")
    except ValueError:
        print("Error: las medidas deben ser números.")


main()
````

## Ejercicio 7.9 — Potencia reactiva y pruebas con `assert`

En un sistema eléctrico de corriente alterna, las potencias **activa** $P$ (kW), **reactiva** $Q$ (kVAR) y **aparente** $S$ (kVA) forman el llamado *triángulo de potencias*:

$$Q = \sqrt{S^2 - P^2} \qquad fp = \frac{P}{S}$$

Escriba un programa en Python con las siguientes funciones:

A. `calcular_potencia_reactiva(aparente: float, activa: float) -> float` (use `** 0.5`).

B. `calcular_factor_potencia(aparente: float, activa: float) -> float`

C. `probar_funciones() -> None`: verifica ambas funciones con `assert` usando casos cuyo resultado se conoce: el triángulo **3-4-5** ($S = 5$, $P = 4 \Rightarrow Q = 3$, $fp = 0.8$), un caso de **carga puramente resistiva** ($S = P \Rightarrow Q = 0$, $fp = 1$) y un caso de **carga puramente reactiva** ($P = 0$). Al final imprime `Pruebas superadas.`

D. `main() -> None`: primero invoca `probar_funciones()`; luego solicita $S$ y $P$, muestra $Q$ y $fp$ (2 decimales) y, si $fp < 0.9$, calcula la potencia reactiva que debería **compensar** un banco de capacitores para llevar el factor de potencia a 0.95: $Q_c = Q - P \cdot \frac{\sqrt{1 - 0.95^2}}{0.95}$.

Si las potencias no son números, muestre `Error: las potencias deben ser números.` y **finalice** el programa (no las vuelva a solicitar). Si $S$ no es mayor que cero o es **menor** que $P$, muestre `Error: S debe ser mayor que cero y no menor que P.`, porque ese triángulo de potencias no existe.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Pruebas superadas.

Potencia aparente S (kVA): <b>250</b>
Potencia activa P (kW): <b>195</b>
Potencia reactiva: 156.44 kVAR
Factor de potencia: 0.78
Se recomienda un banco de capacitores de 92.35 kVAR.
</pre>

````python
#@title Solución
FACTOR_OBJETIVO = 0.95


def calcular_potencia_reactiva(aparente: float, activa: float) -> float:
    """Calcula la potencia reactiva (kVAR) a partir de S (kVA) y P (kW)."""
    return (aparente ** 2 - activa ** 2) ** 0.5


def calcular_factor_potencia(aparente: float, activa: float) -> float:
    """Calcula el factor de potencia P / S."""
    return activa / aparente


def probar_funciones() -> None:
    """Verifica las funciones con casos de resultado conocido."""
    assert calcular_potencia_reactiva(5, 4) == 3
    assert calcular_factor_potencia(5, 4) == 0.8
    assert calcular_potencia_reactiva(10, 10) == 0
    assert calcular_factor_potencia(10, 10) == 1
    assert calcular_potencia_reactiva(7, 0) == 7
    assert calcular_factor_potencia(7, 0) == 0
    print("Pruebas superadas.")


def main() -> None:
    probar_funciones()

    try:
        aparente = float(input("\nPotencia aparente S (kVA): "))
        activa = float(input("Potencia activa P (kW): "))

        if aparente <= 0 or activa > aparente:
            print("Error: S debe ser mayor que cero y no menor que P.")
        else:
            reactiva = calcular_potencia_reactiva(aparente, activa)
            factor = calcular_factor_potencia(aparente, activa)

            print(f"Potencia reactiva: {reactiva:.2f} kVAR")
            print(f"Factor de potencia: {factor:.2f}")

            if factor < 0.9:
                reactiva_objetivo = activa * (1 - FACTOR_OBJETIVO ** 2) ** 0.5 / FACTOR_OBJETIVO
                compensacion = reactiva - reactiva_objetivo
                print(f"Se recomienda un banco de capacitores de {compensacion:.2f} kVAR.")
    except ValueError:
        print("Error: las potencias deben ser números.")


main()
````

**Explicación**

Las pruebas se ejecutan **antes** de usar las funciones con datos reales: si alguna falla, el programa se detiene con un `AssertionError` y no se reportan resultados equivocados. Los casos de prueba se eligieron porque su resultado se conoce **sin calculadora**, incluidos dos casos extremos (carga puramente resistiva y puramente reactiva). La validación de $S$ y $P$ es indispensable: si $P > S$, la raíz cuadrada de un número negativo no produce un error, sino un número **complejo**, y el programa reportaría un disparate en lugar de avisar del problema.

## Ejercicio 7.10 — Evaluación económica de un proyecto

Una empresa evalúa instalar un **sistema de recuperación de calor** en una caldera. El sistema cuesta una inversión inicial y genera un **ahorro anual** constante durante su vida útil. Para saber si conviene, se calcula el **valor presente neto (VPN)**: la suma de los ahorros de cada año traídos a valor presente, menos la inversión.

$$VP_t = \frac{\text{ahorro}}{(1 + i)^t} \qquad VPN = \sum_{t=1}^{n} VP_t - \text{inversión}$$

Escriba un programa en Python con las siguientes funciones:

A. `calcular_valor_presente(monto: float, tasa: float, anio: int) -> float`: use la función integrada `pow()`.

B. `calcular_vpn(inversion: float, ahorro_anual: float, tasa: float, vida_util: int) -> float`: recorre los años con un ciclo, **invocando** la función anterior, e imprime el valor presente de cada año.

C. `main() -> None`: solicita la inversión, el ahorro anual, la tasa de descuento (en %) y la vida útil (años). Muestra el VPN con separador de miles y 2 decimales, y `El proyecto es rentable.` si el VPN es positivo o `El proyecto no es rentable.` en caso contrario. Si alguno de los datos no es un número (la vida útil debe ser un entero), muestra `Error: los datos deben ser numéricos.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

Ahorro anual ($): <b>11500</b>

VPN: $2,085.50
El proyecto es rentable.

````python
#@title Solución
def calcular_valor_presente(monto: float, tasa: float, anio: int) -> float:
    """Trae a valor presente un monto recibido en el año indicado."""
    return monto / pow(1 + tasa, anio)


def calcular_vpn(inversion: float, ahorro_anual: float, tasa: float, vida_util: int) -> float:
    """Calcula el valor presente neto de un proyecto con ahorro anual constante."""
    suma_valores_presentes = 0
    for anio in range(1, vida_util + 1):
        valor_presente = calcular_valor_presente(ahorro_anual, tasa, anio)
        suma_valores_presentes += valor_presente
        print(f"  Año {anio}: ${valor_presente:,.2f}")
    return suma_valores_presentes - inversion


def main() -> None:
    try:
        inversion = float(input("Inversión inicial ($): "))
        ahorro_anual = float(input("Ahorro anual ($): "))
        tasa_porcentaje = float(input("Tasa de descuento (%): "))
        vida_util = int(input("Vida útil (años): "))

        print("\nValor presente de los ahorros:")
        vpn = calcular_vpn(inversion, ahorro_anual, tasa_porcentaje / 100, vida_util)

        print(f"\nVPN: ${vpn:,.2f}")
        if vpn > 0:
            print("El proyecto es rentable.")
        else:
            print("El proyecto no es rentable.")
    except ValueError:
        print("Error: los datos deben ser numéricos.")


main()
````

Un dólar ahorrado dentro de 6 años vale menos que un dólar hoy: por eso los valores presentes **disminuyen** año con año. ## Ejercicio 7.11 — Potencia de una central hidroeléctrica

La potencia que genera una central hidroeléctrica es:

$$P = \rho \cdot g \cdot Q \cdot H \cdot \eta$$

con $\rho = 1000 \text{ kg/m}^3$, $g = 9.81 \text{ m/s}^2$, $Q$ el caudal turbinado (m³/s), $H$ la altura neta (m) y $\eta$ la eficiencia del conjunto turbina-generador.

Escriba un programa en Python que defina $\rho$ y $g$ como **constantes globales** (en mayúsculas) y las siguientes funciones:

A. `calcular_potencia_mw(caudal: float, altura: float, eficiencia: float) -> float`: usa las constantes globales (solo las **lee**, no las modifica).

B. `calcular_energia_anual_gwh(potencia_mw: float, factor_planta: float) -> float`: energía generada en un año (8760 horas), considerando que la central opera en promedio al `factor_planta` de su capacidad. 1 GWh = 1000 MWh.

C. `calcular_hogares_abastecidos(energia_gwh: float, consumo_hogar_kwh_mes: float = 250) -> int`

D. `main() -> None`: solicita el caudal, la altura, la eficiencia (%) y el factor de planta (%), invoca las funciones y muestra la potencia (MW, 2 decimales), la energía anual (GWh, 1 decimal) y la cantidad de hogares que podría abastecer (separador de miles). Si alguno de los datos no es un número, muestra `Error: todos los datos deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

</pre>

**Pregunta de respuesta corta**

Si dentro de `calcular_potencia_mw` se escribiera `GRAVEDAD = 9.8` antes de usarla, ¿cambiaría el valor de la constante global? ¿Por qué no es buena idea modificar variables globales desde una función?

````python
#@title Solución
DENSIDAD_AGUA = 1000
GRAVEDAD = 9.81
HORAS_POR_ANIO = 8760


def calcular_potencia_mw(caudal: float, altura: float, eficiencia: float) -> float:
    """Calcula la potencia eléctrica (MW) de la central."""
    potencia_w = DENSIDAD_AGUA * GRAVEDAD * caudal * altura * eficiencia
    return potencia_w / 1e6


def calcular_energia_anual_gwh(potencia_mw: float, factor_planta: float) -> float:
    """Calcula la energía generada en un año (GWh)."""
    energia_mwh = potencia_mw * HORAS_POR_ANIO * factor_planta
    return energia_mwh / 1000


def calcular_hogares_abastecidos(energia_gwh: float, consumo_hogar_kwh_mes: float = 250) -> int:
    """Estima cuántos hogares se abastecen con la energía anual indicada."""
    energia_kwh = energia_gwh * 1e6
    consumo_hogar_anual = consumo_hogar_kwh_mes * 12
    return int(energia_kwh // consumo_hogar_anual)


def main() -> None:
    try:
        caudal = float(input("Caudal turbinado (m³/s): "))
        altura = float(input("Altura neta (m): "))
        eficiencia = float(input("Eficiencia (%): ")) / 100
        factor_planta = float(input("Factor de planta (%): ")) / 100

        potencia = calcular_potencia_mw(caudal, altura, eficiencia)
        energia = calcular_energia_anual_gwh(potencia, factor_planta)
        hogares = calcular_hogares_abastecidos(energia)

        print(f"\nPotencia instalada: {potencia:.2f} MW")
        print(f"Energía anual: {energia:.1f} GWh")
        print(f"Hogares abastecidos: {hogares:,}")
    except ValueError:
        print("Error: todos los datos deben ser números.")


main()
````

**Explicación**

**Respuesta corta:** no. Esa asignación crearía una variable **local** llamada `GRAVEDAD`, visible solo dentro de la función; la global seguiría valiendo 9.81 (para modificarla haría falta declarar `global GRAVEDAD`). Modificar variables globales hace que el resultado de una función dependa de lo que hayan hecho **otras** funciones antes, lo que vuelve el programa difícil de entender y de probar.

## Ejercicio 7.12 — Dimensionamiento de un sistema fotovoltaico

Una empresa instala sistemas solares **aislados** (con baterías) en fincas sin acceso a la red eléctrica. El dimensionamiento sigue estos pasos:

1. **Consumo diario** (Wh): la suma, para cada aparato, de su potencia (W) por sus horas de uso diarias.
2. **Paneles:** $N_{paneles} = \left\lceil \frac{\text{consumo diario}}{P_{panel} \cdot HSP \cdot 0.8} \right\rceil$, donde $HSP$ son las horas solares pico y 0.8 un factor de pérdidas.
3. **Baterías:** la energía que deben almacenar es el consumo diario por los **días de autonomía**, dividida entre 0.5 (solo se descargan hasta el 50 %). Cada batería almacena $V \cdot Ah$ Wh.

Escriba un programa en Python, diseñado de **arriba hacia abajo**, con las funciones:

A. `pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float`: muestra el mensaje y lee un número; si no es numérico o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar** (la misma función del ejercicio 7.5).

B. `pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int`: igual que la anterior, pero con `int()`, para los datos que deben ser enteros.

C. `redondear_hacia_arriba(valor: float) -> int`: sin usar bibliotecas.

D. `calcular_consumo_diario() -> float`: pregunta cuántos aparatos hay (de 1 a 20) y, para cada uno, su nombre, su potencia (de 1 a 10 000 W) y sus horas de uso (de 0 a 24); retorna el consumo total. Todos los números se leen con las funciones de lectura segura, de modo que un dato mal digitado se **vuelva a solicitar** sin perder los aparatos ya registrados.

E. `calcular_paneles(consumo_diario: float, potencia_panel: float, horas_solares: float) -> int`

F. `calcular_baterias(consumo_diario: float, dias_autonomia: int, voltaje: float, capacidad_ah: float) -> int`

G. `main() -> None`: coordina todo con paneles de 450 W, 4.5 horas solares pico, 2 días de autonomía y baterías de 12 V y 200 Ah, y muestra el resumen del diseño.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Cantidad de aparatos: <b>3</b>

Aparato 1: <b>Refrigeradora</b>
  Potencia de Refrigeradora (W): <b>150</b>
  Horas de uso diarias: <b>30</b>
Error: el valor debe estar entre 0 y 24.
  Horas de uso diarias: <b>10</b>

Aparato 2: <b>Bomba de agua</b>
  Potencia de Bomba de agua (W): <b>750</b>
  Horas de uso diarias: <b>1.5</b>

Aparato 3: <b>Iluminación</b>
  Potencia de Iluminación (W): <b>60</b>
  Horas de uso diarias: <b>6</b>

--- Diseño del sistema ---
Consumo diario: 2,985 Wh
Paneles de 450 W: 2
Baterías de 12 V 200 Ah: 5
</pre>

````python
#@title Solución
POTENCIA_PANEL_W = 450
HORAS_SOLARES_PICO = 4.5
FACTOR_PERDIDAS = 0.8
DIAS_AUTONOMIA = 2
PROFUNDIDAD_DESCARGA = 0.5
VOLTAJE_BATERIA = 12
CAPACIDAD_BATERIA_AH = 200


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def redondear_hacia_arriba(valor: float) -> int:
    """Redondea un número positivo al entero superior."""
    entero = int(valor)
    if valor > entero:
        entero += 1
    return entero


def calcular_consumo_diario() -> float:
    """Solicita los aparatos de la instalación y retorna el consumo diario (Wh)."""
    cantidad_aparatos = pedir_entero_seguro("Cantidad de aparatos: ", 1, 20)
    consumo_total = 0
    for numero in range(1, cantidad_aparatos + 1):
        nombre = input(f"\nAparato {numero}: ")
        potencia = pedir_numero_seguro(f"  Potencia de {nombre} (W): ", 1, 10000)
        horas = pedir_numero_seguro("  Horas de uso diarias: ", 0, 24)
        consumo_total += potencia * horas
    return consumo_total


def calcular_paneles(consumo_diario: float, potencia_panel: float, horas_solares: float) -> int:
    """Calcula la cantidad de paneles necesarios."""
    energia_por_panel = potencia_panel * horas_solares * FACTOR_PERDIDAS
    return redondear_hacia_arriba(consumo_diario / energia_por_panel)


def calcular_baterias(consumo_diario: float, dias_autonomia: int, voltaje: float,
                      capacidad_ah: float) -> int:
    """Calcula la cantidad de baterías necesarias para la autonomía indicada."""
    energia_a_almacenar = consumo_diario * dias_autonomia / PROFUNDIDAD_DESCARGA
    energia_por_bateria = voltaje * capacidad_ah
    return redondear_hacia_arriba(energia_a_almacenar / energia_por_bateria)


def main() -> None:
    consumo_diario = calcular_consumo_diario()
    paneles = calcular_paneles(consumo_diario, POTENCIA_PANEL_W, HORAS_SOLARES_PICO)
    baterias = calcular_baterias(consumo_diario, DIAS_AUTONOMIA, VOLTAJE_BATERIA,
                                 CAPACIDAD_BATERIA_AH)

    print("\n--- Diseño del sistema ---")
    print(f"Consumo diario: {consumo_diario:,.0f} Wh")
    print(f"Paneles de {POTENCIA_PANEL_W} W: {paneles}")
    print(f"Baterías de {VOLTAJE_BATERIA} V {CAPACIDAD_BATERIA_AH} Ah: {baterias}")


main()
````

**Explicación**

En el diseño de arriba hacia abajo se escribe primero `main()` como una lista de pasos de alto nivel y luego cada paso se convierte en una función. Observe que `redondear_hacia_arriba` se **reutiliza** en dos funciones distintas: identificar tareas repetidas es una de las ventajas de descomponer el problema. Las dos funciones de lectura segura son un caso extremo de reutilización: se escriben una vez y sirven para todos los datos numéricos del programa, cada uno con su propio rango.

## Ejercicio 7.13 — Enfriamiento de una pieza forjada

Después del forjado, una pieza de acero se deja enfriar al aire. Según la **ley de enfriamiento de Newton**, su temperatura después de $t$ minutos es:

$$T(t) = T_{amb} + (T_0 - T_{amb}) \cdot e^{-k t}$$

donde $T_0$ es la temperatura inicial, $T_{amb}$ la temperatura ambiente, $k$ la constante de enfriamiento (1/min) y $e \approx 2.718281828$.

Escriba un programa en Python con las siguientes funciones:

A. `calcular_temperatura(tiempo: float, temperatura_inicial: float, temperatura_ambiente: float, constante: float) -> float`: use `pow()`.

B. `calcular_tiempo_seguro(temperatura_inicial: float, temperatura_ambiente: float, constante: float, temperatura_segura: float) -> int`: avanza **minuto a minuto** con un ciclo `while`, invocando la función anterior, hasta que la pieza esté a la temperatura segura de manipulación o menos; imprime la temperatura cada 15 minutos (redondeada a 1 decimal con `round()`) y retorna los minutos transcurridos.

C. `main() -> None`: solicita los datos y muestra el tiempo necesario en horas y minutos (use `divmod`). Si la temperatura segura es menor o igual a la ambiente, muestra un error, porque la pieza **nunca** la alcanzaría. Si alguno de los datos no es un número, muestra `Error: todos los datos deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Temperatura ambiente (°C): <b>28</b>
Constante de enfriamiento k (1/min): <b>0.045</b>
Temperatura segura (°C): <b>60</b>

  Minuto 15: 599.3 °C
  Minuto 30: 318.9 °C
  Minuto 45: 176.1 °C
  Minuto 60: 103.4 °C
  Minuto 75: 66.4 °C

Tiempo hasta la temperatura segura: 1 h 20 min
</pre>

<pre>
Temperatura inicial (°C): <b>1150</b>
Temperatura ambiente (°C): <b>28</b>
Constante de enfriamiento k (1/min): <b>0.045</b>
Temperatura segura (°C): <b>25</b>
Error: la pieza nunca llegará a una temperatura igual o menor a la ambiente.
</pre>

````python
#@title Solución
NUMERO_E = 2.718281828


def calcular_temperatura(tiempo: float, temperatura_inicial: float,
                         temperatura_ambiente: float, constante: float) -> float:
    """Calcula la temperatura de la pieza después de un tiempo (min)."""
    diferencia_inicial = temperatura_inicial - temperatura_ambiente
    return temperatura_ambiente + diferencia_inicial * pow(NUMERO_E, -constante * tiempo)


def calcular_tiempo_seguro(temperatura_inicial: float, temperatura_ambiente: float,
                           constante: float, temperatura_segura: float) -> int:
    """Retorna los minutos necesarios para llegar a la temperatura segura."""
    minutos = 0
    temperatura = temperatura_inicial
    while temperatura > temperatura_segura:
        minutos += 1
        temperatura = calcular_temperatura(minutos, temperatura_inicial,
                                           temperatura_ambiente, constante)
        if minutos % 15 == 0:
            print(f"  Minuto {minutos}: {round(temperatura, 1)} °C")
    return minutos


def main() -> None:
    try:
        temperatura_inicial = float(input("Temperatura inicial (°C): "))
        temperatura_ambiente = float(input("Temperatura ambiente (°C): "))
        constante = float(input("Constante de enfriamiento k (1/min): "))
        temperatura_segura = float(input("Temperatura segura (°C): "))

        if temperatura_segura <= temperatura_ambiente:
            print("Error: la pieza nunca llegará a una temperatura igual o menor a la ambiente.")
        else:
            print()
            minutos = calcular_tiempo_seguro(temperatura_inicial, temperatura_ambiente,
                                             constante, temperatura_segura)
            horas, minutos_restantes = divmod(minutos, 60)
            print(f"\nTiempo hasta la temperatura segura: {horas} h {minutos_restantes} min")
    except ValueError:
        print("Error: todos los datos deben ser números.")


main()
````

**Explicación**

La validación en `main()` es indispensable: la temperatura se acerca a la ambiente **sin llegar nunca** a ella, así que si se pidiera una temperatura segura menor o igual a la ambiente, el ciclo `while` de `calcular_tiempo_seguro` sería **infinito**.

## Ejercicio 7.14 — Volumen de tanques de almacenamiento

Una planta química tiene tanques de tres formas. Sus volúmenes son ($\pi = 3.1416$):

* **Cilíndrico vertical:** $V = \pi r^2 h$
* **Esférico:** $V = \frac{4}{3} \pi r^3$ (la altura no se usa)
* **Cónico** (tolva): $V = \frac{1}{3} \pi r^2 h$

Escriba un programa en Python con las siguientes funciones:

A. `calcular_volumen(radio: float, altura: float = 0, forma: str = "cilindro") -> float`: usa `match` según la forma; si la forma no existe, retorna `-1`.

B. `metros_cubicos_a_litros(volumen: float) -> float`

C. `main() -> None`: solicita la forma del tanque (`cilindro`, `esfera` o `cono`), el radio y —**solo si la forma no es esfera**— la altura. Invoca `calcular_volumen` pasando la forma **por nombre** y muestra el volumen en m³ (2 decimales) y en litros (separador de miles, 0 decimales), o `Forma de tanque desconocida.` Si el radio o la altura no son números, muestra `Error: el radio y la altura deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplos de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Radio (m): <b>3.5</b>

</pre>

<pre>
Forma del tanque (cilindro/esfera/cono): <b>cono</b>
Radio (m): <b>1.2</b>
Altura (m): <b>2.8</b>
Volumen: 4.22 m³ = 4,222 L
</pre>

<pre>

Radio (m): <b>2</b>
Altura (m): <b>4</b>
Forma de tanque desconocida.
</pre>

````python
#@title Solución
PI = 3.1416


def calcular_volumen(radio: float, altura: float = 0, forma: str = "cilindro") -> float:
    """Calcula el volumen (m³) de un tanque; retorna -1 si la forma no existe."""
    match forma:
        case "cilindro":
            volumen = PI * radio ** 2 * altura
        case "esfera":
            volumen = 4 / 3 * PI * radio ** 3
        case "cono":
            volumen = PI * radio ** 2 * altura / 3
        case _:
            volumen = -1
    return volumen


def metros_cubicos_a_litros(volumen: float) -> float:
    """Convierte un volumen de m³ a litros."""
    return volumen * 1000


def main() -> None:
    forma = input("Forma del tanque (cilindro/esfera/cono): ").lower()

    try:
        radio = float(input("Radio (m): "))

        if forma == "esfera":
            volumen = calcular_volumen(radio, forma=forma)
        else:
            altura = float(input("Altura (m): "))
            volumen = calcular_volumen(radio, altura, forma=forma)

        if volumen < 0:
            print("Forma de tanque desconocida.")
        else:
            print(f"Volumen: {volumen:.2f} m³ = {metros_cubicos_a_litros(volumen):,.0f} L")
    except ValueError:
        print("Error: el radio y la altura deben ser números.")


main()
````

**Explicación**

Como la esfera no necesita altura, el valor por defecto `altura = 0` permite invocar la función sin ese argumento. Pasar `forma=forma` **por nombre** es indispensable en ese caso: si se escribiera `calcular_volumen(radio, forma)`, la forma se asignaría al parámetro `altura`.

Potencia total: 7.80 kW
</pre>

````python
#@title Solución
COEFICIENTE_UTILIZACION = 0.65
FACTOR_MANTENIMIENTO = 0.8
POTENCIA_LUMINARIA_W = 150


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def obtener_iluminancia(actividad: str) -> int:
    """Retorna la iluminancia (lux) requerida por la actividad, o 0 si no existe."""
    if actividad == "bodega":
        iluminancia = 300
    elif actividad == "general":
        iluminancia = 500
    elif actividad == "precision":
        iluminancia = 1000
    else:
        iluminancia = 0
    return iluminancia


def calcular_luminarias(iluminancia: float, area: float, flujo: float) -> int:
    """Calcula las luminarias necesarias con el método de los lúmenes."""
    flujo_util = flujo * COEFICIENTE_UTILIZACION * FACTOR_MANTENIMIENTO
    return int(-(-iluminancia * area // flujo_util))


def main() -> None:
    largo = pedir_numero_seguro("Largo de la nave (m): ", 1, 500)
    ancho = pedir_numero_seguro("Ancho de la nave (m): ", 1, 500)

    iluminancia = 0
    while iluminancia == 0:
        actividad = input("Actividad (bodega/general/precision): ").lower()
        iluminancia = obtener_iluminancia(actividad)
        if iluminancia == 0:
            print("Actividad desconocida.")

    flujo = pedir_numero_seguro("Flujo de la luminaria (lm): ", 500, 60000)

    area = largo * ancho
    luminarias = calcular_luminarias(iluminancia, area, flujo)

    print(f"\nÁrea: {area:.1f} m²")
    print(f"Iluminancia requerida: {iluminancia} lux")
    print(f"Luminarias necesarias: {luminarias}")
    print(f"Potencia total: {luminarias * POTENCIA_LUMINARIA_W / 1000:.2f} kW")


main()
````

**Explicación**

`-(-x // y)` es un truco para dividir redondeando **hacia arriba**: `//` redondea hacia abajo, así que al negar el numerador y luego el resultado se obtiene el techo de la división. En iluminación siempre se redondea hacia arriba: con menos luminarias no se alcanzaría la iluminancia requerida. Observe que la actividad se valida con su propio ciclo `while`: `pedir_numero_seguro` sirve para los datos **numéricos**, pero la misma idea (repetir mientras el dato sea inválido) se aplica a cualquier tipo de entrada.

## Ejercicio 7.16 — Registros binarios de un microcontrolador

Un microcontrolador reporta el estado de sus sensores en un **registro de 8 bits**, escrito como una cadena de ceros y unos (por ejemplo, `10110010`). El bit de la **izquierda** es el más significativo (vale $2^7$) y el de la derecha el menos significativo (vale $2^0$). Cada bit en `1` indica un sensor **activo**.

Escriba un programa en Python con las siguientes funciones:

A. `es_registro_valido(registro: str) -> bool`: `True` si tiene exactamente 8 caracteres y todos son `0` o `1`.

B. `binario_a_decimal(registro: str) -> int`: convierte el registro a decimal **sin usar** `int(registro, 2)`; use `pow(2, posicion)`.

C. `contar_sensores_activos(registro: str) -> int`

D. `main() -> None`: solicita registros **hasta** que se ingrese una cadena vacía (solo presionar Enter). Para cada registro válido muestra su valor decimal y la cantidad de sensores activos, y si el **bit 7** (el de la izquierda, que indica una alarma) está en 1, muestra `¡ALARMA ACTIVA!`. Si el registro no es válido, muestra `  Error: el registro debe tener 8 bits (solo 0 y 1).` y **vuelve a solicitar** otro registro: como el programa ya está en un ciclo, no hay que terminarlo por un dato mal digitado.

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Registro (Enter para terminar): <b>10110010</b>
  Valor decimal: 178
  Sensores activos: 4
  ¡ALARMA ACTIVA!
Registro (Enter para terminar): <b>00000101</b>
  Valor decimal: 5
  Sensores activos: 2
Registro (Enter para terminar): <b>1012</b>
  Error: el registro debe tener 8 bits (solo 0 y 1).
Registro (Enter para terminar): <b></b>
Monitoreo finalizado.
</pre>

````python
#@title Solución
CANTIDAD_BITS = 8


def es_registro_valido(registro: str) -> bool:
    """Indica si el registro tiene 8 caracteres y todos son 0 o 1."""
    valido = len(registro) == CANTIDAD_BITS
    for bit in registro:
        if bit != "0" and bit != "1":
            valido = False
    return valido


def binario_a_decimal(registro: str) -> int:
    """Convierte una cadena binaria a su valor decimal."""
    valor = 0
    for indice in range(len(registro)):
        posicion = len(registro) - 1 - indice
        if registro[indice] == "1":
            valor += pow(2, posicion)
    return valor


def contar_sensores_activos(registro: str) -> int:
    """Cuenta los bits en 1 del registro."""
    activos = 0
    for bit in registro:
        if bit == "1":
            activos += 1
    return activos


def main() -> None:
    registro = input("Registro (Enter para terminar): ")
    while registro != "":
        if es_registro_valido(registro):
            print(f"  Valor decimal: {binario_a_decimal(registro)}")
            print(f"  Sensores activos: {contar_sensores_activos(registro)}")
            if registro[0] == "1":
                print("  ¡ALARMA ACTIVA!")
        else:
            print("  Error: el registro debe tener 8 bits (solo 0 y 1).")
        registro = input("Registro (Enter para terminar): ")
    print("Monitoreo finalizado.")


main()
````

**Explicación**

El bit en el índice `i` de la cadena corresponde a la potencia `len(registro) - 1 - i`, porque la cadena se lee de izquierda a derecha pero las potencias de 2 **crecen** de derecha a izquierda. Por ejemplo, `10110010` = 128 + 32 + 16 + 2 = **178**.

## Ejercicio 7.17 — Tren de engranajes

Un reductor de velocidad está formado por varias **etapas** de engranajes. En cada etapa, un piñón de $z_1$ dientes mueve una rueda de $z_2$ dientes, y:

$$n_{salida} = n_{entrada} \cdot \frac{z_1}{z_2} \qquad T_{salida} = T_{entrada} \cdot \frac{z_2}{z_1} \cdot \eta$$

donde $n$ es la velocidad (rpm), $T$ el torque (N·m) y $\eta = 0.97$ la eficiencia de cada etapa.

Escriba un programa en Python con las siguientes funciones:

A. `pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float`: muestra el mensaje y lee un número; si no es numérico o está fuera del rango `[minimo, maximo]`, muestra un error y lo **vuelve a solicitar** (la misma función del ejercicio 7.5).

B. `pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int`: igual que la anterior, pero con `int()`, para los datos que deben ser enteros.

C. `calcular_etapa(velocidad: float, torque: float, dientes_pinon: int, dientes_rueda: int) -> tuple[float, float]`: retorna la velocidad y el torque de salida de la etapa.

D. `calcular_potencia_kw(velocidad: float, torque: float) -> float`: $P = \frac{T \cdot n \cdot 2\pi}{60 \cdot 1000}$ ($\pi = 3.1416$).

E. `main() -> None`: solicita, con las funciones de lectura segura, la velocidad (de 1 a 10 000 rpm) y el torque del motor (de 0.1 a 10 000 N·m) y la cantidad de etapas (de 1 a 5). Para cada etapa solicita los dientes del piñón y de la rueda (de 8 a 200), invoca `calcular_etapa` **desempaquetando** su resultado y muestra la velocidad y el torque de salida. Al final muestra la **relación de reducción total** y la potencia de entrada y de salida (2 decimales).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Torque del motor (N·m): <b>40</b>

Etapa 1 - dientes de la rueda: <b>72</b>
  Salida: 437.5 rpm, 155.2 N·m

Etapa 2 - dientes del piñón: <b>20</b>
Etapa 2 - dientes de la rueda: <b>60</b>
  Salida: 145.8 rpm, 451.6 N·m

Relación de reducción total: 12.00 : 1
Potencia de entrada: 7.33 kW
Potencia de salida: 6.90 kW
</pre>

**Pregunta de respuesta corta**

¿Por qué la potencia de salida es menor que la de entrada si el torque aumentó tanto?

````python
#@title Solución
EFICIENCIA_ETAPA = 0.97
PI = 3.1416


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def calcular_etapa(velocidad: float, torque: float, dientes_pinon: int,
                   dientes_rueda: int) -> tuple[float, float]:
    """Retorna la velocidad (rpm) y el torque (N·m) a la salida de una etapa."""
    relacion = dientes_rueda / dientes_pinon
    velocidad_salida = velocidad / relacion
    torque_salida = torque * relacion * EFICIENCIA_ETAPA
    return velocidad_salida, torque_salida


def calcular_potencia_kw(velocidad: float, torque: float) -> float:
    """Calcula la potencia mecánica (kW) a partir de la velocidad y el torque."""
    return torque * velocidad * 2 * PI / (60 * 1000)


def main() -> None:
    velocidad_motor = pedir_numero_seguro("Velocidad del motor (rpm): ", 1, 10000)
    torque_motor = pedir_numero_seguro("Torque del motor (N·m): ", 0.1, 10000)
    cantidad_etapas = pedir_entero_seguro("Cantidad de etapas: ", 1, 5)

    velocidad = velocidad_motor
    torque = torque_motor
    for etapa in range(1, cantidad_etapas + 1):
        dientes_pinon = pedir_entero_seguro(f"\nEtapa {etapa} - dientes del piñón: ", 8, 200)
        dientes_rueda = pedir_entero_seguro(f"Etapa {etapa} - dientes de la rueda: ", 8, 200)
        velocidad, torque = calcular_etapa(velocidad, torque, dientes_pinon, dientes_rueda)
        print(f"  Salida: {velocidad:.1f} rpm, {torque:.1f} N·m")

    print(f"\nRelación de reducción total: {velocidad_motor / velocidad:.2f} : 1")
    print(f"Potencia de entrada: {calcular_potencia_kw(velocidad_motor, torque_motor):.2f} kW")
    print(f"Potencia de salida: {calcular_potencia_kw(velocidad, torque):.2f} kW")


main()
````

**Explicación**

**Respuesta corta:** un reductor **intercambia** velocidad por torque, pero no puede crear energía: la potencia es el producto de ambos, y el torque aumenta en la misma proporción en que disminuye la velocidad. La potencia de salida es menor solo por las **pérdidas** de cada etapa ($0.97^2 \approx 0.94$).

## Ejercicio 7.18 — Tiempo de ciclo de una celda robotizada

Una celda de manufactura tiene **tres robots en serie**: cada pieza pasa por el robot 1, luego por el 2 y luego por el 3. Como trabajan en paralelo sobre piezas distintas, el **tiempo de ciclo** de la celda lo determina el robot **más lento** (el *cuello de botella*).

Escriba un programa en Python con las siguientes funciones:

A. `formatear_duracion(segundos: int) -> str`: retorna la duración en formato `HH:MM:SS` usando `divmod` dos veces.

B. `calcular_produccion(tiempo_ciclo: float, horas_turno: float) -> int`: piezas completas producidas en un turno.

C. `main() -> None`: solicita el tiempo de operación de cada robot (s) y las horas del turno. Muestra el **cuello de botella** (número de robot y su tiempo, usando `max()`), la producción del turno, el tiempo que la **pieza tarda** en atravesar toda la celda (suma de los tres tiempos) y el tiempo total del turno, ambos formateados con `formatear_duracion`. Muestra también la **utilización** de cada robot (su tiempo entre el tiempo de ciclo, en %). Si alguno de los datos no es un número, muestra `Error: los tiempos y las horas deben ser números.` y **finaliza** el programa (no los vuelve a solicitar).

**Ejemplo de ejecución**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>

Tiempo del robot 2 (s): <b>58</b>

Tiempo para atravesar la celda: 00:02:17
Duración del turno: 08:00:00
Utilización: R1 73%, R2 100%, R3 64%
</pre>

````python
#@title Solución
def formatear_duracion(segundos: int) -> str:
    """Retorna una duración en segundos con el formato HH:MM:SS."""
    minutos_totales, segundos_restantes = divmod(segundos, 60)
    horas, minutos = divmod(minutos_totales, 60)
    return f"{horas:02d}:{minutos:02d}:{segundos_restantes:02d}"


def calcular_produccion(tiempo_ciclo: float, horas_turno: float) -> int:
    """Calcula las piezas completas producidas en un turno."""
    return int(horas_turno * 3600 // tiempo_ciclo)


def main() -> None:
    try:
        tiempo_robot_1 = float(input("Tiempo del robot 1 (s): "))
        tiempo_robot_2 = float(input("Tiempo del robot 2 (s): "))
        tiempo_robot_3 = float(input("Tiempo del robot 3 (s): "))
        horas_turno = float(input("Horas del turno: "))

        tiempo_ciclo = max(tiempo_robot_1, tiempo_robot_2, tiempo_robot_3)
        if tiempo_ciclo == tiempo_robot_1:
            cuello_botella = 1
        elif tiempo_ciclo == tiempo_robot_2:
            cuello_botella = 2
        else:
            cuello_botella = 3

        produccion = calcular_produccion(tiempo_ciclo, horas_turno)
        tiempo_atravesar = int(tiempo_robot_1 + tiempo_robot_2 + tiempo_robot_3)

        print(f"\nCuello de botella: robot {cuello_botella} ({tiempo_ciclo} s)")
        print(f"Producción del turno: {produccion} piezas")
        print(f"Tiempo para atravesar la celda: {formatear_duracion(tiempo_atravesar)}")
        print(f"Duración del turno: {formatear_duracion(int(horas_turno * 3600))}")
        print(f"Utilización: R1 {tiempo_robot_1 / tiempo_ciclo:.0%}, "
              f"R2 {tiempo_robot_2 / tiempo_ciclo:.0%}, R3 {tiempo_robot_3 / tiempo_ciclo:.0%}")
    except ValueError:
        print("Error: los tiempos y las horas deben ser números.")


main()
````

**Explicación**

La función `formatear_duracion` retorna un `str` en lugar de imprimirlo, de modo que `main()` decide **dónde** y **cómo** mostrarlo; así se reutiliza para dos duraciones distintas. Mejorar cualquier robot que **no** sea el cuello de botella no aumenta la producción.

Si por error la segunda condición fuera `resistencia <= 17`, ¿cuál de las pruebas fallaría?

````python
#@title Solución
def clasificar_concreto(resistencia: float) -> str:
    """Retorna la clase del concreto según su resistencia (MPa)."""
    if resistencia < 0:
        clase = "Inválida"
    elif resistencia < 17:
        clase = "No estructural"
    elif resistencia < 28:
        clase = "Normal"
    elif resistencia < 42:
        clase = "Alta"
    else:
        clase = "Muy alta"
    return clase


def probar_clasificacion() -> None:
    """Prueba la clasificación en las fronteras y en casos especiales."""
    pruebas = 0
    assert clasificar_concreto(-5) == "Inválida"
    pruebas += 1
    assert clasificar_concreto(0) == "No estructural"
    pruebas += 1
    assert clasificar_concreto(16.9) == "No estructural"
    pruebas += 1
    assert clasificar_concreto(17) == "Normal"
    pruebas += 1
    assert clasificar_concreto(27.9) == "Normal"
    pruebas += 1
    assert clasificar_concreto(28) == "Alta"
    pruebas += 1
    assert clasificar_concreto(41.9) == "Alta"
    pruebas += 1
    assert clasificar_concreto(42) == "Muy alta"
    pruebas += 1
    print(f"{pruebas} pruebas superadas.\n")


def main() -> None:
    probar_clasificacion()

    no_estructural = 0
    normal = 0
    alta = 0
    muy_alta = 0

    entrada = input("Resistencia (MPa) o 'fin': ")
    while entrada != "fin":
        try:
            clase = clasificar_concreto(float(entrada))
            print(f"  Clase: {clase}")
            if clase == "No estructural":
                no_estructural += 1
            elif clase == "Normal":
                normal += 1
            elif clase == "Alta":
                alta += 1
            elif clase == "Muy alta":
                muy_alta += 1
        except ValueError:
            print("  Error: ingrese un número o 'fin'.")
        entrada = input("Resistencia (MPa) o 'fin': ")

    print(f"\nNo estructural: {no_estructural}, Normal: {normal}, Alta: {alta}, Muy alta: {muy_alta}")


main()
````

**Explicación**

**Respuesta corta:** fallaría `assert clasificar_concreto(17) == "Normal"`, porque 17 se clasificaría como "No estructural". Las pruebas en las **fronteras exactas** son las únicas capaces de detectar la confusión entre `<` y `<=`.

El `try`/`except` va **dentro** del ciclo: así un dato mal digitado solo descarta ese ensayo y el programa sigue pidiendo resistencias, en lugar de terminar y perder la cuenta de los anteriores.

````python
#@title Solución
LIMITE_EJE_SIMPLE = 10
LIMITE_EJE_DOBLE = 18
MULTA_POR_TONELADA = 65000


def pedir_numero_seguro(mensaje: str, minimo: float, maximo: float) -> float:
    """Pide un número dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = float(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número.")
    return numero


def pedir_entero_seguro(mensaje: str, minimo: int, maximo: int) -> int:
    """Pide un número entero dentro de [minimo, maximo], repitiendo hasta que sea válido."""
    numero_valido = False
    while not numero_valido:
        try:
            numero = int(input(mensaje))
            if minimo <= numero <= maximo:
                numero_valido = True
            else:
                print(f"Error: el valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: ingrese un número entero.")
    return numero


def calcular_exceso(peso: float, tipo_eje: str) -> float:
    """Retorna las toneladas que un eje excede su límite legal."""
    if tipo_eje == "s":
        limite = LIMITE_EJE_SIMPLE
    else:
        limite = LIMITE_EJE_DOBLE
    return max(peso - limite, 0)


def calcular_multa(exceso_total: float) -> int:
    """Calcula la multa redondeando el exceso hacia arriba."""
    toneladas_cobradas = int(-(-exceso_total // 1))
    return toneladas_cobradas * MULTA_POR_TONELADA


def pesar_camion() -> float:
    """Pesa los ejes de un camión, muestra el detalle y retorna la multa."""
    placa = input("Placa: ").upper()
    cantidad_ejes = pedir_entero_seguro("Cantidad de ejes: ", 2, 6)

    exceso_total = 0
    for eje in range(1, cantidad_ejes + 1):
        tipo_eje = input(f"Eje {eje} - tipo (s/d): ").lower()
        while tipo_eje != "s" and tipo_eje != "d":
            tipo_eje = input(f"Eje {eje} - tipo inválido, use s o d: ").lower()
        peso = pedir_numero_seguro(f"Eje {eje} - peso (t): ", 0.1, 40)
        exceso = calcular_exceso(peso, tipo_eje)
        exceso_total += exceso
        if exceso > 0:
            print(f"  Exceso de {exceso:.2f} t")

    multa = calcular_multa(exceso_total)
    if multa > 0:
        print(f"Camión {placa}: MULTA de ₡{multa:,}")
    else:
        print(f"Camión {placa}: dentro de los límites.")
    return multa


def main() -> None:
    camiones = 0
    multados = 0
    recaudado = 0

    opcion = ""
    while opcion != "2":
        opcion = input("\n1. Pesar camión  2. Cerrar turno\nOpción: ")
        if opcion == "1":
            multa = pesar_camion()
            camiones += 1
            recaudado += multa
            if multa > 0:
                multados += 1
        elif opcion != "2":
            print("Opción inválida.")

    print(f"\nCamiones pesados: {camiones}")
    print(f"Camiones multados: {multados}")
    print(f"Total recaudado: ₡{recaudado:,}")


main()
````

**Explicación**

`max(peso - limite, 0)` es una forma compacta de decir "el exceso, pero nunca negativo". El programa sigue la estructura de arriba hacia abajo: `main()` solo coordina el menú y los totales, `pesar_camion()` se encarga de un camión completo, y las funciones más pequeñas resuelven cada cálculo.
