## Ejercicio 4.1 — Velocidad media de un tren

Un ingeniero ferroviario escribió este programa para calcular la velocidad media de un tren en **km/h**:

```python
distancia_km = float(input("Distancia recorrida (km): "))
tiempo_min = float(input("Tiempo de viaje (min): "))

velocidad_kmh = distancia_km / tiempo_min

print(f"Velocidad media: {velocidad_kmh:.2f} km/h")
```

Con 84 km en 70 minutos, el programa reporta `1.20 km/h`, lo cual es absurdo para un tren.

1. ¿Es un error lógico o de ejecución? Justifique.
2. Corrija el programa.
3. Proponga un caso de prueba **fácil de verificar mentalmente** que habría detectado el error.

**Ejemplos de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Distancia recorrida (km): <b>84</b>
Tiempo de viaje (min): <b>70</b>
Velocidad media: 72.00 km/h
</pre>

<pre>
Distancia recorrida (km): <b>60</b>
Tiempo de viaje (min): <b>60</b>
Velocidad media: 60.00 km/h
</pre>

````python
#@title Solución
distancia_km = float(input("Distancia recorrida (km): "))
tiempo_min = float(input("Tiempo de viaje (min): "))

tiempo_horas = tiempo_min / 60
velocidad_kmh = distancia_km / tiempo_horas

print(f"Velocidad media: {velocidad_kmh:.2f} km/h")
````

**Explicación**

Es un **error lógico**: el programa termina sin detenerse, pero el resultado es incorrecto porque divide kilómetros entre **minutos**, lo que da km/min. La corrección convierte el tiempo a horas.

Un buen caso de prueba es **60 km en 60 minutos**: cualquiera sabe que la respuesta debe ser 60 km/h, y el programa original habría mostrado 1.00 km/h.

Un programa de monitoreo de un reactor lee la temperatura desde el teclado:

```python
temperatura = int(input("Temperatura del reactor (°C): "))
temperatura_kelvin = temperatura + 273.15
print(f"Temperatura: {temperatura_kelvin} K")
```

El operador escribe `351.7` y el programa se detiene.

1. ¿Qué **excepción** se produce y por qué?

## Ejercicio 4.3 — Reporte de un transformador

Un técnico de una subestación necesita un reporte rápido de la carga de un transformador trifásico. La potencia aparente es $S = \sqrt{3} \cdot V \cdot I$ (VA) y el porcentaje de carga es $\frac{S}{S_{nominal}} \cdot 100$. El técnico escribió este programa, pero se detiene en la penúltima línea:

```python
voltaje = float(input("Voltaje de línea (V): "))
corriente = float(input("Corriente de línea (A): "))
potencia_nominal_kva = float(input("Potencia nominal (kVA): "))

potencia_aparente_kva = 3 ** 0.5 * voltaje * corriente / 1000
porcentaje_carga = potencia_aparente_kva / potencia_nominal_kva * 100

print("Potencia aparente: " + potencia_aparente_kva + " kVA")
print(f"Carga: {porcentaje_carga:.1f}%")
```

A. Indique qué **excepción** se produce y por qué.

B. Escriba el programa corregido, mostrando la potencia con 2 decimales.

C. Agregue un booleano `sobrecargado` que sea `True` si la carga **supera el 100 %**.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Voltaje de línea (V): <b>480</b>
Corriente de línea (A): <b>130</b>
Potencia nominal (kVA): <b>112.5</b>
Potencia aparente: 108.08 kVA
Carga: 96.1%
¿Sobrecargado?: False
</pre>

**Pregunta de respuesta corta**

Además de la *f-string*, ¿de qué otra forma se podría corregir la línea que falla?

````python
#@title Solución
voltaje = float(input("Voltaje de línea (V): "))
corriente = float(input("Corriente de línea (A): "))
potencia_nominal_kva = float(input("Potencia nominal (kVA): "))

potencia_aparente_kva = 3 ** 0.5 * voltaje * corriente / 1000
porcentaje_carga = potencia_aparente_kva / potencia_nominal_kva * 100
sobrecargado = porcentaje_carga > 100

print(f"Potencia aparente: {potencia_aparente_kva:.2f} kVA")
print(f"Carga: {porcentaje_carga:.1f}%")
print(f"¿Sobrecargado?: {sobrecargado}")
````

**Explicación**

Se produce **`TypeError: can only concatenate str (not "float") to str`**: el operador `+` no puede unir un texto con un número.

**Respuesta corta:** convirtiendo el número a texto con `str()`: `"Potencia aparente: " + str(potencia_aparente_kva) + " kVA"` (aunque así no se controlan los decimales).

## Ejercicio 4.4 — Energía cinética de un vehículo

En un laboratorio de seguridad vial se estudia la energía que debe disipar el sistema de frenos de un vehículo. La energía cinética es $E_c = \frac{1}{2} m v^2$, con la velocidad en m/s ($1 \text{ km/h} = \frac{1}{3.6} \text{ m/s}$). Un estudiante escribió:

```python
masa_kg = float(input("Masa del vehículo (kg): "))
velocidad_kmh = float(input("Velocidad (km/h): "))

velocidad_ms = velocidad_kmh / 3.6
energia_cinetica = 0.5 * masa_kg * velocidad_ms ** 2
energia_kj = energia_cinética / 1000

print(f"Energía a disipar: {energia_kj:.1f} kJ")
```

A. Indique qué **excepción** se produce, en qué línea y por qué Python dice que la variable "no existe".

B. Escriba el programa corregido.

C. Agregue el cálculo de la energía si el vehículo fuera al **doble** de velocidad y muestre **cuántas veces** mayor es.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Masa del vehículo (kg): <b>1450</b>
Velocidad (km/h): <b>100</b>
Energía a disipar: 559.4 kJ
Energía al doble de velocidad: 2237.7 kJ
Es 4.0 veces mayor
</pre>

````python
#@title Solución
masa_kg = float(input("Masa del vehículo (kg): "))
velocidad_kmh = float(input("Velocidad (km/h): "))

velocidad_ms = velocidad_kmh / 3.6
energia_cinetica = 0.5 * masa_kg * velocidad_ms ** 2
energia_kj = energia_cinetica / 1000

energia_doble_velocidad_kj = 0.5 * masa_kg * (2 * velocidad_ms) ** 2 / 1000
razon = energia_doble_velocidad_kj / energia_kj

print(f"Energía a disipar: {energia_kj:.1f} kJ")
print(f"Energía al doble de velocidad: {energia_doble_velocidad_kj:.1f} kJ")
print(f"Es {razon:.1f} veces mayor")
````

**Explicación**

Se produce **`NameError: name 'energia_cinética' is not defined`** en la línea de `energia_kj`. Para Python, `energia_cinetica` y `energia_cinética` (con tilde) son **nombres distintos**. Por eso se recomienda escribir los nombres de variables **sin tildes ni ñ**.

La energía crece con el **cuadrado** de la velocidad: al duplicarla, la energía que deben disipar los frenos se multiplica por 4.

## Ejercicio 4.5 — Termostato con precedencia incorrecta

El firmware de un termostato convierte de grados Fahrenheit a Celsius con esta línea:

```python
celsius = fahrenheit - 32 * 5 / 9
```

1. Evalúe la línea para `fahrenheit = 212` y compare con el valor correcto (el agua hierve a 100 °C).
2. Explique el error.
3. Escriba el programa corregido, que solicite la temperatura en °F y muestre los °C, y agregue un `assert` que verifique el punto de congelación (32 °F = 0 °C).

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Temperatura (°F): <b>212</b>
Temperatura: 100.00 °C
</pre>

````python
#@title Solución
fahrenheit = float(input("Temperatura (°F): "))

celsius = (fahrenheit - 32) * 5 / 9

print(f"Temperatura: {celsius:.2f} °C")

# Prueba con un valor conocido: el punto de congelación del agua
assert (32 - 32) * 5 / 9 == 0
````

**Explicación**

Con la línea original: `32 * 5 / 9 = 17.78`, y `212 - 17.78 = 194.22`, en lugar de `100`. Como `*` y `/` tienen **mayor precedencia** que `-`, la resta se hace al final. Es un **error lógico**: el programa no se detiene, solo da un valor equivocado. Los paréntesis `(fahrenheit - 32)` fuerzan el orden correcto.

## Ejercicio 4.6 — Ficha técnica de una bomba

Un ingeniero hidráulico calcula la **potencia** que requiere una bomba:

$$P = \frac{\rho \cdot g \cdot Q \cdot H}{\eta}$$

con $\rho = 1000 \text{ kg/m}^3$, $g = 9.81 \text{ m/s}^2$, $Q$ el caudal (m³/s), $H$ la altura (m) y $\eta$ la eficiencia (fracción). El código de la bomba tiene el formato `BC-0457`, y la **serie** de la bomba son sus dos primeras letras. Este es el programa, que tiene **tres errores de ejecución**:

```python
codigo = input("Código de la bomba: ")
caudal_ls = float(input("Caudal (L/s): "))
altura_m = float(input("Altura de bombeo (m): "))
eficiencia_porcentaje = input("Eficiencia (%): ")

serie = codigo[0] + codigo[1] + codigo[7]
caudal_m3s = caudal_ls / 1000
eficiencia = eficiencia_porcentaje / 100
potencia_w = 1000 * 9.81 * caudal_m3s * altura_m / eficiencia
print("Serie: " + serie + ", potencia: " + potencia_w / 1000 + " kW")
```

A. Identifique los tres errores y nombre la excepción que produce cada uno.

B. Escriba el programa corregido, mostrando la serie y la potencia en kW con 2 decimales.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Código de la bomba: <b>BC-0457</b>
Caudal (L/s): <b>35</b>
Altura de bombeo (m): <b>42</b>
Eficiencia (%): <b>72</b>
Serie: BC, potencia: 20.03 kW
</pre>

**Pregunta de respuesta corta**

Clasifique cada fragmento como **error de ejecución** (nombre la excepción), **error lógico** o **correcto**:

* `area = 3.1416 * 0.2 * 2` (área de un círculo de radio 0.2)
* `presion = float("3,2")`
* `eficiencia = 450 / 500 * 100`

````python
#@title Solución
DENSIDAD_AGUA = 1000
GRAVEDAD = 9.81

codigo = input("Código de la bomba: ")
caudal_ls = float(input("Caudal (L/s): "))
altura_m = float(input("Altura de bombeo (m): "))
eficiencia_porcentaje = float(input("Eficiencia (%): "))

serie = codigo[0:2]
caudal_m3s = caudal_ls / 1000
eficiencia = eficiencia_porcentaje / 100
potencia_w = DENSIDAD_AGUA * GRAVEDAD * caudal_m3s * altura_m / eficiencia
potencia_kw = potencia_w / 1000

print(f"Serie: {serie}, potencia: {potencia_kw:.2f} kW")
````

**Explicación**

1. `codigo[7]`: **`IndexError`**, el código tiene 7 caracteres (índices 0 a 6). Además, la serie son solo las dos primeras letras: `codigo[0:2]`.
2. `eficiencia_porcentaje / 100`: **`TypeError`**, porque la eficiencia nunca se convirtió de `str` a número.
3. `"..." + potencia_w / 1000 + "..."`: **`TypeError`**, no se puede concatenar `str` con `float`.

Python se detiene en el **primer** error que encuentra; los demás solo aparecen después de corregir el anterior. Por eso conviene revisar todo el código y no solo la línea que reporta el *traceback*.

**Respuesta corta:** el primero es un **error lógico** (`* 2` en lugar de `** 2`); el segundo, un **`ValueError`**; el tercero es **correcto** (`90.0`).

## Ejercicio 4.7 — Leer un *traceback*

Al ejecutar un programa que procesa la etiqueta de un medidor de agua, aparece este mensaje:

```text
Traceback (most recent call last):
  File "medidor.py", line 4, in <module>
    digito_control = etiqueta[11]
                     ~~~~~~~~^^^^
IndexError: string index out of range
```

El programa completo es:

```python
etiqueta = input("Etiqueta del medidor: ")
zona = etiqueta[0:2]
numero = int(etiqueta[3:9])
digito_control = etiqueta[11]
print(f"Zona {zona}, medidor {numero}, control {digito_control}")
```

La etiqueta ingresada fue `SJ-004512-7`. El programador usó el índice 11 porque el dígito de control "es el carácter número 11".

1. ¿En qué línea está el error y qué significa el mensaje?
2. ¿Cuál es el índice correcto del último carácter? Corrija el programa de forma que funcione **aunque** la etiqueta tenga otra longitud.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Etiqueta del medidor: <b>SJ-004512-7</b>
Zona SJ, medidor 4512, control 7
</pre>

````python
#@title Solución
etiqueta = input("Etiqueta del medidor: ")
zona = etiqueta[0:2]
numero = int(etiqueta[3:9])
digito_control = etiqueta[-1]
print(f"Zona {zona}, medidor {numero}, control {digito_control}")
````

**Explicación**

El *traceback* se lee **de abajo hacia arriba**: la última línea dice **qué** pasó (`IndexError`: índice fuera de rango) y las anteriores dicen **dónde** (línea 4). La etiqueta tiene 11 caracteres, así que sus índices válidos van del **0 al 10**: el carácter número 11 está en el índice 10, porque los índices **empiezan en 0**. Se podría corregir con `etiqueta[10]`, pero `etiqueta[-1]` es mejor: toma el último carácter sin depender de la longitud de la etiqueta.

## Ejercicio 4.8 — Postes de una línea eléctrica

Una compañía eléctrica calcula cuántos postes necesita para una línea recta con postes en **ambos extremos**:

```python
longitud_m = int(input("Longitud de la línea (m): "))
separacion_m = int(input("Separación entre postes (m): "))
postes = longitud_m // separacion_m
print(f"Postes necesarios: {postes}")
```

1. Haga una **prueba de escritorio** con una línea de 100 m y postes cada 25 m, dibujando los postes. ¿Cuántos se necesitan realmente?
2. Corrija el error.
3. Agregue dos `assert` con casos de prueba: uno normal y uno **de borde** (una línea de longitud igual a la separación).

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Longitud de la línea (m): <b>100</b>
Separación entre postes (m): <b>25</b>
Postes necesarios: 5
</pre>

````python
#@title Solución
longitud_m = int(input("Longitud de la línea (m): "))
separacion_m = int(input("Separación entre postes (m): "))

postes = longitud_m // separacion_m + 1

print(f"Postes necesarios: {postes}")

# Casos de prueba
assert 100 // 25 + 1 == 5   # normal: postes en 0, 25, 50, 75 y 100 m
assert 40 // 40 + 1 == 2    # borde: un solo tramo necesita dos postes
````

**Explicación**

Con 100 m y separación de 25 m hay postes en 0, 25, 50, 75 y 100 m: **5 postes**, pero el programa reporta 4. La división cuenta los **tramos**, no los postes; siempre hay un poste más que tramos. Este tipo de error, en el que el resultado difiere en uno, se conoce como error *off-by-one* y es muy común.

## Ejercicio 4.9 — Promedio de galgas extensométricas

Tres galgas extensométricas miden la deformación de una viga (en microdeformaciones, με). Un programa calcula el promedio así:

```python
galga_1 = int(input("Galga 1 (με): "))
galga_2 = int(input("Galga 2 (με): "))
galga_3 = int(input("Galga 3 (με): "))
promedio = galga_1 + galga_2 + galga_3 // 3
print(f"Deformación promedio: {promedio} με")
```

Con las lecturas `412`, `398` y `405` el programa muestra `945 με`.

1. Identifique los **dos** errores lógicos.
2. Corrija el programa para que muestre el promedio con 2 decimales.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Galga 1 (με): <b>412</b>
Galga 2 (με): <b>398</b>
Galga 3 (με): <b>405</b>
Deformación promedio: 405.00 με
</pre>

**Explicación**

Una prueba sencilla que revela ambos errores: tres lecturas **iguales** (por ejemplo, 100, 100, 100) deben dar exactamente ese mismo valor como promedio.

## Ejercicio 4.11 — Tolerancia de un eje

Un torno CNC mecaniza un eje en **tres pasadas**. El control de calidad suma los diámetros aportados por cada pasada y verifica si el eje cumple con el diámetro nominal. El programa actual es:

```python
pasada_1 = float(input("Pasada 1 (mm): "))
pasada_2 = float(input("Pasada 2 (mm): "))
pasada_3 = float(input("Pasada 3 (mm): "))
diametro_nominal = float(input("Diámetro nominal (mm): "))

diametro = pasada_1 + pasada_2 + pasada_3
cumple = diametro == diametro_nominal
print(f"Diámetro: {diametro} mm, ¿cumple?: {cumple}")
```

Con las pasadas `10.0`, `0.1` y `0.2` y un diámetro nominal de `10.3`, el programa responde `¿cumple?: False`.

A. Explique por qué, si matemáticamente $10.0 + 0.1 + 0.2 = 10.3$.

B. Corrija la verificación: solicite además la **tolerancia** (mm) y considere que el eje cumple si la **diferencia absoluta** con el nominal no supera la tolerancia (use `abs()`).

## Ejercicio 4.14 — Caja negra: potencia disipada

Un equipo de electrónica le entrega una función que calcula la potencia disipada por una resistencia ($P = V^2 / R$, en watts) y le pide **certificarla** antes de usarla en el diseño de una fuente de poder. Trátela como una **caja negra**:

```python
def calcular_potencia(voltaje: float, resistencia: float) -> float:
    return voltaje ** 2 / resistencia
```

Escriba un programa en Python que pruebe la función con los siguientes **5 casos**, imprimiendo para cada uno el valor **esperado**, el **obtenido** y si la prueba **pasa**, y que además verifique cada caso con `assert`:

| Tipo de caso | Voltaje (V) | Resistencia (Ω) | Esperado (W) |
| :-- | :-: | :-: | :-: |
| Normal | 10 | 5 | 20 |
| Normal | 12 | 144 | 1 |
| Borde: sin voltaje | 0 | 100 | 0 |
| Voltaje negativo | -10 | 5 | 20 |

Para el caso con decimales, compare con una **tolerancia** de $10^{-9}$ usando `abs()`.

**Ejemplo de ejecución**

<pre>
Normal (10 V, 5 Ω): esperado 20, obtenido 20.0 -&gt; True
Normal (12 V, 144 Ω): esperado 1, obtenido 1.0 -&gt; True

Negativo (-10 V, 5 Ω): esperado 20, obtenido 20.0 -&gt; True
Decimales (1.5 V, 0.3 Ω): esperado 7.5, obtenido 7.5 -&gt; True
✅ La función pasó todas las pruebas
</pre>

**Pregunta de respuesta corta**

¿Qué **debería** ocurrir si se llama `calcular_potencia(5, 0)`? ¿Es un caso de prueba útil? ¿Por qué?

````python
#@title Solución
def calcular_potencia(voltaje: float, resistencia: float) -> float:
    return voltaje ** 2 / resistencia

TOLERANCIA = 1e-9

obtenido = calcular_potencia(10, 5)
print(f"Normal (10 V, 5 Ω): esperado 20, obtenido {obtenido} -> {obtenido == 20}")
assert obtenido == 20

obtenido = calcular_potencia(12, 144)
print(f"Normal (12 V, 144 Ω): esperado 1, obtenido {obtenido} -> {obtenido == 1}")
assert obtenido == 1

obtenido = calcular_potencia(0, 100)
print(f"Borde (0 V, 100 Ω): esperado 0, obtenido {obtenido} -> {obtenido == 0}")
assert obtenido == 0

obtenido = calcular_potencia(-10, 5)
print(f"Negativo (-10 V, 5 Ω): esperado 20, obtenido {obtenido} -> {obtenido == 20}")
assert obtenido == 20

obtenido = calcular_potencia(1.5, 0.3)
pasa = abs(obtenido - 7.5) < TOLERANCIA
print(f"Decimales (1.5 V, 0.3 Ω): esperado 7.5, obtenido {obtenido} -> {pasa}")
assert pasa

print("✅ La función pasó todas las pruebas")
````

**Explicación**

En las pruebas de caja negra no importa **cómo** está programada la función, sino si su salida coincide con la **esperada**. Los valores "redondos" permiten calcular la respuesta a mano, y los casos de borde (0 V, voltaje negativo) son los que más errores revelan. En el caso con decimales se usa una tolerancia porque las operaciones con `float` pueden acumular errores de redondeo muy pequeños: aunque aquí se obtenga `7.5`, una prueba con `==` sobre flotantes calculados es frágil.

**Respuesta corta:** debe producirse un **`ZeroDivisionError`**. Sí es útil: es un **caso de error** que documenta cómo se comporta la función con una entrada inválida (una resistencia de 0 Ω sería un cortocircuito).

````python
#@title Solución
NEWTONS_POR_LBF = 4.44822

impulso_a = float(input("Impulso del propulsor A (N·s): "))
impulso_b_lbf = float(input("Impulso del propulsor B (lbf·s): "))

impulso_b = impulso_b_lbf * NEWTONS_POR_LBF
impulso_total = impulso_a + impulso_b
impulso_original = impulso_a + impulso_b_lbf
error_porcentual = abs(impulso_original - impulso_total) / impulso_total * 100

print(f"Impulso total: {impulso_total:.2f} N·s")
print(f"Error del programa original: {error_porcentual:.2f}%")
````

**Explicación**

Es un **error lógico**: para Python ambos valores son simplemente `float`, y no tiene forma de saber que representan unidades distintas. Por esta misma confusión de unidades se perdió en 1999 la sonda *Mars Climate Orbiter* de la NASA. Nombrar las variables con su unidad (`impulso_b_lbf`) ayuda a evitar este tipo de errores.

## Ejercicio 4.19 — Sección de una tubería

Para calcular el caudal de una tubería ($Q = A \cdot v$) se necesita el área de su sección circular, $A = \frac{\pi d^2}{4}$. Un programa lo implementó así:

```python
PI = 3.1416
diametro_m = float(input("Diámetro interno (m): "))
velocidad_ms = float(input("Velocidad del agua (m/s): "))
area = PI * (diametro_m / 2) * 2
caudal = area * velocidad_ms
print(f"Caudal: {caudal:.4f} m³/s")
```

1. Verifique la fórmula con un caso fácil: un diámetro de **2 m** debería dar un área de $\pi$ m². ¿Qué obtiene el programa?
2. Corrija la fórmula y agregue un `assert` con ese caso de prueba.
3. Muestre el caudal también en **litros por segundo**.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Diámetro interno (m): <b>0.15</b>
Velocidad del agua (m/s): <b>1.8</b>
Caudal: 0.0318 m³/s
Caudal: 31.81 L/s
</pre>

````python
#@title Solución
PI = 3.1416
LITROS_POR_M3 = 1000

diametro_m = float(input("Diámetro interno (m): "))
velocidad_ms = float(input("Velocidad del agua (m/s): "))

area = PI * (diametro_m / 2) ** 2
caudal_m3s = area * velocidad_ms
caudal_ls = caudal_m3s * LITROS_POR_M3

print(f"Caudal: {caudal_m3s:.4f} m³/s")
print(f"Caudal: {caudal_ls:.2f} L/s")

# Caso de prueba: un diámetro de 2 m da un área de PI
assert PI * (2 / 2) ** 2 == PI
````

**Explicación**

Con `d = 2`, el programa original calcula `PI * 1 * 2 = 6.2832`, el doble del valor correcto. Se escribió `* 2` (multiplicar por dos) en lugar de `** 2` (elevar al cuadrado): un error de **un solo carácter** que Python no puede detectar. Elegir un caso de prueba cuyo resultado se conoce de antemano es la forma más rápida de revelarlo.

## Ejercicio 4.20 — Rendimiento de una flota

El siguiente programa calcula el **rendimiento** (km/L) de un camión de una flota y el **costo por kilómetro**, pero tiene **tres errores**: uno de ejecución y dos lógicos.

```python
odometro_inicial = int(input("Odómetro inicial (km): "))
odometro_final = int(input("Odómetro final (km): "))
litros = input("Litros de diésel cargados: ")
precio_litro = float(input("Precio por litro (₡): "))

distancia = odometro_inicial - odometro_final
rendimiento = distancia / litros
costo_total = litros + precio_litro
costo_por_km = costo_total / distancia

print(f"Distancia: {distancia} km")
print(f"Rendimiento: {rendimiento:.2f} km/L")
print(f"Costo por km: ₡{costo_por_km:.2f}")
```

1. Encuentre y clasifique los tres errores.
2. Corrija el programa.
3. Verifique el programa corregido con `assert` usando un caso fácil: 100 km recorridos con 25 L a ₡800/L deben dar **4 km/L** y **₡200 por km**.

**Ejemplo de ejecución del programa corregido**

*Lo que aparece en **negrita** corresponde a lo que digita la persona usuaria.*

<pre>
Odómetro inicial (km): <b>152300</b>
Odómetro final (km): <b>152718</b>
Litros de diésel cargados: <b>110.5</b>
Precio por litro (₡): <b>745</b>
Distancia: 418 km
Rendimiento: 3.78 km/L
Costo por km: ₡196.94
</pre>

````python
#@title Solución
odometro_inicial = int(input("Odómetro inicial (km): "))
odometro_final = int(input("Odómetro final (km): "))
litros = float(input("Litros de diésel cargados: "))
precio_litro = float(input("Precio por litro (₡): "))

distancia = odometro_final - odometro_inicial
rendimiento = distancia / litros
costo_total = litros * precio_litro
costo_por_km = costo_total / distancia

print(f"Distancia: {distancia} km")
print(f"Rendimiento: {rendimiento:.2f} km/L")
print(f"Costo por km: ₡{costo_por_km:.2f}")

# Caso de prueba fácil de verificar a mano
assert 100 / 25 == 4
assert 25 * 800 / 100 == 200
````

**Explicación**

1. **Ejecución (`TypeError`):** `litros` nunca se convirtió a número, así que `distancia / litros` intenta dividir un `int` entre un `str`.
2. **Lógico:** la distancia se calcula al revés (`inicial - final`) y queda **negativa**.
3. **Lógico:** el costo total **suma** litros y precio, cuando debe **multiplicarlos**.

Observe que, si se corrige solo el error de ejecución, el programa "funciona" pero reporta rendimientos negativos: los errores lógicos solo se detectan **probando** con casos cuyo resultado se conoce.
