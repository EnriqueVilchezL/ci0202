## Ejercicio 12.1 — Curva de carga de una subestación

El operador de una **subestación eléctrica** necesita visualizar la demanda (MW) de cada hora del día para identificar en qué momentos se supera la **capacidad firme** del transformador, que es de **42 MW**.

```python
DEMANDA_MW = [21, 19, 18, 18, 19, 23, 30, 36, 39, 41, 43, 45,
              44, 42, 41, 40, 42, 46, 48, 45, 39, 33, 27, 23]
```

Escriba un programa en Python que use Matplotlib (interfaz de *Axes*) y NumPy, con las funciones:

A. `horas_sobre_capacidad(demanda: np.ndarray, capacidad: float) -> np.ndarray`: retorna las horas en que la demanda supera la capacidad.

B. `graficar_curva_de_carga(demanda: np.ndarray, capacidad: float) -> None`: dibuja la demanda con marcadores, una línea horizontal **discontinua** roja con la capacidad (`ax.axhline`), y **sombrea** en rojo las zonas donde la demanda supera la capacidad (`ax.fill_between` con el parámetro `where`). Incluya título, etiquetas con unidades, marcas del eje x cada 2 horas, cuadrícula y leyenda.

C. `main() -> None`: imprime las horas sobre capacidad y la energía total del día (MWh), y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Horas sobre la capacidad: [10 11 12 17 18 19]
Energía del día: 822 MWh
</pre>

![Gráfico esperado del ejercicio 12.1](imgs/ejercicios_resueltos/ejercicio_12_1_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

DEMANDA_MW = [21, 19, 18, 18, 19, 23, 30, 36, 39, 41, 43, 45,
              44, 42, 41, 40, 42, 46, 48, 45, 39, 33, 27, 23]
CAPACIDAD_MW = 42


def horas_sobre_capacidad(demanda: np.ndarray, capacidad: float) -> np.ndarray:
    """Retorna las horas en que la demanda supera la capacidad."""
    return np.arange(24)[demanda > capacidad]


def graficar_curva_de_carga(demanda: np.ndarray, capacidad: float) -> None:
    """Dibuja la curva de carga y resalta los periodos de sobrecarga."""
    horas = np.arange(24)
    fig, ax = plt.subplots(figsize=(9, 4.5))
    ax.plot(horas, demanda, marker="o", color="tab:blue", label="Demanda")
    ax.axhline(capacidad, color="red", linestyle="--", label=f"Capacidad firme ({capacidad} MW)")
    ax.fill_between(horas, demanda, capacidad, where=demanda > capacidad,
                    color="red", alpha=0.25, interpolate=True, label="Sobrecarga")
    ax.set_title("Curva de carga diaria de la subestación")
    ax.set_xlabel("Hora del día")
    ax.set_ylabel("Demanda (MW)")
    ax.set_xticks(range(0, 24, 2))
    ax.grid(alpha=0.3)
    ax.legend()


def main() -> None:
    demanda = np.array(DEMANDA_MW)
    print(f"Horas sobre la capacidad: {horas_sobre_capacidad(demanda, CAPACIDAD_MW)}")
    print(f"Energía del día: {demanda.sum()} MWh")
    graficar_curva_de_carga(demanda, CAPACIDAD_MW)
    plt.show()


main()
````

**Explicación**

Con la interfaz de *Axes*, `plt.subplots()` retorna la figura y los ejes, y todo se dibuja y personaliza a través de `ax`. El parámetro `where` de `fill_between` recibe una máscara booleana: solo se rellena donde es `True`, e `interpolate=True` hace que el sombreado empiece y termine exactamente donde las curvas se cruzan.

## Ejercicio 12.2 — Curva esfuerzo-deformación de un acero

Un laboratorio de materiales ensayó a tracción una probeta de acero y obtuvo estos puntos de la curva **esfuerzo-deformación**:

```python
DEFORMACION = [0, 0.0005, 0.001, 0.0015, 0.002, 0.004, 0.01, 0.03, 0.06, 0.09, 0.12, 0.15, 0.17]
ESFUERZO_MPA = [0, 100, 200, 300, 400, 420, 425, 470, 520, 545, 550, 540, 500]
```

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `calcular_propiedades(deformacion: np.ndarray, esfuerzo: np.ndarray) -> tuple[float, float, float]`: retorna el **módulo de elasticidad** (pendiente entre los dos primeros puntos, en GPa), la **resistencia última** (esfuerzo máximo) y la deformación en la que ocurre.

B. `graficar_ensayo(deformacion: np.ndarray, esfuerzo: np.ndarray) -> None`: dibuja la curva con la deformación en **porcentaje**, marca la resistencia última con un punto rojo y una **anotación con flecha** (`ax.annotate`) que indique su valor, y marca con una línea vertical punteada la deformación de **fluencia** (0.2 %). Título, ejes con unidades y cuadrícula.

C. `main() -> None`: imprime las tres propiedades y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Módulo de elasticidad: 200 GPa
Resistencia última: 550 MPa a una deformación de 12%
</pre>

![Gráfico esperado del ejercicio 12.2](imgs/ejercicios_resueltos/ejercicio_12_2_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

DEFORMACION = [0, 0.0005, 0.001, 0.0015, 0.002, 0.004, 0.01, 0.03, 0.06, 0.09, 0.12, 0.15, 0.17]
ESFUERZO_MPA = [0, 100, 200, 300, 400, 420, 425, 470, 520, 545, 550, 540, 500]
DEFORMACION_FLUENCIA = 0.002


def calcular_propiedades(deformacion: np.ndarray, esfuerzo: np.ndarray) -> tuple[float, float, float]:
    """Retorna el módulo de elasticidad (GPa), la resistencia última y su deformación."""
    modulo_gpa = (esfuerzo[1] - esfuerzo[0]) / (deformacion[1] - deformacion[0]) / 1000
    indice = np.argmax(esfuerzo)
    return modulo_gpa, esfuerzo[indice], deformacion[indice]


def graficar_ensayo(deformacion: np.ndarray, esfuerzo: np.ndarray) -> None:
    """Dibuja la curva esfuerzo-deformación con la resistencia última anotada."""
    _, ultima, deformacion_ultima = calcular_propiedades(deformacion, esfuerzo)
    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.plot(deformacion * 100, esfuerzo, color="tab:gray", linewidth=2)
    ax.plot(deformacion_ultima * 100, ultima, "o", color="red")
    ax.annotate(f"Resistencia última: {ultima} MPa", xy=(deformacion_ultima * 100, ultima),
                xytext=(4, 300), arrowprops={"arrowstyle": "->"})
    ax.axvline(DEFORMACION_FLUENCIA * 100, color="tab:blue", linestyle=":", label="Fluencia (0.2 %)")
    ax.set_title("Curva esfuerzo-deformación - acero de refuerzo")
    ax.set_xlabel("Deformación (%)")
    ax.set_ylabel("Esfuerzo (MPa)")
    ax.grid(alpha=0.3)
    ax.legend(loc="lower right")


def main() -> None:
    deformacion = np.array(DEFORMACION)
    esfuerzo = np.array(ESFUERZO_MPA)
    modulo, ultima, deformacion_ultima = calcular_propiedades(deformacion, esfuerzo)
    print(f"Módulo de elasticidad: {modulo:.0f} GPa")
    print(f"Resistencia última: {ultima} MPa a una deformación de {deformacion_ultima:.0%}")
    graficar_ensayo(deformacion, esfuerzo)
    plt.show()


main()
````

**Explicación**

`ax.annotate` recibe el texto, el punto que se señala (`xy`) y la posición del texto (`xytext`); `arrowprops` dibuja la flecha entre ambos. Multiplicar la deformación por 100 **solo** al graficar mantiene los cálculos en unidades adimensionales y muestra el eje en porcentaje, que es como se reporta en los ensayos.

## Ejercicio 12.3 — Generación eléctrica por fuente

El operador del sistema eléctrico publica la **generación mensual** (GWh) por fuente. Se quiere un gráfico de **barras apiladas** que muestre, para cada mes, cuánto aporta cada fuente al total:

```python
MESES = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
GENERACION_GWH = {
    "Hidroeléctrica": [640, 560, 520, 510, 690, 780],
    "Geotérmica": [135, 125, 138, 132, 136, 130],
    "Eólica": [190, 175, 160, 120, 80, 65],
    "Solar": [12, 13, 15, 14, 10, 8],
}
```

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `porcentaje_renovable_variable(generacion: dict[str, list[float]]) -> np.ndarray`: porcentaje del total mensual que aportan las fuentes **eólica y solar**.

B. `graficar_generacion(meses: list[str], generacion: dict[str, list[float]]) -> None`: dibuja las barras apiladas recorriendo el diccionario; cada fuente se apila **sobre** las anteriores usando el parámetro `bottom` (un arreglo que acumula las alturas). Escriba sobre cada barra el **total** del mes con `ax.text`. Incluya título, etiqueta del eje y y leyenda.

C. `main() -> None`: imprime el total del semestre por fuente y el porcentaje eólico + solar de cada mes, y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Hidroeléctrica: 3700 GWh
Geotérmica: 796 GWh
Eólica: 790 GWh
Solar: 72 GWh
Eólica + solar por mes (%): [20.7 21.5 21.  17.3  9.8  7.4]
</pre>

![Gráfico esperado del ejercicio 12.3](imgs/ejercicios_resueltos/ejercicio_12_3_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

MESES = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
GENERACION_GWH = {
    "Hidroeléctrica": [640, 560, 520, 510, 690, 780],
    "Geotérmica": [135, 125, 138, 132, 136, 130],
    "Eólica": [190, 175, 160, 120, 80, 65],
    "Solar": [12, 13, 15, 14, 10, 8],
}
COLORES = {"Hidroeléctrica": "tab:blue", "Geotérmica": "tab:red", "Eólica": "tab:green", "Solar": "gold"}


def porcentaje_renovable_variable(generacion: dict[str, list[float]]) -> np.ndarray:
    """Porcentaje mensual del total que aportan la eólica y la solar."""
    total = np.sum([np.array(valores) for valores in generacion.values()], axis=0)
    variable = np.array(generacion["Eólica"]) + np.array(generacion["Solar"])
    return variable / total * 100


def graficar_generacion(meses: list[str], generacion: dict[str, list[float]]) -> None:
    """Dibuja la generación mensual en barras apiladas por fuente."""
    fig, ax = plt.subplots(figsize=(8, 4.5))
    base = np.zeros(len(meses))
    for fuente, valores in generacion.items():
        ax.bar(meses, valores, bottom=base, label=fuente, color=COLORES[fuente])
        base = base + np.array(valores)
    for posicion, total in enumerate(base):
        ax.text(posicion, total + 10, f"{total:.0f}", ha="center")
    ax.set_title("Generación eléctrica mensual por fuente")
    ax.set_ylabel("Generación (GWh)")
    ax.set_ylim(0, base.max() * 1.12)
    ax.legend(loc="upper left", ncol=2)


def main() -> None:
    for fuente, valores in GENERACION_GWH.items():
        print(f"{fuente}: {sum(valores)} GWh")
    print(f"Eólica + solar por mes (%): {np.round(porcentaje_renovable_variable(GENERACION_GWH), 1)}")
    graficar_generacion(MESES, GENERACION_GWH)
    plt.show()


main()
````

**Explicación**

La variable `base` acumula la altura de las barras ya dibujadas: cada fuente nueva empieza donde terminó la anterior. Al final del ciclo, `base` contiene el **total** de cada mes, que se reutiliza para escribir las etiquetas. El gráfico hace evidente la complementariedad estacional: cuando baja el viento, aumenta la generación hidroeléctrica por las lluvias.

## Ejercicio 12.4 — Intensidad energética de edificios

Una universidad quiere comunicar, en una presentación a sus autoridades, qué edificios tienen una **intensidad energética** (kWh/m² al año) mayor que el promedio del campus.

```python
EDIFICIOS = ["Ingeniería", "Biblioteca", "Ciencias Sociales", "Lab. de Química",
             "Educación", "Centro de Cómputo", "Medicina"]
CONSUMO_KWH = [2150000, 1380000, 890000, 1520000, 610000, 1105000, 1840000]
AREA_M2 = [18500, 9200, 12400, 6100, 8800, 2300, 11600]
```

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `calcular_intensidades(consumo: np.ndarray, area: np.ndarray) -> np.ndarray`

B. `graficar_ranking(nombres: np.ndarray, intensidades: np.ndarray) -> None`: dibuja **barras horizontales** (`ax.barh`) **ordenadas** de menor a mayor (use `np.argsort`), en color rojo las que superan el promedio y en gris las demás; agregue una línea vertical con el promedio y el valor de cada barra al final de la misma. Título y etiqueta del eje x.

C. `main() -> None`: imprime el promedio y los edificios sobre el promedio, y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Intensidad promedio: 185.1 kWh/m²
Sobre el promedio: [&#x27;Lab. de Química&#x27;, &#x27;Centro de Cómputo&#x27;]
</pre>

![Gráfico esperado del ejercicio 12.4](imgs/ejercicios_resueltos/ejercicio_12_4_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

EDIFICIOS = ["Ingeniería", "Biblioteca", "Ciencias Sociales", "Lab. de Química",
             "Educación", "Centro de Cómputo", "Medicina"]
CONSUMO_KWH = [2150000, 1380000, 890000, 1520000, 610000, 1105000, 1840000]
AREA_M2 = [18500, 9200, 12400, 6100, 8800, 2300, 11600]


def calcular_intensidades(consumo: np.ndarray, area: np.ndarray) -> np.ndarray:
    """Intensidad energética de cada edificio (kWh/m² al año)."""
    return consumo / area


def graficar_ranking(nombres: np.ndarray, intensidades: np.ndarray) -> None:
    """Dibuja el ranking de intensidad energética en barras horizontales."""
    orden = np.argsort(intensidades)
    promedio = intensidades.mean()
    colores = np.where(intensidades[orden] > promedio, "tab:red", "tab:gray")

    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.barh(nombres[orden], intensidades[orden], color=colores)
    ax.axvline(promedio, color="black", linestyle="--", label=f"Promedio: {promedio:.0f} kWh/m²")
    for posicion, valor in enumerate(intensidades[orden]):
        ax.text(valor + 5, posicion, f"{valor:.0f}", va="center")
    ax.set_title("Intensidad energética de los edificios del campus")
    ax.set_xlabel("Intensidad energética (kWh/m² al año)")
    ax.set_xlim(0, intensidades.max() * 1.12)
    ax.legend(loc="lower right")


def main() -> None:
    nombres = np.array(EDIFICIOS)
    intensidades = calcular_intensidades(np.array(CONSUMO_KWH), np.array(AREA_M2))
    print(f"Intensidad promedio: {intensidades.mean():.1f} kWh/m²")
    print(f"Sobre el promedio: {nombres[intensidades > intensidades.mean()].tolist()}")
    graficar_ranking(nombres, intensidades)
    plt.show()


main()
````

**Explicación**

Las barras **horizontales** son la mejor opción cuando las categorías tienen nombres largos, y ordenarlas convierte el gráfico en un ranking que se lee de un vistazo. Usar color **solo** para lo que se quiere destacar (los edificios sobre el promedio) dirige la atención del lector hacia el mensaje principal.

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

RELACION_AC = [0.40, 0.42, 0.45, 0.48, 0.50, 0.52, 0.55, 0.58, 0.60, 0.65, 0.70]
RESISTENCIA_MPA = [41.2, 39.5, 37.1, 35.8, 33.2, 32.5, 29.9, 28.4, 26.1, 23.8, 20.9]


def ajustar_recta(x: np.ndarray, y: np.ndarray) -> tuple[float, float]:
    """Pendiente e intercepto de la recta de mínimos cuadrados."""
    pendiente, intercepto = np.polyfit(x, y, 1)
    return float(pendiente), float(intercepto)


def graficar_ajuste(x: np.ndarray, y: np.ndarray, pendiente: float, intercepto: float,
                    objetivo: float) -> None:
    """Dibuja los ensayos, la recta de ajuste y la resistencia objetivo."""
    x_recta = np.linspace(x.min(), x.max(), 50)
    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.scatter(x, y, color="tab:blue", label="Ensayos")
    ax.plot(x_recta, pendiente * x_recta + intercepto, color="tab:orange", label="Ajuste lineal")
    ax.axhline(objetivo, color="green", linestyle="--", label=f"Objetivo: {objetivo} MPa")
    ax.text(0.58, 38, f"f'c = {pendiente:.1f}·(a/c) + {intercepto:.1f}", fontsize=11)
    ax.set_title("Ley de Abrams: resistencia vs. relación agua/cemento")
    ax.set_xlabel("Relación agua/cemento")
    ax.set_ylabel("Resistencia a 28 días (MPa)")
    ax.grid(alpha=0.3)
    ax.legend(loc="lower left")


def main() -> None:
    x = np.array(RELACION_AC)
    y = np.array(RESISTENCIA_MPA)
    objetivo = float(input("Resistencia objetivo (MPa): "))

    pendiente, intercepto = ajustar_recta(x, y)
    relacion_maxima = (objetivo - intercepto) / pendiente
    print(f"Recta de ajuste: f'c = {pendiente:.2f} * (a/c) + {intercepto:.2f}")
    print(f"Relación a/c máxima para {objetivo} MPa: {relacion_maxima:.3f}")

    graficar_ajuste(x, y, pendiente, intercepto, objetivo)
    plt.show()


main()
````

**Explicación**

`np.polyfit(x, y, 1)` ajusta un polinomio de grado 1 (una recta) por mínimos cuadrados y retorna sus coeficientes, empezando por el de mayor grado. Los datos se dibujan como **puntos** (son mediciones) y el modelo como **línea** (es una función continua): distinguirlos visualmente es una convención importante en los gráficos técnicos.

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `simular_diametros(cantidad: int) -> np.ndarray`

B. `graficar_histograma(diametros: np.ndarray, minimo: float, maximo: float) -> None`: dibuja un **histograma** de 30 clases con bordes negros, dos líneas verticales rojas en los límites de tolerancia, una línea verde en el valor nominal y una línea discontinua en la **media** del lote. Título, ejes y leyenda.

C. `main() -> None`: imprime la media, la desviación estándar, el porcentaje de piezas fuera de tolerancia y el **índice de capacidad** $C_{pk} = \frac{\min(LSE - \bar{x}, \bar{x} - LIE)}{3\sigma}$ (un proceso es capaz si $C_{pk} \ge 1.33$), y muestra el gráfico.

**Ejemplo de ejecución**

<pre>

Fuera de tolerancia: 4.25%
Cpk: 0.60 -&gt; proceso NO capaz
</pre>

![Gráfico esperado del ejercicio 12.6](imgs/ejercicios_resueltos/ejercicio_12_6_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

NOMINAL_MM = 10.00
TOLERANCIA_MM = 0.05
CPK_MINIMO = 1.33


def simular_diametros(cantidad: int) -> np.ndarray:
    """Simula los diámetros medidos de un lote de pernos."""
    np.random.seed(5)
    return np.random.normal(10.012, 0.021, cantidad)


def graficar_histograma(diametros: np.ndarray, minimo: float, maximo: float) -> None:
    """Dibuja el histograma de diámetros con los límites de tolerancia."""
    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.hist(diametros, bins=30, color="tab:blue", edgecolor="black", alpha=0.8)
    ax.axvline(minimo, color="red", linewidth=2, label="Límites de tolerancia")
    ax.axvline(maximo, color="red", linewidth=2)
    ax.axvline(NOMINAL_MM, color="green", linewidth=2, label="Nominal")
    ax.axvline(diametros.mean(), color="black", linestyle="--", label="Media del lote")
    ax.set_title("Distribución del diámetro de pernos M10 (n = 400)")
    ax.set_xlabel("Diámetro (mm)")
    ax.set_ylabel("Cantidad de pernos")
    ax.legend()


def main() -> None:
    diametros = simular_diametros(400)
    minimo = NOMINAL_MM - TOLERANCIA_MM
    maximo = NOMINAL_MM + TOLERANCIA_MM

    media = diametros.mean()
    desviacion = diametros.std()
    fuera = np.mean((diametros < minimo) | (diametros > maximo))
    cpk = min(maximo - media, media - minimo) / (3 * desviacion)

    print(f"Media: {media:.4f} mm  Desviación estándar: {desviacion:.4f} mm")
    print(f"Fuera de tolerancia: {fuera:.2%}")
    print(f"Cpk: {cpk:.2f} -> {'proceso capaz' if cpk >= CPK_MINIMO else 'proceso NO capaz'}")

    graficar_histograma(diametros, minimo, maximo)
    plt.show()


main()
````

**Explicación**

El histograma muestra algo que los números por sí solos esconden: el proceso está **descentrado** hacia el límite superior. Por eso el $C_{pk}$ es bajo aunque la dispersión sea pequeña; centrar el proceso en el nominal (ajustando la máquina) mejoraría la capacidad sin necesidad de reducir la variabilidad.

Un ingeniero químico ajustó un **modelo** del rendimiento (%) de una reacción en función de la temperatura $T$ (°C) y la presión $P$ (bar), a partir de experimentos:

válido para $120 \le T \le 180$ °C y $4 \le P \le 16$ bar.

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `rendimiento(temperatura: np.ndarray, presion: np.ndarray) -> np.ndarray`

B. `graficar_superficie(t_malla: np.ndarray, p_malla: np.ndarray, y_malla: np.ndarray) -> None`: crea los ejes 3D con `plt.subplots(subplot_kw={"projection": "3d"})`, dibuja la superficie con `ax.plot_surface` y el mapa de colores `"viridis"`, agrega una barra de color y rotula los tres ejes.

C. `main() -> None`: genera una malla de 60 × 60 puntos con `np.linspace` y `np.meshgrid`, evalúa el modelo, imprime las **condiciones óptimas** (temperatura, presión y rendimiento máximo de la malla) y muestra el gráfico.

![Gráfico esperado del ejercicio 12.8](imgs/ejercicios_resueltos/ejercicio_12_8_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np


def rendimiento(temperatura: np.ndarray, presion: np.ndarray) -> np.ndarray:
    """Modelo del rendimiento (%) de la reacción."""
    dt = temperatura - 150
    dp = presion - 10
    return 72 + 0.5 * dt + 2.8 * dp - 0.012 * dt ** 2 - 0.35 * dp ** 2 + 0.02 * dt * dp


def graficar_superficie(t_malla: np.ndarray, p_malla: np.ndarray, y_malla: np.ndarray) -> None:
    """Dibuja la superficie de respuesta del rendimiento."""
    fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={"projection": "3d"})
    superficie = ax.plot_surface(t_malla, p_malla, y_malla, cmap="viridis")
    fig.colorbar(superficie, ax=ax, shrink=0.6, pad=0.12, label="Rendimiento (%)")
    ax.set_title("Superficie de respuesta del reactor")
    ax.set_xlabel("Temperatura (°C)")
    ax.set_ylabel("Presión (bar)")
    ax.set_zlabel("Rendimiento (%)")


def main() -> None:
    temperaturas = np.linspace(120, 180, 60)
    presiones = np.linspace(4, 16, 60)
    t_malla, p_malla = np.meshgrid(temperaturas, presiones)
    y_malla = rendimiento(t_malla, p_malla)

    fila, columna = np.unravel_index(np.argmax(y_malla), y_malla.shape)
    print(f"Condiciones óptimas: T = {t_malla[fila, columna]:.1f} °C, P = {p_malla[fila, columna]:.1f} bar")
    print(f"Rendimiento máximo: {y_malla.max():.2f} %")

    graficar_superficie(t_malla, p_malla, y_malla)
    plt.show()


main()
````

**Explicación**

`np.meshgrid` combina los dos vectores en dos **matrices** con todas las parejas (T, P) posibles, de modo que el modelo se evalúa en toda la malla con una sola operación vectorizada. Buscar el máximo en la malla da una buena aproximación del óptimo; con una malla más fina, el resultado sería más preciso.

B. `graficar_vibracion(tiempo: np.ndarray, senal: np.ndarray) -> None`: crea una figura con **dos** gráficos lado a lado (`fig, (ax1, ax2) = plt.subplots(1, 2)`): a la izquierda, los primeros **0.2 s** de la señal; a la derecha, el **histograma** de toda la señal. Cada gráfico con su propio título y etiquetas, y un título general con `fig.suptitle`.

</pre>

![Gráfico esperado del ejercicio 12.9](imgs/ejercicios_resueltos/ejercicio_12_9_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

FRECUENCIA_GIRO_HZ = 30
RMS_ALERTA = 1.8


def generar_vibracion() -> tuple[np.ndarray, np.ndarray]:
    """Genera 1 s de la señal de vibración del motor."""
    np.random.seed(1)
    tiempo = np.linspace(0, 1, 2000, endpoint=False)
    senal = (2.0 * np.sin(2 * np.pi * FRECUENCIA_GIRO_HZ * tiempo)
             + 0.8 * np.sin(2 * np.pi * 2 * FRECUENCIA_GIRO_HZ * tiempo)
             + np.random.normal(0, 0.3, tiempo.size))
    return tiempo, senal


def graficar_vibracion(tiempo: np.ndarray, senal: np.ndarray) -> None:
    """Dibuja la señal en el tiempo y su histograma."""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))
    primeros = tiempo <= 0.2
    ax1.plot(tiempo[primeros] * 1000, senal[primeros], linewidth=1)
    ax1.set_title("Señal (primeros 200 ms)")
    ax1.set_xlabel("Tiempo (ms)")
    ax1.set_ylabel("Velocidad de vibración (mm/s)")
    ax1.grid(alpha=0.3)

    ax2.hist(senal, bins=40, color="tab:purple", edgecolor="white")
    ax2.set_title("Distribución de amplitudes")
    ax2.set_xlabel("Velocidad de vibración (mm/s)")
    ax2.set_ylabel("Cantidad de muestras")
    fig.suptitle("Monitoreo de vibración del motor M-07")
    fig.tight_layout()


def main() -> None:
    tiempo, senal = generar_vibracion()
    pico = np.abs(senal).max()
    rms = np.sqrt(np.mean(senal ** 2))

    print(f"Pico: {pico:.2f} mm/s  RMS: {rms:.2f} mm/s  Factor de cresta: {pico / rms:.2f}")
    if rms > RMS_ALERTA:
        print("Advertencia: vibración en zona de alerta, programe una inspección.")

    graficar_vibracion(tiempo, senal)
    plt.show()


main()
````

**Explicación**

`plt.subplots(1, 2)` crea una figura con una fila y dos columnas de ejes, que se desempaquetan en `ax1` y `ax2`; cada uno se personaliza por separado. `fig.tight_layout()` ajusta los espacios para que los títulos y etiquetas no se superpongan. Ver la señal y su distribución **juntas** ayuda a interpretar el factor de cresta.

## Ejercicio 12.10 — Punto de operación de una bomba

Para elegir la **velocidad** de una bomba centrífuga, un ingeniero hidráulico grafica sus **curvas características** (altura $H$ vs. caudal $Q$) junto con la **curva del sistema** de tuberías. El punto donde se cruzan es el **punto de operación**.

* Curva de la bomba a la velocidad nominal (1750 rpm): $H = 48 - 0.012 Q^2$ (m, con $Q$ en L/s).
* A otra velocidad $n$, por las **leyes de afinidad**: $H_n(Q) = 48 r^2 - 0.012 Q^2$, con $r = n / 1750$.
* Curva del sistema: $H_s = 15 + 0.006 Q^2$ (15 m de altura estática más las pérdidas).

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `altura_bomba(caudal: np.ndarray, velocidad: float) -> np.ndarray` y `altura_sistema(caudal: np.ndarray) -> np.ndarray`.

B. `punto_de_operacion(velocidad: float) -> tuple[float, float]`: igualando ambas curvas se obtiene $Q = \sqrt{\frac{48 r^2 - 15}{0.018}}$; retorna $Q$ y $H$.

C. `graficar_curvas(velocidades: list[float]) -> None`: dibuja una curva de la bomba por cada velocidad (con su etiqueta en la leyenda), la curva del sistema en negro y más gruesa, y marca cada punto de operación. Limite el eje y a valores positivos.

D. `main() -> None`: imprime el punto de operación y la **potencia hidráulica** $P = \rho g Q H$ (kW) para 1450, 1600 y 1750 rpm, y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
1450 rpm: Q = 31.6 L/s, H = 21.0 m, P = 6.50 kW
1600 rpm: Q = 37.4 L/s, H = 23.4 m, P = 8.57 kW
1750 rpm: Q = 42.8 L/s, H = 26.0 m, P = 10.92 kW
</pre>

![Gráfico esperado del ejercicio 12.10](imgs/ejercicios_resueltos/ejercicio_12_10_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

VELOCIDAD_NOMINAL = 1750
DENSIDAD = 1000
GRAVEDAD = 9.81


def altura_bomba(caudal: np.ndarray, velocidad: float) -> np.ndarray:
    """Altura (m) que entrega la bomba a una velocidad dada (leyes de afinidad)."""
    r = velocidad / VELOCIDAD_NOMINAL
    return 48 * r ** 2 - 0.012 * caudal ** 2


def altura_sistema(caudal: np.ndarray) -> np.ndarray:
    """Altura (m) que requiere el sistema de tuberías."""
    return 15 + 0.006 * caudal ** 2


def punto_de_operacion(velocidad: float) -> tuple[float, float]:
    """Caudal (L/s) y altura (m) donde se cruzan la curva de la bomba y la del sistema."""
    r = velocidad / VELOCIDAD_NOMINAL
    caudal = np.sqrt((48 * r ** 2 - 15) / 0.018)
    return float(caudal), float(altura_sistema(caudal))


def graficar_curvas(velocidades: list[float]) -> None:
    """Dibuja las curvas de la bomba, la del sistema y los puntos de operación."""
    caudales = np.linspace(0, 65, 200)
    fig, ax = plt.subplots(figsize=(8, 5))
    for velocidad in velocidades:
        ax.plot(caudales, altura_bomba(caudales, velocidad), label=f"Bomba a {velocidad} rpm")
        caudal, altura = punto_de_operacion(velocidad)
        ax.plot(caudal, altura, "o", color="red")
    ax.plot(caudales, altura_sistema(caudales), color="black", linewidth=2.5, label="Sistema")
    ax.set_ylim(0, 55)
    ax.set_title("Curvas de la bomba y del sistema")
    ax.set_xlabel("Caudal (L/s)")
    ax.set_ylabel("Altura (m)")
    ax.grid(alpha=0.3)
    ax.legend()


def main() -> None:
    velocidades = [1450, 1600, 1750]
    for velocidad in velocidades:
        caudal, altura = punto_de_operacion(velocidad)
        potencia_kw = DENSIDAD * GRAVEDAD * caudal / 1000 * altura / 1000
        print(f"{velocidad} rpm: Q = {caudal:.1f} L/s, H = {altura:.1f} m, P = {potencia_kw:.2f} kW")
    graficar_curvas(velocidades)
    plt.show()


main()
````

**Explicación**

Cada llamada a `ax.plot` agrega una curva nueva sobre los mismos ejes, y el parámetro `label` hace que aparezca en la leyenda. El gráfico muestra de inmediato un resultado de ingeniería: reducir la velocidad baja el caudal y, sobre todo, la potencia, porque la potencia disminuye mucho más rápido que el caudal.

## Ejercicio 12.13 — Diagrama de Pareto de fallas

El **principio de Pareto** dice que, a menudo, unas pocas causas explican la mayoría de los problemas. Un ingeniero de calidad registró las causas de los paros de una línea de ensamble durante un mes (en minutos de paro):

```python
CAUSAS = {"Falta de material": 185, "Cambio de molde": 420, "Falla eléctrica": 95,
          "Atasco en banda": 610, "Calibración": 70, "Ausencia de operario": 45}
```

Un **diagrama de Pareto** muestra las causas en barras ordenadas de mayor a menor y, sobre ellas, una línea con el **porcentaje acumulado**.

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `ordenar_causas(causas: dict[str, float]) -> tuple[list[str], np.ndarray, np.ndarray]`: retorna los nombres y los minutos ordenados de mayor a menor, y el porcentaje acumulado (`np.cumsum`).

B. `graficar_pareto(nombres: list[str], minutos: np.ndarray, acumulado: np.ndarray) -> None`: dibuja las barras; cree un **segundo eje y** a la derecha con `ax2 = ax.twinx()` para la línea del porcentaje acumulado (con marcadores, de 0 a 100 %) y dibuje en él una línea horizontal en el 80 %. Rote las etiquetas del eje x para que se lean.

C. `main() -> None`: imprime la tabla con cada causa, sus minutos y el porcentaje acumulado, y las causas que forman el **80 %** de los paros ("las pocas vitales"); luego muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Causa                    Min   Acum.
Atasco en banda          610   42.8%

Calibración               70   96.8%

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

CAUSAS = {"Falta de material": 185, "Cambio de molde": 420, "Falla eléctrica": 95,
          "Atasco en banda": 610, "Calibración": 70, "Ausencia de operario": 45}


def ordenar_causas(causas: dict[str, float]) -> tuple[list[str], np.ndarray, np.ndarray]:
    """Ordena las causas de mayor a menor y calcula el porcentaje acumulado."""
    nombres = sorted(causas, key=causas.get, reverse=True)
    minutos = np.array([causas[nombre] for nombre in nombres])
    acumulado = np.cumsum(minutos) / minutos.sum() * 100
    return nombres, minutos, acumulado


def graficar_pareto(nombres: list[str], minutos: np.ndarray, acumulado: np.ndarray) -> None:
    """Dibuja el diagrama de Pareto de las causas de paro."""
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.bar(nombres, minutos, color="tab:blue")
    ax.set_ylabel("Minutos de paro")
    ax.tick_params(axis="x", rotation=30)

    ax2 = ax.twinx()
    ax2.plot(nombres, acumulado, color="tab:red", marker="o")
    ax2.axhline(80, color="gray", linestyle="--")
    ax2.set_ylim(0, 105)
    ax2.set_ylabel("Porcentaje acumulado (%)")
    ax.set_title("Diagrama de Pareto de los paros de la línea")
    fig.tight_layout()


def main() -> None:
    nombres, minutos, acumulado = ordenar_causas(CAUSAS)
    print(f"{'Causa':<22}{'Min':>6}{'Acum.':>8}")
    for nombre, minuto, porcentaje in zip(nombres, minutos, acumulado):
        print(f"{nombre:<22}{minuto:6d}{porcentaje:7.1f}%")

    vitales = [nombre for nombre, porcentaje in zip(nombres, acumulado) if porcentaje <= 80]
    print(f"\nPocas vitales (80 %): {vitales}")
    graficar_pareto(nombres, minutos, acumulado)
    plt.show()


main()
````

**Explicación**

`ax.twinx()` crea un segundo eje y que **comparte** el eje x con el primero: así se pueden dibujar en el mismo gráfico magnitudes con unidades distintas (minutos y porcentaje). `sorted(causas, key=causas.get, reverse=True)` ordena las **claves** del diccionario según sus valores. En el ejemplo, solo dos causas (atasco y cambio de molde) explican más de dos tercios del tiempo perdido: ahí debe concentrarse el esfuerzo de mejora.

## Ejercicio 12.14 — Gráfico de control de un proceso de llenado

Una embotelladora controla el **volumen de llenado** (mL) de sus botellas con una **carta de control**: se grafica cada medición en orden y se comparan con la línea central (la media de referencia) y los límites de control $\bar{x} \pm 3\sigma$, calculados con las **primeras 20** mediciones del proceso estable.

Las mediciones se simulan: 30 mediciones con media 500 mL y desviación 0.8 mL (semilla 12) a las que, para representar un problema real, se les suma un **desplazamiento** de +4.5 mL a partir de la medición 26.

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `simular_mediciones() -> np.ndarray` y `calcular_limites(referencia: np.ndarray) -> tuple[float, float, float]`.

B. `graficar_carta(mediciones: np.ndarray, media: float, lic: float, lsc: float) -> None`: dibuja las mediciones unidas por líneas, la línea central en verde, los límites en rojo discontinuo con su valor escrito al lado (`ax.text`), y los puntos **fuera de control** como círculos rojos grandes.

C. `main() -> None`: imprime los límites y los números de las mediciones fuera de control, y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Línea central: 500.03 mL  LIC: 497.28 mL  LSC: 502.78 mL
Mediciones fuera de control: [27 28 30]
</pre>

![Gráfico esperado del ejercicio 12.14](imgs/ejercicios_resueltos/ejercicio_12_14_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

MEDICIONES_REFERENCIA = 20


def simular_mediciones() -> np.ndarray:
    """Simula el volumen de llenado de 30 botellas, con un desplazamiento al final."""
    np.random.seed(12)
    mediciones = np.random.normal(500, 0.8, 30)
    mediciones[25:] += 4.5
    return mediciones


def calcular_limites(referencia: np.ndarray) -> tuple[float, float, float]:
    """Retorna la línea central y los límites de control inferior y superior."""
    media = referencia.mean()
    desviacion = referencia.std(ddof=1)
    return media, media - 3 * desviacion, media + 3 * desviacion


def graficar_carta(mediciones: np.ndarray, media: float, lic: float, lsc: float) -> None:
    """Dibuja la carta de control y resalta los puntos fuera de control."""
    numeros = np.arange(1, len(mediciones) + 1)
    fuera = (mediciones < lic) | (mediciones > lsc)

    fig, ax = plt.subplots(figsize=(9, 4.5))
    ax.plot(numeros, mediciones, marker="o", color="tab:blue", label="Mediciones")
    ax.axhline(media, color="green", label="Línea central")
    ax.axhline(lsc, color="red", linestyle="--", label="Límites de control")
    ax.axhline(lic, color="red", linestyle="--")
    ax.text(len(mediciones) + 0.5, lsc, f"LSC = {lsc:.2f}", va="center")
    ax.text(len(mediciones) + 0.5, lic, f"LIC = {lic:.2f}", va="center")
    ax.scatter(numeros[fuera], mediciones[fuera], s=150, facecolors="none", edgecolors="red",
               linewidths=2, label="Fuera de control", zorder=3)
    ax.set_title("Carta de control del volumen de llenado")
    ax.set_xlabel("Número de medición")
    ax.set_ylabel("Volumen (mL)")
    ax.set_xlim(0, len(mediciones) + 5)
    ax.legend(loc="upper left")


def main() -> None:
    mediciones = simular_mediciones()
    media, lic, lsc = calcular_limites(mediciones[:MEDICIONES_REFERENCIA])
    fuera = np.arange(1, len(mediciones) + 1)[(mediciones < lic) | (mediciones > lsc)]

    print(f"Línea central: {media:.2f} mL  LIC: {lic:.2f} mL  LSC: {lsc:.2f} mL")
    print(f"Mediciones fuera de control: {fuera}")
    graficar_carta(mediciones, media, lic, lsc)
    plt.show()


main()
````

**Explicación**

Los puntos fuera de control se dibujan **encima** de la serie con círculos vacíos (`facecolors="none"`) y `zorder=3`, para que resalten sin ocultar el punto original. Los límites se calculan **solo** con la parte estable del proceso (`mediciones[:20]`): si se incluyeran las mediciones desplazadas, los límites se ensancharían y el problema podría pasar desapercibido.

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

NOMINAL_G = 1000
MINIMO_LEGAL_G = 990
BOLSAS_POR_ANIO = 2_000_000


def simular_pesos() -> tuple[np.ndarray, np.ndarray]:
    """Simula el peso de 500 bolsas de cada máquina."""
    np.random.seed(21)
    return np.random.normal(1004.0, 6.0, 500), np.random.normal(1001.0, 2.5, 500)


def graficar_comparacion(pesos_a: np.ndarray, pesos_b: np.ndarray, minimo: float) -> None:
    """Compara las distribuciones de peso de ambas máquinas."""
    intervalos = np.arange(980, 1026, 1.5)
    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.hist(pesos_a, bins=intervalos, color="tab:orange", alpha=0.6, label="Máquina A (actual)")
    ax.hist(pesos_b, bins=intervalos, color="tab:blue", alpha=0.6, label="Máquina B (nueva)")
    ax.axvline(minimo, color="red", linewidth=2, label=f"Mínimo legal ({minimo} g)")
    ax.axvline(NOMINAL_G, color="black", linestyle=":", label=f"Nominal ({NOMINAL_G} g)")
    ax.set_title("Peso de las bolsas por máquina llenadora (500 bolsas c/u)")
    ax.set_xlabel("Peso (g)")
    ax.set_ylabel("Cantidad de bolsas")
    ax.legend()


def main() -> None:
    pesos_a, pesos_b = simular_pesos()
    for nombre, pesos in [("A", pesos_a), ("B", pesos_b)]:
        bajo_minimo = np.mean(pesos < MINIMO_LEGAL_G)
        print(f"Máquina {nombre}: media {pesos.mean():.1f} g, σ {pesos.std():.1f} g, "
              f"bajo el mínimo {bajo_minimo:.1%}, regalado {pesos.mean() - NOMINAL_G:.1f} g/bolsa")

    ahorro_kg = (pesos_a.mean() - pesos_b.mean()) * BOLSAS_POR_ANIO / 1000
    print(f"\nAhorro anual con la máquina B: {ahorro_kg:,.0f} kg de producto")
    graficar_comparacion(pesos_a, pesos_b, MINIMO_LEGAL_G)
    plt.show()


main()
````

**Explicación**

Para comparar histogramas, **ambos** deben usar los mismos intervalos: si Matplotlib los eligiera por separado, las barras tendrían anchos distintos y la comparación sería engañosa. La máquina A debe llenar "de más" (su media es 1004 g) para que su gran dispersión no produzca bolsas bajo el mínimo legal; reducir la dispersión permite bajar la media y regalar menos producto.

B. `graficar_enfriamiento(constantes: dict[str, float]) -> None`: crea una figura con **dos** gráficos lado a lado con las mismas curvas: a la izquierda con el eje y **lineal** y a la derecha con el eje y en escala **logarítmica** (`ax.set_yscale("log")`). Dibuje en ambos una línea horizontal en 5 °C (la diferencia segura para manipular el equipo).

C. `main() -> None`: imprime, para cada diseño, el tiempo necesario para llegar a una diferencia de 5 °C ($t = \tau \ln(\Delta T_0 / 5)$), y muestra la figura.

**Ejemplo de ejecución**

<pre>
Radiadores normales: 8.21 h para llegar a 5 °C de diferencia
Ventilación forzada: 3.59 h para llegar a 5 °C de diferencia
</pre>

![Gráfico esperado del ejercicio 12.16](imgs/ejercicios_resueltos/ejercicio_12_16_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

DIFERENCIA_INICIAL = 65
DIFERENCIA_SEGURA = 5
CONSTANTES_H = {"Radiadores normales": 3.2, "Ventilación forzada": 1.4}


def diferencia_temperatura(tiempo: np.ndarray, constante: float) -> np.ndarray:
    """Diferencia de temperatura aceite-ambiente (°C) en el tiempo (h)."""
    return DIFERENCIA_INICIAL * np.exp(-tiempo / constante)


def graficar_enfriamiento(constantes: dict[str, float]) -> None:
    """Dibuja las curvas de enfriamiento en escala lineal y logarítmica."""
    tiempo = np.linspace(0, 12, 200)
    fig, (ax_lineal, ax_log) = plt.subplots(1, 2, figsize=(11, 4))
    for ax in (ax_lineal, ax_log):
        for nombre, constante in constantes.items():
            ax.plot(tiempo, diferencia_temperatura(tiempo, constante), label=f"{nombre} (τ = {constante} h)")
        ax.axhline(DIFERENCIA_SEGURA, color="red", linestyle="--", label="Diferencia segura")
        ax.set_xlabel("Tiempo desde la desconexión (h)")
        ax.set_ylabel("ΔT aceite-ambiente (°C)")
        ax.grid(alpha=0.3, which="both")
    ax_log.set_yscale("log")
    ax_lineal.set_title("Escala lineal")
    ax_log.set_title("Escala logarítmica")
    ax_lineal.legend()
    fig.suptitle("Enfriamiento del transformador después de la desconexión")
    fig.tight_layout()


def main() -> None:
    for nombre, constante in CONSTANTES_H.items():
        tiempo = constante * np.log(DIFERENCIA_INICIAL / DIFERENCIA_SEGURA)
        print(f"{nombre}: {tiempo:.2f} h para llegar a {DIFERENCIA_SEGURA} °C de diferencia")
    graficar_enfriamiento(CONSTANTES_H)
    plt.show()


main()
````

**Explicación**

En escala logarítmica, una curva **exponencial** se ve como una **recta**, y su pendiente depende de la constante de tiempo: así es muy fácil comparar los dos diseños y leer valores pequeños que en escala lineal quedan "aplastados" contra el eje. Esta es la razón por la que muchos gráficos de ingeniería (decaimiento, respuesta en frecuencia, fatiga) usan escalas logarítmicas.

## Ejercicio 12.17 — Curvas de nivel de un terreno

Un topógrafo modela la elevación (m) de un terreno de 500 m × 400 m donde se construirá una **urbanización**, con una colina y una hondonada:

$$z(x, y) = 1200 + 35 e^{-\frac{(x - 150)^2 + (y - 250)^2}{2 \cdot 80^2}} - 18 e^{-\frac{(x - 380)^2 + (y - 120)^2}{2 \cdot 60^2}} + 0.04x$$

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `elevacion(x: np.ndarray, y: np.ndarray) -> np.ndarray`

B. `graficar_curvas_de_nivel(x_malla: np.ndarray, y_malla: np.ndarray, z_malla: np.ndarray) -> None`: dibuja un mapa de colores relleno con `ax.contourf` (mapa `"terrain"`, 20 niveles) y, encima, **curvas de nivel** negras cada 5 m con `ax.contour`, rotuladas con `ax.clabel`. Marque con un triángulo el punto **más alto** y con una "x" el **más bajo**. Barra de color, ejes con unidades y `ax.set_aspect("equal")` para no deformar el terreno.

C. `main() -> None`: evalúa la elevación en una malla de 100 × 80 puntos, imprime las elevaciones y posiciones máxima y mínima y el **desnivel** total, y muestra el gráfico.

**Ejemplo de ejecución**

<pre>
Punto más alto: 1241.1 m en (157, 248)
Punto más bajo: 1197.2 m en (374, 122)
Desnivel total: 43.9 m
</pre>

![Gráfico esperado del ejercicio 12.17](imgs/ejercicios_resueltos/ejercicio_12_17_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np


def elevacion(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    """Modelo de la elevación del terreno (m)."""
    colina = 35 * np.exp(-((x - 150) ** 2 + (y - 250) ** 2) / (2 * 80 ** 2))
    hondonada = 18 * np.exp(-((x - 380) ** 2 + (y - 120) ** 2) / (2 * 60 ** 2))
    return 1200 + colina - hondonada + 0.04 * x


def graficar_curvas_de_nivel(x_malla: np.ndarray, y_malla: np.ndarray, z_malla: np.ndarray) -> None:
    """Dibuja el mapa de elevaciones con curvas de nivel."""
    fila_max, col_max = np.unravel_index(np.argmax(z_malla), z_malla.shape)
    fila_min, col_min = np.unravel_index(np.argmin(z_malla), z_malla.shape)

    fig, ax = plt.subplots(figsize=(8, 6))
    relleno = ax.contourf(x_malla, y_malla, z_malla, levels=20, cmap="terrain")
    curvas = ax.contour(x_malla, y_malla, z_malla, levels=np.arange(1185, 1250, 5), colors="black",
                        linewidths=0.6)
    ax.clabel(curvas, fontsize=8)
    fig.colorbar(relleno, ax=ax, label="Elevación (m)")
    ax.plot(x_malla[fila_max, col_max], y_malla[fila_max, col_max], "^", color="red", markersize=12,
            label="Punto más alto")
    ax.plot(x_malla[fila_min, col_min], y_malla[fila_min, col_min], "x", color="blue", markersize=12,
            markeredgewidth=3, label="Punto más bajo")
    ax.set_aspect("equal")
    ax.set_title("Curvas de nivel del terreno")
    ax.set_xlabel("Este (m)")
    ax.set_ylabel("Norte (m)")
    ax.legend(loc="lower left")


def main() -> None:
    x_malla, y_malla = np.meshgrid(np.linspace(0, 500, 100), np.linspace(0, 400, 80))
    z_malla = elevacion(x_malla, y_malla)

    fila_max, col_max = np.unravel_index(np.argmax(z_malla), z_malla.shape)
    fila_min, col_min = np.unravel_index(np.argmin(z_malla), z_malla.shape)
    print(f"Punto más alto: {z_malla.max():.1f} m en ({x_malla[fila_max, col_max]:.0f}, "
          f"{y_malla[fila_max, col_max]:.0f})")
    print(f"Punto más bajo: {z_malla.min():.1f} m en ({x_malla[fila_min, col_min]:.0f}, "
          f"{y_malla[fila_min, col_min]:.0f})")
    print(f"Desnivel total: {z_malla.max() - z_malla.min():.1f} m")

    graficar_curvas_de_nivel(x_malla, y_malla, z_malla)
    plt.show()


main()
````

**Explicación**

Las curvas de nivel son la forma **plana** de representar una superficie, y en muchos casos se leen mejor que un gráfico 3D: no dependen del ángulo de visión y permiten medir distancias. Donde las curvas están **más juntas**, la pendiente es mayor. `set_aspect("equal")` es indispensable en un plano topográfico: un metro en el eje x debe medir lo mismo que un metro en el eje y.

## Ejercicio 12.18 — Perfil longitudinal de una tubería

Un ingeniero diseña una **tubería por gravedad** que lleva agua desde un tanque hasta una comunidad. En el **perfil longitudinal** se dibujan el terreno y la **línea de gradiente hidráulico** (LGH), que indica hasta qué altura subiría el agua en cualquier punto. La **presión** en cada punto, en metros de columna de agua, es la diferencia entre la LGH y el terreno. Si la LGH queda **por debajo** del terreno, el agua no llega: la presión es negativa.

```python
DISTANCIA_M = [0, 250, 500, 750, 1000, 1250, 1500, 1750, 2000]
TERRENO_M = [1520, 1498, 1470, 1508, 1455, 1430, 1446, 1412, 1395]
```

La LGH empieza en el nivel del agua del tanque (**1522 m**) y baja **linealmente** con una pérdida de **3.5 m cada 100 m**.

Escriba un programa en Python que use Matplotlib y NumPy, con las funciones:

A. `calcular_presiones(distancia: np.ndarray, terreno: np.ndarray, nivel_tanque: float, perdida_100m: float) -> tuple[np.ndarray, np.ndarray]`: retorna la LGH y la presión en cada punto.

B. `graficar_perfil(distancia: np.ndarray, terreno: np.ndarray, lgh: np.ndarray) -> None`: dibuja el terreno en café con el área bajo él rellena, la LGH en azul discontinuo, rellene en celeste la zona entre ambas **donde la presión es positiva** y en rojo **donde es negativa**, y anote la presión del punto más **bajo** del terreno.

C. `main() -> None`: imprime la presión en cada punto y los puntos con presión menor que **10 m** (mínimo para el servicio) o negativa, y muestra el gráfico.

**Ejemplo de ejecución**

Distancia (m)  Presión (m)
            0          2.0
          250         15.2
          500         34.5
          750        -12.2
         1000         32.0

         1500         23.5

Puntos con presión menor que 10 m: [  0 750]
</pre>

![Gráfico esperado del ejercicio 12.18](imgs/ejercicios_resueltos/ejercicio_12_18_1_1.png)

````python
#@title Solución
import matplotlib.pyplot as plt
import numpy as np

DISTANCIA_M = [0, 250, 500, 750, 1000, 1250, 1500, 1750, 2000]
TERRENO_M = [1520, 1498, 1470, 1508, 1455, 1430, 1446, 1412, 1395]
NIVEL_TANQUE_M = 1522
PERDIDA_POR_100M = 3.5
PRESION_MINIMA_M = 10


def calcular_presiones(distancia: np.ndarray, terreno: np.ndarray, nivel_tanque: float,
                       perdida_100m: float) -> tuple[np.ndarray, np.ndarray]:
    """Retorna la línea de gradiente hidráulico y la presión (m) en cada punto."""
    lgh = nivel_tanque - perdida_100m * distancia / 100
    return lgh, lgh - terreno


def graficar_perfil(distancia: np.ndarray, terreno: np.ndarray, lgh: np.ndarray) -> None:
    """Dibuja el perfil longitudinal del terreno y la línea de gradiente hidráulico."""
    fig, ax = plt.subplots(figsize=(10, 4.5))
    ax.fill_between(distancia, terreno, terreno.min() - 20, color="tan", alpha=0.6)
    ax.plot(distancia, terreno, color="saddlebrown", linewidth=2, label="Terreno")
    ax.plot(distancia, lgh, color="tab:blue", linestyle="--", linewidth=2, label="Línea de gradiente hidráulico")
    ax.fill_between(distancia, terreno, lgh, where=lgh >= terreno, interpolate=True,
                    color="skyblue", alpha=0.5, label="Presión positiva")
    ax.fill_between(distancia, terreno, lgh, where=lgh < terreno, interpolate=True,
                    color="red", alpha=0.5, label="Presión negativa")
    indice = np.argmin(terreno)
    ax.annotate(f"Presión: {lgh[indice] - terreno[indice]:.1f} m", xy=(distancia[indice], terreno[indice]),
                xytext=(distancia[indice] - 600, terreno[indice] - 10), arrowprops={"arrowstyle": "->"})
    ax.set_title("Perfil longitudinal de la conducción")
    ax.set_xlabel("Distancia desde el tanque (m)")
    ax.set_ylabel("Elevación (m s. n. m.)")
    ax.set_ylim(terreno.min() - 20, lgh.max() + 10)
    ax.legend(loc="lower left")


def main() -> None:
    distancia = np.array(DISTANCIA_M)
    terreno = np.array(TERRENO_M)
    lgh, presion = calcular_presiones(distancia, terreno, NIVEL_TANQUE_M, PERDIDA_POR_100M)

    print("Distancia (m)  Presión (m)")
    for punto, valor in zip(distancia, presion):
        print(f"{punto:13d}  {valor:11.1f}")
    criticos = distancia[presion < PRESION_MINIMA_M]
    print(f"\nPuntos con presión menor que {PRESION_MINIMA_M} m: {criticos}")

    graficar_perfil(distancia, terreno, lgh)
    plt.show()


main()
````

**Explicación**

`fill_between` puede rellenar la zona entre **dos** curvas (no solo entre una curva y el eje), y el parámetro `where` permite usar colores distintos según la condición. El gráfico muestra el problema de diseño de un vistazo: alrededor de los 750 m el terreno "sube" por encima de la LGH y la presión se vuelve **negativa**, así que el agua no pasaría; habría que aumentar el diámetro (menos pérdidas, LGH más alta) o cambiar el trazado. La presión baja al inicio es normal: a la salida del tanque la tubería todavía está casi al nivel del agua.

````python
#@title Solución
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


def variables_mas_relacionadas(correlacion: pd.DataFrame, objetivo: str) -> pd.Series:
    """Correlaciones con la variable objetivo, ordenadas por valor absoluto."""
    con_objetivo = correlacion[objetivo].drop(objetivo)
    return con_objetivo.reindex(con_objetivo.abs().sort_values(ascending=False).index)


def graficar_correlacion(correlacion: pd.DataFrame) -> None:
    """Dibuja la matriz de correlación como mapa de calor."""
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(correlacion, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
                square=True, ax=ax)
    ax.set_title("Correlación entre las variables del secado")
    fig.tight_layout()


def main() -> None:
    lotes = pd.read_csv("lotes_secado.csv")
    correlacion = lotes.corr()

    print("Relación con la humedad final:")
    print(variables_mas_relacionadas(correlacion, "humedad_final_pct").round(3))
    graficar_correlacion(correlacion)
    plt.show()


main()
````

**Explicación**

Seaborn está construido sobre Matplotlib: por eso `sns.heatmap` acepta el parámetro `ax` y el gráfico se sigue personalizando con los métodos de siempre (`ax.set_title`).

**Respuesta corta:** no necesariamente: **correlación no implica causalidad**. En estos lotes, la temperatura alta aparece junto con capas más delgadas (las variables explicativas están correlacionadas **entre sí**), así que parte del efecto podría deberse al espesor. Para separar los efectos se necesita un experimento diseñado, en el que se varíe una variable a la vez o se usen combinaciones balanceadas.

````python
#@title Solución
import matplotlib.pyplot as plt
import pandas as pd

POTENCIA_NOMINAL_MW = 5
PR_GARANTIZADO = 78


def calcular_indicadores(datos: pd.DataFrame) -> pd.DataFrame:
    """Agrega el índice de rendimiento diario (%)."""
    datos["pr"] = (datos["energia_mwh"] / (POTENCIA_NOMINAL_MW * datos["irradiacion_kwh_m2"]) * 100).round(1)
    return datos


def graficar_tablero(datos: pd.DataFrame) -> plt.Figure:
    """Crea el tablero de 2 x 2 gráficos de la planta solar."""
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))

    ax1.bar(datos["dia"], datos["energia_mwh"], color="gold", edgecolor="darkorange")
    ax1.axhline(datos["energia_mwh"].mean(), color="black", linestyle="--", label="Promedio")
    ax1.set_title("Energía generada por día")
    ax1.set_xlabel("Día de junio")
    ax1.set_ylabel("Energía (MWh)")
    ax1.legend()

    ax2.plot(datos["dia"], datos["pr"], marker="o", color="tab:green")
    ax2.axhline(PR_GARANTIZADO, color="red", linestyle="--", label=f"Garantizado ({PR_GARANTIZADO} %)")
    ax2.set_title("Índice de rendimiento (PR)")
    ax2.set_xlabel("Día de junio")
    ax2.set_ylabel("PR (%)")
    ax2.legend()

    ax3.hist(datos["irradiacion_kwh_m2"], bins=6, color="tab:orange", edgecolor="black")
    ax3.set_title("Distribución de la irradiación diaria")
    ax3.set_xlabel("Irradiación (kWh/m²)")
    ax3.set_ylabel("Cantidad de días")

    ax4.scatter(datos["temperatura_modulo_c"], datos["pr"], color="tab:red")
    ax4.set_title("PR vs. temperatura de los módulos")
    ax4.set_xlabel("Temperatura de los módulos (°C)")
    ax4.set_ylabel("PR (%)")

    for ax in (ax1, ax2, ax3, ax4):
        ax.grid(alpha=0.3)
    fig.suptitle("Planta solar fotovoltaica de 5 MW - junio 2026", fontsize=14)
    fig.tight_layout()
    return fig


def main() -> None:
    datos = calcular_indicadores(pd.read_csv("planta_solar_junio.csv"))
    bajo_garantia = datos.loc[datos["pr"] < PR_GARANTIZADO, "dia"]

    print(f"Energía total: {datos['energia_mwh'].sum():.1f} MWh")
    print(f"PR promedio: {datos['pr'].mean():.1f} %")
    print(f"Días con PR bajo el garantizado: {list(bajo_garantia)}")

    fig = graficar_tablero(datos)
    fig.savefig("tablero_junio.png", dpi=150)
    print("Tablero guardado en tablero_junio.png")
    plt.show()


main()
````

**Explicación**

`plt.subplots(2, 2)` retorna una matriz de 2 × 2 ejes, que se desempaqueta respetando su forma: `((ax1, ax2), (ax3, ax4))`. Retornar la figura desde la función permite que `main()` decida qué hacer con ella (guardarla, mostrarla o ambas). El gráfico de dispersión revela un efecto físico conocido: los módulos pierden eficiencia cuando se calientan, por eso el PR tiende a bajar en los días más calurosos.
